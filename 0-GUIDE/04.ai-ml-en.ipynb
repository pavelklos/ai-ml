## 01. Introduction to AI: From Theory to Practice

Introduction to machine learning tasks and optimization problems. Learn gradient descent algorithm and Python libraries for optimization.

````python
import numpy as np
from scipy.optimize import minimize

# Basic optimization example
def objective_function(x):
    return x[0]**2 + x[1]**2

# Gradient descent implementation
def gradient_descent(x_start, learning_rate=0.1, iterations=100):
    x = x_start
    for i in range(iterations):
        gradient = 2 * x  # derivative of x^2
        x = x - learning_rate * gradient
    return x

result = gradient_descent(np.array([5.0, 3.0]))
print(f"Optimized result: {result}")
````

## 02. Linear Regression Model: Teaching Computers to Predict

Learn linear regression theory, least squares method, and implementation using scikit-learn for single and multiple variable regression.

````python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np

# Generate sample data
X = np.random.randn(100, 2)
y = 3*X[:, 0] + 2*X[:, 1] + np.random.randn(100)*0.1

# Train linear regression model
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions
predictions = model.predict(X_test)
print(f"Model coefficients: {model.coef_}")
print(f"Model intercept: {model.intercept_}")
````

## 03. Classification: Teaching Computers to Recognize

Master classification tasks using logistic regression and k-nearest neighbors. Learn accuracy metrics and handling imbalanced datasets.

````python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# Sample classification data
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=1000, n_features=4, n_classes=2)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Logistic Regression
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)
log_pred = log_reg.predict(X_test)

# K-Nearest Neighbors
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)

print(f"Logistic Regression Accuracy: {accuracy_score(y_test, log_pred)}")
print(f"KNN Accuracy: {accuracy_score(y_test, knn_pred)}")
````

## 06. Data Transformers: Creating Effective Models

Master data transformers for categorical encoding, polynomials, and numerical transformations. Create custom transformers.

````python
from sklearn.preprocessing import LabelEncoder, PolynomialFeatures
from sklearn.base import BaseEstimator, TransformerMixin

# Custom transformer example
class CustomScaler(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.mean_ = None
        self.std_ = None
    
    def fit(self, X, y=None):
        self.mean_ = np.mean(X, axis=0)
        self.std_ = np.std(X, axis=0)
        return self
    
    def transform(self, X):
        return (X - self.mean_) / self.std_

# Polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X[:, :2])  # Use first 2 features
print(f"Original features: {X.shape[1]}, Polynomial features: {X_poly.shape[1]}")
````

## 07. Data Quality: Foundation of Successful AI

Handle missing values, outliers, and data scaling. Learn identification and correction methods for data problems.

````python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler

# Create data with missing values
data = pd.DataFrame({
    'feature1': [1, 2, np.nan, 4, 5],
    'feature2': [2, np.nan, 3, 4, 100],  # 100 is an outlier
    'target': [1, 0, 1, 0, 1]
})

# Handle missing values
imputer = SimpleImputer(strategy='median')
data[['feature1', 'feature2']] = imputer.fit_transform(data[['feature1', 'feature2']])

# Robust scaling (handles outliers better)
scaler = RobustScaler()
data[['feature1', 'feature2']] = scaler.fit_transform(data[['feature1', 'feature2']])

print("Data after preprocessing:")
print(data)
````

## 08. Cross-Validation and Model Selection Art

Master cross-validation importance for model accuracy and hyperparameter tuning. Learn various cross-validation techniques.

````python
from sklearn.model_selection import cross_val_score, GridSearchCV, KFold

# K-Fold cross-validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(LogisticRegression(), X, y, cv=kf)

print(f"Cross-validation scores: {cv_scores}")
print(f"Average CV score: {cv_scores.mean():.3f} (+/- {cv_scores.std() * 2:.3f})")

# Grid search for hyperparameter tuning
param_grid = {'C': [0.1, 1, 10], 'penalty': ['l1', 'l2']}
grid_search = GridSearchCV(LogisticRegression(solver='liblinear'), param_grid, cv=5)
grid_search.fit(X_train, y_train)

print(f"Best parameters: {grid_search.best_params_}")
print(f"Best cross-validation score: {grid_search.best_score_:.3f}")
````

## 09. Ensemble Models: Power of Diversity

Learn ensemble theory including bagging, boosting, and stacking. Master Random Forest, Gradient Boosting, XGBoost, and LightGBM.

````python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.ensemble import VotingClassifier

# Random Forest (Bagging)
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

# Gradient Boosting
gb = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb.fit(X_train, y_train)
gb_pred = gb.predict(X_test)

# Voting Classifier (Ensemble)
voting_clf = VotingClassifier([
    ('rf', rf), ('gb', gb), ('lr', LogisticRegression())
], voting='hard')

voting_clf.fit(X_train, y_train)
voting_pred = voting_clf.predict(X_test)

print(f"Random Forest Accuracy: {accuracy_score(y_test, rf_pred):.3f}")
print(f"Gradient Boosting Accuracy: {accuracy_score(y_test, gb_pred):.3f}")
print(f"Voting Classifier Accuracy: {accuracy_score(y_test, voting_pred):.3f}")
````

## 10. Time Series: Predicting the Future

Master time series characteristics, prediction models, feature engineering, and specialized cross-validation techniques.

````python
import pandas as pd
from sklearn.metrics import mean_squared_error
from sklearn.ensemble import RandomForestRegressor

# Generate time series data
dates = pd.date_range('2020-01-01', periods=100, freq='D')
ts_data = pd.DataFrame({
    'date': dates,
    'value': np.cumsum(np.random.randn(100)) + np.sin(np.arange(100) * 0.1) * 5
})

# Feature engineering for time series
ts_data['day_of_week'] = ts_data['date'].dt.dayofweek
ts_data['month'] = ts_data['date'].dt.month
ts_data['lag_1'] = ts_data['value'].shift(1)
ts_data['rolling_mean'] = ts_data['value'].rolling(window=7).mean()

# Prepare data for modeling
ts_clean = ts_data.dropna()
X_ts = ts_clean[['day_of_week', 'month', 'lag_1', 'rolling_mean']]
y_ts = ts_clean['value']

# Time series split
split_point = int(len(X_ts) * 0.8)
X_ts_train, X_ts_test = X_ts[:split_point], X_ts[split_point:]
y_ts_train, y_ts_test = y_ts[:split_point], y_ts[split_point:]

# Train model
ts_model = RandomForestRegressor(n_estimators=50)
ts_model.fit(X_ts_train, y_ts_train)
ts_pred = ts_model.predict(X_ts_test)

print(f"Time Series RMSE: {np.sqrt(mean_squared_error(y_ts_test, ts_pred)):.3f}")
````

## 13. ChatGPT and Language Models: Conversing with AI

Master large language model theory, LangChain library, OpenAI API interaction, and Retrieval-Augmented Generation (RAG).

````python
# Note: This example requires openai library and API key
# pip install openai langchain

# Simulated example (replace with actual API calls)
class MockLLM:
    def generate(self, prompt):
        return f"Mock response to: {prompt[:50]}..."

# Basic LLM interaction pattern
def query_llm(prompt, context=None):
    if context:
        enhanced_prompt = f"Context: {context}\n\nQuestion: {prompt}"
    else:
        enhanced_prompt = prompt
    
    # Mock LLM call
    llm = MockLLM()
    response = llm.generate(enhanced_prompt)
    return response

# RAG pattern example
def rag_query(question, knowledge_base):
    # 1. Retrieve relevant documents
    relevant_docs = [doc for doc in knowledge_base if any(word in doc.lower() for word in question.lower().split())]
    
    # 2. Create context from retrieved documents
    context = " ".join(relevant_docs[:3])
    
    # 3. Generate answer with context
    answer = query_llm(question, context)
    return answer

# Example usage
knowledge_base = [
    "Machine learning is a subset of artificial intelligence",
    "Neural networks are inspired by biological neurons",
    "Deep learning uses multiple layers of neural networks"
]

question = "What is machine learning?"
answer = rag_query(question, knowledge_base)
print(f"Question: {question}")
print(f"Answer: {answer}")
````
