## 01. Úvod do AI: Od teorie k praxi

Úvod do úloh strojového učení a optimalizačních problémů. Naučte se algoritmus gradientního sestupu a Python knihovny pro optimalizaci.


````python
import numpy as np
from scipy.optimize import minimize

# Základní příklad optimalizace
def ucilova_funkce(x):
    return x[0]**2 + x[1]**2

# Implementace gradientního sestupu
def gradientni_sestup(x_start, learning_rate=0.1, iterace=100):
    x = x_start
    for i in range(iterace):
        gradient = 2 * x  # derivace x^2
        x = x - learning_rate * gradient
    return x

vysledek = gradientni_sestup(np.array([5.0, 3.0]))
print(f"Optimalizovaný výsledek: {vysledek}")
````

## 02. Model lineární regrese: Naučte počítač předpovídat

Naučte se teorii lineární regrese, metodu nejmenších čtverců a implementaci pomocí scikit-learn pro jednoduchou i vícenásobnou regresi.

````python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import numpy as np

# Generování vzorových dat
X = np.random.randn(100, 2)
y = 3*X[:, 0] + 2*X[:, 1] + np.random.randn(100)*0.1

# Trénování modelu lineární regrese
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = LinearRegression()
model.fit(X_train, y_train)

# Vytvoření predikce
predikce = model.predict(X_test)
print(f"Koeficienty modelu: {model.coef_}")
print(f"Úsek modelu: {model.intercept_}")
````

## 03. Klasifikace: Naučte počítač rozpoznávat

Osvojte si klasifikační úlohy pomocí logistické regrese a k-nearest neighbors. Naučte se metriky přesnosti a řešení nevyvážených datových souborů.

````python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# Vzorová klasifikační data
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=1000, n_features=4, n_classes=2)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Logistická regrese
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)
log_pred = log_reg.predict(X_test)

# K-nejbližších sousedů
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)

print(f"Přesnost logistické regrese: {accuracy_score(y_test, log_pred)}")
print(f"Přesnost KNN: {accuracy_score(y_test, knn_pred)}")
````

## 06. Transformátory dat: Tvorba účinných modelů

Osvojte si datové transformátory pro kódování kategorických proměnných, polynomy a numerické transformace. Vytvořte vlastní transformátory.

````python
from sklearn.preprocessing import LabelEncoder, PolynomialFeatures
from sklearn.base import BaseEstimator, TransformerMixin

# Příklad vlastního transformátoru
class VlastniSkaler(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.prumer_ = None
        self.smerodatna_odchylka_ = None
    
    def fit(self, X, y=None):
        self.prumer_ = np.mean(X, axis=0)
        self.smerodatna_odchylka_ = np.std(X, axis=0)
        return self
    
    def transform(self, X):
        return (X - self.prumer_) / self.smerodatna_odchylka_

# Polynomiální vlastnosti
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X[:, :2])  # Použití prvních 2 vlastností
print(f"Původní vlastnosti: {X.shape[1]}, Polynomiální vlastnosti: {X_poly.shape[1]}")
````

## 08. Cross-validace a umění výběru modelu

Osvojte si význam cross-validace pro přesnost modelu a ladění hyperparametrů. Naučte se různé techniky cross-validace.

````python
from sklearn.model_selection import cross_val_score, GridSearchCV, KFold

# K-Fold cross-validace
kf = KFold(n_splits=5, shuffle=True, random_state=42)
cv_skore = cross_val_score(LogisticRegression(), X, y, cv=kf)

print(f"Cross-validační skóre: {cv_skore}")
print(f"Průměrné CV skóre: {cv_skore.mean():.3f} (+/- {cv_skore.std() * 2:.3f})")

# Grid search pro ladění hyperparametrů
param_grid = {'C': [0.1, 1, 10], 'penalty': ['l1', 'l2']}
grid_search = GridSearchCV(LogisticRegression(solver='liblinear'), param_grid, cv=5)
grid_search.fit(X_train, y_train)

print(f"Nejlepší parametry: {grid_search.best_params_}")
print(f"Nejlepší cross-validační skóre: {grid_search.best_score_:.3f}")
````

## 09. Ensemble modely: Síla diversity

Naučte se teorii ensemble včetně baggingu, boostingu a stackingu. Osvojte si Random Forest, Gradient Boosting, XGBoost a LightGBM.

## 10. Časové řady: Predikce budoucnosti

Osvojte si charakteristiky časových řad, predikční modely, feature engineering a specializované techniky cross-validace.

````python
import pandas as pd
from sklearn.metrics import mean_squared_error
from sklearn.ensemble import RandomForestRegressor

# Generování dat časové řady
datumy = pd.date_range('2020-01-01', periods=100, freq='D')
ts_data = pd.DataFrame({
    'datum': datumy,
    'hodnota': np.cumsum(np.random.randn(100)) + np.sin(np.arange(100) * 0.1) * 5
})

# Feature engineering pro časové řady
ts_data['den_v_tydnu'] = ts_data['datum'].dt.dayofweek
ts_data['mesic'] = ts_data['datum'].dt.month
ts_data['zpozdeni_1'] = ts_data['hodnota'].shift(1)
ts_data['klouzavy_prumer'] = ts_data['hodnota'].rolling(window=7).mean()

# Příprava dat pro modelování
ts_cista = ts_data.dropna()
X_ts = ts_cista[['den_v_tydnu', 'mesic', 'zpozdeni_1', 'klouzavy_prumer']]
y_ts = ts_cista['hodnota']

# Rozdělení časové řady
bod_rozdeleni = int(len(X_ts) * 0.8)
X_ts_train, X_ts_test = X_ts[:bod_rozdeleni], X_ts[bod_rozdeleni:]
y_ts_train, y_ts_test = y_ts[:bod_rozdeleni], y_ts[bod_rozdeleni:]

# Trénování modelu
ts_model = RandomForestRegressor(n_estimators=50)
ts_model.fit(X_ts_train, y_ts_train)
ts_pred = ts_model.predict(X_ts_test)

print(f"RMSE časové řady: {np.sqrt(mean_squared_error(y_ts_test, ts_pred)):.3f}")
````

## 13. ChatGPT a jazykové modely: Rozhovor s AI

Osvojte si teorii velkých jazykových modelů, knihovnu LangChain, interakci s OpenAI API a Retrieval-Augmented Generation (RAG).

````python
# Poznámka: Tento příklad vyžaduje knihovnu openai a API klíč
# pip install openai langchain

# Simulovaný příklad (nahraďte skutečnými API voláními)
class MockLLM:
    def generuj(self, prompt):
        return f"Mock odpověď na: {prompt[:50]}..."

# Základní vzor interakce s LLM
def dotaz_llm(prompt, kontext=None):
    if kontext:
        rozsireny_prompt = f"Kontext: {kontext}\n\nOtázka: {prompt}"
    else:
        rozsireny_prompt = prompt
    
    # Mock LLM volání
    llm = MockLLM()
    odpoved = llm.generuj(rozsireny_prompt)
    return odpoved

# Příklad RAG vzoru
def rag_dotaz(otazka, znalostni_baze):
    # 1. Získání relevantních dokumentů
    relevantni_docs = [doc for doc in znalostni_baze if any(slovo in doc.lower() for slovo in otazka.lower().split())]
    
    # 2. Vytvoření kontextu z získaných dokumentů
    kontext = " ".join(relevantni_docs[:3])
    
    # 3. Generování odpovědi s kontextem
    odpoved = dotaz_llm(otazka, kontext)
    return odpoved

# Příklad použití
znalostni_baze = [
    "Strojové učení je podmnožina umělé inteligence",
    "Neuronové sítě jsou inspirovány biologickými neurony",
    "Hluboké učení používá více vrstev neuronových sítí"
]

otazka = "Co je strojové učení?"
odpoved = rag_dotaz(otazka, znalostni_baze)
print(f"Otázka: {otazka}")
print(f"Odpověď: {odpoved}")
````
