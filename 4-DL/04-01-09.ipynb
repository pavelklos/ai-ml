## 4. PyTorch Computational Graph and Autograd

PyTorch's Autograd is a powerful automatic differentiation system that powers neural network training. It keeps track of operations performed on tensors and allows computing gradients automatically.

```python
# Create tensors with gradients
x = torch.tensor(2.0, requires_grad=True)  # Set requires_grad=True to track operations
y = torch.tensor(3.0, requires_grad=True)

# Perform operations
z = x**2 + 2*x*y + y**2  # z = x^2 + 2xy + y^2 = (x+y)^2

print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")

# Check the computational graph
print("\nRequires gradient:")
print(f"x.requires_grad: {x.requires_grad}")
print(f"y.requires_grad: {y.requires_grad}")
print(f"z.requires_grad: {z.requires_grad}")

# Display the gradient function (the function that will compute gradients)
print(f"\nz.grad_fn: {z.grad_fn}")
```

```python
# Compute gradients
z.backward()  # Compute gradients of z with respect to x and y

# Print gradients
print(f"Gradient of z with respect to x (dz/dx): {x.grad}")
print(f"Gradient of z with respect to y (dz/dy): {y.grad}")

# Let's verify these gradients analytically
# For z = x^2 + 2xy + y^2 = (x+y)^2
# dz/dx = 2x + 2y = 2(x+y) = 2(2+3) = 10
# dz/dy = 2y + 2x = 2(y+x) = 2(3+2) = 10
```

```python
# Example with vector operations
w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
v = 2 * w**2
print(f"w: {w}")
print(f"v: {v}")

# If v is a vector, we need to provide a gradient argument to backward()
# or compute a scalar (like a sum) before calling backward()

# Approach 1: Provide a gradient argument
external_grad = torch.tensor([1.0, 1.0, 1.0])
v.backward(gradient=external_grad)
print(f"Gradients of w (approach 1): {w.grad}")

# Reset gradients
w.grad.zero_()

# Approach 2: Compute scalar output before calling backward()
s = v.sum()
s.backward()
print(f"Gradients of w (approach 2): {w.grad}")
```

```python
# Stopping gradient tracking
a = torch.tensor([2.0, 3.0], requires_grad=True)
print("a requires gradient:", a.requires_grad)

# Temporary disable gradient tracking
with torch.no_grad():
    b = a * 2
    print("b requires gradient:", b.requires_grad)

# An alternative way to disable gradient tracking
c = a * 2
d = c.detach()  # Creates a new tensor detached from the computation graph
print("c requires gradient:", c.requires_grad)
print("d requires gradient:", d.requires_grad)

# Setting requires_grad flag directly
e = a.clone()
e.requires_grad_(False)  # Note the underscore suffix for in-place operation
print("e requires gradient:", e.requires_grad)
```

## 7. Creating a Simple Neural Network

Let's build a simple neural network in PyTorch using the `nn.Module` class.

```python
class SimpleNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        # Define layers
        self.layer1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(hidden_size, output_size)
        
    def forward(self, x):
        # Define forward pass
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        return x
    
# Create model instance
input_size = 10
hidden_size = 20
output_size = 2
model = SimpleNN(input_size, hidden_size, output_size)

# Print model architecture
print(model)

# Generate random input and forward
sample_input = torch.randn(5, input_size)  # 5 samples, 10 features each
output = model(sample_input)
print("\nInput shape:", sample_input.shape)
print("Output shape:", output.shape)
```

```python
# A more complex model with multiple layers
class MultiLayerNN(nn.Module):
    def __init__(self, input_size, hidden_sizes, output_size, dropout_prob=0.2):
        super(MultiLayerNN, self).__init__()
        
        # Create a list to hold all layers
        layers = []
        
        # Input layer to first hidden layer
        layers.append(nn.Linear(input_size, hidden_sizes[0]))
        layers.append(nn.ReLU())
        layers.append(nn.BatchNorm1d(hidden_sizes[0]))
        layers.append(nn.Dropout(dropout_prob))
        
        # Additional hidden layers
        for i in range(len(hidden_sizes) - 1):
            layers.append(nn.Linear(hidden_sizes[i], hidden_sizes[i+1]))
            layers.append(nn.ReLU())
            layers.append(nn.BatchNorm1d(hidden_sizes[i+1]))
            layers.append(nn.Dropout(dropout_prob))
        
        # Output layer
        layers.append(nn.Linear(hidden_sizes[-1], output_size))
        
        # Combine all layers into a sequential model
        self.model = nn.Sequential(*layers)
    
    def forward(self, x):
        return self.model(x)

# Create a more complex model
model = MultiLayerNN(
    input_size=20,
    hidden_sizes=[64, 32, 16],
    output_size=5,
    dropout_prob=0.3
)

print(model)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nTotal parameters: {total_params}")
print(f"Trainable parameters: {trainable_params}")

# Forward pass
x = torch.randn(10, 20)  # 10 samples, 20 features
output = model(x)
print(f"\nOutput shape: {output.shape}")
```

## 8. Using PyTorch Dataset and DataLoader

PyTorch provides elegant abstractions for handling data through the `Dataset` and `DataLoader` classes. These help with loading, batching, and preprocessing data efficiently.

```python
# Create a simple custom dataset
class SimpleDataset(Dataset):
    def __init__(self, num_samples=1000, feature_dim=10):
        # Generate random data for demonstration
        self.features = torch.randn(num_samples, feature_dim)
        # Generate random binary labels (0 or 1)
        self.labels = torch.randint(0, 2, (num_samples,))
        
    def __len__(self):
        return len(self.features)
    
    def __getitem__(self, idx):
        return self.features[idx], self.labels[idx]
    
# Create dataset instance
dataset = SimpleDataset(num_samples=1000, feature_dim=5)
print(f"Dataset size: {len(dataset)}")

# Access a single data point
features, label = dataset[0]
print(f"Single sample features shape: {features.shape}")
print(f"Single sample label: {label}")
```

```python
# Create DataLoaders for efficient data handling
batch_size = 32
train_loader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=True,  # Shuffling is important for training
    num_workers=0   # Number of subprocesses for data loading
)

# Iterate through the DataLoader
for batch_idx, (data, targets) in enumerate(train_loader):
    print(f"Batch {batch_idx}:")
    print(f"  Data shape: {data.shape}")
    print(f"  Targets shape: {targets.shape}")
    
    # Just show the first few batches
    if batch_idx >= 2:
        break
```

```python
# Advanced DataLoader features

# Create train/test split manually
total_samples = len(dataset)
train_size = int(0.8 * total_samples)  # 80% for training
test_size = total_samples - train_size  # 20% for testing

train_dataset, test_dataset = torch.utils.data.random_split(dataset, [train_size, test_size])

print(f"Train dataset size: {len(train_dataset)}")
print(f"Test dataset size: {len(test_dataset)}")

# Create data loaders for each set
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)  # No need to shuffle test data

print(f"\nNumber of batches in train_loader: {len(train_loader)}")
print(f"Number of batches in test_loader: {len(test_loader)}")
```

## 9. Training a Basic Model in PyTorch

Now, let's put everything together and train a simple neural network model.

```python
# Set random seed for reproducibility
torch.manual_seed(42)

# Create a synthetic dataset for binary classification
n_samples = 1000
n_features = 10

# Generate random data
X = torch.randn(n_samples, n_features)
# Generate target: if sum of features > 0, class = 1, else class = 0
y = (X.sum(dim=1) > 0).float()

# Split into train and test sets
train_size = int(0.8 * n_samples)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

print(f"Training samples: {X_train.shape[0]}")
print(f"Testing samples: {X_test.shape[0]}")

# Create DataLoaders
train_dataset = torch.utils.data.TensorDataset(X_train, y_train)
test_dataset = torch.utils.data.TensorDataset(X_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

```python
# Define a binary classification model
class BinaryClassifier(nn.Module):
    def __init__(self, input_dim):
        super(BinaryClassifier, self).__init__()
        self.layer1 = nn.Linear(input_dim, 16)
        self.layer2 = nn.Linear(16, 8)
        self.layer3 = nn.Linear(8, 1)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()  # For binary classification
        self.dropout = nn.Dropout(0.2)
        
    def forward(self, x):
        x = self.relu(self.layer1(x))
        x = self.dropout(x)
        x = self.relu(self.layer2(x))
        x = self.dropout(x)
        x = self.layer3(x)
        return self.sigmoid(x)  # Output probability between 0 and 1

# Initialize the model
model = BinaryClassifier(n_features)
print(model)

# Define loss function and optimizer
criterion = nn.BCELoss()  # Binary Cross Entropy
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
# Training function
def train(model, train_loader, criterion, optimizer, epochs):
    model.train()  # Set model to training mode
    
    train_losses = []
    for epoch in range(epochs):
        running_loss = 0.0
        correct = 0
        total = 0
        
        for inputs, targets in train_loader:
            # Zero the parameter gradients
            optimizer.zero_grad()
            
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs.squeeze(), targets)
            
            # Backward pass and optimize
            loss.backward()
            optimizer.step()
            
            # Calculate statistics
            running_loss += loss.item() * inputs.size(0)
            predicted = (outputs.squeeze() > 0.5).float()
            total += targets.size(0)
            correct += (predicted == targets).sum().item()
        
        # Calculate epoch statistics
        epoch_loss = running_loss / len(train_loader.dataset)
        epoch_acc = correct / total
        train_losses.append(epoch_loss)
        
        print(f"Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.4f}, Accuracy: {epoch_acc:.4f}")
    
    return train_losses

# Evaluation function
def evaluate(model, test_loader, criterion):
    model.eval()  # Set model to evaluation mode
    
    test_loss = 0.0
    correct = 0
    total = 0
    
    # Disable gradient calculation for evaluation
    with torch.no_grad():
        for inputs, targets in test_loader:
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs.squeeze(), targets)
            
            # Calculate statistics
            test_loss += loss.item() * inputs.size(0)
            predicted = (outputs.squeeze() > 0.5).float()
            total += targets.size(0)
            correct += (predicted == targets).sum().item()
    
    # Calculate overall statistics
    avg_loss = test_loss / len(test_loader.dataset)
    accuracy = correct / total
    
    print(f"Test Loss: {avg_loss:.4f}, Accuracy: {accuracy:.4f}")
    return avg_loss, accuracy
```

```python
# Train the model
epochs = 10
train_losses = train(model, train_loader, criterion, optimizer, epochs)

# Evaluate the model on test data
test_loss, test_acc = evaluate(model, test_loader, criterion)

# Plot the training loss
plt.figure(figsize=(10, 5))
plt.plot(range(1, epochs + 1), train_losses, marker='o')
plt.title('Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.show()
```

```python
# Let's make predictions on new data
# Generate some new random data
new_data = torch.randn(5, n_features)

# Make predictions
model.eval()
with torch.no_grad():
    predictions = model(new_data)

# Display results
print("New data samples:")
print(new_data)
print("\nPredicted probabilities:")
print(predictions.squeeze())
print("\nPredicted classes:")
print((predictions.squeeze() > 0.5).float())

# The "true" labels for our synthetic dataset would be:
true_labels = (new_data.sum(dim=1) > 0).float()
print("\nTrue labels based on our data generation rule:")
print(true_labels)
```
