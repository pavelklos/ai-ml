## 2. Understanding TensorFlow and Keras

TensorFlow is an open-source machine learning framework developed by Google, while Keras is a high-level neural networks API that can run on top of TensorFlow. As of TensorFlow 2.0, Keras has been integrated as TensorFlow's official high-level API.

### Relationship between TensorFlow and Keras

- **TensorFlow**: A comprehensive machine learning platform with lower-level operations for tensor manipulation and computational graph definitions
- **Keras**: A user-friendly API focused on simplifying the model building and training process

### API Levels in TensorFlow

1. **Low-level API**: Core TensorFlow operations (tensors, variables, gradients)
2. **Mid-level API**: Layers, metrics, optimizers, callbacks
3. **High-level API**: Keras (model building, training, evaluation)

```python
# Demonstrate the relationship between TensorFlow and Keras
print("TensorFlow and Keras are now integrated:")
print(f"tf.keras version: {tf.keras.__version__}")

# Show different API level examples
print("\nAPI Levels Example:")
print("1. Low-level (Core TensorFlow):")
x = tf.constant([[1, 2], [3, 4]])
y = tf.constant([[5, 6], [7, 8]])
print(f"Matrix multiplication: \n{tf.matmul(x, y)}")

print("\n2. Mid-level (Layers):")
layer = tf.keras.layers.Dense(10, activation='relu')
print(f"Dense layer: {layer}")

print("\n3. High-level (Keras Models):")
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(1)
])
print(f"Sequential model: {model}")
```

## 3. TensorFlow Basics

TensorFlow's core concept is the tensor - a multi-dimensional array. Understanding how to work with tensors is fundamental to using TensorFlow effectively.

### Core TensorFlow Concepts

1. **Tensors**: Multi-dimensional arrays with a uniform type
2. **Variables**: Special tensors that maintain state across TensorFlow operations
3. **Operations**: Mathematical computations performed on tensors
4. **Computational Graph**: A series of TensorFlow operations arranged in a graph
5. **Eager Execution**: Evaluates operations immediately (default in TF 2.x)

```python
# Creating tensors
scalar = tf.constant(5)  # 0D tensor (scalar)
vector = tf.constant([1, 2, 3])  # 1D tensor (vector)
matrix = tf.constant([[1, 2], [3, 4]])  # 2D tensor (matrix)
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])  # 3D tensor

print("Tensors of different shapes:")
print(f"Scalar: {scalar}, shape: {scalar.shape}")
print(f"Vector: {vector}, shape: {vector.shape}")
print(f"Matrix: {matrix}, shape: {matrix.shape}")
print(f"3D Tensor: \n{tensor_3d}, \nshape: {tensor_3d.shape}")

# Tensor operations
print("\nTensor Operations:")
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

print(f"Addition: \n{a + b}")
print(f"Multiplication (element-wise): \n{a * b}")
print(f"Matrix multiplication: \n{tf.matmul(a, b)}")

# TensorFlow Variables
print("\nTensorFlow Variables:")
variable = tf.Variable([1.0, 2.0, 3.0])
print(f"Variable: {variable}")

# Update variable
variable.assign([4.0, 5.0, 6.0])
print(f"Updated variable: {variable}")
```

```python
# Demonstrate shape transformations
print("Shape Transformations:")
t = tf.constant([[1, 2, 3], [4, 5, 6]])
print(f"Original tensor: \n{t}, shape: {t.shape}")

# Reshaping
reshaped = tf.reshape(t, [3, 2])
print(f"Reshaped tensor: \n{reshaped}, shape: {reshaped.shape}")

# Transposing
transposed = tf.transpose(t)
print(f"Transposed tensor: \n{transposed}, shape: {transposed.shape}")

# Expanding dimensions
expanded = tf.expand_dims(t, axis=0)  # Add batch dimension
print(f"Expanded tensor shape: {expanded.shape}")

# Slicing tensors
print("\nSlicing Operations:")
print(f"First row: {t[0]}")
print(f"Second column: {t[:, 1]}")
print(f"Submatrix: \n{t[0:1, 0:2]}")
```

## 5. Building Models with Keras

Keras offers two main APIs for building models:

1. **Sequential API**: Simple, linear stack of layers
2. **Functional API**: More flexible, allows for models with multiple inputs/outputs or shared layers

Let's explore both approaches to model building.

```python
# Building models with the Sequential API
print("Sequential API Example:")

# Create a simple sequential model for binary classification
sequential_model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(10,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Display model summary
sequential_model.summary()

# Compile the model
sequential_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Building models with the Functional API
print("Functional API Example:")

# Define inputs
inputs = tf.keras.Input(shape=(10,))

# Define the network
x = tf.keras.layers.Dense(64, activation='relu')(inputs)
x = tf.keras.layers.Dense(32, activation='relu')(x)
x = tf.keras.layers.Dense(16, activation='relu')(x)
outputs = tf.keras.layers.Dense(1, activation='sigmoid')(x)

# Create and compile the model
functional_model = tf.keras.Model(inputs=inputs, outputs=outputs)
functional_model.summary()

functional_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# More complex model with multiple inputs using the Functional API

# Define two inputs
input_a = tf.keras.Input(shape=(20,), name='input_a')
input_b = tf.keras.Input(shape=(10,), name='input_b')

# Process input_a
x_a = tf.keras.layers.Dense(32, activation='relu')(input_a)
x_a = tf.keras.layers.Dense(16, activation='relu')(x_a)

# Process input_b
x_b = tf.keras.layers.Dense(16, activation='relu')(input_b)

# Combine the processed inputs
concat = tf.keras.layers.Concatenate()([x_a, x_b])

# Add more layers to the combined result
x = tf.keras.layers.Dense(32, activation='relu')(concat)
x = tf.keras.layers.Dropout(0.2)(x)
output = tf.keras.layers.Dense(1, activation='sigmoid')(x)

# Create the model
multi_input_model = tf.keras.Model(inputs=[input_a, input_b], outputs=output)

# Display model summary
multi_input_model.summary()

# Visualize the model (this requires pydot and graphviz to be installed)
try:
    tf.keras.utils.plot_model(multi_input_model, show_shapes=True, to_file='multi_input_model.png')
    print("Model visualization saved to 'multi_input_model.png'")
except ImportError:
    print("Could not visualize model - pydot or graphviz not installed")
```

### Custom Layers and Models

Keras allows you to create custom layers and models by subclassing `tf.keras.layers.Layer` and `tf.keras.Model`.

```python
# Creating a custom layer
class CustomDense(tf.keras.layers.Layer):
    def __init__(self, units, activation=None):
        super(CustomDense, self).__init__()
        self.units = units
        self.activation = tf.keras.activations.get(activation)
        
    def build(self, input_shape):
        # Create trainable weights
        self.w = self.add_weight(
            shape=(input_shape[-1], self.units),
            initializer='random_normal',
            trainable=True,
        )
        self.b = self.add_weight(
            shape=(self.units,),
            initializer='zeros',
            trainable=True,
        )
        
    def call(self, inputs):
        # Define the computation
        output = tf.matmul(inputs, self.w) + self.b
        if self.activation is not None:
            output = self.activation(output)
        return output

# Creating a custom model
class CustomModel(tf.keras.Model):
    def __init__(self):
        super(CustomModel, self).__init__()
        self.dense1 = tf.keras.layers.Dense(64, activation='relu')
        self.dense2 = CustomDense(32, activation='relu')
        self.dense3 = tf.keras.layers.Dense(1, activation='sigmoid')
        
    def call(self, inputs, training=False):
        x = self.dense1(inputs)
        x = self.dense2(x)
        return self.dense3(x)

# Instantiate custom model
custom_model = CustomModel()

# Compile the model
custom_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Build the model (define input shape)
sample_input = tf.zeros((1, 10))
_ = custom_model(sample_input)

# Display model summary
custom_model.summary()
```

## 6. Training and Evaluation

Once a model is built and compiled, the next steps are training and evaluation. Keras provides simple methods for these tasks:

1. **Training**: `model.fit()` for batch training or `model.fit_generator()` for generator-based training
2. **Evaluation**: `model.evaluate()` to measure model performance on test data
3. **Prediction**: `model.predict()` to generate predictions on new data

Let's create a toy dataset and train our model.

```python
# Create a synthetic dataset for classification
np.random.seed(42)
x_train = np.random.random((1000, 10))
y_train = np.random.randint(0, 2, size=(1000, 1))

x_val = np.random.random((200, 10))
y_val = np.random.randint(0, 2, size=(200, 1))

x_test = np.random.random((200, 10))
y_test = np.random.randint(0, 2, size=(200, 1))

print(f"Training data shape: {x_train.shape}, {y_train.shape}")
print(f"Validation data shape: {x_val.shape}, {y_val.shape}")
print(f"Testing data shape: {x_test.shape}, {y_test.shape}")

# Train the sequential model we created earlier
history = sequential_model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=32,
    validation_data=(x_val, y_val),
    verbose=1
)
```

```python
# Visualize the training history
plt.figure(figsize=(12, 4))

# Plot training & validation loss values
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(loc='upper right')

# Plot training & validation accuracy values
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(loc='lower right')

plt.tight_layout()
plt.show()

# Evaluate the model on test data
test_results = sequential_model.evaluate(x_test, y_test, verbose=1)
print(f"Test loss: {test_results[0]:.4f}, Test accuracy: {test_results[1]:.4f}")

# Make predictions
predictions = sequential_model.predict(x_test)
print("First 5 predictions (probabilities):")
print(predictions[:5])

# Convert probabilities to binary predictions
binary_predictions = (predictions > 0.5).astype(int)
print("First 5 binary predictions:")
print(binary_predictions[:5])
```

### Using Callbacks

Callbacks in Keras allow you to perform actions during training, such as:
- Saving checkpoints
- Early stopping to prevent overfitting
- Adjusting learning rates
- Logging metrics to TensorBoard

```python
# Create a new model to demonstrate callbacks
callback_model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(10,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

callback_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Define callbacks
callbacks = [
    # Save the best model based on validation loss
    tf.keras.callbacks.ModelCheckpoint(
        filepath='best_model.h5',
        monitor='val_loss',
        save_best_only=True,
        verbose=1
    ),
    
    # Stop training when validation loss stops improving
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=3,
        verbose=1,
        restore_best_weights=True
    ),
    
    # Reduce learning rate when validation loss plateaus
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=2,
        verbose=1,
        min_lr=1e-6
    ),
    
    # Add TensorBoard logging
    tf.keras.callbacks.TensorBoard(
        log_dir='./logs',
        histogram_freq=1
    )
]

# Train with callbacks
history_with_callbacks = callback_model.fit(
    x_train, y_train,
    epochs=20,  # We set more epochs, but early stopping might trigger before reaching all of them
    batch_size=32,
    validation_data=(x_val, y_val),
    callbacks=callbacks,
    verbose=1
)
```

## 7. Saving and Loading Models

TensorFlow/Keras provides multiple ways to save and load models:

1. **HDF5 Format**: Save the entire model (architecture, weights, optimizer state)
2. **SavedModel Format**: TensorFlow's recommended format for saving models
3. **Model Weights**: Save only the model weights
4. **Model Architecture**: Save only the model architecture as JSON or YAML

```python
# Save and load models in different formats

# Create a directory for saving models
os.makedirs('saved_models', exist_ok=True)

# 1. HDF5 Format (Save entire model)
sequential_model.save('saved_models/full_model.h5')
print("Model saved in HDF5 format")

# Load the model from HDF5
loaded_h5_model = tf.keras.models.load_model('saved_models/full_model.h5')
print("Model loaded from HDF5 format")
print(f"Loaded model structure: {loaded_h5_model.layers}")

# 2. SavedModel Format (TensorFlow format)
sequential_model.save('saved_models/tf_savedmodel')
print("Model saved in SavedModel format")

# Load the model from SavedModel
loaded_savedmodel = tf.keras.models.load_model('saved_models/tf_savedmodel')
print("Model loaded from SavedModel format")

# 3. Save only the weights
sequential_model.save_weights('saved_models/model_weights')
print("Model weights saved")

# Create a new model with the same architecture
new_model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(10,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Load the weights into the new model
new_model.load_weights('saved_models/model_weights')
print("Weights loaded into new model")

# 4. Save only the model architecture (as JSON)
model_json = sequential_model.to_json()
with open('saved_models/model_architecture.json', 'w') as json_file:
    json_file.write(model_json)
print("Model architecture saved as JSON")

# Load the model architecture from JSON
with open('saved_models/model_architecture.json', 'r') as json_file:
    loaded_model_json = json_file.read()
    
json_model = tf.keras.models.model_from_json(loaded_model_json)
print("Model architecture loaded from JSON")
```

## 8. TensorFlow Datasets

TensorFlow provides built-in datasets and utilities for working with data efficiently:

1. **tf.data API**: Create data pipelines for efficient data loading and preprocessing
2. **TensorFlow Datasets (TFDS)**: Library of ready-to-use datasets
3. **Built-in datasets in Keras**: Popular datasets like MNIST, CIFAR-10, etc.

Let's explore how to work with these datasets.

```python
# Load the MNIST dataset using Keras
(x_mnist_train, y_mnist_train), (x_mnist_test, y_mnist_test) = tf.keras.datasets.mnist.load_data()

print(f"MNIST training data shape: {x_mnist_train.shape}")
print(f"MNIST training labels shape: {y_mnist_train.shape}")

# Normalize pixel values to be between 0 and 1
x_mnist_train = x_mnist_train.astype('float32') / 255.0
x_mnist_test = x_mnist_test.astype('float32') / 255.0

# Visualize some MNIST examples
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_mnist_train[i], cmap='gray')
    plt.title(f"Label: {y_mnist_train[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Using tf.data API to create efficient data pipelines

# Create a dataset from tensors
train_dataset = tf.data.Dataset.from_tensor_slices((x_mnist_train, y_mnist_train))

# Apply transformations to the dataset
train_dataset = train_dataset.shuffle(10000)  # Shuffle the data
train_dataset = train_dataset.batch(32)       # Create batches of 32 samples
train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)  # Prefetch for efficiency

# Create a test dataset
test_dataset = tf.data.Dataset.from_tensor_slices((x_mnist_test, y_mnist_test)).batch(32)

print("Dataset created with tf.data API:")
for images, labels in train_dataset.take(1):
    print(f"Batch shape: {images.shape}")
    print(f"Labels shape: {labels.shape}")

# Define a simple CNN model for MNIST
mnist_model = tf.keras.Sequential([
    tf.keras.layers.Reshape((28, 28, 1), input_shape=(28, 28)),
    tf.keras.layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(pool_size=(2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compile the model
mnist_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model using the dataset API
mnist_history = mnist_model.fit(
    train_dataset,
    epochs=5,
    validation_data=test_dataset
)
```

```python
# Data augmentation demonstration
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
])

# Apply data augmentation to an example image
plt.figure(figsize=(10, 5))
image = x_mnist_train[0]
plt.subplot(1, 6, 1)
plt.imshow(image, cmap='gray')
plt.title("Original")
plt.axis('off')

for i in range(5):
    augmented_image = data_augmentation(image[tf.newaxis, ..., tf.newaxis])
    plt.subplot(1, 6, i+2)
    plt.imshow(augmented_image[0, ..., 0], cmap='gray')
    plt.title(f"Augmented {i+1}")
    plt.axis('off')
    
plt.tight_layout()
plt.show()
```

## 9. Using Pre-trained Models

TensorFlow/Keras provides access to many pre-trained models that can be used for:
1. **Feature extraction**: Using the pre-trained model as a feature extractor
2. **Transfer learning**: Fine-tuning a pre-trained model on your specific task
3. **Model customization**: Modifying parts of pre-trained architectures

Let's explore how to use pre-trained models from Keras Applications.

```python
# Load a pre-trained model (MobileNetV2)
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,  # Don't include the classification head
    weights='imagenet'  # Pre-trained on ImageNet
)

# Freeze the base model layers to prevent them from being updated during training
base_model.trainable = False

print("Pre-trained MobileNetV2 model loaded:")
print(f"Number of layers: {len(base_model.layers)}")
print(f"Input shape: {base_model.input_shape}")
print(f"Output shape: {base_model.output_shape}")

# Define a new model that uses the pre-trained model as a feature extractor
inputs = tf.keras.Input(shape=(224, 224, 3))
x = base_model(inputs, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dense(256, activation='relu')(x)
x = tf.keras.layers.Dropout(0.2)(x)
outputs = tf.keras.layers.Dense(10, activation='softmax')(x)  # For a 10-class classification problem

transfer_model = tf.keras.Model(inputs, outputs)

# Compile the model
transfer_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display the model summary (showing the frozen base model and our added layers)
transfer_model.summary(line_length=100)
```

```python
# Fine-tuning approach
fine_tune_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# First, freeze all layers
fine_tune_model.trainable = False

# Now, unfreeze only the last few layers for fine-tuning
for layer in fine_tune_model.layers[-10:]:
    layer.trainable = True

# Count how many layers are trainable vs non-trainable
trainable_count = sum(1 for layer in fine_tune_model.layers if layer.trainable)
non_trainable_count = sum(1 for layer in fine_tune_model.layers if not layer.trainable)

print(f"Total layers: {len(fine_tune_model.layers)}")
print(f"Trainable layers: {trainable_count}")
print(f"Non-trainable layers: {non_trainable_count}")

# Create a new model for fine-tuning
inputs = tf.keras.Input(shape=(224, 224, 3))
x = fine_tune_model(inputs)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dense(256, activation='relu')(x)
x = tf.keras.layers.Dropout(0.2)(x)
outputs = tf.keras.layers.Dense(10, activation='softmax')(x)

fine_tuned_model = tf.keras.Model(inputs, outputs)

# Compile with a lower learning rate for fine-tuning
fine_tuned_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```
