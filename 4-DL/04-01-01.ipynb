## Deep Learning Foundations

### What is Deep Learning?

Deep learning is a subset of machine learning that uses neural networks with multiple layers (deep neural networks) to analyze various forms of data. Unlike traditional machine learning algorithms that require feature engineering, deep learning models can automatically discover the representations needed for feature detection or classification from raw data.

### Traditional ML vs. Deep Learning

| Traditional Machine Learning | Deep Learning |
|------------------------------|---------------|
| Often requires manual feature engineering | Automatically extracts features |
| Works well with structured data | Excels with unstructured data (images, text, audio) |
| Generally needs less data | Usually requires large amounts of data |
| Typically more interpretable | Often considered a "black box" |
| Less computationally intensive | Requires significant computational resources |

## Neural Network Components

### Building Blocks

The fundamental components of neural networks include:

1. **Neurons**: The basic computational units
2. **Weights**: Parameters that determine the strength of connections between neurons
3. **Biases**: Offset values that control neuron activation thresholds
4. **Layers**: Collections of neurons organized in a structured manner
   - Input layer: Receives the raw data
   - Hidden layers: Intermediate processing layers
   - Output layer: Produces the final prediction

### Mathematical Representation

For a single neuron, the output is calculated as:

$$y = f(\sum_{i=1}^{n} w_i x_i + b)$$

Where:
- $x_i$ are the inputs
- $w_i$ are the weights
- $b$ is the bias
- $f$ is the activation function
- $y$ is the output

Let's implement a single neuron and visualize its components:

```python
# Implementing a simple neuron
class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias
        
    def activate(self, inputs, activation_function):
        # Calculate weighted sum of inputs plus bias
        z = np.dot(self.weights, inputs) + self.bias
        # Apply activation function
        return activation_function(z), z
    
# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)

# Create a neuron with 3 inputs
weights = np.array([0.5, -0.4, 0.2])
bias = 0.1
my_neuron = Neuron(weights, bias)

# Sample input
inputs = np.array([1.0, 2.0, 0.5])

# Calculate outputs with different activation functions
sigmoid_output, sigmoid_z = my_neuron.activate(inputs, sigmoid)
relu_output, relu_z = my_neuron.activate(inputs, relu)
tanh_output, tanh_z = my_neuron.activate(inputs, tanh)

print(f"Inputs: {inputs}")
print(f"Weights: {weights}")
print(f"Bias: {bias}")
print(f"Weighted sum (z): {sigmoid_z}")
print(f"Sigmoid output: {sigmoid_output}")
print(f"ReLU output: {relu_output}")
print(f"Tanh output: {tanh_output}")

# Visualize the computation
fig, ax = plt.subplots(figsize=(8, 6))

# Set up the canvas
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Draw inputs
for i, val in enumerate(inputs):
    ax.text(1, 8-i, f"x{i+1} = {val}", fontsize=12)
    
# Draw weights
for i, w in enumerate(weights):
    ax.text(3, 8-i, f"w{i+1} = {w}", fontsize=12, color='blue')
    
# Draw bias
ax.text(3, 4, f"b = {bias}", fontsize=12, color='green')

# Draw weighted sum
ax.text(5, 6, f"z = Σ(w·x) + b = {sigmoid_z:.2f}", fontsize=12)

# Draw activation function outputs
ax.text(8, 8, f"sigmoid(z) = {sigmoid_output:.4f}", fontsize=12, color='red')
ax.text(8, 7, f"relu(z) = {relu_output:.4f}", fontsize=12, color='purple')
ax.text(8, 6, f"tanh(z) = {tanh_output:.4f}", fontsize=12, color='orange')

plt.title("Single Neuron Computation", fontsize=14)
plt.tight_layout()
plt.show()
```

### Neural Network Layers

A neural network consists of multiple layers of neurons:

1. **Input Layer**: Contains the raw input features
2. **Hidden Layers**: Intermediate layers that process the data
3. **Output Layer**: Produces the final prediction

The power of deep learning comes from using multiple hidden layers, which progressively extract higher-level features from the data.

Let's visualize a simple feed-forward neural network:

```python
# Visualize a simple neural network architecture
def draw_neural_net(ax, left, right, bottom, top, layer_sizes):
    '''
    Draw a neural network cartoon using matplotlib
    
    Args:
        ax: matplotlib.axes.Axes
        left: float, the left margin of the diagram
        right: float, the right margin of the diagram
        bottom: float, the bottom margin of the diagram
        top: float, the top margin of the diagram
        layer_sizes: list of int, list containing the number of neurons in each layer
    '''
    n_layers = len(layer_sizes)
    v_spacing = (top - bottom) / max(layer_sizes)
    h_spacing = (right - left) / n_layers
    
    # Nodes
    for n, layer_size in enumerate(layer_sizes):
        layer_top = v_spacing * (layer_size - 1) / 2. + (top + bottom) / 2.
        for m in range(layer_size):
            circle = plt.Circle((n * h_spacing + left, layer_top - m * v_spacing), 
                                v_spacing/4, 
                                color='w', 
                                ec='k', 
                                zorder=4)
            ax.add_artist(circle)
            if n == 0:
                ax.text(left - 0.3, layer_top - m * v_spacing, 
                        f"x{m+1}", 
                        fontsize=12)
            elif n == n_layers - 1:
                ax.text(n * h_spacing + left + 0.3, layer_top - m * v_spacing, 
                        f"y{m+1}", 
                        fontsize=12)
    
    # Edges
    for n, (layer_size_a, layer_size_b) in enumerate(zip(layer_sizes[:-1], layer_sizes[1:])):
        layer_top_a = v_spacing * (layer_size_a - 1) / 2. + (top + bottom) / 2.
        layer_top_b = v_spacing * (layer_size_b - 1) / 2. + (top + bottom) / 2.
        for m in range(layer_size_a):
            for o in range(layer_size_b):
                line = plt.Line2D([n * h_spacing + left, (n + 1) * h_spacing + left],
                                 [layer_top_a - m * v_spacing, layer_top_b - o * v_spacing], 
                                 c='k', 
                                 alpha=0.5)
                ax.add_artist(line)

# Define the architecture
fig, ax = plt.subplots(figsize=(12, 8))
draw_neural_net(ax, 0.1, 0.9, 0.1, 0.9, [3, 5, 4, 2])

# Add labels
plt.text(0.2, 0.95, 'Input\nLayer', ha='center', fontsize=14)
plt.text(0.4, 0.95, 'Hidden\nLayer 1', ha='center', fontsize=14)
plt.text(0.6, 0.95, 'Hidden\nLayer 2', ha='center', fontsize=14)
plt.text(0.8, 0.95, 'Output\nLayer', ha='center', fontsize=14)

plt.axis('off')
plt.title('Feed-Forward Neural Network Architecture', fontsize=16)
plt.tight_layout()
plt.show()
```

## Activation Functions

Activation functions introduce non-linearity into the network, enabling it to learn complex patterns. Common activation functions include:

1. **Sigmoid**: Maps values to the range (0,1), useful for binary classification
   $$\sigma(x) = \frac{1}{1 + e^{-x}}$$

2. **Tanh**: Maps values to the range (-1,1), zero-centered
   $$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

3. **ReLU (Rectified Linear Unit)**: Returns x if x > 0, else 0
   $$\text{ReLU}(x) = \max(0, x)$$

4. **Leaky ReLU**: Allows a small gradient when x < 0
   $$\text{LeakyReLU}(x) = \max(αx, x)$$, where α is a small constant

5. **Softmax**: Used for multi-class classification, converts values to probabilities
   $$\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$

Let's implement and visualize these activation functions:

```python
# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.maximum(alpha * x, x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))  # Subtract max for numerical stability
    return exp_x / exp_x.sum()

# Visualize activation functions
x = np.linspace(-5, 5, 1000)

fig, axs = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Common Activation Functions', fontsize=16)

# Sigmoid
axs[0, 0].plot(x, sigmoid(x), 'b-', linewidth=2)
axs[0, 0].set_title('Sigmoid')
axs[0, 0].set_xlabel('x')
axs[0, 0].set_ylabel('σ(x)')
axs[0, 0].grid(True)
axs[0, 0].text(-4, 0.8, r'$\sigma(x) = \frac{1}{1 + e^{-x}}$', fontsize=12)
axs[0, 0].axhline(y=0, color='k', linestyle='-', alpha=0.3)
axs[0, 0].axhline(y=1, color='k', linestyle='-', alpha=0.3)
axs[0, 0].axvline(x=0, color='k', linestyle='-', alpha=0.3)

# Tanh
axs[0, 1].plot(x, tanh(x), 'g-', linewidth=2)
axs[0, 1].set_title('Tanh')
axs[0, 1].set_xlabel('x')
axs[0, 1].set_ylabel('tanh(x)')
axs[0, 1].grid(True)
axs[0, 1].text(-4, 1.5, r'$tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$', fontsize=12)
axs[0, 1].axhline(y=0, color='k', linestyle='-', alpha=0.3)
axs[0, 1].axhline(y=1, color='k', linestyle='-', alpha=0.3)
axs[0, 1].axhline(y=-1, color='k', linestyle='-', alpha=0.3)
axs[0, 1].axvline(x=0, color='k', linestyle='-', alpha=0.3)

# ReLU vs Leaky ReLU
axs[1, 0].plot(x, relu(x), 'r-', linewidth=2, label='ReLU')
axs[1, 0].plot(x, leaky_relu(x, 0.1), 'orange', linestyle='--', linewidth=2, label='Leaky ReLU (α=0.1)')
axs[1, 0].set_title('ReLU and Leaky ReLU')
axs[1, 0].set_xlabel('x')
axs[1, 0].set_ylabel('f(x)')
axs[1, 0].grid(True)
axs[1, 0].legend()
axs[1, 0].text(2, 1, r'$ReLU(x) = max(0, x)$', fontsize=12)
axs[1, 0].text(-4, -1.5, r'$LeakyReLU(x) = max(\alpha x, x)$', fontsize=12)
axs[1, 0].axhline(y=0, color='k', linestyle='-', alpha=0.3)
axs[1, 0].axvline(x=0, color='k', linestyle='-', alpha=0.3)

# Softmax demonstration
softmax_demo = np.array([2.0, 1.0, 0.1])
softmax_output = softmax(softmax_demo)

axs[1, 1].bar(['Class 1', 'Class 2', 'Class 3'], softmax_output, color='purple')
axs[1, 1].set_title('Softmax Example')
axs[1, 1].set_xlabel('Classes')
axs[1, 1].set_ylabel('Probability')
axs[1, 1].grid(True, axis='y')
for i, v in enumerate(softmax_output):
    axs[1, 1].text(i, v + 0.01, f'{v:.3f}', ha='center')
axs[1, 1].text(0.5, 0.8, r'$softmax(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$', fontsize=12, transform=axs[1, 1].transAxes)
axs[1, 1].set_ylim(0, 1)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

### Choosing Activation Functions

Selecting the right activation function depends on the specific task:

| Activation Function | Use Case |
|---------------------|----------|
| **ReLU** | Default choice for most hidden layers in modern networks |
| **Sigmoid** | Binary classification output layer |
| **Softmax** | Multi-class classification output layer |
| **Tanh** | Alternative to ReLU in some recurrent networks |
| **Leaky ReLU** | When dealing with "dying ReLU" problems |

Practical considerations:
- ReLU is computationally efficient but can suffer from "dying ReLU" (neurons that always output 0)
- Sigmoid and tanh can suffer from vanishing gradient problems in deeper networks
- Leaky ReLU and variants like ELU (Exponential Linear Unit) aim to address ReLU limitations

## Types of Neural Networks

Different neural network architectures are designed to handle specific types of problems:

### 1. Feedforward Neural Networks (FNN)
- The most basic type of neural network
- Information flows in one direction from input to output
- Used for tabular data classification/regression

### 2. Convolutional Neural Networks (CNN)
- Specialized for processing grid-like data (e.g., images)
- Uses convolutional layers to detect spatial patterns
- Examples: ResNet, VGG, Inception

### 3. Recurrent Neural Networks (RNN)
- Process sequential data with "memory" of previous inputs
- Information cycles through loops within the network
- Examples: LSTM, GRU networks for text/time series

### 4. Transformers
- Advanced architecture using attention mechanisms
- Revolutionized NLP and other domains
- Examples: BERT, GPT models, Vision Transformers

Let's create simple examples of different neural network architectures:

```python
# Create simple examples of different neural network architectures using TensorFlow/Keras

# 1. Simple Feedforward Neural Network
feedforward_model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(10,)),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

# 2. Convolutional Neural Network (for image classification)
cnn_model = keras.Sequential([
    keras.layers.Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
    keras.layers.MaxPooling2D(pool_size=(2, 2)),
    keras.layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPooling2D(pool_size=(2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

# 3. Recurrent Neural Network (for sequence data)
rnn_model = keras.Sequential([
    keras.layers.Embedding(input_dim=10000, output_dim=32, input_length=100),
    keras.layers.LSTM(64, return_sequences=True),
    keras.layers.LSTM(32),
    keras.layers.Dense(1, activation='sigmoid')
])

# Print model summaries
print("Feedforward Neural Network:")
feedforward_model.summary()

print("\nConvolutional Neural Network:")
cnn_model.summary()

print("\nRecurrent Neural Network:")
rnn_model.summary()

# Visualize the architecture of the CNN model
keras.utils.plot_model(cnn_model, 
                       to_file='cnn_model.png', 
                       show_shapes=True, 
                       show_dtype=False,
                       show_layer_names=True,
                       rankdir='TB',
                       expand_nested=True,
                       dpi=96)

# Display the CNN architecture diagram if available
try:
    from IPython.display import Image
    Image('cnn_model.png')
except:
    print("Unable to display the CNN architecture diagram")
```

## Loss Functions

Loss functions measure how well the model is performing by calculating the difference between predicted outputs and actual target values. Different tasks require different loss functions:

### Regression Loss Functions:

1. **Mean Squared Error (MSE)**:
   $$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$
   - Penalizes larger errors more heavily
   - Standard choice for regression problems

2. **Mean Absolute Error (MAE)**:
   $$MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$
   - Less sensitive to outliers than MSE
   - Produces median predictions

3. **Huber Loss**:
   - Combines MSE and MAE
   - Less sensitive to outliers than MSE
   - Parameter δ determines the transition point

### Classification Loss Functions:

1. **Binary Cross-Entropy**:
   $$BCE = -\frac{1}{n}\sum_{i=1}^{n}[y_i \log(\hat{y}_i) + (1-y_i) \log(1-\hat{y}_i)]$$
   - Used for binary classification
   - Works with sigmoid activation in the output layer

2. **Categorical Cross-Entropy**:
   $$CCE = -\sum_{i=1}^{n}\sum_{j=1}^{m}y_{ij} \log(\hat{y}_{ij})$$
   - Used for multi-class classification
   - Works with softmax activation in the output layer

Let's implement and visualize these loss functions:

```python
# Implement common loss functions
def mse_loss(y_true, y_pred):
    return np.mean(np.square(y_true - y_pred))

def mae_loss(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

def huber_loss(y_true, y_pred, delta=1.0):
    error = y_true - y_pred
    is_small_error = np.abs(error) <= delta
    squared_loss = 0.5 * np.square(error)
    linear_loss = delta * (np.abs(error) - 0.5 * delta)
    return np.mean(np.where(is_small_error, squared_loss, linear_loss))

def binary_crossentropy(y_true, y_pred):
    # Add small epsilon to avoid log(0)
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

# Visualize regression loss functions
y_true = 0.5
y_pred_range = np.linspace(-0.5, 1.5, 1000)

mse_values = [mse_loss(y_true, y_pred) for y_pred in y_pred_range]
mae_values = [mae_loss(y_true, y_pred) for y_pred in y_pred_range]
huber_values = [huber_loss(y_true, y_pred) for y_pred in y_pred_range]

plt.figure(figsize=(12, 6))
plt.plot(y_pred_range, mse_values, 'b-', label='Mean Squared Error')
plt.plot(y_pred_range, mae_values, 'r-', label='Mean Absolute Error')
plt.plot(y_pred_range, huber_values, 'g-', label='Huber Loss (delta=1.0)')
plt.axvline(x=y_true, color='k', linestyle='--', label='True Value')
plt.xlabel('Predicted Value')
plt.ylabel('Loss')
plt.title('Comparison of Regression Loss Functions')
plt.legend()
plt.grid(True)
plt.show()

# Visualize binary cross-entropy loss
y_true_binary = np.array([0, 1])
y_pred_range = np.linspace(0.01, 0.99, 100)

bce_class0 = [-np.log(1 - y_pred) for y_pred in y_pred_range]  # For y_true = 0
bce_class1 = [-np.log(y_pred) for y_pred in y_pred_range]      # For y_true = 1

plt.figure(figsize=(12, 6))
plt.plot(y_pred_range, bce_class0, 'b-', label='BCE Loss (y_true = 0)')
plt.plot(y_pred_range, bce_class1, 'r-', label='BCE Loss (y_true = 1)')
plt.xlabel('Predicted Probability (ŷ)')
plt.ylabel('Binary Cross-Entropy Loss')
plt.title('Binary Cross-Entropy Loss Function')
plt.legend()
plt.grid(True)
plt.xlim(0, 1)
plt.show()
```

## Optimization Algorithms

Optimization algorithms adjust the network's weights and biases to minimize the loss function. The most common optimization algorithms include:

### 1. Gradient Descent

The basic algorithm that updates weights in the opposite direction of the gradient:

$$w_{t+1} = w_t - \eta \nabla_w L(w_t)$$

Where:
- $w_t$ is the weight at time step t
- $\eta$ is the learning rate
- $\nabla_w L(w_t)$ is the gradient of the loss function with respect to the weights

### 2. Stochastic Gradient Descent (SGD)

Updates weights using the gradient computed from a single sample or small batch:

- More efficient than standard gradient descent
- Introduces noise that can help escape local minima
- Can be enhanced with momentum to accelerate learning

### 3. Advanced Optimizers

- **Adam**: Adaptive Moment Estimation, combines momentum and RMSprop
- **RMSprop**: Adapts learning rates for each parameter
- **Adagrad**: Adapts learning rates based on parameter's history

Let's visualize how different optimizers work:

```python
# Define a simple 2D loss function for visualization
def loss_function(x, y):
    return (x**2 + y**2) * (1 + 0.5 * np.sin(x*3) * np.cos(y*2))

# Create a grid of points
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = loss_function(X, Y)

# Implement optimizers
def gradient(x, y):
    # Numerical approximation of gradient
    eps = 1e-7
    dx = (loss_function(x + eps, y) - loss_function(x - eps, y)) / (2 * eps)
    dy = (loss_function(x, y + eps) - loss_function(x, y - eps)) / (2 * eps)
    return np.array([dx, dy])

def sgd(init_x, init_y, learning_rate=0.1, iterations=50):
    path = np.zeros((iterations, 2))
    x, y = init_x, init_y
    path[0] = [x, y]
    
    for i in range(1, iterations):
        grad = gradient(x, y)
        x -= learning_rate * grad[0]
        y -= learning_rate * grad[1]
        path[i] = [x, y]
    
    return path

def sgd_with_momentum(init_x, init_y, learning_rate=0.1, momentum=0.9, iterations=50):
    path = np.zeros((iterations, 2))
    x, y = init_x, init_y
    path[0] = [x, y]
    velocity = np.zeros(2)
    
    for i in range(1, iterations):
        grad = gradient(x, y)
        velocity = momentum * velocity - learning_rate * grad
        x += velocity[0]
        y += velocity[1]
        path[i] = [x, y]
    
    return path

def adam(init_x, init_y, learning_rate=0.1, beta1=0.9, beta2=0.999, epsilon=1e-8, iterations=50):
    path = np.zeros((iterations, 2))
    x, y = init_x, init_y
    path[0] = [x, y]
    
    m = np.zeros(2)  # First moment
    v = np.zeros(2)  # Second moment
    
    for i in range(1, iterations):
        grad = gradient(x, y)
        
        # Update biased first moment estimate
        m = beta1 * m + (1 - beta1) * grad
        
        # Update biased second raw moment estimate
        v = beta2 * v + (1 - beta2) * np.square(grad)
        
        # Bias correction
        m_corrected = m / (1 - beta1**(i+1))
        v_corrected = v / (1 - beta2**(i+1))
        
        # Update parameters
        x -= learning_rate * m_corrected[0] / (np.sqrt(v_corrected[0]) + epsilon)
        y -= learning_rate * m_corrected[1] / (np.sqrt(v_corrected[1]) + epsilon)
        path[i] = [x, y]
    
    return path

# Starting point
init_x, init_y = 2.5, 2.5

# Run optimizers
sgd_path = sgd(init_x, init_y)
momentum_path = sgd_with_momentum(init_x, init_y)
adam_path = adam(init_x, init_y)

# Create a figure
fig, ax = plt.subplots(figsize=(12, 10))

# Plot contour of the loss function
contour = ax.contour(X, Y, Z, 20, cmap='viridis', alpha=0.6)
fig.colorbar(contour, ax=ax)

# Plot optimizer paths
ax.plot(sgd_path[:, 0], sgd_path[:, 1], 'r.-', label='SGD', linewidth=2, markersize=8)
ax.plot(momentum_path[:, 0], momentum_path[:, 1], 'g.-', label='SGD with Momentum', linewidth=2, markersize=8)
ax.plot(adam_path[:, 0], adam_path[:, 1], 'b.-', label='Adam', linewidth=2, markersize=8)

# Mark the starting point
ax.plot(init_x, init_y, 'ko', markersize=10, label='Starting Point')

# Mark the global minimum
ax.plot(0, 0, 'r*', markersize=15, label='Global Minimum')

ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('Comparison of Optimization Algorithms', fontsize=14)
ax.legend(fontsize=12)
ax.grid(True)

plt.show()

# Compare convergence rates
sgd_loss = [loss_function(x, y) for x, y in sgd_path]
momentum_loss = [loss_function(x, y) for x, y in momentum_path]
adam_loss = [loss_function(x, y) for x, y in adam_path]

plt.figure(figsize=(10, 6))
plt.plot(sgd_loss, 'r-', label='SGD')
plt.plot(momentum_loss, 'g-', label='SGD with Momentum')
plt.plot(adam_loss, 'b-', label='Adam')
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Convergence Rate of Different Optimizers')
plt.legend()
plt.grid(True)
plt.yscale('log')
plt.show()
```

## Deep Learning Workflow

A complete deep learning workflow typically involves the following steps:

1. **Data Preparation**:
   - Data collection and cleaning
   - Feature engineering and normalization
   - Train/validation/test split

2. **Model Definition**:
   - Choose architecture
   - Define layers and activation functions
   - Select loss function and optimizer

3. **Training**:
   - Feed data in batches
   - Forward propagation
   - Calculate loss
   - Backpropagation and weight updates

4. **Validation and Evaluation**:
   - Monitor training progress
   - Avoid overfitting
   - Evaluate on test data

5. **Inference**:
   - Deploy model
   - Make predictions on new data

Let's implement a simple end-to-end workflow using the MNIST dataset:

```python
# Complete deep learning workflow example using MNIST dataset

# 1. Data Preparation
# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize the data
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255

# Reshape for CNN input
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# One-hot encode the labels
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print(f"Training data shape: {x_train.shape}")
print(f"Testing data shape: {x_test.shape}")

# 2. Model Definition
model = keras.Sequential([
    keras.layers.Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),
    keras.layers.MaxPooling2D(pool_size=(2, 2)),
    keras.layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),
    keras.layers.MaxPooling2D(pool_size=(2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dropout(0.5),  # Prevent overfitting
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# 3. Training
# Use a small subset for quick demonstration
train_subset = 5000
val_subset = 1000
x_train_subset = x_train[:train_subset]
y_train_subset = y_train[:train_subset]
x_val_subset = x_test[:val_subset]
y_val_subset = y_test[:val_subset]

# Early stopping to prevent overfitting
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

# Train the model
history = model.fit(
    x_train_subset, y_train_subset,
    epochs=5,
    batch_size=64,
    validation_data=(x_val_subset, y_val_subset),
    callbacks=[early_stopping],
    verbose=1
)

# 4. Evaluation
# Evaluate on test data
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest accuracy: {test_accuracy:.4f}")
print(f"Test loss: {test_loss:.4f}")

# Visualize training history
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0, 1])
plt.legend()
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Training and Validation Loss')

plt.tight_layout()
plt.show()

# 5. Inference
# Make predictions on test data
predictions = model.predict(x_test[:9])
predicted_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(y_test[:9], axis=1)

# Display some predictions
plt.figure(figsize=(12, 12))
for i in range(9):
    plt.subplot(3, 3, i+1)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.title(f"True: {true_classes[i]}, Pred: {predicted_classes[i]}")
    plt.axis('off')

plt.tight_layout()
plt.show()

# Create a confusion matrix
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Use a subset of test data
y_pred = np.argmax(model.predict(x_test[:1000]), axis=1)
y_true = np.argmax(y_test[:1000], axis=1)

# Create confusion matrix
cm = confusion_matrix(y_true, y_pred)

# Plot confusion matrix
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
# Compare implementations of the same model using different frameworks

# Define a simple network architecture
# Input → Dense(64) → ReLU → Dense(32) → ReLU → Dense(10) → Softmax

# 1. TensorFlow/Keras Implementation
print("TensorFlow/Keras Implementation:")

tf_model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(784,)),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

tf_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

tf_model.summary()

# 2. PyTorch Implementation
print("\nPyTorch Implementation:")

try:
    import torch
    import torch.nn as nn
    import torch.nn.functional as F
    
    class PytorchModel(nn.Module):
        def __init__(self):
            super(PytorchModel, self).__init__()
            self.fc1 = nn.Linear(784, 64)
            self.fc2 = nn.Linear(64, 32)
            self.fc3 = nn.Linear(32, 10)
        
        def forward(self, x):
            x = F.relu(self.fc1(x))
            x = F.relu(self.fc2(x))
            x = F.softmax(self.fc3(x), dim=1)
            return x
    
    pytorch_model = PytorchModel()
    print(pytorch_model)
    
    # Example of how training would be done
    print("\nPyTorch Training Loop Example:")
    print("optimizer = torch.optim.Adam(model.parameters(), lr=0.001)")
    print("loss_fn = nn.CrossEntropyLoss()")
    print("for epoch in range(epochs):")
    print("    for batch in dataloader:")
    print("        inputs, targets = batch")
    print("        optimizer.zero_grad()")
    print("        outputs = model(inputs)")
    print("        loss = loss_fn(outputs, targets)")
    print("        loss.backward()")
    print("        optimizer.step()")
    
except ImportError:
    print("PyTorch is not installed. Cannot show PyTorch implementation.")


# 3. Keras (Low-Level) Implementation for comparison
print("\nKeras with Functional API Implementation:")

inputs = keras.Input(shape=(784,))
x = keras.layers.Dense(64, activation='relu')(inputs)
x = keras.layers.Dense(32, activation='relu')(x)
outputs = keras.layers.Dense(10, activation='softmax')(x)

functional_model = keras.Model(inputs=inputs, outputs=outputs)
functional_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

functional_model.summary()

# Compare key differences
print("\n\n--- Framework Comparison ---")
print("TensorFlow/Keras:")
print("✓ Eager execution by default (immediate evaluation)")
print("✓ Static computational graph when using @tf.function")
print("✓ Integrated with TensorFlow ecosystem")
print("✓ Built-in deployment options (TensorFlow Serving, TF Lite)")

print("\nPyTorch:")
print("✓ Dynamic computational graph")
print("✓ More pythonic API and debugging")
print("✓ Popular in research")
print("✓ TorchScript for production deployment")

print("\nKey Consideration:")
print("- TensorFlow/Keras: Better for production deployment")
print("- PyTorch: More flexible for research and experimentation")
print("- Both: Active development and community support")
```
