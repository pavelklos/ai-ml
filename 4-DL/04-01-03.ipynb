### 5.4 Managing GPU Memory

Both TensorFlow and PyTorch allow control over GPU memory allocation:

````python
# TensorFlow: Allow memory growth (prevents taking all GPU memory at once)
try:
    import tensorflow as tf
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print("Memory growth set to True for all GPUs")
except ImportError:
    print("TensorFlow not installed")
except Exception as e:
    print(f"Error setting memory growth: {e}")
````

## 6. Verifying Installations

Let's verify both frameworks are working with simple examples.

### 6.1 TensorFlow Basic Example

````python
try:
    import tensorflow as tf
    import numpy as np
    
    # Create a simple neural network
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(128, activation='relu', input_shape=(784,)),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(10, activation='softmax')
    ])
    
    model.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    
    # Create dummy data
    x_train = np.random.random((1000, 784))
    y_train = np.random.randint(10, size=1000)
    
    # Train for just 1 epoch to verify functionality
    model.fit(x_train, y_train, epochs=1, batch_size=32)
    print("TensorFlow model training successful!")
    
except ImportError:
    print("TensorFlow not installed")
except Exception as e:
    print(f"Error testing TensorFlow: {e}")
````

### 6.2 PyTorch Basic Example

````python
try:
    import torch
    import torch.nn as nn
    import torch.optim as optim
    
    # Determine device
    device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")
    
    # Create a simple neural network
    class SimpleNet(nn.Module):
        def __init__(self):
            super(SimpleNet, self).__init__()
            self.fc1 = nn.Linear(784, 128)
            self.dropout = nn.Dropout(0.2)
            self.fc2 = nn.Linear(128, 10)
            
        def forward(self, x):
            x = torch.relu(self.fc1(x))
            x = self.dropout(x)
            x = self.fc2(x)
            return x
    
    model = SimpleNet().to(device)
    
    # Create dummy data
    x_train = torch.randn(1000, 784, device=device)
    y_train = torch.randint(0, 10, (1000,), device=device)
    
    # Define loss function and optimizer
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters())
    
    # Train for just 1 batch to verify functionality
    model.train()
    optimizer.zero_grad()
    outputs = model(x_train)
    loss = criterion(outputs, y_train)
    loss.backward()
    optimizer.step()
    print(f"PyTorch training successful! Loss: {loss.item()}")
    
except ImportError:
    print("PyTorch not installed")
except Exception as e:
    print(f"Error testing PyTorch: {e}")
````

## 8. Common Issues and Troubleshooting

### 8.1 General Issues

1. **Dependency conflicts**: Use separate virtual environments
2. **Out of memory errors**: Reduce batch size, use gradient accumulation
3. **Version mismatches**: Check compatibility matrices

### 8.2 TensorFlow-specific Issues

1. **GPU not detected**: Check CUDA/cuDNN compatibility
2. **Memory allocation errors**: Enable memory growth
3. **Undefined symbol errors**: Mismatched CUDA versions

````python
# TensorFlow: Common fixes for GPU issues
import os

# Allow memory growth
os.environ['TF_FORCE_GPU_ALLOW_GROWTH'] = 'true'

# Use older allocator if you have issues
# os.environ['TF_GPU_ALLOCATOR'] = 'cuda_malloc_async'

# Log device placement
# os.environ['TF_CPP_MIN_LOG_LEVEL'] = '0'
# tf.debugging.set_log_device_placement(True)
````

### 8.3 PyTorch-specific Issues

1. **CUDA error: out of memory**: Reduce batch size, try `torch.cuda.empty_cache()`
2. **Wrong CUDA version**: Reinstall PyTorch with correct CUDA version
3. **Dataloader bottlenecks**: Increase num_workers, use pin_memory

````python
# PyTorch: Common fixes for memory issues
try:
    import torch
    
    # Clear cache
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        print("CUDA cache cleared")
        
    # Check memory usage
    if torch.cuda.is_available():
        print(f"Memory allocated: {torch.cuda.memory_allocated() / 1e9:.2f} GB")
        print(f"Memory reserved: {torch.cuda.memory_reserved() / 1e9:.2f} GB")
except ImportError:
    print("PyTorch not installed")
````

### 8.4 Diagnostic Tool

````python
def diagnose_environment():
    """Run diagnostics on the deep learning environment"""
    import sys, platform
    
    print("========== SYSTEM INFO ==========")
    print(f"System: {platform.system()} {platform.release()}")
    print(f"Python: {sys.version}")
    print(f"Executable: {sys.executable}")
    
    print("\n========== NVIDIA DRIVER ==========")
    try:
        import subprocess
        nvidia_output = subprocess.check_output(["nvidia-smi"], universal_newlines=True)
        for line in nvidia_output.split('\n')[:10]:  # Print first few lines
            print(line)
        print("...")
    except:
        print("NVIDIA driver not found or nvidia-smi not available")
    
    print("\n========== TENSORFLOW ==========")
    try:
        import tensorflow as tf
        print(f"TensorFlow: {tf.__version__}")
        print(f"  GPU available: {len(tf.config.list_physical_devices('GPU')) > 0}")
        print(f"  Devices: {tf.config.list_physical_devices()}")
    except ImportError:
        print("TensorFlow not installed")
    except Exception as e:
        print(f"TensorFlow error: {e}")
    
    print("\n========== PYTORCH ==========")
    try:
        import torch
        print(f"PyTorch: {torch.__version__}")
        print(f"  CUDA available: {torch.cuda.is_available()}")
        if torch.cuda.is_available():
            print(f"  CUDA version: {torch.version.cuda}")
            print(f"  Device count: {torch.cuda.device_count()}")
            print(f"  Current device: {torch.cuda.current_device()}")
            print(f"  Device name: {torch.cuda.get_device_name(torch.cuda.current_device())}")
    except ImportError:
        print("PyTorch not installed")
    except Exception as e:
        print(f"PyTorch error: {e}")

# Run diagnostics
diagnose_environment()
````
