## Understanding Learning Paradigms

Machine learning algorithms can be categorized based on how they learn from data. The two primary learning paradigms are:

1. **Supervised Learning**: Algorithms learn from labeled data, where the desired output is known.
2. **Unsupervised Learning**: Algorithms learn from unlabeled data, discovering patterns and relationships without guidance.

Other paradigms include:
- **Semi-supervised Learning**: Uses both labeled and unlabeled data
- **Reinforcement Learning**: Learns through interaction with an environment and feedback

In this notebook, we'll focus on the differences between supervised and unsupervised learning, with examples of each.

## Supervised Learning

Supervised learning uses labeled training data to learn the mapping function from input variables (X) to the output variable (Y). The goal is to approximate the mapping function so well that when you have new input data, you can predict the output variables for that data.

### Key Characteristics:
- **Training data is labeled**: Each example has input features and the correct output
- **Direct feedback**: The algorithm can compare its predictions with the correct answers
- **Output-driven**: Learns to predict specific outputs based on input patterns

### Common Supervised Learning Algorithms:

#### Classification Algorithms:
- Logistic Regression
- Support Vector Machines (SVM)
- Decision Trees and Random Forests
- K-Nearest Neighbors (KNN)
- Neural Networks

#### Regression Algorithms:
- Linear Regression
- Ridge Regression
- Lasso Regression
- Decision Tree Regression
- Random Forest Regression

### Evaluation Metrics:
- **Classification**: Accuracy, Precision, Recall, F1 Score, ROC-AUC
- **Regression**: Mean Squared Error (MSE), R-squared, Mean Absolute Error (MAE)

## Unsupervised Learning

Unsupervised learning works with unlabeled data, finding structure or patterns in the input data without any explicit output variable to predict. The system tries to learn the patterns and underlying distribution in the data on its own.

### Key Characteristics:
- **Training data is unlabeled**: Only input data is available, no corresponding output
- **No feedback**: No external guidance on whether the algorithm is finding the right patterns
- **Structure-driven**: Focuses on finding inherent structures in the data

### Common Unsupervised Learning Algorithms:

#### Clustering Algorithms:
- K-Means Clustering
- Hierarchical Clustering
- DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
- Gaussian Mixture Models

#### Dimensionality Reduction:
- Principal Component Analysis (PCA)
- t-Distributed Stochastic Neighbor Embedding (t-SNE)
- Autoencoders
- UMAP (Uniform Manifold Approximation and Projection)

#### Association Rule Learning:
- Apriori Algorithm
- Eclat Algorithm
- FP-Growth

### Evaluation Approaches:
- Silhouette Score
- Calinski-Harabasz Index
- Davies-Bouldin Index
- Visual inspection of clusters

## Semi-Supervised and Reinforcement Learning

While we're focusing on supervised vs. unsupervised learning, it's worth briefly mentioning other learning paradigms for context:

### Semi-Supervised Learning
- Combines labeled and unlabeled data for training
- Particularly useful when labeled data is limited or expensive to obtain
- Uses the structure learned from unlabeled data to improve supervised tasks
- Examples: Semi-supervised SVM, label propagation, generative models

### Reinforcement Learning
- An agent learns by interacting with an environment
- Receives feedback through rewards or penalties
- Goal is to learn a policy that maximizes cumulative rewards
- Examples: Q-learning, Deep Q Networks (DQN), Policy Gradient methods
- Applications: Game playing (AlphaGo), robotics, autonomous vehicles

These approaches blend elements of supervised and unsupervised learning or introduce new learning paradigms entirely.

## Implementing Supervised Learning Example

Let's implement a simple supervised learning example using the Iris dataset. This dataset contains measurements of iris flowers and the task is to classify them into one of three species.

```python
# Load the iris dataset from scikit-learn
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# Create a DataFrame for better visualization
iris_df = pd.DataFrame(X, columns=feature_names)
iris_df['species'] = [target_names[i] for i in y]

# Display the first few rows of the dataset
print("Iris Dataset Shape:", iris_df.shape)
iris_df.head()
```

```python
# Basic exploratory data analysis
print("Dataset Overview:")
print(f"Number of samples: {X.shape[0]}")
print(f"Number of features: {X.shape[0]}")
print(f"Number of classes: {len(np.unique(y))}")
print(f"Class distribution: {np.bincount(y)}")

# Visualize the data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=[target_names[i] for i in y], palette='viridis')
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
plt.title('Sepal Features')

plt.subplot(1, 2, 2)
sns.scatterplot(x=X[:, 2], y=X[:, 3], hue=[target_names[i] for i in y], palette='viridis')
plt.xlabel(feature_names[2])
plt.ylabel(feature_names[3])
plt.title('Petal Features')

plt.tight_layout()
plt.show()
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Create and train multiple supervised learning models
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'SVM': SVC(random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=3)
}

# Dictionary to store results
results = {}

# Train and evaluate each model
for name, model in models.items():
    # Train the model
    model.fit(X_train_scaled, y_train)
    
    # Make predictions
    y_pred = model.predict(X_test_scaled)
    
    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)
    results[name] = accuracy
    
    print(f"{name} Accuracy: {accuracy:.4f}")
```

```python
# Visualize the results
plt.figure(figsize=(10, 6))
models_df = pd.DataFrame({'Model': list(results.keys()), 'Accuracy': list(results.values())})
sns.barplot(x='Model', y='Accuracy', data=models_df)
plt.title('Supervised Learning Model Comparison')
plt.ylim(0.8, 1.0)  # Set y-axis limits for better visualization
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Let's examine the best performing model
best_model_name = max(results, key=results.get)
best_model = models[best_model_name]

# Get predictions from the best model
best_y_pred = best_model.predict(X_test_scaled)

# Create a confusion matrix
cm = confusion_matrix(y_test, best_y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title(f'Confusion Matrix for {best_model_name}')
plt.tight_layout()
plt.show()
```

### Supervised Learning Workflow Summary

The key steps in our supervised learning example were:

1. **Data Collection**: We used the Iris dataset with labeled species.
2. **Data Preprocessing**: We split the data and standardized the features.
3. **Model Selection**: We tried multiple classification algorithms.
4. **Model Training**: Each model learned patterns from the labeled training data.
5. **Model Evaluation**: We assessed performance using accuracy and visualized with a confusion matrix.
6. **Model Comparison**: We compared different algorithms to find the best performer.

This is a classic supervised learning workflow where the algorithm learns from known examples to make predictions on new, unseen data.

```python
import os
import multiprocessing  # Make sure to import the multiprocessing module
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import pandas as pd

# For unsupervised learning, we'll use the same data but ignore the labels
# We'll scale the features for better algorithm performance
X_scaled = StandardScaler().fit_transform(X)

# Let's apply K-Means clustering
# Since we know there are 3 species, we'll use 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

# Apply PCA for dimensionality reduction
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Apply t-SNE for non-linear dimensionality reduction
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

# Get CPU count using multiprocessing (avoids subprocess run)
cpu_cores = multiprocessing.cpu_count()

# Create a DataFrame for visualization
pca_df = pd.DataFrame({
    'PCA1': X_pca[:, 0],
    'PCA2': X_pca[:, 1],
    'KMeans_Cluster': kmeans_labels,
    'True_Species': [target_names[i] for i in y]
})

tsne_df = pd.DataFrame({
    'TSNE1': X_tsne[:, 0],
    'TSNE2': X_tsne[:, 1],
    'KMeans_Cluster': kmeans_labels,
    'True_Species': [target_names[i] for i in y]
})

# Optionally, print out CPU core information
print(f"CPU Cores Detected: {cpu_cores}")
```

```python
# Visualize the K-Means clustering results
plt.figure(figsize=(16, 6))

# Plot with true labels for reference
plt.subplot(1, 3, 1)
sns.scatterplot(x='PCA1', y='PCA2', hue='True_Species', data=pca_df, palette='viridis')
plt.title('PCA - True Species Labels')

# Plot with K-Means clusters
plt.subplot(1, 3, 2)
sns.scatterplot(x='PCA1', y='PCA2', hue='KMeans_Cluster', data=pca_df, palette='viridis')
plt.title('PCA - K-Means Clusters')

# Visualize cluster centers
centers = pca.transform(kmeans.cluster_centers_)
plt.subplot(1, 3, 3)
sns.scatterplot(x='PCA1', y='PCA2', hue='KMeans_Cluster', data=pca_df, palette='viridis', alpha=0.5)
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=100, alpha=0.8, marker='X')
plt.title('PCA - K-Means Clusters with Centroids')

plt.tight_layout()
plt.show()

# Visualize t-SNE results
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.scatterplot(x='TSNE1', y='TSNE2', hue='True_Species', data=tsne_df, palette='viridis')
plt.title('t-SNE - True Species Labels')

plt.subplot(1, 2, 2)
sns.scatterplot(x='TSNE1', y='TSNE2', hue='KMeans_Cluster', data=tsne_df, palette='viridis')
plt.title('t-SNE - K-Means Clusters')

plt.tight_layout()
plt.show()
```

```python
# Evaluate clustering performance using external metrics
# (In real unsupervised learning, we wouldn't have these labels to compare with)
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score, silhouette_score

# Calculate the agreement between the clusters and the true labels
ari = adjusted_rand_score(y, kmeans_labels)
nmi = normalized_mutual_info_score(y, kmeans_labels)
silhouette = silhouette_score(X_scaled, kmeans_labels)

print("Clustering Performance Metrics:")
print(f"Adjusted Rand Index: {ari:.4f}")
print(f"Normalized Mutual Information: {nmi:.4f}")
print(f"Silhouette Score: {silhouette:.4f}")

# Create a contingency table to see how clusters map to actual species
contingency_table = pd.crosstab(
    pd.Series(y, name='True Species'),
    pd.Series(kmeans_labels, name='Cluster')
)
contingency_table.index = target_names

print("\nContingency Table (Species vs Cluster):")
print(contingency_table)
```

```python
# Let's try a different clustering algorithm: Hierarchical Clustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Apply hierarchical clustering
linkage_matrix = linkage(X_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(12, 6))
dendrogram(linkage_matrix, labels=[target_names[i] for i in y])
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()

# Cut the dendrogram to get 3 clusters
from scipy.cluster.hierarchy import fcluster
hierarchical_labels = fcluster(linkage_matrix, 3, criterion='maxclust') - 1

# Compare with true labels
ari_hierarchical = adjusted_rand_score(y, hierarchical_labels)
print(f"Hierarchical Clustering - Adjusted Rand Index: {ari_hierarchical:.4f}")

# Create a contingency table for hierarchical clustering
hierarchical_contingency = pd.crosstab(
    pd.Series(y, name='True Species'),
    pd.Series(hierarchical_labels, name='Cluster')
)
hierarchical_contingency.index = target_names

print("\nContingency Table (Species vs Hierarchical Cluster):")
print(hierarchical_contingency)
```

## When to Use Each Approach

Choosing between supervised and unsupervised learning depends on several factors:

### Use Supervised Learning When:

1. **You have labeled data**: When you have a substantial amount of data with known outcomes
2. **You have a specific prediction task**: When you need to predict a specific target variable
3. **You need precise outputs**: When accuracy and specific predictions are important
4. **Your problem is well-defined**: When you know exactly what you're trying to predict
5. **You have resources for data labeling**: When you can invest in creating labeled datasets

### Use Unsupervised Learning When:

1. **You lack labeled data**: When obtaining labeled data is expensive, time-consuming, or impossible
2. **You're exploring unknown patterns**: When you want to discover hidden structures in your data
3. **You need dimensionality reduction**: When working with high-dimensional data that needs simplification
4. **You want to identify customer segments**: When you need to group similar entities together
5. **You're detecting anomalies**: When you need to find outliers or unusual patterns

### Hybrid Approaches:

Sometimes the best approach is to combine both:
- Use unsupervised learning to discover patterns and create features
- Then use supervised learning to make predictions with these new features
- Semi-supervised learning leverages both labeled and unlabeled data

## Conclusion

In this notebook, we've explored the fundamental differences between supervised and unsupervised learning:

- **Supervised learning** requires labeled data and aims to predict specific outcomes. It's ideal for classification and regression problems where the target variable is known.

- **Unsupervised learning** works with unlabeled data to discover hidden patterns and structures. It's perfect for clustering, dimensionality reduction, and anomaly detection tasks.

We demonstrated both approaches with the Iris dataset:
- Using supervised learning, we built classification models that achieved high accuracy in predicting flower species.
- Using unsupervised learning, we discovered natural groupings in the data that largely corresponded to the actual species.

The choice between supervised and unsupervised learning depends on your data availability, problem definition, and ultimate goals. In many real-world applications, a combination of both approaches may yield the best results.

As machine learning continues to evolve, understanding these fundamental paradigms provides a strong foundation for more advanced techniques like semi-supervised learning, transfer learning, and reinforcement learning.
