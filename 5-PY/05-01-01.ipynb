# Installing Python and Setting Up the Environment

This notebook provides a comprehensive guide to installing Python and setting up a development environment across different operating systems. We'll cover Python installation options, package management, virtual environments, Jupyter setup, and IDE recommendations.

## 1. Python Installation Options

Python can be installed in several ways depending on your needs and operating system. Here are the main options:

### Official Python Installer
- Available from [python.org](https://www.python.org/downloads/)
- Provides a clean, standard Python installation
- Allows fine-grained control over installation options
- Best for developers who need specific Python versions

### Anaconda Distribution
- Available from [anaconda.com](https://www.anaconda.com/products/distribution)
- Includes Python, hundreds of packages, and Conda package manager
- Provides tools like Jupyter Notebook, Spyder IDE
- Best for data scientists and researchers
- Requires more disk space (~3GB)

### Miniconda
- A minimal version of Anaconda containing only Python and conda
- Much smaller installation size
- You can install only the packages you need
- Good balance between control and convenience

### System Package Managers
- Linux: `apt`, `yum`, `pacman`
- macOS: `brew`
- Windows: `winget`, `chocolatey`
- Simplest integration with your OS, but may not have the latest Python versions

### Comparison Table

| Method | Pros | Cons | Best For |
|--------|------|------|----------|
| Official Installer | Clean, standard installation<br>Latest versions | Manual PATH setup<br>Package management limited to pip | Python-focused developers |
| Anaconda | Pre-installed data science packages<br>Conda environment management | Large size<br>Can conflict with other Python installs | Data scientists, beginners |
| Miniconda | Conda without bloat<br>Flexible | Manual package installation<br>Initial setup time | Experienced users who want control |
| System Package Manager | System integration<br>Easy updates | May lag behind latest versions<br>Potential system conflicts | System administrators, casual users |

## 2. Setting Up Python on Windows

### Using the Official Installer

1. Download the latest installer from [python.org](https://www.python.org/downloads/)
2. Run the installer with these important options:
   - ✅ Add Python to PATH
   - ✅ Install pip
   - ✅ Install for all users (if you have admin rights)
3. Verify the installation by opening Command Prompt and running:
```
python --version
```

### Using Anaconda on Windows

1. Download the Anaconda installer from [anaconda.com](https://www.anaconda.com/products/distribution)
2. Run the installer and follow the prompts
3. Optionally add Anaconda to the PATH (not recommended by default)
4. Launch Anaconda Navigator or use Anaconda Prompt

### Using Windows Subsystem for Linux (WSL)

For developers who prefer a Linux environment on Windows:

1. Install WSL from Windows features or PowerShell:
```
wsl --install
```
2. Install Python within the Linux distribution using apt:
```
sudo apt update
sudo apt install python3 python3-pip
```

## 3. Setting Up Python on macOS

## 5. Package Management with pip

`pip` is Python's default package manager. It allows you to install, update, and remove Python packages.

### Basic pip Commands

```bash
# Install a package
pip install package_name

# Install a specific version
pip install package_name==1.2.3

# Upgrade a package
pip install --upgrade package_name

# Uninstall a package
pip uninstall package_name

# List installed packages
pip list

# Show information about a package
pip show package_name
```

### Using Requirements Files

Requirements files help you maintain consistent package versions across environments.

```bash
# Create a requirements.txt file with current packages
pip freeze > requirements.txt

# Install packages from requirements.txt
pip install -r requirements.txt
```

### Pip Configuration

You can configure pip using a `pip.conf` file (Linux/macOS) or `pip.ini` file (Windows) in:
- Global: `/etc/pip.conf`
- User: `~/.config/pip/pip.conf` or `%APPDATA%\pip\pip.ini`
- Virtual Environment: `venv/pip.conf`

Example configuration to use a different package index:
```ini
[global]
index-url = https://your-private-pypi.org/simple
trusted-host = your-private-pypi.org
```

## 6. Creating Virtual Environments

Virtual environments allow you to create isolated Python environments for different projects. This prevents package version conflicts and keeps your global Python installation clean.

### Using venv (Built into Python 3.3+)

```bash
# Create a virtual environment
python -m venv myproject_env

# Activate the environment
# On Windows:
myproject_env\Scripts\activate

# On macOS/Linux:
source myproject_env/bin/activate

# Deactivate when finished
deactivate
```

### Using virtualenv (Works with Python 2 and 3)

```bash
# Install virtualenv
pip install virtualenv

# Create a virtual environment
virtualenv myproject_env

# Activate and deactivate as with venv
```

### Using conda (Anaconda/Miniconda)

```bash
# Create a conda environment
conda create --name myproject_env python=3.10

# Activate the environment
conda activate myproject_env

# Deactivate
conda deactivate

# List environments
conda env list
```

### Using virtualenvwrapper (Enhanced Management)

```bash
# Install
pip install virtualenvwrapper

# On Windows:
pip install virtualenvwrapper-win

# Set up (add to .bashrc or .zshrc)
export WORKON_HOME=$HOME/.virtualenvs
source /usr/local/bin/virtualenvwrapper.sh

# Create environment
mkvirtualenv myproject_env

# Work on environment
workon myproject_env

# Deactivate
deactivate
```

````python
# Example of creating a virtual environment programmatically
import sys
import subprocess
import os

def create_venv(env_name):
    """Create a Python virtual environment with the given name."""
    try:
        subprocess.run([sys.executable, '-m', 'venv', env_name], check=True)
        print(f"Virtual environment '{env_name}' created successfully")
        
        # Show activation command based on OS
        if os.name == 'nt':  # Windows
            print(f"\nTo activate, run: {env_name}\\Scripts\\activate")
        else:  # Unix-like
            print(f"\nTo activate, run: source {env_name}/bin/activate")
    except subprocess.CalledProcessError:
        print(f"Error creating virtual environment '{env_name}'")

# Uncomment to create a virtual environment
# create_venv('example_env')
````

## 8. Setting Environment Variables

Environment variables affect how Python and its packages behave. Properly configuring them is important for development.

### Common Python Environment Variables

- **PYTHONPATH**: Directories to add to the module search path
- **PYTHONHOME**: Change the location of the standard Python libraries
- **PYTHONSTARTUP**: Path to a file executed on interactive startup
- **PYTHONUNBUFFERED**: Force stdin, stdout, and stderr to be unbuffered (set to 1)
- **PYTHONDEBUG**: Enable debug output (set to 1)

### Setting Environment Variables

#### Windows

```cmd
:: Temporarily in Command Prompt
set PYTHONPATH=C:\my_python_libs

:: Permanently via System Properties
setx PYTHONPATH "C:\my_python_libs"
```

#### macOS/Linux

```bash
# Temporarily in terminal
export PYTHONPATH=/path/to/my_python_libs

# Permanently in .bashrc, .zshrc, etc.
echo 'export PYTHONPATH=/path/to/my_python_libs' >> ~/.bashrc
```

### Using .env Files

For project-specific environment variables:

```bash
# Install python-dotenv
pip install python-dotenv
```

Create a `.env` file in your project directory:
```
DATABASE_URL=postgresql://user:password@localhost/dbname
API_KEY=your_api_key_here
DEBUG=True
```

Load it in your Python code:
```python
from dotenv import load_dotenv
import os

load_dotenv()  # take environment variables from .env

database_url = os.getenv("DATABASE_URL")
api_key = os.getenv("API_KEY")
```

````python
# View current Python environment variables
import os
import sys

# Get Python-related environment variables
python_env_vars = {k: v for k, v in os.environ.items() if 'PYTHON' in k}
print("Python environment variables:")
for k, v in python_env_vars.items():
    print(f"{k} = {v}")

# Display Python path
print("\nPython sys.path (module search paths):")
for path in sys.path:
    print(f"- {path}")
````

## 10. Verifying Your Python Installation

Let's verify that your Python environment is correctly set up and check its configuration.

### Basic Verification

From your terminal or command prompt:

```bash
# Check Python version
python --version

# Start Python interpreter
python

# Check pip installation
pip --version
```

````python
# Check Python system information
import sys
import platform

print(f"Python version: {sys.version}")
print(f"Python implementation: {platform.python_implementation()}")
print(f"Python executable path: {sys.executable}")
print(f"Python path: {sys.path[0]}")
print(f"Platform: {platform.platform()}")
````

````python
# Check for essential data science packages
def check_package(package_name):
    """Check if a package is installed and get its version."""
    try:
        package = __import__(package_name)
        version = getattr(package, '__version__', 'unknown version')
        return f"{package_name}: {version} (installed)"
    except ImportError:
        return f"{package_name}: Not installed"

# List of common packages to check
packages = [
    'numpy', 'pandas', 'matplotlib', 'scipy', 'sklearn',
    'seaborn', 'jupyter', 'jupyterlab', 'ipywidgets'
]

for package in packages:
    print(check_package(package))
````

### Comprehensive Environment Test

This test will check for common issues in your Python environment:

````python
def test_environment():
    """Run a series of tests to verify your Python environment."""
    import sys
    import os
    import platform
    import subprocess
    from pathlib import Path
    
    print("======= PYTHON ENVIRONMENT VERIFICATION =======\n")
    
    # 1. Python version
    print(f"Python version: {platform.python_version()}")
    if sys.version_info < (3, 7):
        print("⚠️ WARNING: Python version is older than 3.7. Consider upgrading.")
    else:
        print("✅ Python version is 3.7 or newer.")
    
    # 2. Check pip
    try:
        pip_version = subprocess.check_output([sys.executable, '-m', 'pip', '--version']).decode().strip()
        print(f"Pip version: {pip_version}")
        print("✅ pip is installed and working.")
    except Exception as e:
        print(f"⚠️ ERROR: pip check failed: {e}")
    
    # 3. Check virtual environment
    in_venv = sys.prefix != sys.base_prefix
    if in_venv:
        print(f"✅ Running in a virtual environment: {sys.prefix}")
    else:
        print("📝 NOTE: Not running in a virtual environment. Consider using one for project isolation.")
    
    # 4. Check PATH configuration
    python_in_path = False
    python_path = Path(sys.executable).parent
    
    path_entries = os.environ.get('PATH', '').split(os.pathsep)
    for entry in path_entries:
        if Path(entry).resolve() == python_path.resolve():
            python_in_path = True
            break
    
    if python_in_path:
        print("✅ Python directory is in PATH.")
    else:
        print(f"⚠️ WARNING: Python directory ({python_path}) is not in PATH.")
    
    # 5. Check file encoding settings
    file_encoding = sys.getfilesystemencoding()
    print(f"File system encoding: {file_encoding}")
    if file_encoding.lower().startswith(('utf', 'ascii')):
        print("✅ File system encoding supports Unicode.")
    else:
        print("⚠️ WARNING: File system encoding may not fully support Unicode.")
        
    # 6. Overall assessment
    print("\n======= ENVIRONMENT SUMMARY =======")
    print(f"Python {platform.python_version()} at {sys.executable}")
    print(f"Operating System: {platform.platform()}")
    print("Your Python environment appears to be properly set up.")
    print("To install packages, use: pip install package_name")
    
# Run the environment test
test_environment()
````
