## 1. Understanding Modules and Packages

### What are Modules?

A **module** in Python is a file containing Python code that can be imported and used in other Python scripts. Modules help organize code into reusable components.

### What are Packages?

A **package** is a collection of modules organized in directories. Packages provide a way to hierarchically organize modules for better code management.

````python
# Let's explore some built-in modules
import math
import random
import os

print(f"Value of pi: {math.pi}")
print(f"Random number between 0 and 1: {random.random()}")
print(f"Current working directory: {os.getcwd()}")
````

### Module Structure

A module can contain:
- Variables
- Functions
- Classes
- Executable code

### Package Structure

```
my_package/
    __init__.py
    module1.py
    module2.py
    subpackage/
        __init__.py
        module3.py
```

## 2. Creating and Importing Modules

Let's create a simple module and import it.

````python
# Writing content to a module file
%%writefile my_module.py

# This is a simple module
def greet(name):
    """Return a greeting message"""
    return f"Hello, {name}!"

def add(a, b):
    """Add two numbers"""
    return a + b

PI = 3.14159

class Calculator:
    """A simple calculator class"""
    def __init__(self, value=0):
        self.value = value
        
    def add(self, x):
        self.value += x
        return self.value
    
# Code that runs when the module is executed directly
if __name__ == "__main__":
    print("This module is being run directly.")
else:
    print("This module has been imported.")
````

````python
# Import the entire module
import my_module

# Using module components with dot notation
print(my_module.greet("Python User"))
print(f"Sum of 5 and 7: {my_module.add(5, 7)}")
print(f"PI value from module: {my_module.PI}")

# Create an instance of the Calculator class
calc = my_module.Calculator(10)
print(f"Calculator value after adding 5: {calc.add(5)}")
````

````python
# Import specific components from a module
from my_module import greet, PI

# Now we can use these components directly
print(greet("Direct Import User"))
print(f"PI directly: {PI}")
````

````python
# Import with an alias
import my_module as mm

print(mm.greet("Alias User"))
````

````python
# Import all components (generally not recommended due to namespace pollution)
from my_module import *

print(greet("Wildcard Import User"))
print(f"Direct access to PI: {PI}")
````

### Module Search Path

When you import a module, Python looks for it in several locations in a specific order.

## 3. Module Search Path and sys.path

The `sys.path` is a list of directories that Python searches when importing modules.

````python
import sys

# Print the module search path
print("Module search paths:")
for path in sys.path:
    print(f"  - {path}")
````

### Modifying the Module Search Path

You can add directories to `sys.path` to make Python look for modules in additional locations.

````python
import sys
import os

# Create a directory for custom modules
os.makedirs("custom_modules", exist_ok=True)

# Create a module in that directory
with open("custom_modules/custom_math.py", "w") as f:
    f.write("""\
def square(x):
    """Return the square of a number"""
    return x * x

def cube(x):
    """Return the cube of a number"""
    return x * x * x
""")

# Add the directory to the module search path
custom_path = os.path.abspath("custom_modules")
sys.path.append(custom_path)
print(f"Added {custom_path} to sys.path")

# Now we can import from that directory
import custom_math

print(f"Square of 4: {custom_math.square(4)}")
print(f"Cube of 3: {custom_math.cube(3)}")
````

### Environmental variable PYTHONPATH

You can also set the `PYTHONPATH` environment variable to add directories to the module search path. In a Jupyter notebook, we can simulate this by modifying `sys.path` directly, but in a real environment, you might do:

```bash
# On Unix/Linux/macOS
export PYTHONPATH=/path/to/custom/modules:$PYTHONPATH

# On Windows
set PYTHONPATH=C:\path\to\custom\modules;%PYTHONPATH%
```

## 4. Working with Packages

Let's create a simple package structure.

````python
import os

# Create package directory structure
os.makedirs("my_package", exist_ok=True)
os.makedirs("my_package/subpackage", exist_ok=True)

# Create __init__.py files
with open("my_package/__init__.py", "w") as f:
    f.write("""\
# my_package initialization
print("Initializing my_package")

# Define what gets imported with 'from my_package import *'
__all__ = ['module1', 'module2']
""")

with open("my_package/subpackage/__init__.py", "w") as f:
    f.write("""\
# subpackage initialization
print("Initializing subpackage")
""")

# Create module files
with open("my_package/module1.py", "w") as f:
    f.write("""\
# my_package.module1
def function1():
    return "This is function1 from module1"
""")

with open("my_package/module2.py", "w") as f:
    f.write("""\
# my_package.module2
def function2():
    return "This is function2 from module2"
""")

with open("my_package/subpackage/module3.py", "w") as f:
    f.write("""\
# my_package.subpackage.module3
def function3():
    return "This is function3 from subpackage.module3"
""")

print("Package structure created successfully!")
````

````python
# Import the package
import my_package
````

````python
# Import specific modules from the package
from my_package import module1

print(module1.function1())
````

````python
# Import from a subpackage
from my_package.subpackage import module3

print(module3.function3())
````

## 5. Import Variations and Best Practices

There are several ways to import modules in Python, each with its own pros and cons.

````python
# Different import approaches

# Option 1: Import the entire module
import math
result1 = math.sqrt(16)

# Option 2: Import specific functions/objects
from math import sqrt, pi
result2 = sqrt(16)

# Option 3: Import with alias
import numpy as np
result3 = np.array([1, 2, 3])

# Option 4: Import all (generally not recommended)
from math import *
result4 = cos(0)

print(f"Result 1: {result1}")
print(f"Result 2: {result2}")
print(f"Result 3: {result3}")
print(f"Result 4: {result4}")
````

### Import Best Practices

1. **Be explicit**: Prefer `import math` or `from math import sqrt, pi` over `from math import *`
2. **Use aliases for common libraries**: `import numpy as np`, `import pandas as pd`
3. **Order imports properly**:
   - Standard library imports first
   - Third-party library imports second
   - Local application imports last
4. **Group imports**: Separate groups with blank lines
5. **Avoid circular imports**: When two modules import each other

### Example of well-organized imports

````python
# Standard library imports
import os
import sys
from datetime import datetime, timedelta

# Third-party imports
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

# Local application imports
from my_package import module1
from my_package.subpackage import module3

print("Imports organized properly!")
````

## 7. Virtual Environments

Virtual environments are isolated Python environments that allow you to install packages without affecting the global Python installation or other projects.

### Creating Virtual Environments

There are several ways to create virtual environments in Python:

1. **Using venv** (built-in since Python 3.3):

```bash
# Create a virtual environment
python -m venv myenv

# Activate it (Windows)
myenv\Scripts\activate

# Activate it (Unix/macOS)
source myenv/bin/activate

# Deactivate
deactivate
```

2. **Using conda** (Anaconda/Miniconda):

```bash
# Create a virtual environment
conda create -n myenv python=3.9

# Activate it
conda activate myenv

# Deactivate
conda deactivate
```

3. **Using virtualenv** (needs to be installed):

```bash
# Install virtualenv
pip install virtualenv

# Create a virtual environment
virtualenv myenv

# Activate/deactivate (same as venv)
```

### Benefits of Virtual Environments

1. **Dependency Isolation**: Different projects can use different versions of the same package
2. **Clean Testing Environment**: Test your package in a clean environment
3. **Easy Dependency Management**: Requirements files work well with virtual environments
4. **Reproducibility**: Others can recreate your exact environment
5. **No sudo/admin privileges**: Install packages locally without system-wide permissions

### Virtual Environment Demo

Let's demonstrate creating a virtual environment (note: this runs in the system, not the notebook):

## 8. Creating Your Own Packages

Let's create a simple package that could be distributed to others.

````python
import os

# Create package directory structure
os.makedirs("mathutils", exist_ok=True)
os.makedirs("mathutils/algebra", exist_ok=True)
os.makedirs("mathutils/geometry", exist_ok=True)

# Create package files
with open("mathutils/__init__.py", "w") as f:
    f.write("""\
# MathUtils package
"""
    )

with open("mathutils/algebra/__init__.py", "w") as f:
    f.write("""\
# Algebra subpackage
"""
    )

with open("mathutils/geometry/__init__.py", "w") as f:
    f.write("""\
# Geometry subpackage
"""
    )

# Add module files
with open("mathutils/algebra/equations.py", "w") as f:
    f.write("""\
def solve_linear(a, b):
    """Solve the linear equation ax + b = 0"""
    if a == 0:
        raise ValueError("Coefficient 'a' cannot be zero")
    return -b / a

def solve_quadratic(a, b, c):
    """Solve the quadratic equation ax^2 + bx + c = 0"""
    if a == 0:
        return [solve_linear(b, c)]
    
    discriminant = b**2 - 4*a*c
    if discriminant < 0:
        return []
    elif discriminant == 0:
        return [-b/(2*a)]
    else:
        x1 = (-b + discriminant**0.5) / (2*a)
        x2 = (-b - discriminant**0.5) / (2*a)
        return [x1, x2]
"""
    )

with open("mathutils/geometry/shapes.py", "w") as f:
    f.write("""\
import math

def circle_area(radius):
    """Calculate the area of a circle"""
    return math.pi * radius**2

def rectangle_area(width, height):
    """Calculate the area of a rectangle"""
    return width * height

def triangle_area(base, height):
    """Calculate the area of a triangle"""
    return 0.5 * base * height
"""
    )

# Create a README file
with open("mathutils/README.md", "w") as f:
    f.write("""\
# MathUtils

A simple Python package for mathematical utilities.

## Installation

```
pip install mathutils
```

## Usage

```python
from mathutils.algebra import equations
from mathutils.geometry import shapes

# Solve equations
x = equations.solve_linear(2, -4)  # 2x - 4 = 0
print(f"x = {x}")  # x = 2.0

# Calculate areas
area = shapes.circle_area(5)
print(f"Circle area = {area}")  # Circle area = 78.54...
```
"""
    )

print("Package structure created successfully!")
````

### Using the package

````python
# Add the current directory to the path
import sys
sys.path.append('.')

# Import from our package
from mathutils.algebra import equations
from mathutils.geometry import shapes

# Use the functions
x = equations.solve_linear(2, -4)  # 2x - 4 = 0
print(f"Solution to 2x - 4 = 0: x = {x}")

roots = equations.solve_quadratic(1, -3, 2)  # x^2 - 3x + 2 = 0
print(f"Solutions to x^2 - 3x + 2 = 0: x = {roots}")

area = shapes.circle_area(5)
print(f"Area of circle with radius 5: {area:.2f}")

area = shapes.rectangle_area(3, 4)
print(f"Area of rectangle 3×4: {area}")
````

## 9. Working with \_\_init\_\_.py

The `__init__.py` file serves several important purposes in Python packages:

### Basic Usage

- **Identifies directories as Python packages**
- **Initializes package-level variables and imports**
- **Can be empty (just a marker)**
- **Controls what's exported with `__all__`**

````python
# Let's modify our mathutils package to use __init__.py more effectively
with open("mathutils/__init__.py", "w") as f:
    f.write("""\
# MathUtils package
__version__ = '0.1.0'

# Import common functions directly to the package namespace
from mathutils.algebra.equations import solve_linear, solve_quadratic
from mathutils.geometry.shapes import circle_area, rectangle_area, triangle_area

# Define what gets imported with 'from mathutils import *'
__all__ = ['solve_linear', 'solve_quadratic', 'circle_area', 'rectangle_area', 'triangle_area']
"""
    )
````

````python
# Also improve the subpackage init files
with open("mathutils/algebra/__init__.py", "w") as f:
    f.write("""\
# Algebra subpackage
from mathutils.algebra.equations import solve_linear, solve_quadratic

__all__ = ['solve_linear', 'solve_quadratic']
"""
    )

with open("mathutils/geometry/__init__.py", "w") as f:
    f.write("""\
# Geometry subpackage
from mathutils.geometry.shapes import circle_area, rectangle_area, triangle_area

__all__ = ['circle_area', 'rectangle_area', 'triangle_area']
"""
    )
````

````python
# Reload the modules to apply changes
import importlib
import mathutils
importlib.reload(mathutils)

# Now we can import directly from the package
from mathutils import circle_area, solve_quadratic

print(f"Package version: {mathutils.__version__}")
print(f"Circle area: {circle_area(3):.2f}")
print(f"Quadratic equation roots: {solve_quadratic(1, -5, 6)}")
````

### Advanced \_\_init\_\_.py Techniques

- **Lazy loading**: Import modules only when needed
- **Package configuration**: Set up configuration on import
- **Automatic submodule discovery**: Automatically import all submodules
- **Deprecated warnings**: Warn users about deprecated features

````python
# Example of a more advanced __init__.py (not writing it to file)
advanced_init = """\
# Advanced package initialization
__version__ = '0.2.0'

# Lazy loading with properties
class _LazyLoader:
    def __init__(self):
        self._algebra = None
        self._geometry = None
        
    @property
    def algebra(self):
        if self._algebra is None:
            import mathutils.algebra
            self._algebra = mathutils.algebra
        return self._algebra
        
    @property
    def geometry(self):
        if self._geometry is None:
            import mathutils.geometry
            self._geometry = mathutils.geometry
        return self._geometry

# Create and export the lazy loader
lazy = _LazyLoader()

# Set up logging
import logging
logger = logging.getLogger(__name__)
logger.addHandler(logging.NullHandler())

# Define shortcuts for common functions
from mathutils.algebra.equations import solve_linear, solve_quadratic
from mathutils.geometry.shapes import circle_area, rectangle_area

__all__ = ['lazy', 'solve_linear', 'solve_quadratic', 'circle_area', 'rectangle_area']
"""

print(advanced_init)
````

## 10. Distributing Python Packages

To share your packages with others, you can publish them on the Python Package Index (PyPI) or distribute them directly.

### Package Distribution Files

Let's create the necessary files to distribute our package:

````python
# Create setup.py
with open("setup.py", "w") as f:
    f.write("""\
from setuptools import setup, find_packages

setup(
    name="mathutils",
    version="0.1.0",
    author="Your Name",
    author_email="your.email@example.com",
    description="A simple math utilities package",
    long_description=open("mathutils/README.md").read(),
    long_description_content_type="text/markdown",
    url="https://github.com/yourusername/mathutils",
    packages=find_packages(),
    classifiers=[
        "Programming Language :: Python :: 3",
        "License :: OSI Approved :: MIT License",
        "Operating System :: OS Independent",
    ],
    python_requires=">=3.6",
)
"""
    )
````

### Steps to Distribute a Package

1. **Build the package**:
```bash
pip install setuptools wheel
python setup.py sdist bdist_wheel
```

2. **Upload to PyPI**:
```bash
pip install twine
twine upload dist/*
```

3. **Install from PyPI**:
```bash
pip install mathutils
```

### Alternative Distribution Methods

1. **Direct from GitHub**:
```bash
pip install git+https://github.com/username/mathutils.git
```

2. **Local install (development mode)**:
```bash
pip install -e .
```

### Using modern packaging tools

The standard `setup.py` approach is gradually being replaced by newer tools and approaches:

````python
# Example pyproject.toml
with open("pyproject.toml", "w") as f:
    f.write("""\
[build-system]
requires = ["setuptools>=42", "wheel"]
build-backend = "setuptools.build_meta"

[project]
name = "mathutils"
version = "0.1.0"
authors = [
    {name = "Your Name", email = "your.email@example.com"},
]
description = "A simple math utilities package"
readme = "mathutils/README.md"
requires-python = ">=3.6"
classifiers = [
    "Programming Language :: Python :: 3",
    "License :: OSI Approved :: MIT License",
    "Operating System :: OS Independent",
]

[project.urls]
"Homepage" = "https://github.com/yourusername/mathutils"
"Bug Tracker" = "https://github.com/yourusername/mathutils/issues"
"""
    )
````
