## 1. Project Structure Basics

A well-organized Python project typically follows established conventions for directory structure and file organization.

````python
import os

def display_project_structure(base_dir="my_project"):
    """Display a typical Python project structure"""
    structure = {
        "": ["README.md", "setup.py", "requirements.txt", ".gitignore", "pyproject.toml"],
        "my_package": ["__init__.py", "main.py", "config.py"],
        "my_package/utils": ["__init__.py", "helpers.py"],
        "tests": ["__init__.py", "test_main.py"],
        "docs": ["index.md", "usage.md"],
        "data": [".gitkeep"]
    }
    
    # Print the structure
    print(f"Project structure for '{base_dir}':\n")
    for directory, files in structure.items():
        if directory:
            dir_path = os.path.join(base_dir, directory)
            print(f"📂 {dir_path}/")
        else:
            dir_path = base_dir
            print(f"📂 {dir_path}/")
            
        for file in files:
            print(f"  📄 {os.path.join(dir_path, file)}")
        print()
    
    return "Each component has a specific purpose in a well-organized project."

display_project_structure()
````

### Common Project Structure Components

| Component | Purpose |
|-----------|--------|
| `README.md` | Project documentation and overview |
| `setup.py` | Installation script for the package |
| `requirements.txt` | Lists dependencies |
| `pyproject.toml` | Modern Python packaging configuration |
| `my_package/` | Main package directory |
| `tests/` | Test files |
| `docs/` | Documentation files |
| `data/` | Data files used by the project |

## 2. Creating a Basic Python Package

Python packages are directories containing Python modules, with a special `__init__.py` file to indicate it's a package.

````python
def create_basic_package(package_name="sample_package"):
    """Create the basic structure for a Python package"""
    
    # Define package content
    init_content = '''"""
{} - A sample Python package
"""

__version__ = "0.1.0"
'''
    
    main_content = '''"""
Main functionality for the package
"""

def hello():
    """Return a greeting"""
    return "Hello from {}!"
'''
    
    setup_content = '''from setuptools import setup, find_packages

setup(
    name="{}",
    version="0.1.0",
    packages=find_packages(),
    install_requires=[
        # Dependencies go here
    ],
    author="Your Name",
    author_email="your.email@example.com",
    description="A short description of the package",
    keywords="sample, package",
    url="http://github.com/yourusername/{}",
    classifiers=[
        "Development Status :: 3 - Alpha",
        "Programming Language :: Python :: 3",
    ],
)
'''
    
    # Print what we would create
    print(f"Creating package '{package_name}':\n")
    
    print(f"📄 {package_name}/__init__.py:")
    print("```python")
    print(init_content.format(package_name))
    print("```\n")
    
    print(f"📄 {package_name}/main.py:")
    print("```python")
    print(main_content.format(package_name))
    print("```\n")
    
    print(f"📄 setup.py:")
    print("```python")
    print(setup_content.format(package_name, package_name))
    print("```\n")
    
    return "These are the minimum files needed for a distributable Python package."

create_basic_package("example_pkg")
````

## 6. Documentation Best Practices

Good documentation is essential for project maintainability and onboarding new developers.

````python
def show_documentation_practices():
    """Demonstrates documentation best practices"""
    
    # Example docstrings
    docstring_example = '''
def calculate_discount(price, discount_percentage, max_discount=100):
    """Calculate the discounted price of an item.
    
    This function applies a percentage discount to a given price,
    ensuring that the discount doesn't exceed the maximum allowed.
    
    Args:
        price (float): The original price of the item.
        discount_percentage (float): The discount percentage (0-100).
        max_discount (float, optional): The maximum allowed discount amount. 
            Defaults to 100.
    
    Returns:
        float: The price after applying the discount.
        
    Raises:
        ValueError: If price or discount_percentage is negative.
        
    Examples:
        >>> calculate_discount(100, 20)
        80.0
        >>> calculate_discount(100, 20, 15)
        85.0
    """
    if price < 0 or discount_percentage < 0:
        raise ValueError("Price and discount percentage must be non-negative")
        
    discount_amount = price * (discount_percentage / 100)
    discount_amount = min(discount_amount, max_discount)  # Cap at max_discount
    
    return price - discount_amount
'''
    
    # Example README.md
    readme_example = '''
# My Project

[![PyPI version](https://badge.fury.io/py/myproject.svg)](https://badge.fury.io/py/myproject)
[![Build Status](https://travis-ci.org/username/myproject.svg?branch=main)](https://travis-ci.org/username/myproject)
[![Documentation Status](https://readthedocs.io/en/latest/?badge=latest)](https://docs.readthedocs.io/en/latest/?badge=latest)

A Python package for processing data and generating reports.

## Features

- Process CSV and Excel files
- Generate PDF reports
- Export data to multiple formats
- Command-line interface

## Installation

```bash
pip install myproject
```

## Quick Start

```python
from myproject import process_file

# Process a file
result = process_file("data.csv")
print(result.summary())
```

## Documentation

Full documentation is available at [https://myproject.readthedocs.io](https://myproject.readthedocs.io).

## License

This project is licensed under the MIT License - see the LICENSE file for details.
'''
    
    print("1. Python Docstrings:")
    print("```python")
    print(docstring_example)
    print("```\n")
    
    print("2. Example README.md:")
    print("```markdown")
    print(readme_example)
    print("```\n")
    
    print("3. Documentation Tools:")
    tools = [
        "Sphinx - Python documentation generator",
        "Read the Docs - Documentation hosting",
        "MkDocs - Project documentation with Markdown",
        "pdoc - Auto-generate API documentation",
        "Doctest - Verify examples in docstrings",
    ]
    for tool in tools:
        print(f"  • {tool}")
        
    print("\n4. Documentation Best Practices:")
    practices = [
        "Document the 'why', not just the 'what'",
        "Keep documentation close to the code",
        "Include examples in docstrings",
        "Update documentation with code changes",
        "Use type annotations to improve clarity",
        "Provide installation and quickstart guides",
        "Document command-line options"
    ]
    for practice in practices:
        print(f"  • {practice}")

show_documentation_practices()
````

## 8. Testing Structure

Testing is crucial for code quality and reliability.

````python
def show_testing_structure():
    """Demonstrates testing structure best practices"""
    
    # Example test file
    test_example = '''
# tests/test_utils.py
import unittest
from my_package.utils import calculate_discount

class TestUtils(unittest.TestCase):
    def test_calculate_discount_normal(self):
        """Test normal discount calculation"""
        result = calculate_discount(100, 20)
        self.assertEqual(result, 80)
        
    def test_calculate_discount_max(self):
        """Test with maximum discount cap"""
        result = calculate_discount(100, 20, 15)
        self.assertEqual(result, 85)
        
    def test_calculate_discount_invalid(self):
        """Test with invalid inputs"""
        with self.assertRaises(ValueError):
            calculate_discount(-100, 20)
            
        with self.assertRaises(ValueError):
            calculate_discount(100, -20)
            
if __name__ == "__main__":
    unittest.main()
'''
    
    # Example pytest file
    pytest_example = '''
# tests/test_utils_pytest.py
import pytest
from my_package.utils import calculate_discount

def test_calculate_discount_normal():
    """Test normal discount calculation"""
    result = calculate_discount(100, 20)
    assert result == 80
    
def test_calculate_discount_max():
    """Test with maximum discount cap"""
    result = calculate_discount(100, 20, 15)
    assert result == 85
    
def test_calculate_discount_invalid():
    """Test with invalid inputs"""
    with pytest.raises(ValueError):
        calculate_discount(-100, 20)
        
    with pytest.raises(ValueError):
        calculate_discount(100, -20)
'''
    
    # Example conftest.py
    conftest_example = '''
# tests/conftest.py
import pytest
import tempfile
import os

@pytest.fixture
def sample_data():
    """Provide sample data for tests"""
    return {
        "id": 1,
        "name": "Test Product",
        "price": 99.99,
        "categories": ["electronics", "gadgets"]
    }
    
@pytest.fixture
def temp_file():
    """Create a temporary file for testing"""
    fd, path = tempfile.mkstemp()
    yield path
    os.close(fd)
    os.unlink(path)
'''
    
    print("1. Example unittest Structure:")
    print("```python")
    print(test_example)
    print("```\n")
    
    print("2. Example pytest Structure:")
    print("```python")
    print(pytest_example)
    print("```\n")
    
    print("3. Example pytest Fixtures:")
    print("```python")
    print(conftest_example)
    print("```\n")
    
    print("4. Running Tests:")
    print("```bash")
    print("# Using unittest")
    print("python -m unittest discover tests  # Run all tests")
    print("python -m unittest tests/test_utils.py  # Run specific test file")
    print("")
    print("# Using pytest")
    print("pytest  # Run all tests")
    print("pytest tests/test_utils_pytest.py  # Run specific test file")
    print("pytest -v  # Verbose output")
    print("pytest --cov=my_package  # Run with coverage")
    print("```\n")
    
    print("5. Testing Best Practices:")
    practices = [
        "Write tests that are isolated and independent",
        "Use clear test names that describe what's being tested",
        "Follow the Arrange-Act-Assert pattern",
        "Test both normal and edge cases",
        "Use fixtures for common test data",
        "Mock external dependencies",
        "Aim for high test coverage, but focus on critical code"
    ]
    for practice in practices:
        print(f"  • {practice}")

show_testing_structure()
````

## 10. Building Command-Line Interfaces

Command-line interfaces allow users to interact with your project from the terminal.

````python
def show_cli_development():
    """Demonstrates building command-line interfaces"""
    
    # Using argparse
    argparse_example = '''
# cli.py
import argparse

def main():
    """Main CLI function"""
    parser = argparse.ArgumentParser(
        description="Process some data files.",
        epilog="Example: python cli.py process data.csv --format json"
    )
    
    # Add commands
    subparsers = parser.add_subparsers(dest="command", help="Commands")
    
    # Process command
    process_parser = subparsers.add_parser("process", help="Process a data file")
    process_parser.add_argument("file", help="File to process")
    process_parser.add_argument(
        "--format", 
        choices=["csv", "json", "excel"], 
        default="csv", 
        help="Output format"
    )
    process_parser.add_argument(
        "--output", 
        help="Output file"
    )
    
    # Analyze command
    analyze_parser = subparsers.add_parser("analyze", help="Analyze data")
    analyze_parser.add_argument("file", help="File to analyze")
    analyze_parser.add_argument(
        "--type", 
        choices=["summary", "detailed"], 
        default="summary", 
        help="Analysis type"
    )
    
    # Parse arguments
    args = parser.parse_args()
    
    # Execute commands
    if args.command == "process":
        print(f"Processing {args.file} to {args.format} format")
        if args.output:
            print(f"Output will be saved to {args.output}")
    elif args.command == "analyze":
        print(f"Analyzing {args.file} with {args.type} analysis")
    else:
        parser.print_help()

if __name__ == "__main__":
    main()
'''
    
    # Using Click
    click_example = '''
# cli_click.py
import click

@click.group()
def cli():
    """Command line tool for data processing."""
    pass

@cli.command()
@click.argument("file")
@click.option(
    "--format", 
    type=click.Choice(["csv", "json", "excel"]), 
    default="csv",
    help="Output format"
)
@click.option(
    "--output", 
    help="Output file"
)
def process(file, format, output):
    """Process a data file."""
    click.echo(f"Processing {file} to {format} format")
    if output:
        click.echo(f"Output will be saved to {output}")

@cli.command()
@click.argument("file")
@click.option(
    "--type", 
    type=click.Choice(["summary", "detailed"]), 
    default="summary",
    help="Analysis type"
)
def analyze(file, type):
    """Analyze data."""
    click.echo(f"Analyzing {file} with {type} analysis")

if __name__ == "__main__":
    cli()
'''
    
    # Setting up entry points in setup.py
    entry_points_example = '''
# setup.py
from setuptools import setup, find_packages

setup(
    name="my_package",
    version="0.1.0",
    packages=find_packages(),
    install_requires=[
        "click>=8.0.0",
    ],
    entry_points={
        "console_scripts": [
            "my-command=my_package.cli:cli",
        ],
    },
)
'''
    
    # Example rich integration
    rich_example = '''
# cli_rich.py
import click
from rich.console import Console
from rich.table import Table
from rich.progress import track
import time

console = Console()

@click.command()
@click.argument("files", nargs=-1, required=True)
@click.option("--verbose", is_flag=True, help="Show detailed output")
def process(files, verbose):
    """Process multiple files with a nice UI."""
    console.print(f"[bold blue]Processing {len(files)} files...[/bold blue]")
    
    # Show progress bar
    for file in track(files, description="Processing"):
        # Simulate processing
        time.sleep(0.5)
        if verbose:
            console.print(f"Processed [green]{file}[/green]")
    
    # Show results in a table
    table = Table(title="Processing Results")
    
    table.add_column("File", style="cyan")
    table.add_column("Status", style="green")
    table.add_column("Size", justify="right", style="magenta")
    
    for file in files:
        table.add_row(file, "Success", f"{len(file) * 100} bytes")
    
    console.print(table)

if __name__ == "__main__":
    process()
'''
    
    print("1. Using argparse:")
    print("```python")
    print(argparse_example)
    print("```\n")
    
    print("2. Using Click:")
    print("```python")
    print(click_example)
    print("```\n")
    
    print("3. Setting Up Entry Points:")
    print("```python")
    print(entry_points_example)
    print("```\n")
    
    print("4. Creating Rich CLI Interfaces:")
    print("```python")
    print(rich_example)
    print("```\n")
    
    print("5. CLI Development Best Practices:")
    practices = [
        "Follow the Unix philosophy (do one thing well)",
        "Provide clear help and documentation",
        "Support both verbose and quiet modes",
        "Use sensible defaults",
        "Provide useful error messages",
        "Support configuration files",
        "Consider color and formatting for better UX",
        "Add progress indicators for long-running operations"
    ]
    for practice in practices:
        print(f"  • {practice}")

show_cli_development()
````
