# Basic File I/O Operations in Python

This notebook provides a comprehensive introduction to file input/output (I/O) operations in Python. File I/O is a fundamental aspect of programming that allows us to persist data by reading from and writing to files on disk.

## Introduction to File Operations in Python

Python provides robust built-in functions and methods for working with files. File operations generally involve:

1. **Opening** a file
2. **Reading from** or **writing to** the file
3. **Closing** the file

Python makes these operations straightforward and offers various methods to handle different file formats and scenarios.

````python
# Import necessary libraries
import os
import json
import csv
import pickle
import shutil
from pathlib import Path
````

## Opening and Closing Files

In Python, we open files using the `open()` function, which returns a file object. Always close files after operations to free up system resources.

````python
# Basic opening and closing of a file

# Create a sample text file for demonstration
with open('sample.txt', 'w') as f:
    f.write('This is a sample text file.\n')
    f.write('It contains multiple lines.\n')
    f.write('We will use it for demonstration.\n')

# Method 1: Traditional way of opening and closing files
print("Method 1: Traditional open/close")
file = open('sample.txt', 'r')
content = file.read()
print(content)
file.close()  # Important to close the file

# Method 2: Using 'with' statement (context manager) - recommended
print("\nMethod 2: Using with statement (context manager)")
with open('sample.txt', 'r') as file:
    content = file.read()
    print(content)  # File is automatically closed when the block exits
````

The `open()` function takes two main parameters:

- **filename**: The name/path of the file to open
- **mode**: A string indicating how the file should be opened (read, write, etc.)

Key modes include:
- `'r'`: Read (default)
- `'w'`: Write (creates new file or truncates existing file)
- `'a'`: Append (adds to end of file)
- `'b'`: Binary mode (e.g., 'rb', 'wb')
- `'t'`: Text mode (default)
- `'+'`: Update mode (read and write)

## Reading from Files

Python provides multiple methods for reading file content.

````python
# Different ways to read files

# 1. Read the entire file content
with open('sample.txt', 'r') as file:
    content = file.read()
    print("1. Reading entire file:")
    print(content)

# 2. Read file line by line
with open('sample.txt', 'r') as file:
    print("\n2. Reading line by line:")
    for line in file:
        print(f"  Line: {line.strip()}")  # strip() removes newline characters

# 3. Reading all lines into a list
with open('sample.txt', 'r') as file:
    lines = file.readlines()
    print("\n3. Reading all lines into a list:")
    print(f"  Number of lines: {len(lines)}")
    print(f"  Lines: {[line.strip() for line in lines]}")

# 4. Reading specific number of characters
with open('sample.txt', 'r') as file:
    chunk = file.read(20)  # Read first 20 characters
    print("\n4. Reading first 20 characters:")
    print(f"  '{chunk}'")
````

## Writing to Files

Python provides several methods for writing content to files.

````python
# Different ways to write to files

# 1. Writing a string to a file
with open('output1.txt', 'w') as file:
    file.write("This is a single line of text.\n")
    file.write("This is another line of text.\n")

# 2. Writing multiple lines at once
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']
with open('output2.txt', 'w') as file:
    file.writelines(lines)

# 3. Appending to a file
with open('output1.txt', 'a') as file:
    file.write("This line is appended to the end.\n")

# Display the content of the files we created
print("Content of output1.txt:")
with open('output1.txt', 'r') as file:
    print(file.read())

print("\nContent of output2.txt:")
with open('output2.txt', 'r') as file:
    print(file.read())
````

## Working with File Modes

The mode argument in the `open()` function specifies how you want to interact with the file.

````python
# Demonstrating different file modes

print("Common File Modes in Python:")
print("-------------------------")
print("'r' - Read mode (default)")
print("'w' - Write mode (creates new file or truncates existing file)")
print("'a' - Append mode (adds to end of file)")
print("'x' - Exclusive creation (fails if file already exists)")
print("'b' - Binary mode (e.g., 'rb', 'wb')")
print("'t' - Text mode (default)")
print("'+' - Update mode (read and write)")

# Examples

# 1. Binary mode for working with non-text files
data = bytes([0, 1, 2, 3, 4, 5])
with open('binary_file.bin', 'wb') as file:
    file.write(data)

with open('binary_file.bin', 'rb') as file:
    binary_data = file.read()
    print(f"\nBinary data read: {binary_data}")

# 2. Read and write mode (r+)
with open('sample.txt', 'r+') as file:
    content = file.read(10)  # Read first 10 characters
    print(f"\nFirst 10 characters: '{content}'")
    file.write("\nThis line is added using r+ mode.")

# Display the updated content
print("\nUpdated content of sample.txt:")
with open('sample.txt', 'r') as file:
    print(file.read())
````

````python
# File pointers and seeking

# Create a simple file for demonstration
with open('seek_demo.txt', 'wb') as file:  # Write in binary mode
    file.write(b"abcdefghijklmnopqrstuvwxyz")  # Write bytes

# Working with file pointers
with open('seek_demo.txt', 'rb+') as file:  # Open in binary read/write mode
    # 1. Get current position
    position = file.tell()
    print(f"Initial position: {position}")
    
    # 2. Read 5 bytes
    data = file.read(5)
    print(f"Read data: {data.decode()}")  # Decode bytes to string
    print(f"New position: {file.tell()}")
    
    # 3. Move to position 10
    file.seek(10)
    print(f"After seeking to position 10: {file.tell()}")
    print(f"Reading from position 10: {file.read(5).decode()}")
    
    # 4. Move 3 bytes from current position
    file.seek(3, 1)  # 1 means relative to current position
    print(f"After seeking 3 bytes from current: {file.tell()}")
    print(f"Reading: {file.read(5).decode()}")
    
    # 5. Seek from end of file
    file.seek(-5, 2)  # 2 means relative to end of file
    print(f"After seeking 5 bytes from end: {file.tell()}")
    print(f"Reading last 5 characters: {file.read().decode()}")
````

The `seek()` method takes two arguments:
- **offset**: The number of bytes to move
- **whence**: The reference position (optional)
  - 0 = from beginning of file (default)
  - 1 = from current position
  - 2 = from end of file

## Using Context Managers

Context managers (`with` statement) are the recommended way to handle files in Python. They ensure proper resource management by automatically closing files when operations are complete.

````python
# Benefits of using context managers for file I/O

# 1. Traditional approach (not recommended)
try:
    file = open('sample.txt', 'r')
    content = file.read()
    print("Traditional approach content length:", len(content))
finally:
    file.close()
    print("File closed manually")

# 2. Context manager approach (recommended)
with open('sample.txt', 'r') as file:
    content = file.read()
    print("\nContext manager content length:", len(content))
print("File automatically closed")

# 3. Working with multiple files
with open('input.txt', 'w') as input_file:
    input_file.write("This is input data")

with open('input.txt', 'r') as input_file, open('output3.txt', 'w') as output_file:
    data = input_file.read()
    output_file.write(data.upper())
    print("\nData copied and converted to uppercase")
    
# Verify output
with open('output3.txt', 'r') as file:
    print(f"Output file content: '{file.read()}'")
````

## Working with Different File Formats

Python provides libraries for working with various file formats such as CSV, JSON, and binary files.

````python
# Working with different file formats

# 1. CSV Files
print("1. Working with CSV files")
# Writing to CSV
data = [
    ['Name', 'Age', 'Country'],
    ['Alice', 28, 'USA'],
    ['Bob', 32, 'Canada'],
    ['Charlie', 25, 'UK']
]

with open('data.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)
    print("  CSV file written successfully")

# Reading from CSV
with open('data.csv', 'r', newline='') as file:
    reader = csv.reader(file)
    print("  CSV content:")
    for row in reader:
        print(f"    {row}")

# 2. JSON Files
print("\n2. Working with JSON files")
person = {
    'name': 'John Doe',
    'age': 30,
    'is_employed': True,
    'languages': ['Python', 'JavaScript', 'SQL'],
    'address': {
        'city': 'New York',
        'zip': '10001'
    }
}

# Writing JSON
with open('data.json', 'w') as file:
    json.dump(person, file, indent=2)
    print("  JSON file written successfully")

# Reading JSON
with open('data.json', 'r') as file:
    loaded_data = json.load(file)
    print("  JSON content:")
    print(f"    {loaded_data}")

# 3. Binary Files with pickle
print("\n3. Working with binary files using pickle")
complex_data = {
    'array': [1, 2, 3, 4, 5],
    'dict': {'a': 1, 'b': 2},
    'tuple': (10, 20, 30)
}

# Serializing with pickle
with open('data.pickle', 'wb') as file:
    pickle.dump(complex_data, file)
    print("  Data serialized with pickle")

# Deserializing with pickle
with open('data.pickle', 'rb') as file:
    loaded_data = pickle.load(file)
    print("  Deserialized data:")
    print(f"    {loaded_data}")
````

## Error Handling in File Operations

File operations can fail due to various reasons (file doesn't exist, permission issues, etc.). It's important to handle these errors properly.

````python
# Error handling in file operations

# 1. Handling non-existent files
try:
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("1. Error: The file doesn't exist")

# 2. Creating a file safely
file_path = 'test_file.txt'
try:
    # Check if file exists before opening in write mode
    if os.path.exists(file_path):
        print(f"\n2. Warning: {file_path} already exists and will be overwritten")
    with open(file_path, 'w') as file:
        file.write("This file was created safely.")
    print(f"   File '{file_path}' created successfully")
except PermissionError:
    print(f"\n2. Error: No permission to write to {file_path}")
except IOError as e:
    print(f"\n2. I/O error: {e}")

# 3. Using 'x' mode for exclusive creation
try:
    with open('test_file.txt', 'x') as file:
        file.write("This won't work if file already exists")
except FileExistsError:
    print("\n3. Error: Cannot create file that already exists using 'x' mode")

# 4. Using a finally block to ensure cleanup
temp_file = 'temp_file.txt'
file = None
try:
    file = open(temp_file, 'w')
    file.write("Temporary content")
    print("\n4. File opened and written to successfully")
except IOError as e:
    print(f"\n4. Error: {e}")
finally:
    if file and not file.closed:
        file.close()
        print("   File closed in finally block")
        os.remove(temp_file)  # Cleanup
        print("   Temporary file removed")
````

## Working with Directories

Python provides modules like `os` and `pathlib` for working with directories and file paths.

````python
# Working with directories

# 1. Getting current working directory
current_dir = os.getcwd()
print(f"1. Current working directory: {current_dir}")

# 2. Creating directories
try:
    # Create a new directory
    os.mkdir('new_directory')
    print("\n2. Directory 'new_directory' created")
except FileExistsError:
    print("\n2. Directory already exists")

# Create nested directories
os.makedirs('parent/child/grandchild', exist_ok=True)
print("   Nested directories created")

# 3. Listing directory contents
print("\n3. Files in current directory:")
for item in os.listdir('.'):
    if os.path.isfile(item):
        print(f"   File: {item}")
    elif os.path.isdir(item):
        print(f"   Directory: {item}")

# 4. Working with file paths
file_path = os.path.join('new_directory', 'file_in_dir.txt')
with open(file_path, 'w') as f:
    f.write("This file is inside the new directory.")
print(f"\n4. Created file at: {file_path}")

# 5. Checking file properties
if os.path.exists(file_path):
    print(f"\n5. File size: {os.path.getsize(file_path)} bytes")
    print(f"   Last modified: {os.path.getmtime(file_path)}")

# 6. Using pathlib (more modern approach)
print("\n6. Using pathlib for path operations:")
path = Path('new_directory/another_file.txt')
path.write_text("Content written using pathlib")
print(f"   File created: {path}")
print(f"   File exists: {path.exists()}")
print(f"   Parent directory: {path.parent}")

# 7. Walking directory tree
print("\n7. Walking directory tree:")
for root, dirs, files in os.walk('parent'):
    print(f"   Directory: {root}")
    if dirs:
        print(f"     Subdirectories: {dirs}")
    if files:
        print(f"     Files: {files}")

# 8. Cleaning up (removing files and directories)
# Comment out the next section if you want to keep the created files
try:
    os.remove(file_path)
    os.remove(str(path))
    os.rmdir('new_directory')
    shutil.rmtree('parent')  # Recursively remove directory tree
    print("\n8. Cleanup complete - removed test files and directories")
except Exception as e:
    print(f"\n8. Cleanup error: {e}")
````
