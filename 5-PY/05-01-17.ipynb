# Python Debugging Techniques

This notebook explores various debugging techniques in Python programming, from simple print statements to advanced tools and approaches. Effective debugging is an essential skill for any programmer to locate and fix issues in their code.

## 1. Understanding Debugging

Debugging is the process of identifying, analyzing, and removing errors (bugs) in software code. It's a critical skill in software development that helps ensure code functions as expected.

### What is debugging?

Debugging is a methodical process of:
- Finding the source of an error
- Understanding why it occurs
- Fixing the issue while ensuring the solution doesn't introduce new problems

### The Debugging Process

1. **Identify the bug**: Recognize that something isn't working correctly
2. **Reproduce the bug**: Find a reliable way to make the error occur
3. **Isolate the problem**: Narrow down where in the code the issue might be
4. **Analyze the code**: Understand what's happening vs. what should be happening
5. **Fix the bug**: Make the necessary code changes
6. **Test the solution**: Verify the fix works without breaking anything else

## 2. Using Print Statements

One of the simplest and most widely used debugging techniques is adding `print()` statements to your code. This allows you to:

- Display variable values at specific points in execution
- Trace program flow
- Confirm whether certain code blocks are being executed

Let's see this in practice:

```python
def calculate_average(numbers):
    print(f"Function called with: {numbers}")
    
    if not numbers:
        print("Empty list detected, returning 0")
        return 0
    
    total = sum(numbers)
    print(f"Sum calculated: {total}")
    
    average = total / len(numbers)
    print(f"Average calculated: {average}")
    
    return average

# Test with different inputs
print("Test 1:")
calculate_average([1, 2, 3, 4, 5])

print("\nTest 2:")
calculate_average([])

print("\nTest 3:")
try:
    calculate_average("not a list")
except Exception as e:
    print(f"Error occurred: {e}")
```

### Formatting Print Output

When debugging complex data structures, it's helpful to format the output to make it more readable:

```python
import json
from pprint import pprint

def debug_print(data, label=None):
    """Helper function to print debug information in different formats"""
    if label:
        print(f"\n--- {label} ---")
    
    # For dictionaries and complex structures
    if isinstance(data, (dict, list)) and len(str(data)) > 100:
        print("Pretty print:")
        pprint(data)
        print("\nJSON format:")
        print(json.dumps(data, indent=2))
    else:
        print(f"Value: {data}")
        print(f"Type: {type(data)}")
        if hasattr(data, "__len__"):
            print(f"Length: {len(data)}")

# Test with different data types
debug_print(42, "Integer")
debug_print("Hello, world!", "String")
debug_print([1, 2, 3, 4, 5], "List")
debug_print({"name": "John", "age": 30, "city": "New York", 
             "skills": ["Python", "JavaScript", "SQL"],
             "job": {"title": "Developer", "level": "Senior"}}, "Dictionary")
```

## 3. Python's Built-in Debugger (pdb)

Python includes a powerful debugger in the standard library called `pdb`. It allows you to:

- Set breakpoints in your code
- Step through code execution line by line
- Inspect variables at any point
- Modify variables during execution
- Evaluate expressions in the current context

Let's look at how to use `pdb`:

```python
import pdb

def complex_calculation(a, b, c):
    first_step = a * b
    second_step = first_step / c
    final_result = second_step ** 2
    return final_result

def debug_example():
    x = 10
    y = 5
    z = 0  # This will cause a division by zero error
    
    # Insert a breakpoint
    pdb.set_trace()  # Execution stops here when running the function
    
    result = complex_calculation(x, y, z)
    print(f"Result: {result}")

# Note: When running this in a notebook, the debugger may behave differently 
# than in a regular Python script. To fully experience pdb, try this in a 
# standalone Python file.

# Commenting out to prevent halting notebook execution
# debug_example()
```

### Common pdb Commands

When in the pdb prompt:

- `h` (help): Display available commands
- `n` (next): Execute the current line and move to the next one (doesn't step into functions)
- `s` (step): Step into functions called at the current line
- `c` (continue): Continue execution until the next breakpoint or end
- `q` (quit): Exit the debugger
- `p expression` (print): Evaluate and print an expression
- `l` (list): Show the current line of code in context
- `w` (where): Show the current call stack
- `b line` (break): Set a breakpoint at the specified line
- `r` (return): Continue execution until the current function returns

Let's see a cleaner example of how to use pdb in a typical debugging scenario:

```python
def find_average_of_positive_numbers(numbers):
    """Calculate the average of positive numbers in a list."""
    positive_nums = [num for num in numbers if num > 0]
    if not positive_nums:
        return 0
    return sum(positive_nums) / len(positive_nums)

def demonstrate_debugger():
    """This function demonstrates how you would use pdb in a normal Python script."""
    test_cases = [
        [1, 2, 3, 4, 5],
        [-1, -2, 0, 3, 4],
        [],
        "not a list"  # This will cause an error
    ]
    
    results = []
    for i, test in enumerate(test_cases):
        print(f"Processing test case {i+1}: {test}")
        # In a script, you would add: import pdb; pdb.set_trace()
        # Then you could examine variables, step through code, etc.
        try:
            result = find_average_of_positive_numbers(test)
            results.append(result)
        except Exception as e:
            print(f"Error in test case {i+1}: {e}")
    
    print("Results:", results)

demonstrate_debugger()
```

## 4. Using the Breakpoint() Function

Python 3.7 introduced the built-in `breakpoint()` function, which provides a more convenient way to enter the debugger. It's essentially a shorthand for `import pdb; pdb.set_trace()`.

By default, `breakpoint()` enters the pdb debugger, but you can configure it to use other debuggers by setting the `PYTHONBREAKPOINT` environment variable.

```python
def calculate_fibonacci(n):
    """Calculate the nth Fibonacci number."""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    
    # In normal Python script, you would add a breakpoint here
    # breakpoint()  
    
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Example usage
for i in range(10):
    result = calculate_fibonacci(i)
    print(f"Fibonacci({i}) = {result}")
```

### Configuring Breakpoint Behavior

You can customize the behavior of `breakpoint()` using environment variables:

1. **Disable debugger**: Set `PYTHONBREAKPOINT=0` to make `breakpoint()` do nothing
2. **Use a different debugger**: Set `PYTHONBREAKPOINT=module.function` to use a custom debugging function
3. **Advanced configurations**: Different debuggers might have their own environment variables

This flexibility makes `breakpoint()` a powerful tool in your debugging arsenal.

## 5. Exception Handling for Debugging

Exception handling is not just for making your code robust; it's also a powerful debugging technique. By catching exceptions and examining their details, you can gain insights into what's going wrong.

```python
def debug_with_exceptions(func, *args, **kwargs):
    """
    Executes a function and provides detailed exception information if it fails.
    """
    try:
        result = func(*args, **kwargs)
        print(f"Function executed successfully!")
        print(f"Result: {result}")
        return result
    except Exception as e:
        import traceback
        import sys
        
        print(f"Exception type: {type(e).__name__}")
        print(f"Exception message: {e}")
        print("\nTraceback:")
        traceback.print_exc()
        
        # Get detailed information about the exception
        exc_type, exc_value, exc_traceback = sys.exc_info()
        print("\nException Details:")
        print(f"  File: {exc_traceback.tb_frame.f_code.co_filename}")
        print(f"  Function: {exc_traceback.tb_frame.f_code.co_name}")
        print(f"  Line: {exc_traceback.tb_lineno}")
        
        # Print local variables at the point of exception
        print("\nLocal variables at the point of exception:")
        if exc_traceback.tb_frame.f_locals:
            for key, value in exc_traceback.tb_frame.f_locals.items():
                print(f"  {key} = {value}")

# Test function with a deliberate error
def divide_numbers(a, b):
    return a / b

# Test our debug wrapper
debug_with_exceptions(divide_numbers, 10, 2)  # Should work
print("\n")
debug_with_exceptions(divide_numbers, 10, 0)  # Should fail with ZeroDivisionError
```

### Creating Custom Exceptions for Better Debugging

Custom exceptions can provide more context and make debugging easier:

```python
class ValidationError(Exception):
    """Exception raised for validation errors in input."""
    def __init__(self, message, value=None, expected_type=None):
        self.message = message
        self.value = value
        self.expected_type = expected_type
        super().__init__(self.message)
    
    def __str__(self):
        base_msg = self.message
        if self.value is not None:
            base_msg += f"\n- Received value: {self.value}"
            base_msg += f"\n- Value type: {type(self.value).__name__}"
        if self.expected_type is not None:
            base_msg += f"\n- Expected type: {self.expected_type.__name__}"
        return base_msg

def process_age(age):
    if not isinstance(age, int):
        raise ValidationError("Age must be an integer", age, int)
    if age < 0 or age > 120:
        raise ValidationError(f"Age must be between 0 and 120", age)
    return f"Processing age: {age}"

# Test with different inputs
try:
    print(process_age(25))
except ValidationError as e:
    print(f"Validation Error: {e}")

try:
    print(process_age("twenty"))
except ValidationError as e:
    print(f"Validation Error: {e}")

try:
    print(process_age(150))
except ValidationError as e:
    print(f"Validation Error: {e}")
```

## 6. Logging for Debugging

While print statements are useful, a more structured approach to tracking program execution is using Python's `logging` module. It offers several advantages:

- Different log levels (DEBUG, INFO, WARNING, ERROR, CRITICAL)
- Formatted output with timestamps
- Ability to log to different destinations (console, files, network)
- Can be enabled/disabled without changing code

```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# Create a logger for this module
logger = logging.getLogger('debugging_demo')

def calculate_statistics(data):
    """Calculate basic statistics for a list of numbers."""
    logger.debug(f"calculate_statistics called with data: {data}")
    
    if not data:
        logger.warning("Empty data list provided")
        return {
            "count": 0,
            "sum": 0,
            "mean": None,
            "min": None,
            "max": None
        }
    
    try:
        # Calculate statistics
        count = len(data)
        total = sum(data)
        mean = total / count
        minimum = min(data)
        maximum = max(data)
        
        logger.info(f"Successfully calculated statistics for {count} items")
        logger.debug(f"Sum: {total}, Mean: {mean}, Min: {minimum}, Max: {maximum}")
        
        return {
            "count": count,
            "sum": total,
            "mean": mean,
            "min": minimum,
            "max": maximum
        }
    except Exception as e:
        logger.error(f"Error calculating statistics: {e}", exc_info=True)
        raise

# Test with different inputs
test_data = [
    [1, 2, 3, 4, 5],
    [],
    [10],
    [-1, -5, 20, 15]
]

for data in test_data:
    print(f"\nProcessing: {data}")
    try:
        result = calculate_statistics(data)
        print(f"Result: {result}")
    except Exception as e:
        print(f"An error occurred: {e}")
        
# Try with invalid data
try:
    print("\nProcessing invalid data:")
    calculate_statistics(["a", "b", "c"])
except Exception as e:
    print(f"Expected error with invalid data: {e}")
```

### Advanced Logging Techniques

For more complex applications, you might want to configure multiple loggers with different handlers:

```python
import logging
import io
import sys

# Create a string buffer to capture log output for demonstration
log_capture = io.StringIO()

# Create and configure loggers
def setup_logger(name, level=logging.INFO):
    """Set up a logger with console and string buffer handlers."""
    # Create logger
    logger = logging.getLogger(name)
    logger.setLevel(level)
    logger.propagate = False  # Don't pass messages to ancestor loggers
    
    # Clear any existing handlers
    logger.handlers = []
    
    # Create console handler
    console = logging.StreamHandler(sys.stdout)
    console.setLevel(level)
    
    # Create string buffer handler for our demonstration
    string_handler = logging.StreamHandler(log_capture)
    string_handler.setLevel(level)
    
    # Create formatters
    detailed_formatter = logging.Formatter(
        '%(asctime)s - %(name)s - %(levelname)s - %(filename)s:%(lineno)d - %(message)s'
    )
    simple_formatter = logging.Formatter('%(levelname)s: %(message)s')
    
    # Add formatters to handlers
    console.setFormatter(simple_formatter)
    string_handler.setFormatter(detailed_formatter)
    
    # Add handlers to logger
    logger.addHandler(console)
    logger.addHandler(string_handler)
    
    return logger

# Set up loggers for different components
main_logger = setup_logger('app.main', logging.INFO)
data_logger = setup_logger('app.data', logging.DEBUG)
utils_logger = setup_logger('app.utils', logging.WARNING)

def process_data(items):
    """A function that processes data items."""
    main_logger.info(f"Processing {len(items)} items")
    
    results = []
    for i, item in enumerate(items):
        data_logger.debug(f"Processing item {i}: {item}")
        
        try:
            # Simulate some processing
            if isinstance(item, (int, float)):
                result = item * 2
                data_logger.debug(f"Calculated result: {result}")
            else:
                utils_logger.warning(f"Non-numeric item encountered: {item}")
                result = str(item) + "_processed"
            
            results.append(result)
            
        except Exception as e:
            utils_logger.error(f"Error processing item {i}: {e}", exc_info=True)
    
    main_logger.info("Processing complete")
    return results

# Test our logging setup
mixed_data = [1, 5.5, "hello", 10, None, 7]
results = process_data(mixed_data)
print(f"\nResults: {results}")

# Display captured logs
print("\n--- Captured Log Output ---")
print(log_capture.getvalue())
```

## 8. Debugging with Assertions

Assertions are statements that check if a condition is True, raising an `AssertionError` if it's False. They are a way to catch logical errors and validate assumptions during development.

Key points about assertions:

- They are development tools, not meant for production error handling
- They can be disabled when running Python with the `-O` optimization flag
- They help catch bugs early by ensuring your code's assumptions hold true

```python
def calculate_discount(price, discount_rate):
    """Calculate the discounted price.
    
    Args:
        price: Original price (positive number)
        discount_rate: Discount as a percentage (0-100)
    
    Returns:
        The price after applying the discount
    """
    # Validate inputs using assertions
    assert isinstance(price, (int, float)), f"Price must be a number, got {type(price)}"
    assert price >= 0, f"Price must be non-negative, got {price}"
    assert isinstance(discount_rate, (int, float)), f"Discount must be a number, got {type(discount_rate)}"
    assert 0 <= discount_rate <= 100, f"Discount must be between 0 and 100, got {discount_rate}"
    
    # Calculate the discount
    discount_amount = price * (discount_rate / 100)
    discounted_price = price - discount_amount
    
    # Verify the result is logical
    assert discounted_price <= price, "Discounted price should be less than or equal to original price"
    
    return discounted_price

# Test with valid inputs
try:
    print(f"$100 with 20% discount: ${calculate_discount(100, 20)}")
    print(f"$50 with 0% discount: ${calculate_discount(50, 0)}")
    print(f"$75.50 with 10% discount: ${calculate_discount(75.50, 10)}")
except AssertionError as e:
    print(f"Assertion failed: {e}")

# Test with invalid inputs
print("\nTesting invalid inputs:")
for test_case in [
    ("Price as string", ("100", 20)),
    ("Negative price", (-50, 10)),
    ("Discount as string", (100, "20")),
    ("Discount above 100%", (100, 150)),
    ("Discount below 0%", (100, -10))
]:
    print(f"\nTest: {test_case[0]}")
    try:
        result = calculate_discount(*test_case[1])
        print(f"Result: {result}")
    except AssertionError as e:
        print(f"Assertion failed as expected: {e}")
```

## 9. Common Bugs and How to Find Them

Let's explore common types of bugs in Python and specific techniques to identify and fix them.

```python
def demonstrate_common_bugs():
    """Demonstrate common Python bugs and how to find them."""
    print("Common Python bugs and how to debug them:")
    
    # 1. Off-by-one errors
    print("\n1. Off-by-one errors:")
    def print_first_n_numbers(n):
        print(f"Attempting to print first {n} numbers:")
        # Bug: This will print n+1 numbers (0 through n)
        for i in range(n + 1):  # should be range(n)
            print(i, end=' ')
        print()
    
    print("- Buggy version:")
    print_first_n_numbers(5)
    
    # Fix:
    def print_first_n_numbers_fixed(n):
        print(f"First {n} numbers (fixed):")
        for i in range(n):  # Correct range
            print(i + 1, end=' ')  # Start from 1
        print()
    
    print("- Fixed version:")
    print_first_n_numbers_fixed(5)
    
    # 2. Mutable default arguments
    print("\n2. Mutable default arguments:")
    
    def add_item_buggy(item, items=[]):
        items.append(item)
        return items
    
    print("- Buggy version:")
    print("First call:", add_item_buggy(1))
    print("Second call:", add_item_buggy(2))  # Will contain both 1 and 2!
    print("Third call:", add_item_buggy(3))   # Will contain 1, 2, and 3!
    
    # Fix:
    def add_item_fixed(item, items=None):
        if items is None:
            items = []
        items.append(item)
        return items
    
    print("- Fixed version:")
    print("First call:", add_item_fixed(1))
    print("Second call:", add_item_fixed(2))  # Only contains 2
    print("Third call:", add_item_fixed(3))   # Only contains 3
    
    # 3. Incorrect string comparison
    print("\n3. Incorrect string comparison:")
    def is_yes_buggy(response):
        # Bug: This only checks for "YES" in uppercase
        return response == "YES"
    
    # Fix:
    def is_yes_fixed(response):
        # This checks for "yes" in any case
        return response.upper() == "YES"
    
    test_inputs = ["YES", "yes", "Yes", "yEs"]
    
    print("- Buggy version:")
    for input_str in test_inputs:
        print(f"'{input_str}' is yes? {is_yes_buggy(input_str)}")
    
    print("- Fixed version:")
    for input_str in test_inputs:
        print(f"'{input_str}' is yes? {is_yes_fixed(input_str)}")
    
    # 4. Improper error handling
    print("\n4. Improper error handling:")
    
    def divide_buggy(a, b):
        # Bug: No error handling for division by zero
        return a / b
    
    # Fix:
    def divide_fixed(a, b):
        try:
            return a / b
        except ZeroDivisionError:
            print("Error: Division by zero")
            return None
    
    print("- Buggy version:")
    try:
        print(f"10 / 2 = {divide_buggy(10, 2)}")
        print(f"10 / 0 = {divide_buggy(10, 0)}")  # Will crash
    except ZeroDivisionError as e:
        print(f"Error caught: {e}")
    
    print("- Fixed version:")
    print(f"10 / 2 = {divide_fixed(10, 2)}")
    print(f"10 / 0 = {divide_fixed(10, 0)}")  # Will handle the error

# Run the demonstration
demonstrate_common_bugs()
```

### More Complex Bug Scenarios

Let's look at some more subtle bugs that are harder to detect and debug:

```python
import time
import random

def complex_bug_scenarios():
    """Demonstrate more complex bugs that are harder to detect."""
    
    # 1. Shared state between function calls
    print("\n1. Shared state bugs:")
    
    class UserManager:
        active_users = []  # Class variable shared across all instances
        
        def add_user(self, username):
            self.active_users.append(username)
            return f"Added user: {username}"
        
        def list_users(self):
            return self.active_users
    
    print("- Buggy implementation with shared state:")
    manager1 = UserManager()
    manager2 = UserManager()
    
    print("Manager 1 adds 'Alice':", manager1.add_user("Alice"))
    print("Manager 2 adds 'Bob':", manager2.add_user("Bob"))
    
    print("Manager 1 users:", manager1.list_users())  # Will show both Alice and Bob
    print("Manager 2 users:", manager2.list_users())  # Will show both Alice and Bob
    
    # Fix: Use instance variables instead
    class FixedUserManager:
        def __init__(self):
            self.active_users = []  # Instance variable
        
        def add_user(self, username):
            self.active_users.append(username)
            return f"Added user: {username}"
        
        def list_users(self):
            return self.active_users
    
    print("\n- Fixed implementation with instance variables:")
    fixed_manager1 = FixedUserManager()
    fixed_manager2 = FixedUserManager()
    
    print("Manager 1 adds 'Alice':", fixed_manager1.add_user("Alice"))
    print("Manager 2 adds 'Bob':", fixed_manager2.add_user("Bob"))
    
    print("Manager 1 users:", fixed_manager1.list_users())  # Only Alice
    print("Manager 2 users:", fixed_manager2.list_users())  # Only Bob
    
    
    # 2. Race condition bug (simulated)
    print("\n2. Race condition bugs:")
    
    balance = 100
    
    def withdraw(amount):
        nonlocal balance
        if balance >= amount:
            # Simulate some processing time where another thread could intervene
            time.sleep(0.1)
            balance -= amount
            return True
        return False
    
    print("- Simulating race condition:")
    print(f"Initial balance: {balance}")
    
    # Simulate concurrent withdrawals
    print("Withdraw 60:", withdraw(60))
    print(f"Balance after first withdrawal: {balance}")
    print("Withdraw 60:", withdraw(60))
    print(f"Final balance: {balance}")
    
    print("\n- In a real concurrent scenario, both withdrawals might succeed")
    print("  even though there's not enough money for both, leading to a negative balance.")
    
    
    # 3. Floating point precision issues
    print("\n3. Floating point precision bugs:")
    
    def sum_to_one():
        """Add 0.1 ten times, which should equal 1.0"""
        total = 0
        for _ in range(10):
            total += 0.1
        return total
    
    result = sum_to_one()
    print(f"0.1 + 0.1 + ... (10 times) = {result}")
    print(f"Result == 1.0: {result == 1.0}")
    print(f"Rounded result: {round(result, 10)}")
    
    # Fix: Use decimal for financial calculations
    from decimal import Decimal
    
    def decimal_sum_to_one():
        """Add 0.1 ten times using Decimal for precision"""
        total = Decimal('0')
        for _ in range(10):
            total += Decimal('0.1')
        return total
    
    decimal_result = decimal_sum_to_one()
    print(f"\nUsing Decimal: {decimal_result}")
    print(f"Decimal result == 1.0: {decimal_result == Decimal('1.0')}")

# Run complex bug demonstrations
complex_bug_scenarios()
```

## 10. Debugging Best Practices

Let's conclude with some best practices for effective debugging:

### 1. Understand the Problem Before Fixing

Before attempting to fix a bug:
- Reproduce the issue reliably
- Understand what the code is supposed to do
- Identify what it's actually doing
- Look for patterns in when the bug occurs

### 2. Isolate the Problem

- Use binary search methodology: Comment out half the code to see if the bug persists
- Create a minimal reproducible example
- Test components individually

### 3. Check Your Assumptions

- Don't assume your understanding of how something works is correct
- Verify variable types, values, and function behaviors
- Test API behaviors with simple examples

### 4. Use the Scientific Method

1. Observe the bug
2. Form a hypothesis about what's causing it
3. Test your hypothesis
4. Analyze the results and refine your hypothesis

### 5. Tools and Techniques

- Use appropriate tools for the job (print statements, debugger, logging)
- Add instrumentation to see what's happening inside the code
- Use version control to track changes and isolate when bugs were introduced

### 6. Documentation

- Document bugs and their fixes
- Write tests to prevent regression
- Update documentation based on discoveries

### 7. Prevention

- Write testable code
- Use defensive programming
- Add assertions for important assumptions
- Follow coding standards
- Code reviews
