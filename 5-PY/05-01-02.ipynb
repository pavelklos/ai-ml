# Python Variables, Data Types, and Basic Operations

This notebook provides a comprehensive introduction to Python variables, data types, and basic operations. We'll explore how variables work in Python, the various built-in data types, and the operations you can perform on them.

## Introduction to Python Variables

In Python, variables are used to store data values. Unlike statically-typed languages, Python is dynamically typed, which means you don't need to declare variable types explicitly - the interpreter infers the type based on the assigned value.

````python
# Creating variables in Python
name = "Alice"
age = 30
height = 5.7
is_student = False

# Printing variables
print(name)
print(age)
print(height)
print(is_student)
````

Python variables are essentially references or pointers to objects in memory. When you create a variable, Python:
1. Creates an object to represent the value
2. Determines the appropriate data type
3. Sets the variable name to point to that object in memory

Let's check the type and memory location of our variables:

````python
# Checking variable types
print(f"Type of name: {type(name)}")
print(f"Type of age: {type(age)}")
print(f"Type of height: {type(height)}")
print(f"Type of is_student: {type(is_student)}")

# Memory location (id) of variables
print(f"Memory address of name: {id(name)}")
print(f"Memory address of age: {id(age)}")
````

Python variables can be reassigned to different values and different types:

````python
x = 10        # x is an integer
print(f"x = {x}, type: {type(x)}")

x = "hello"   # x is now a string
print(f"x = {x}, type: {type(x)}")

x = [1, 2, 3] # x is now a list
print(f"x = {x}, type: {type(x)}")
````

Variables can also share references to the same object:

````python
a = [1, 2, 3]
b = a  # b references the same list as a

print(f"a: {a}, id(a): {id(a)}")
print(f"b: {b}, id(b): {id(b)}")

# Modifying a also modifies b because they reference the same object
a.append(4)
print(f"After modifying a: a = {a}, b = {b}")
````

## Numeric Data Types

Python has several built-in numeric data types:

1. **Integers (int)**: Whole numbers without decimal places
2. **Floating-point numbers (float)**: Numbers with decimal places
3. **Complex numbers (complex)**: Numbers with real and imaginary parts

````python
# Integers
a = 10
b = -5
c = 0
d = 10_000_000  # Underscores for readability in large numbers

print(f"a = {a}, type: {type(a)}")
print(f"b = {b}, type: {type(b)}")
print(f"c = {c}, type: {type(c)}")
print(f"d = {d}, type: {type(d)}")

# Integer conversions - decimal, binary, octal, and hexadecimal
decimal_num = 42
print(f"Decimal: {decimal_num}")
print(f"Binary: {bin(decimal_num)}")
print(f"Octal: {oct(decimal_num)}")
print(f"Hexadecimal: {hex(decimal_num)}")
````

````python
# Floating point numbers
a = 3.14
b = -0.001
c = 2.0
d = 1.23e6  # Scientific notation: 1.23 × 10^6

print(f"a = {a}, type: {type(a)}")
print(f"b = {b}, type: {type(b)}")
print(f"c = {c}, type: {type(c)}")
print(f"d = {d}, type: {type(d)}")

# Floating point precision
print(0.1 + 0.2)  # Not exactly 0.3 due to floating point precision limitations
print((0.1 + 0.2) == 0.3)  # False
````

````python
# Complex numbers (real + imaginary)
a = 3 + 4j
b = complex(2, 5)  # 2 + 5j

print(f"a = {a}, type: {type(a)}")
print(f"b = {b}, type: {type(b)}")

# Real and imaginary parts
print(f"Real part of a: {a.real}")
print(f"Imaginary part of a: {a.imag}")

# Complex operations
print(f"Absolute value (magnitude) of a: {abs(a)}")
print(f"Conjugate of a: {a.conjugate()}")
````

## String Data Type and Operations

Strings in Python are sequences of characters enclosed in quotes (single, double, or triple quotes). They are immutable, meaning their values cannot be changed after creation.

````python
# Creating strings
s1 = 'Hello'               # Single quotes
s2 = "World"               # Double quotes
s3 = '''Multiline
string'''                  # Triple quotes for multi-line strings
s4 = """Another
multiline string"""        # Triple double quotes also work

print(s1)
print(s2)
print(s3)
print(s4)
````

````python
# String indexing and slicing
text = "Python Programming"

# Indexing (0-based)
print(f"First character: {text[0]}")
print(f"Fifth character: {text[4]}")
print(f"Last character: {text[-1]}")  # Negative index counts from the end

# Slicing [start:end:step] - end is exclusive
print(f"First five characters: {text[0:5]}")
print(f"From index 7 to the end: {text[7:]}")
print(f"From beginning to index 6: {text[:6]}")
print(f"Every second character: {text[::2]}")
print(f"Reversed string: {text[::-1]}")
````

````python
# Common string operations and methods
s = "hello world"

# Length
print(f"Length of '{s}': {len(s)}")

# Case conversion
print(f"Uppercase: {s.upper()}")
print(f"Lowercase: {s.lower()}")
print(f"Title case: {s.title()}")
print(f"Capitalize: {s.capitalize()}")

# Checking content
print(f"Starts with 'hello': {s.startswith('hello')}")
print(f"Ends with 'world': {s.endswith('world')}")
print(f"Contains 'lo wo': {'lo wo' in s}")

# Finding substrings
print(f"Position of 'world': {s.find('world')}")
print(f"Count of 'l': {s.count('l')}")

# Removing whitespace
s_with_spaces = "   text with spaces   "
print(f"Original: '{s_with_spaces}'")
print(f"Left strip: '{s_with_spaces.lstrip()}'")
print(f"Right strip: '{s_with_spaces.rstrip()}'")
print(f"Strip both sides: '{s_with_spaces.strip()}'")
````

````python
# String concatenation and repetition
first = "Hello"
last = "World"

# Concatenation using + operator
full = first + " " + last
print(full)

# String repetition using * operator
separator = "-" * 20
print(separator)

# Joining strings
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(sentence)
````

````python
# String formatting
name = "Alice"
age = 30

# Using % formatting (older style)
print("My name is %s and I am %d years old." % (name, age))

# Using str.format() method
print("My name is {} and I am {} years old.".format(name, age))
print("My name is {0} and I am {1} years old.".format(name, age))
print("My name is {n} and I am {a} years old.".format(n=name, a=age))

# Using f-strings (Python 3.6+) - most modern and recommended
print(f"My name is {name} and I am {age} years old.")
print(f"My name is {name.upper()} and next year I'll be {age + 1} years old.")

# Formatting numbers in f-strings
pi = 3.14159
print(f"Pi to 2 decimal places: {pi:.2f}")
print(f"Pi in scientific notation: {pi:.2e}")
print(f"Percentage: {0.25:.0%}")
````

## Boolean Data Type

Boolean values in Python are either `True` or `False`. They're often used in conditional expressions and logical operations.

````python
# Boolean values
x = True
y = False

print(f"x = {x}, type: {type(x)}")
print(f"y = {y}, type: {type(y)}")

# Boolean operations
print(f"NOT x: {not x}")
print(f"x AND y: {x and y}")
print(f"x OR y: {x or y}")
````

````python
# Converting other types to boolean
# Use the bool() function

# Numbers - 0 is False, any other number is True
print(f"bool(0) = {bool(0)}")
print(f"bool(1) = {bool(1)}")
print(f"bool(-10) = {bool(-10)}")

# Strings - empty string is False, any other string is True
print(f"bool('') = {bool('')}")
print(f"bool('hello') = {bool('hello')}")

# Collections - empty collections are False, non-empty are True
print(f"bool([]) = {bool([])}")
print(f"bool([1, 2]) = {bool([1, 2])}")
print(f"bool({{}}) = {bool({})}")
print(f"bool(set()) = {bool(set())}")

# None is False
print(f"bool(None) = {bool(None)}")
````

````python
# Boolean expressions in conditional statements
x = 10
y = 5

if x > y:
    print("x is greater than y")
else:
    print("x is not greater than y")

# Truthy and falsy values with if statements
name = ""
if name:  # Empty string is falsy
    print(f"Hello, {name}!")
else:
    print("Name is empty")

numbers = [1, 2, 3]
if numbers:  # Non-empty list is truthy
    print(f"The list contains {len(numbers)} elements")
else:
    print("The list is empty")
````

## Collection Data Types

Python offers several built-in collection data types for storing multiple items:

1. **Lists**: Ordered, mutable collections (can be modified after creation)
2. **Tuples**: Ordered, immutable collections
3. **Dictionaries**: Unordered collections of key-value pairs
4. **Sets**: Unordered collections of unique elements

````python
# Lists - ordered, mutable collections
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, [1, 2]]  # Lists can contain different data types

print(f"fruits: {fruits}, type: {type(fruits)}")
print(f"mixed: {mixed}, type: {type(mixed)}")

# Accessing elements
print(f"First fruit: {fruits[0]}")
print(f"Last number: {numbers[-1]}")
print(f"Nested list element: {mixed[4][1]}")  # Access the second element of the nested list

# Modifying lists
fruits.append("orange")  # Add to the end
print(f"After append: {fruits}")

fruits.insert(1, "mango")  # Insert at specific position
print(f"After insert: {fruits}")

fruits.remove("banana")  # Remove specific element
print(f"After remove: {fruits}")

popped = fruits.pop()  # Remove and return last element
print(f"Popped element: {popped}")
print(f"After pop: {fruits}")

# List operations
combined = fruits + numbers  # Concatenation
print(f"Combined list: {combined}")

repeated = fruits * 2  # Repetition
print(f"Repeated list: {repeated}")

# List comprehensions - powerful way to create lists
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

even_numbers = [x for x in range(10) if x % 2 == 0]
print(f"Even numbers: {even_numbers}")
````

````python
# Tuples - ordered, immutable collections
coordinates = (10, 20)
person = ("John", 30, "New York")
singleton = (42,)  # Note the comma for a single-element tuple

print(f"coordinates: {coordinates}, type: {type(coordinates)}")
print(f"person: {person}, type: {type(person)}")

# Accessing elements (same as lists)
print(f"X-coordinate: {coordinates[0]}")
print(f"Person's name: {person[0]}")

# Trying to modify a tuple raises an error
try:
    coordinates[0] = 15
except TypeError as e:
    print(f"Error: {e}")

# Tuple unpacking
name, age, city = person
print(f"Unpacked: name={name}, age={age}, city={city}")

# Tuple operations
combined = coordinates + person  # Concatenation
print(f"Combined tuple: {combined}")

repeated = coordinates * 3  # Repetition
print(f"Repeated tuple: {repeated}")
````

````python
# Dictionaries - key-value pairs
person = {
    "name": "John",
    "age": 30,
    "city": "New York",
    "is_student": False
}

print(f"person: {person}, type: {type(person)}")

# Accessing values using keys
print(f"Name: {person['name']}")
print(f"Age: {person['age']}")

# Alternative access method with .get() (safer if key might not exist)
print(f"Occupation: {person.get('occupation', 'Not specified')}")

# Modifying dictionaries
person["age"] = 31  # Update existing key
person["occupation"] = "Engineer"  # Add new key-value pair
print(f"Updated person: {person}")

# Dictionary methods
keys = person.keys()
values = person.values()
items = person.items()  # Returns key-value pairs as tuples

print(f"Keys: {keys}")
print(f"Values: {values}")
print(f"Items: {items}")

# Removing items
removed_age = person.pop("age")
print(f"Removed age: {removed_age}")
print(f"After removal: {person}")

# Dictionary comprehensions
squares_dict = {x: x**2 for x in range(1, 6)}
print(f"Squares dictionary: {squares_dict}")
````

````python
# Sets - unordered collections of unique elements
fruits = {"apple", "banana", "cherry", "apple"}  # Duplicate "apple" will be ignored
numbers = set([1, 2, 3, 2, 1])  # Creating a set from a list

print(f"fruits: {fruits}, type: {type(fruits)}")
print(f"numbers: {numbers}, type: {type(numbers)}")

# Set operations
fruits.add("orange")
print(f"After add: {fruits}")

fruits.remove("banana")  # Raises an error if element doesn't exist
print(f"After remove: {fruits}")

fruits.discard("mango")  # No error if element doesn't exist
print(f"After discard: {fruits}")

# Mathematical set operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Union: {set_a | set_b}")  # Or set_a.union(set_b)
print(f"Intersection: {set_a & set_b}")  # Or set_a.intersection(set_b)
print(f"Difference (a-b): {set_a - set_b}")  # Or set_a.difference(set_b)
print(f"Symmetric difference: {set_a ^ set_b}")  # Or set_a.symmetric_difference(set_b)

# Set comprehensions
even_set = {x for x in range(10) if x % 2 == 0}
print(f"Even numbers set: {even_set}")
````

## Type Conversion

Python allows both implicit and explicit conversion between different data types.

````python
# Explicit type conversion (type casting)

# Converting to integers
print(f"int('100') = {int('100')}")
print(f"int(12.34) = {int(12.34)}")  # Truncates (doesn't round)
print(f"int(True) = {int(True)}")
print(f"int(False) = {int(False)}")

# Converting to floats
print(f"float('12.34') = {float('12.34')}")
print(f"float(100) = {float(100)}")
print(f"float(True) = {float(True)}")

# Converting to strings
print(f"str(100) = {str(100)}")
print(f"str(12.34) = {str(12.34)}")
print(f"str(True) = {str(True)}")

# Converting to boolean
print(f"bool(1) = {bool(1)}")
print(f"bool(0) = {bool(0)}")
print(f"bool('hello') = {bool('hello')}")
print(f"bool('') = {bool('')}")

# Converting to lists, tuples, and sets
print(f"list('hello') = {list('hello')}")
print(f"tuple([1, 2, 3]) = {tuple([1, 2, 3])}")
print(f"set([1, 2, 2, 3, 3, 3]) = {set([1, 2, 2, 3, 3, 3])}")
````

````python
# Implicit type conversion (coercion)

# Integer and float in arithmetic operations
result = 10 + 3.5  # Integer 10 is implicitly converted to float
print(f"10 + 3.5 = {result}, type: {type(result)}")

# With strings, + means concatenation, not addition
try:
    result = "10" + 5  # This will raise a TypeError
except TypeError as e:
    print(f"Error: {e}")
    
    # Correct way is to explicitly convert
    result = "10" + str(5)
    print(f"'10' + str(5) = {result}")
    
    # Or convert string to number
    result = int("10") + 5
    print(f"int('10') + 5 = {result}")
````

## Basic Arithmetic Operations

Python provides a full set of arithmetic operators for mathematical calculations.

````python
a = 10
b = 3

# Basic arithmetic operations
print(f"{a} + {b} = {a + b}")  # Addition
print(f"{a} - {b} = {a - b}")  # Subtraction
print(f"{a} * {b} = {a * b}")  # Multiplication
print(f"{a} / {b} = {a / b}")  # Division (always returns float)
print(f"{a} // {b} = {a // b}")  # Integer division (floor division)
print(f"{a} % {b} = {a % b}")  # Modulus (remainder)
print(f"{a} ** {b} = {a ** b}")  # Exponentiation (a to the power of b)

# Order of operations follows mathematical precedence (PEMDAS)
# Parentheses, Exponents, Multiplication/Division, Addition/Subtraction
result = 10 + 3 * 2  # Multiplication happens before addition
print(f"10 + 3 * 2 = {result}")

result = (10 + 3) * 2  # Parentheses change the order
print(f"(10 + 3) * 2 = {result}")

# Complex expression
result = 2 + 3 * 4 ** 2 - 6 / 2
print(f"2 + 3 * 4 ** 2 - 6 / 2 = {result}")

# Breaking it down:
# 1. 4 ** 2 = 16 (exponentiation first)
# 2. 3 * 16 = 48 (multiplication next)
# 3. 6 / 2 = 3.0 (division next)
# 4. 2 + 48 - 3.0 = 47.0 (addition and subtraction last, from left to right)
````

````python
# Shorthand assignment operators
x = 10

# Instead of x = x + 5
x += 5
print(f"After x += 5, x = {x}")

# Other shorthand operators
x -= 3  # x = x - 3
print(f"After x -= 3, x = {x}")

x *= 2  # x = x * 2
print(f"After x *= 2, x = {x}")

x /= 4  # x = x / 4
print(f"After x /= 4, x = {x}")

x //= 2  # x = x // 2
print(f"After x //= 2, x = {x}")

x %= 3  # x = x % 3
print(f"After x %= 3, x = {x}")

x **= 3  # x = x ** 3
print(f"After x **= 3, x = {x}")
````

````python
# Working with built-in math functions
import math

# Basic functions
print(f"Absolute value of -10: {abs(-10)}")
print(f"Rounded 3.7: {round(3.7)}")
print(f"Rounded 3.7 to 0 decimal places: {round(3.7, 0)}")
print(f"Rounded 3.14159 to 3 decimal places: {round(3.14159, 3)}")

# Functions from the math module
print(f"Square root of 16: {math.sqrt(16)}")
print(f"Floor of 3.7: {math.floor(3.7)}")
print(f"Ceiling of 3.2: {math.ceil(3.2)}")
print(f"Sin of 0: {math.sin(0)}")
print(f"Cos of 0: {math.cos(0)}")
print(f"Pi: {math.pi}")
print(f"e: {math.e}")
````

## Comparison and Logical Operations

Comparison operators compare values and return boolean results. Logical operators combine boolean expressions.

````python
# Comparison operators
a = 10
b = 5
c = 10

print(f"{a} > {b}: {a > b}")    # Greater than
print(f"{a} < {b}: {a < b}")    # Less than
print(f"{a} >= {c}: {a >= c}")  # Greater than or equal to
print(f"{a} <= {b}: {a <= b}")  # Less than or equal to
print(f"{a} == {c}: {a == c}")  # Equal to
print(f"{a} != {b}: {a != b}")  # Not equal to

# Comparing different data types
print(f"5 == '5': {5 == '5'}")  # False - different types
print(f"0 == False: {0 == False}")  # True - special case
print(f"1 == True: {1 == True}")    # True - special case

# Identity comparison (is) - checks if objects are the same, not just equal
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b: {a == b}")  # True - they have equal values
print(f"a is b: {a is b}")  # False - they are different objects in memory
print(f"a is c: {a is c}")  # True - c references the same object as a

# Identity comparison with None
x = None
print(f"x is None: {x is None}")  # Correct way to check for None
print(f"x == None: {x == None}")  # Works but not recommended
````

````python
# Logical operators

x = True
y = False

print(f"{x} and {y}: {x and y}")  # True if both are True
print(f"{x} or {y}: {x or y}")    # True if at least one is True
print(f"not {x}: {not x}")        # True if x is False, False if x is True

# Chaining logical operators
result = (x or y) and not (x and y)
print(f"(x or y) and not (x and y): {result}")

# Short-circuit evaluation:
# 'and' stops as soon as it finds a False
# 'or' stops as soon as it finds a True
# print(f"False and print('This won't be executed'): {False and print('This won't be executed')}")
# print(f"True or print('This won't be executed'): {True or print('This won't be executed')}")

# Using non-boolean values with logical operators
print(f"0 and 5: {0 and 5}")  # Returns 0 (first falsy value)
print(f"1 and 5: {1 and 5}")  # Returns 5 (last value if all are truthy)
print(f"0 or 5: {0 or 5}")    # Returns 5 (first truthy value)
print(f"3 or 5: {3 or 5}")    # Returns 3 (first truthy value)
````

## Variable Assignment and Naming Conventions

Python provides flexible ways to assign values to variables and has specific conventions for naming them.

````python
# Basic variable assignment
x = 10

# Multiple assignment
a, b, c = 1, 2, 3
print(f"a = {a}, b = {b}, c = {c}")

# Assigning the same value to multiple variables
x = y = z = 0
print(f"x = {x}, y = {y}, z = {z}")

# Swapping variables
a, b = 5, 10
print(f"Before swap: a = {a}, b = {b}")
a, b = b, a
print(f"After swap: a = {a}, b = {b}")

# Unpacking collections
coordinates = (3, 4, 5)
x, y, z = coordinates
print(f"x = {x}, y = {y}, z = {z}")

# Extended unpacking (Python 3.x)
first, *middle, last = [1, 2, 3, 4, 5]
print(f"first = {first}, middle = {middle}, last = {last}")
````

````python
# Exercise 3: String Manipulation
# Practice working with string operations

text = "Python is a powerful programming language"

# 1. Split the string into a list of words
words = text.split()
print(f"Words: {words}")

# 2. Join the words back together with dashes
dashed = "-".join(words)
print(f"Dashed text: {dashed}")

# 3. Convert the original text to uppercase
upper_text = text.upper()
print(f"Uppercase: {upper_text}")

# 4. Replace "powerful" with "versatile"
new_text = text.replace("powerful", "versatile")
print(f"New text: {new_text}")

# 5. Check if the string contains "programming" and "script"
contains_programming = "programming" in text
contains_script = "script" in text
print(f"Contains 'programming': {contains_programming}")
print(f"Contains 'script': {contains_script}")
````

````python
# Exercise 4: Type Conversion Challenge
# Convert between different data types

# 1. Convert a string representation of a list to an actual list
list_str = "[1, 2, 3, 4, 5]"
# This requires using the eval() function (be careful with eval in real code)
actual_list = eval(list_str)
print(f"Converted list: {actual_list}, type: {type(actual_list)}")

# 2. Convert a list of strings to a list of integers
str_numbers = ["10", "20", "30", "40"]
int_numbers = [int(num) for num in str_numbers]
print(f"Integer list: {int_numbers}")

# 3. Convert a tuple of various types to all strings
mixed_tuple = (1, 2.5, True, "hello")
str_tuple = tuple(str(item) for item in mixed_tuple)
print(f"String tuple: {str_tuple}")

# 4. Create a dictionary from two lists (keys and values)
keys = ["a", "b", "c", "d"]
values = [1, 2, 3, 4]
result_dict = dict(zip(keys, values))
print(f"Dictionary from lists: {result_dict}")
````

````python
# Exercise 5: Logical Expressions
# Practice with complex logical expressions

age = 25
income = 45000
has_degree = True
years_experience = 3

# Determine eligibility for different job positions
# Junior position: age < 30 and (has degree or years_experience >= 2)
junior_eligible = age < 30 and (has_degree or years_experience >= 2)

# Senior position: age >= 25 and years_experience >= 5 and has_degree
senior_eligible = age >= 25 and years_experience >= 5 and has_degree

# Management position: age >= 30 and years_experience >= 7 and has_degree and income > 60000
management_eligible = age >= 30 and years_experience >= 7 and has_degree and income > 60000

print(f"Eligible for junior position: {junior_eligible}")
print(f"Eligible for senior position: {senior_eligible}")
print(f"Eligible for management position: {management_eligible}")

# Create a complex expression that combines multiple conditions
special_program = ((age < 30 and income < 50000) or 
                   (has_degree and years_experience >= 5) or
                   (age >= 35 and income < 40000))

print(f"Eligible for special program: {special_program}")
````
