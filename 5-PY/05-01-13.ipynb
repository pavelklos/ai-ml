# Basic OOP Concepts in Python (Classes and Objects)

This notebook covers the fundamental concepts of Object-Oriented Programming (OOP) in Python, focusing on classes and objects.

## 1. Understanding Object-Oriented Programming

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of "objects", which can contain data and code: data in the form of fields (attributes), and code in the form of procedures (methods).

**Key Concepts in OOP:**

1. **Classes**: Blueprints for creating objects
2. **Objects**: Instances of classes
3. **Encapsulation**: Bundling data and methods that work on the data
4. **Inheritance**: Creating new classes from existing ones
5. **Polymorphism**: Different classes can be treated through the same interface

OOP helps organize code, making it more reusable, modular, and easier to maintain.

## 2. Creating Classes in Python

In Python, we define a class using the `class` keyword, followed by the class name and a colon. The class body is indented.

```python
# Basic syntax for defining a class
class Dog:
    # Class attribute (shared by all instances)
    species = "Canis familiaris"
    
    # The __init__ method is called when a new object is created
    # It's similar to a constructor in other languages
    def __init__(self, name, age):
        # Instance attributes (specific to each object)
        self.name = name
        self.age = age
    
    # Instance method
    def description(self):
        return f"{self.name} is {self.age} years old"
    
    # Another instance method
    def speak(self, sound):
        return f"{self.name} says {sound}"
```

Let's break down the components of a class definition:

- **Class Name**: Should follow the CapWords convention (also called CamelCase)
- **Class Attributes**: Defined directly inside the class (not within methods)
- **`__init__` Method**: Special method that initializes a new object (constructor)
- **`self` Parameter**: Represents the instance of the class
- **Instance Attributes**: Defined within methods, specific to each instance
- **Instance Methods**: Functions defined inside the class that operate on instances

## 3. Working with Objects

After defining a class, we can create objects (instances) of that class.

```python
# Creating instances of the Dog class
buddy = Dog("Buddy", 9)
miles = Dog("Miles", 4)

# Accessing instance attributes
print(f"{buddy.name} is {buddy.age} years old.")
print(f"{miles.name} is {miles.age} years old.")

# Accessing class attributes
print(f"Both dogs are {buddy.species}")

# Calling instance methods
print(buddy.description())
print(miles.speak("Woof Woof"))
```

Objects are created by calling the class name followed by parentheses. Arguments required by the `__init__` method are passed into the parentheses.

## 4. Class Attributes vs Instance Attributes

Understanding the difference between class attributes and instance attributes is crucial in Python OOP:

```python
class Car:
    # Class attribute - shared by all instances
    wheels = 4
    
    def __init__(self, make, model, color):
        # Instance attributes - unique to each instance
        self.make = make
        self.model = model
        self.color = color
        
# Create two Car objects
car1 = Car("Toyota", "Corolla", "Blue")
car2 = Car("Honda", "Civic", "Red")

# Accessing class attributes
print(f"Car1 has {car1.wheels} wheels")
print(f"Car2 has {car2.wheels} wheels")

# Changing a class attribute affects all instances
Car.wheels = 3
print(f"After change: Car1 has {car1.wheels} wheels")
print(f"After change: Car2 has {car2.wheels} wheels")

# However, you can override the class attribute for a specific instance
car1.wheels = 6
print(f"Car1 now has {car1.wheels} wheels")
print(f"Car2 still has {car2.wheels} wheels")
# Note this doesn't change the class attribute, only the instance
print(f"Car class still defines {Car.wheels} wheels by default")
```

**Key differences:**

- **Class attributes**:
  - Defined at the class level
  - Shared among all instances
  - Accessed via the class name or instance name
  - Memory efficient for shared data

- **Instance attributes**:
  - Defined in the `__init__` method
  - Unique to each instance
  - Accessed via instance name
  - Stores data specific to an object

## 5. Methods in Python Classes

Python classes can have three types of methods:

1. **Instance Methods**: Most common; operate on instance data
2. **Class Methods**: Operate on class data; decorated with `@classmethod`
3. **Static Methods**: Don't operate on instance or class data; decorated with `@staticmethod`

```python
class Employee:
    # Class attribute
    company = "Tech Corp"
    employee_count = 0
    
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.employee_count += 1
    
    # Instance method - uses self
    def get_salary(self):
        return f"{self.name}'s salary is ${self.salary}"
    
    # Class method - uses cls
    @classmethod
    def set_company(cls, new_company):
        cls.company = new_company
        return f"Company name changed to {cls.company}"
    
    # Another class method
    @classmethod
    def get_employee_count(cls):
        return f"Total employees: {cls.employee_count}"
    
    # Static method - uses neither self nor cls
    @staticmethod
    def is_workday(day):
        # assuming day is a datetime.date object
        if day.weekday() == 5 or day.weekday() == 6:  # 5 is Saturday, 6 is Sunday
            return False
        return True
```

```python
# Creating instances
emp1 = Employee("John", 50000)
emp2 = Employee("Alice", 60000)

# Using instance method
print(emp1.get_salary())

# Using class methods
print(Employee.get_employee_count())
print(Employee.set_company("New Tech Corp"))
print(f"Company name for emp1: {emp1.company}")

# Using static method
import datetime
today = datetime.date.today()
print(f"Is today a workday? {Employee.is_workday(today)}")
```

**When to use each type of method:**

- Use **instance methods** when you need to access or modify instance data
- Use **class methods** when you need to access or modify class data
- Use **static methods** when you need a utility function related to the class but that doesn't need instance or class data

## 6. Special Methods and Magic Methods

Python classes have special methods (sometimes called "magic methods") that are surrounded by double underscores (`__`). These methods allow classes to work with built-in Python functionality.

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.current_page = 0
        
    # String representation for end users
    def __str__(self):
        return f"{self.title} by {self.author}"
    
    # String representation for developers
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    
    # Called when using len() function on an object
    def __len__(self):
        return self.pages
    
    # Makes the object callable like a function
    def __call__(self, page):
        if 0 <= page <= self.pages:
            self.current_page = page
            return f"Turned to page {page}"
        return "Invalid page number"
    
    # Defines behavior for the '==' operator
    def __eq__(self, other):
        if not isinstance(other, Book):
            return False
        return self.title == other.title and self.author == other.author
```

```python
# Creating Book objects
book1 = Book("Python Crash Course", "Eric Matthes", 544)
book2 = Book("Python Cookbook", "David Beazley", 706)
book3 = Book("Python Crash Course", "Eric Matthes", 544)

# Using __str__ (implicitly called by print())
print(book1)

# Using __repr__
print(repr(book2))

# Using __len__
print(f"The book has {len(book1)} pages")

# Using __call__
print(book1(42))  # Calling the book object like a function

# Using __eq__
print(f"Are book1 and book2 the same? {book1 == book2}")
print(f"Are book1 and book3 the same? {book1 == book3}")
```

**Common magic methods:**

- `__init__`: Constructor, called when an instance is created
- `__str__`: String representation for humans (str(), print())
- `__repr__`: String representation for developers (debugging)
- `__len__`: Return length with len() function
- `__call__`: Make object callable like a function
- `__eq__`, `__lt__`, `__gt__`: Comparison operators (==, <, >)
- `__add__`, `__sub__`: Arithmetic operators (+, -)
- `__getitem__`, `__setitem__`: Indexing and slicing ([]) operations

## 7. Inheritance Basics

Inheritance is a way to form new classes using classes that have already been defined. The new class inherits attributes and behaviors from its parent class while potentially adding new functionality.

```python
# Base class (Parent class)
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
        
    def make_sound(self):
        return "Some generic animal sound"
    
    def __str__(self):
        return f"{self.name} is a {self.species}"

# Derived class (Child class)
class Cat(Animal):  # Cat inherits from Animal
    def __init__(self, name, breed, toy):
        # Call the parent class's __init__ method
        super().__init__(name, species="Cat")
        self.breed = breed
        self.toy = toy
        
    # Override the parent class's method
    def make_sound(self):
        return "Meow!"
    
    # Add a new method
    def play(self):
        return f"{self.name} plays with {self.toy}"

# Another derived class
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, species="Dog")
        self.breed = breed
        
    def make_sound(self):
        return "Woof!"
```

```python
# Create instances of the derived classes
whiskers = Cat("Whiskers", "Tabby", "String")
rex = Dog("Rex", "German Shepherd")

# Using inherited and overridden methods
print(whiskers)  # Uses Animal's __str__
print(rex)  # Uses Animal's __str__

print(f"{whiskers.name} says: {whiskers.make_sound()}")  # Uses Cat's make_sound
print(f"{rex.name} says: {rex.make_sound()}")  # Uses Dog's make_sound

# Using a method specific to the Cat class
print(whiskers.play())

# Check inheritance relationships
print(f"Is whiskers a Cat? {isinstance(whiskers, Cat)}")
print(f"Is whiskers an Animal? {isinstance(whiskers, Animal)}")
print(f"Is whiskers a Dog? {isinstance(whiskers, Dog)}")
```

**Key concepts in inheritance:**

1. **Parent/Base Class**: The original class being extended
2. **Child/Derived Class**: The new class that inherits from the base class
3. **`super()` function**: Used to call methods from the parent class
4. **Method Overriding**: Redefining a method from the parent class
5. **Multiple Inheritance**: A class can inherit from multiple parent classes

```python
# Example of multiple inheritance
class Flying:
    def fly(self):
        return "I can fly!"
    
class Swimming:
    def swim(self):
        return "I can swim!"
    
# Duck inherits from both Flying and Swimming
class Duck(Flying, Swimming, Animal):
    def __init__(self, name):
        Animal.__init__(self, name, species="Duck")
        
    def make_sound(self):
        return "Quack!"

# Create a Duck instance
donald = Duck("Donald")

# Duck can use methods from all parent classes
print(donald.fly())     # From Flying class
print(donald.swim())    # From Swimming class
print(donald.make_sound())  # Overridden from Animal class
print(donald)           # From Animal class
```

### Example 1: Bank Account System

```python
class BankAccount:
    interest_rate = 0.02  # Class attribute: 2% interest rate
    
    def __init__(self, account_number, owner_name, balance=0):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance
        self.transaction_history = []
        
    def deposit(self, amount):
        if amount <= 0:
            return "Amount must be positive"
        self.balance += amount
        self.transaction_history.append(f"Deposit: +${amount}")
        return f"${amount} deposited. New balance: ${self.balance}"
    
    def withdraw(self, amount):
        if amount <= 0:
            return "Amount must be positive"
        if amount > self.balance:
            return "Insufficient funds"
        self.balance -= amount
        self.transaction_history.append(f"Withdrawal: -${amount}")
        return f"${amount} withdrawn. New balance: ${self.balance}"
    
    def get_balance(self):
        return f"Current balance: ${self.balance}"
    
    def apply_interest(self):
        interest = self.balance * BankAccount.interest_rate
        self.balance += interest
        self.transaction_history.append(f"Interest: +${interest:.2f}")
        return f"Interest applied: ${interest:.2f}. New balance: ${self.balance:.2f}"
    
    def print_statement(self):
        print(f"Statement for Account {self.account_number} ({self.owner_name})")
        print("-" * 40)
        for transaction in self.transaction_history[-5:]:  # Show last 5 transactions
            print(transaction)
        print("-" * 40)
        print(f"Current balance: ${self.balance:.2f}")
        
# Child class that extends BankAccount
class SavingsAccount(BankAccount):
    interest_rate = 0.04  # Higher interest rate for savings accounts
    
    def __init__(self, account_number, owner_name, balance=0):
        super().__init__(account_number, owner_name, balance)
        self.withdrawal_limit = 3
        self.withdrawals_this_month = 0
        
    def withdraw(self, amount):
        if self.withdrawals_this_month >= self.withdrawal_limit:
            return "Withdrawal limit reached for this month"
        
        result = super().withdraw(amount)
        
        # Check if withdrawal was successful before incrementing counter
        if "withdrawn" in result:
            self.withdrawals_this_month += 1
            remaining = self.withdrawal_limit - self.withdrawals_this_month
            return f"{result} (Remaining withdrawals this month: {remaining})"
        
        return result
    
    def reset_withdrawals(self):
        self.withdrawals_this_month = 0
        return "Monthly withdrawal count has been reset"
```

```python
# Create regular bank account
account1 = BankAccount("123456", "John Doe", 1000)

# Perform transactions
print(account1.deposit(500))
print(account1.withdraw(200))
print(account1.apply_interest())
print(account1.deposit(1000))
print(account1.withdraw(300))

# Print statement
account1.print_statement()

print("\n" + "-"*50 + "\n")

# Create savings account
savings = SavingsAccount("987654", "Jane Smith", 2000)

# Perform transactions with withdrawal limits
print(savings.deposit(1000))
print(savings.withdraw(300))  # 1st withdrawal
print(savings.withdraw(200))  # 2nd withdrawal
print(savings.withdraw(100))  # 3rd withdrawal
print(savings.withdraw(50))   # Should hit the limit
print(savings.apply_interest())  # Should use the 4% interest rate

# Print statement
savings.print_statement()
```

### Example 2: Online Shopping System

```python
import datetime

class Product:
    def __init__(self, product_id, name, price, category, stock=0):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.category = category
        self.stock = stock
        
    def __str__(self):
        return f"{self.name} (${self.price:.2f})"
    
    def restock(self, quantity):
        if quantity < 0:
            return "Quantity must be positive"
        self.stock += quantity
        return f"{self.name} restocked. New stock: {self.stock}"
    
    def is_available(self, quantity=1):
        return self.stock >= quantity

class ShoppingCart:
    def __init__(self):
        self.items = {}  # Dictionary: product: quantity
        
    def add_item(self, product, quantity=1):
        if not product.is_available(quantity):
            return f"Sorry, only {product.stock} units of {product.name} available"
        
        if product in self.items:
            self.items[product] += quantity
        else:
            self.items[product] = quantity
        return f"{quantity} x {product.name} added to cart"
    
    def remove_item(self, product, quantity=None):
        if product not in self.items:
            return f"{product.name} is not in your cart"
        
        if quantity is None or quantity >= self.items[product]:
            del self.items[product]
            return f"Removed all {product.name} from cart"
        else:
            self.items[product] -= quantity
            return f"Removed {quantity} x {product.name} from cart"
    
    def get_total(self):
        total = sum(product.price * quantity for product, quantity in self.items.items())
        return total
    
    def checkout(self):
        if not self.items:
            return "Your cart is empty"
        
        # Check stock availability again
        for product, quantity in self.items.items():
            if not product.is_available(quantity):
                return f"Sorry, only {product.stock} units of {product.name} available now"
        
        # Process the order
        total = self.get_total()
        order = Order(self.items.copy())
        
        # Update stock
        for product, quantity in self.items.items():
            product.stock -= quantity
        
        # Clear cart
        self.items.clear()
        
        return order
        
    def __str__(self):
        if not self.items:
            return "Shopping Cart: Empty"
        
        cart_str = "Shopping Cart:\n" + "-"*30 + "\n"
        
        for product, quantity in self.items.items():
            subtotal = product.price * quantity
            cart_str += f"{quantity} x {product.name} @ ${product.price:.2f} = ${subtotal:.2f}\n"
            
        cart_str += "-"*30 + f"\nTotal: ${self.get_total():.2f}"
        return cart_str

class Order:
    order_count = 0
    
    def __init__(self, items):
        Order.order_count += 1
        self.order_id = f"ORD-{Order.order_count:06d}"
        self.items = items
        self.order_date = datetime.datetime.now()
        self.status = "Placed"
        
        # Calculate total
        self.total = sum(product.price * quantity for product, quantity in self.items.items())
        
    def update_status(self, new_status):
        valid_statuses = ["Placed", "Processing", "Shipped", "Delivered", "Cancelled"]
        if new_status not in valid_statuses:
            return f"Invalid status. Valid options are: {', '.join(valid_statuses)}"
        self.status = new_status
        return f"Order {self.order_id} status updated to: {self.status}"
    
    def __str__(self):
        order_str = f"Order {self.order_id}\n"
        order_str += f"Date: {self.order_date.strftime('%Y-%m-%d %H:%M')}\n"
        order_str += f"Status: {self.status}\n"
        order_str += "-"*40 + "\n"
        
        for product, quantity in self.items.items():
            subtotal = product.price * quantity
            order_str += f"{quantity} x {product.name} @ ${product.price:.2f} = ${subtotal:.2f}\n"
            
        order_str += "-"*40 + f"\nTotal: ${self.total:.2f}"
        return order_str
```

```python
# Create some products
laptop = Product("P001", "Laptop", 999.99, "Electronics", 10)
mouse = Product("P002", "Wireless Mouse", 29.99, "Electronics", 20)
headphones = Product("P003", "Headphones", 79.99, "Electronics", 15)
t_shirt = Product("P004", "T-Shirt", 19.99, "Clothing", 50)

# Create a shopping cart and add items
cart = ShoppingCart()
print(cart.add_item(laptop, 1))  # Add 1 laptop
print(cart.add_item(mouse, 2))   # Add 2 mice
print(cart.add_item(headphones)) # Add 1 headphone (default quantity)
print(cart.add_item(laptop, 15)) # Should fail because not enough stock

# Display the cart
print("\n" + str(cart) + "\n")

# Remove an item
print(cart.remove_item(mouse, 1))  # Remove 1 mouse
print("\n" + str(cart) + "\n")

# Checkout
order = cart.checkout()
print("After checkout - Cart:")  # Should be empty
print(str(cart))

print("\n" + "-"*50 + "\n")

# Display the order
print("Order details:")
print(order)

# Update order status
print("\n" + order.update_status("Processing"))
print(order.update_status("Invalid"))  # Should fail
print(order.update_status("Shipped"))

# Check remaining stock
print(f"\nLaptop remaining stock: {laptop.stock}")
print(f"Mouse remaining stock: {mouse.stock}")
print(f"Headphones remaining stock: {headphones.stock}")
```

## Summary

In this notebook, we've covered the basic concepts of object-oriented programming in Python:

1. **Classes and Objects**: Classes are blueprints for objects, which are instances of the class.

2. **Attributes**: There are two types: class attributes (shared by all instances) and instance attributes (specific to each object).

3. **Methods**: Functions defined within a class that can operate on objects of the class. These include instance methods, class methods, and static methods.

4. **Special/Magic Methods**: Methods with double underscores that allow objects to work with Python's built-in functions and operators.

5. **Inheritance**: A mechanism where a new class can be derived from an existing class, inheriting its attributes and methods.

OOP helps organize code by grouping related data and behaviors together, promoting code reusability and modularity. By separating concerns and encapsulating data, OOP can make complex systems more manageable and easier to maintain.
