# Tuples and Their Uses in Python

This notebook explores the concept of tuples in Python, their characteristics, operations, and practical applications in programming.

## Introduction to Tuples

Tuples are one of Python's built-in sequence data types. They are similar to lists but with one key difference: tuples are **immutable**, meaning once created, their elements cannot be modified.

Tuples are defined by enclosing elements in parentheses `()`, separated by commas.

Key characteristics of tuples:
- Immutable (cannot be changed after creation)
- Ordered (elements maintain a defined order)
- Can contain elements of different data types
- Allow duplicate elements
- Can be indexed and sliced like other sequences

Tuples are commonly used for:
- Representing fixed collections of items
- Returning multiple values from functions
- Dictionary keys (when containing only immutable elements)
- Data that shouldn't be changed throughout program execution

```python
# Basic tuple example
simple_tuple = (1, 2, 3, 'a', 'b', True)
print(f"Tuple: {simple_tuple}")
print(f"Type: {type(simple_tuple)}")
```

## Creating and Accessing Tuples

Tuples can be created in several ways:

```python
# 1. Using parentheses (most common)
tuple1 = (1, 2, 3, 4, 5)
print("Using parentheses:", tuple1)

# 2. Without parentheses (values are separated by commas)
tuple2 = 1, 2, 3, 4, 5
print("Without parentheses:", tuple2)

# 3. Creating an empty tuple
empty_tuple = ()
print("Empty tuple:", empty_tuple)

# 4. Creating a single-element tuple (note the trailing comma)
single_tuple = (42,)  # Comma is necessary!
print("Single element tuple:", single_tuple)
not_a_tuple = (42)    # This is just an integer in parentheses
print("Not a tuple:", not_a_tuple, type(not_a_tuple))

# 5. Using the tuple() constructor
tuple_from_list = tuple([1, 2, 3])
tuple_from_string = tuple("Python")
print("From list:", tuple_from_list)
print("From string:", tuple_from_string)

# 6. Nested tuples
nested_tuple = (1, 2, ("a", "b"), (3, 4))
print("Nested tuple:", nested_tuple)
```

### Accessing Tuple Elements

Tuple elements are accessed using indexing and slicing, just like lists:

```python
sample_tuple = ('P', 'y', 't', 'h', 'o', 'n', 2023)

# Accessing by index (0-based)
print("First element:", sample_tuple[0])
print("Last element:", sample_tuple[-1])

# Slicing
print("First three elements:", sample_tuple[:3])
print("Elements from index 2 onward:", sample_tuple[2:])
print("Elements from index 1 to 5:", sample_tuple[1:6])
print("Every other element:", sample_tuple[::2])

# Accessing nested elements
nested = (1, 2, ('a', 'b', 'c'), 3)
print("Nested element 'b':", nested[2][1])

# Try to modify a tuple (will cause an error)
try:
    sample_tuple[0] = 'J'
except TypeError as e:
    print("Error when trying to modify tuple:", e)
```

## Tuple Operations and Methods

Although tuples are immutable (their elements cannot be modified after creation), you can still perform various operations on them.

```python
# Concatenation with + operator
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined = tuple1 + tuple2
print("Concatenated tuples:", combined)

# Repetition with * operator
repeated = tuple1 * 3
print("Repeated tuple:", repeated)

# Membership testing with 'in'
print("Is 2 in tuple1?", 2 in tuple1)
print("Is 7 in tuple1?", 7 in tuple1)

# Length of a tuple
print("Length of tuple1:", len(tuple1))

# Min and max values
numbers = (5, 2, 8, 1, 9, 3)
print("Minimum value:", min(numbers))
print("Maximum value:", max(numbers))

# Sum of values
print("Sum of values:", sum(numbers))

# Built-in tuple methods
mixed_tuple = ('a', 'b', 'c', 'a', 'd', 'a')

# count() - counts occurrences of an element
print("Count of 'a':", mixed_tuple.count('a'))

# index() - returns the index of the first occurrence
print("Index of 'b':", mixed_tuple.index('b'))

# The following methods do not exist for tuples because they would modify the tuple
try:
    mixed_tuple.append('e')
except AttributeError as e:
    print("Error when trying to append:", e)
    
try:
    mixed_tuple.remove('a')
except AttributeError as e:
    print("Error when trying to remove:", e)
```

## Tuples vs Lists

Tuples and lists are both sequence types in Python, but they have important differences that make each suitable for different scenarios.

```python
import sys
import timeit

# Create equivalently sized list and tuple
sample_list = [1, 2, 3, 4, 5, 'a', 'b', 'c']
sample_tuple = (1, 2, 3, 4, 5, 'a', 'b', 'c')

# Compare memory usage
print("Memory usage comparison:")
print(f"List size: {sys.getsizeof(sample_list)} bytes")
print(f"Tuple size: {sys.getsizeof(sample_tuple)} bytes")

# Compare creation time
list_creation = timeit.timeit(stmt="[1, 2, 3, 4, 5]", number=1000000)
tuple_creation = timeit.timeit(stmt="(1, 2, 3, 4, 5)", number=1000000)

print("\nCreation time comparison (1,000,000 iterations):")
print(f"List creation time: {list_creation:.6f} seconds")
print(f"Tuple creation time: {tuple_creation:.6f} seconds")
print(f"Tuples are {list_creation/tuple_creation:.2f}x faster to create")

# Compare access time
list_access = timeit.timeit(stmt="a[2]", setup="a = [1, 2, 3, 4, 5]", number=10000000)
tuple_access = timeit.timeit(stmt="a[2]", setup="a = (1, 2, 3, 4, 5)", number=10000000)

print("\nElement access time comparison (10,000,000 iterations):")
print(f"List access time: {list_access:.6f} seconds")
print(f"Tuple access time: {tuple_access:.6f} seconds")
print(f"Tuples are {list_access/tuple_access:.2f}x faster to access")
```

### Key Differences Between Tuples and Lists

| Aspect | Tuples | Lists |
|--------|--------|-------|
| **Mutability** | Immutable (cannot be changed) | Mutable (can be modified) |
| **Syntax** | Uses parentheses `()` | Uses square brackets `[]` |
| **Performance** | Generally faster | Slightly slower |
| **Memory** | More memory-efficient | Requires more memory |
| **Usage** | For fixed data | For collections that may change |
| **Methods** | Only `count()` and `index()` | Many methods for modification |
| **As Dictionary Keys** | Can be used (if elements are immutable) | Cannot be used |

### When to use tuples vs lists:

**Use tuples when:**
- Data shouldn't change throughout program execution
- You need a hashable sequence (for dict keys or set elements)
- You want to convey that the data won't be modified
- Performance and memory optimization are important
- Returning multiple values from a function

**Use lists when:**
- You need to modify the sequence (add/remove elements)
- You need in-place sorting
- You need a dynamic collection of items
- You need sequence-specific operations like `.append()` or `.extend()`

## Tuple Packing and Unpacking

Python allows packing multiple values into a tuple and unpacking tuple elements into separate variables. This is a powerful feature that enables clean, readable code.

```python
# Tuple packing - combining multiple values into a tuple
coordinates = 3, 4    # Equivalent to coordinates = (3, 4)
print("Packed tuple:", coordinates)

person_info = "John", "Doe", 35, "Software Developer"
print("Packed person info:", person_info)

# Basic tuple unpacking - extracting values to variables
x, y = coordinates
print(f"Unpacked coordinates: x = {x}, y = {y}")

first_name, last_name, age, occupation = person_info
print(f"Person: {first_name} {last_name}, Age: {age}, Job: {occupation}")

# Value swapping using tuple packing and unpacking
a = 10
b = 20
print(f"Before swap: a = {a}, b = {b}")

a, b = b, a  # Swap values using tuple packing/unpacking
print(f"After swap: a = {a}, b = {b}")
```

```python
# Extended unpacking with * operator (Python 3.x feature)

# Unpacking specific elements, with the rest going into a list
first, *middle, last = (1, 2, 3, 4, 5)
print(f"first = {first}, middle = {middle}, last = {last}")

# Using * to collect remaining items
head, *tail = ["a", "b", "c", "d"]
print(f"head = {head}, tail = {tail}")

*beginning, end = [1, 2, 3, 4, 5]
print(f"beginning = {beginning}, end = {end}")

# Using * in the middle
start, *middle, end = [1, 2, 3, 4, 5, 6, 7]
print(f"start = {start}, middle = {middle}, end = {end}")

# Using * with empty sequences
first, *rest = [1]
print(f"first = {first}, rest = {rest}")  # rest will be an empty list

# Ignoring values during unpacking (using underscore)
a, _, b = (1, 2, 3)
print(f"Extracted values: a = {a}, b = {b}")

# Multiple _ does not work as expected; use * for discarding multiple values
a, *_, b = (1, 2, 3, 4, 5, 6)
print(f"First and last: a = {a}, b = {b}")
```

## Tuples as Dictionary Keys

One of the significant advantages of tuples over lists is that tuples can be used as dictionary keys, provided they contain only immutable elements (numbers, strings, other tuples with immutable elements). This is because dictionary keys must be hashable, and immutable types are hashable by default.

```python
# Creating dictionaries with tuple keys

# Using tuples of coordinates as keys
coordinates = {
    (0, 0): "Origin",
    (0, 1): "North",
    (1, 0): "East",
    (0, -1): "South",
    (-1, 0): "West"
}

print("Dictionary with tuple keys:")
for key, value in coordinates.items():
    print(f"  {key}: {value}")

# Accessing values using tuple keys
position = (0, 1)
print(f"Position {position} is {coordinates[position]}")

# Using nested tuples as keys
student_grades = {
    ("John", "Doe"): {"Math": 85, "Science": 92, "History": 78},
    ("Jane", "Smith"): {"Math": 92, "Science": 88, "History": 95},
    ("Bob", "Johnson"): {"Math": 76, "Science": 85, "History": 82}
}

print("\nStudent grades:")
for name, grades in student_grades.items():
    print(f"  {name[0]} {name[1]}: {grades}")

# Why lists can't be used as dictionary keys
try:
    # This will fail because lists are mutable
    invalid_dict = {[1, 2]: "value"}
except TypeError as e:
    print("\nError when using list as dict key:", e)
    
# But this works fine with tuples
valid_dict = {(1, 2): "value"}
print("Dictionary with tuple key:", valid_dict)
```

### Real-world Application: Geographic Information Systems (GIS)

In GIS applications, tuples are commonly used as keys for storing data associated with specific coordinates.

```python
# Example: Storing elevation data for specific coordinates
elevation_data = {
    (40.7128, -74.0060): 10,    # New York
    (34.0522, -118.2437): 93,   # Los Angeles
    (51.5074, -0.1278): 20,     # London
    (48.8566, 2.3522): 35,      # Paris
    (35.6762, 139.6503): 40     # Tokyo
}

# Finding elevation for a given coordinate
nyc_coords = (40.7128, -74.0060)
print(f"Elevation at coordinates {nyc_coords} is {elevation_data[nyc_coords]} meters")

# Example: Storing city information using coordinates as keys
city_data = {
    (40.7128, -74.0060): {
        "name": "New York City",
        "country": "USA",
        "population": 8.4,
        "landmarks": ["Statue of Liberty", "Empire State Building", "Central Park"]
    },
    (51.5074, -0.1278): {
        "name": "London",
        "country": "UK",
        "population": 8.9,
        "landmarks": ["Big Ben", "Tower Bridge", "Buckingham Palace"]
    }
}

# Accessing city information
london_info = city_data[(51.5074, -0.1278)]
print(f"\nCity: {london_info['name']}, Country: {london_info['country']}")
print(f"Population: {london_info['population']} million")
print(f"Famous landmarks: {', '.join(london_info['landmarks'])}")
```

## Returning Multiple Values from Functions

One common and practical use of tuples is for functions to return multiple values. Since Python functions can only return a single object, tuples provide a convenient way to package multiple return values.

```python
# Function that returns multiple values as a tuple
def get_user_info():
    name = "Alice"
    age = 30
    email = "alice@example.com"
    is_active = True
    return name, age, email, is_active  # Implicit tuple packing

# Unpacking the returned tuple
user_name, user_age, user_email, user_active = get_user_info()
print(f"User: {user_name}, Age: {user_age}")
print(f"Email: {user_email}, Active: {user_active}")

# We can also keep the returned values as a tuple
user_data = get_user_info()
print(f"User data tuple: {user_data}")
print(f"Type of returned data: {type(user_data)}")
```

```python
# Example: Function to calculate statistics for a set of values
def calculate_statistics(numbers):
    """Calculate basic statistics for a sequence of numbers"""
    if not numbers:
        return 0, 0, 0, 0  # Handle empty sequence
    
    count = len(numbers)
    total = sum(numbers)
    average = total / count
    minimum = min(numbers)
    maximum = max(numbers)
    
    return count, minimum, maximum, average

# Using the function
data = [23, 45, 12, 67, 89, 34, 56]
count, min_val, max_val, avg = calculate_statistics(data)

print(f"Data: {data}")
print(f"Count: {count}")
print(f"Minimum: {min_val}")
print(f"Maximum: {max_val}")
print(f"Average: {avg:.2f}")
```

```python
# Example: Function that divides two numbers and returns both the quotient and remainder
def divide_with_remainder(dividend, divisor):
    """Return both quotient and remainder"""
    if divisor == 0:
        return None  # Handle division by zero
    
    quotient = dividend // divisor
    remainder = dividend % divisor
    return quotient, remainder

# Using the function
result = divide_with_remainder(17, 5)
print(f"Result tuple: {result}")

q, r = divide_with_remainder(17, 5)
print(f"17 divided by 5 equals {q} with remainder {r}")
```

## Named Tuples

Python's `collections` module provides `namedtuple`, which creates tuple subclasses with named fields. Named tuples allow accessing elements by name (using dot notation) as well as by index, providing more readable and self-documenting code while maintaining tuple's immutability.

```python
from collections import namedtuple

# Creating a named tuple type
Person = namedtuple('Person', ['name', 'age', 'gender', 'email'])

# Creating instances of the named tuple
alice = Person(name='Alice', age=30, gender='Female', email='alice@example.com')
bob = Person('Bob', 25, 'Male', 'bob@example.com')  # Positional arguments also work

# Accessing by name (attribute)
print(f"{alice.name} is {alice.age} years old")
print(f"{bob.name}'s email is {bob.email}")

# Accessing by index (like regular tuples)
print(f"First person: {alice[0]}, {alice[1]}, {alice[2]}")

# Unpacking works just like regular tuples
name, age, gender, email = bob
print(f"Unpacked: {name}, {age}, {gender}, {email}")

# Named tuples are immutable
try:
    alice.age = 31
except AttributeError as e:
    print("Error when trying to modify named tuple:", e)

# Converting to regular dictionary
alice_dict = alice._asdict()
print(f"Dictionary: {alice_dict}")

# Creating a new instance with updated fields
alice_updated = alice._replace(age=31, email='alice.new@example.com')
print(f"Updated Alice: {alice_updated}")
print(f"Original Alice: {alice}")  # Original remains unchanged
```

```python
# More examples of named tuples

# Representing a point in 2D space
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(f"Point: ({p.x}, {p.y})")

# Computing distance from origin
distance = (p.x**2 + p.y**2)**0.5
print(f"Distance from origin: {distance}")

# Representing a circle
Circle = namedtuple('Circle', ['center', 'radius'])
c = Circle(Point(0, 0), 5)
print(f"Circle: center=({c.center.x}, {c.center.y}), radius={c.radius}")

# Representing a student record
Student = namedtuple('Student', 'id name gpa courses')
s1 = Student(1234, 'John Doe', 3.7, ['Math', 'Physics', 'CS'])
print(f"Student: {s1.name}, ID: {s1.id}")
print(f"GPA: {s1.gpa}, Courses: {', '.join(s1.courses)}")

# Creating with defaults
# Python 3.7+ can use defaults parameter
Employee = namedtuple('Employee', ['name', 'id', 'role', 'salary'], defaults=[0, 'staff', 50000])
e1 = Employee('Sarah Johnson')  # Only providing the required field
e2 = Employee('Mike Davis', 101)  # Providing some fields
e3 = Employee('Anna Smith', 102, 'manager', 75000)  # Providing all fields

print("\nEmployees:")
print(f"e1: {e1}")
print(f"e2: {e2}")
print(f"e3: {e3}")

# Getting field names
print(f"\nField names: {Employee._fields}")
```

## Practical Applications of Tuples

Tuples are widely used in Python for various applications. Here are some practical examples:

```python
# 1. Coordinates in graphics and mapping
def plot_points(points):
    """Simulate plotting points on a graph"""
    for point in points:
        x, y = point
        print(f"Plotting point at coordinates ({x}, {y})")

coordinates = [(3, 4), (6, 8), (2, 5), (9, 3)]
plot_points(coordinates)
```

```python
# 2. Representing database records
# In database operations, tuples often represent rows of data

def fetch_users_from_database():
    """Simulate fetching users from a database"""
    # In real code, this would come from a DB query
    return [
        (1, "John Doe", "john@example.com", "2023-01-15"),
        (2, "Jane Smith", "jane@example.com", "2023-02-20"),
        (3, "Bob Johnson", "bob@example.com", "2023-03-10")
    ]

users = fetch_users_from_database()
print("Users from database:")
for user in users:
    user_id, name, email, join_date = user
    print(f"  ID: {user_id}, Name: {name}, Email: {email}, Joined: {join_date}")
```

```python
# 3. Configuration settings
# Tuples can represent configuration options that shouldn't change

# Server configurations (host, port, protocol)
servers = [
    ("main-server", 8080, "https"),
    ("backup-server", 8081, "https"),
    ("dev-server", 8082, "http")
]

for server in servers:
    host, port, protocol = server
    url = f"{protocol}://{host}:{port}"
    print(f"Server URL: {url}")

# Default application settings
DEFAULT_SETTINGS = (
    1024,  # window width
    768,   # window height
    "dark", # theme
    True,   # notifications enabled
    60      # refresh rate
)

def initialize_app_with_defaults():
    width, height, theme, notifications, refresh = DEFAULT_SETTINGS
    print(f"Initializing app with window size {width}x{height}")
    print(f"Theme: {theme}, Notifications: {notifications}, Refresh rate: {refresh}Hz")

initialize_app_with_defaults()
```

```python
# 4. Fixed data structures (e.g., RGB colors)

def apply_color(element, color):
    """Apply RGB color to an element"""
    r, g, b = color
    print(f"Applying RGB color ({r}, {g}, {b}) to {element}")

colors = {
    "red": (255, 0, 0),
    "green": (0, 255, 0),
    "blue": (0, 0, 255),
    "yellow": (255, 255, 0),
    "purple": (128, 0, 128)
}

apply_color("button", colors["red"])
apply_color("background", colors["blue"])
apply_color("text", colors["yellow"])
```

```python
# 5. State transitions (current_state, event) -> new_state

# Define state transition rules
state_machine = {
    ("idle", "start"): "running",
    ("running", "pause"): "paused",
    ("paused", "resume"): "running",
    ("running", "complete"): "finished",
    ("paused", "stop"): "idle",
    ("running", "stop"): "idle",
    ("finished", "reset"): "idle"
}

def process_event(current_state, event):
    """Process an event based on current state"""
    if (current_state, event) in state_machine:
        new_state = state_machine[(current_state, event)]
        print(f"Transition: {current_state} + {event} -> {new_state}")
        return new_state
    else:
        print(f"Invalid transition: {current_state} + {event}")
        return current_state

# Simulate a process
state = "idle"
state = process_event(state, "start")    # idle -> running
state = process_event(state, "pause")    # running -> paused
state = process_event(state, "stop")     # paused -> idle
state = process_event(state, "complete") # idle + complete (invalid)
```

```python
# 6. Function arguments caching (memoization)
import functools
import time

# Using tuples as cache keys in memoization
@functools.lru_cache(maxsize=128)
def expensive_computation(x, y, z):
    """A simulated expensive computation function"""
    print(f"Computing result for ({x}, {y}, {z})...")
    time.sleep(1)  # Simulate expensive computation
    return x**2 + y**2 + z**2

# First call will compute the result
result1 = expensive_computation(3, 4, 5)
print(f"Result 1: {result1}")

# Second call with same arguments will use cached result
result2 = expensive_computation(3, 4, 5)
print(f"Result 2: {result2}")

# Different arguments will compute new result
result3 = expensive_computation(5, 6, 7)
print(f"Result 3: {result3}")
```
