## 1. Introduction to Dictionaries

Dictionaries are one of Python's most versatile and powerful data structures. Unlike lists which store elements in a specific order and are accessed by index, dictionaries store data as **key-value pairs** and elements are accessed by keys.

Key characteristics of dictionaries:
- Unordered (prior to Python 3.7; ordered since 3.7+)
- Mutable (can be changed after creation)
- Keys must be unique and immutable (strings, numbers, tuples)
- Values can be any data type (including other dictionaries)

### When to Use Dictionaries

Dictionaries are ideal when:
1. You need to associate values with keys, like a real-world dictionary
2. You need fast lookups by a key (O(1) average time complexity)
3. You need a logical mapping between two sets of data
4. You want to represent structured data (like JSON)

```python
# Basic comparison between lists and dictionaries

# Using a list to store student grades (by position)
grades_list = [85, 92, 78, 95, 88]
print(f"Student 3's grade (using list): {grades_list[2]}")  # Have to know the index

# Using a dictionary to store student grades (by name)
grades_dict = {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 95, 'Evan': 88}
print(f"Charlie's grade (using dict): {grades_dict['Charlie']}")  # More intuitive lookup

# Lists preserve order
print("\nLists preserve order:")
for i, grade in enumerate(grades_list):
    print(f"Student {i+1}: {grade}")

# Dictionaries (3.7+) preserve insertion order but are accessed by key
print("\nDictionaries are accessed by key:")
for student, grade in grades_dict.items():
    print(f"{student}: {grade}")
```

<a id="creating-dict"></a>
## 2. Creating and Accessing Dictionaries

There are several ways to create dictionaries in Python:

1. Using curly braces `{}` with key-value pairs
2. Using the `dict()` constructor
3. Using dict comprehensions (covered later)
4. Creating dictionaries from sequences

```python
# 1. Using curly braces
person = {'name': 'John', 'age': 30, 'city': 'New York'}
print(person)

# 2. Using dict() constructor
person2 = dict(name='Jane', age=25, city='Boston')
print(person2)

# 3. Creating an empty dictionary and adding items
person3 = {}  # or person3 = dict()
person3['name'] = 'Mike'
person3['age'] = 35
person3['city'] = 'Chicago'
print(person3)

# 4. Creating dictionary from sequences
names = ['Alice', 'Bob', 'Charlie']
ages = [24, 30, 35]
people = dict(zip(names, ages))
print(people)
```

### Accessing Dictionary Values

Values in a dictionary can be accessed in several ways:
- By using the key directly inside square brackets
- By using the `get()` method (safer as it returns None or a default value if the key doesn't exist)
- Checking if a key exists using the `in` keyword

```python
person = {'name': 'John', 'age': 30, 'city': 'New York'}

# Direct access with key
print(f"Name: {person['name']}")

# What happens if we try to access a key that doesn't exist?
try:
    print(person['email'])
except KeyError as e:
    print(f"KeyError: {e}")

# Using get() method - safer approach
print(f"Email: {person.get('email')}")  # Returns None if key doesn't exist
print(f"Email: {person.get('email', 'Not Available')}")  # Providing a default value

# Checking if a key exists before accessing
if 'age' in person:
    print(f"Age: {person['age']}")
    
if 'phone' not in person:
    print("Phone number not found")
```

<a id="dict-methods"></a>
## 3. Dictionary Methods and Operations

Python dictionaries come with several built-in methods that make them flexible and powerful. Here are the most commonly used methods:

```python
# Create a sample dictionary
student = {
    'name': 'Alex', 
    'age': 20, 
    'grades': {'math': 90, 'science': 85, 'history': 80},
    'active': True
}

# 1. keys() - returns all keys
print("Keys:", list(student.keys()))

# 2. values() - returns all values
print("Values:", list(student.values()))

# 3. items() - returns key-value tuples
print("Items:", list(student.items()))

# 4. update() - update dictionary with another dictionary
student.update({'email': 'alex@example.com', 'age': 21})  # Note: 'age' gets updated
print("\nAfter update():", student)

# 5. pop() - removes a key and returns its value
removed_value = student.pop('active')
print(f"\nRemoved 'active': {removed_value}")
print("After pop():", student)

# 6. popitem() - removes and returns the last inserted key-value pair
last_item = student.popitem()  # In Python 3.7+ removes the last added item
print(f"\nRemoved last item: {last_item}")
print("After popitem():", student)

# 7. clear() - removes all items
student_copy = student.copy()  # Make a copy to demonstrate clear()
student_copy.clear()
print("\nAfter clear():", student_copy)

# 8. setdefault() - returns value of key, sets default if key doesn't exist
gpa = student.setdefault('gpa', 3.5)  # Key doesn't exist, so it gets added with default value
name = student.setdefault('name', 'Unknown')  # Key exists, returns the existing value
print("\nAfter setdefault():", student)
print(f"Retrieved values - gpa: {gpa}, name: {name}")
```

### Iterating Through Dictionaries

Python provides several ways to iterate through dictionaries:

```python
menu = {
    'coffee': 2.50,
    'tea': 2.00,
    'sandwich': 5.75,
    'cake': 3.50,
    'cookie': 1.25
}

# Iterate through keys (default behavior)
print("Menu items:")
for item in menu:
    print(item)

# Iterate through values
print("\nPrices:")
for price in menu.values():
    print(f"${price:.2f}")

# Iterate through key-value pairs
print("\nMenu with prices:")
for item, price in menu.items():
    print(f"{item}: ${price:.2f}")

# Sorted iteration
print("\nMenu sorted alphabetically:")
for item in sorted(menu):
    print(f"{item}: ${menu[item]:.2f}")

# Sorted by value (price in this case)
print("\nMenu sorted by price (low to high):")
for item, price in sorted(menu.items(), key=lambda x: x[1]):
    print(f"{item}: ${price:.2f}")
```

<a id="dict-comp"></a>
## 4. Dictionary Comprehensions

Dictionary comprehensions provide a concise way to create dictionaries using a single line of code. They follow a syntax similar to list comprehensions but create dictionary objects instead.

Syntax: `{key_expr: value_expr for item in iterable}`

Dictionary comprehensions can also include conditions to filter items.

```python
# Basic dictionary comprehension
squares = {x: x**2 for x in range(6)}
print("Squares:", squares)

# Dictionary comprehension with condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print("Even squares:", even_squares)

# Converting two lists to a dictionary
fruits = ['apple', 'banana', 'orange', 'kiwi']
prices = [1.2, 0.5, 0.8, 1.5]
fruit_prices = {fruit: price for fruit, price in zip(fruits, prices)}
print("\nFruit prices:", fruit_prices)

# Transforming keys and values
fruit_inventory = {name.upper(): int(price * 10) for name, price in fruit_prices.items()}
print("Transformed inventory:", fruit_inventory)

# Filtering with more complex conditions
discounted = {k: v * 0.8 for k, v in fruit_prices.items() if v > 1.0}
print("Discounted expensive fruits:", discounted)

# Creating a dictionary from another dictionary
original = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
squared_values = {k: v**2 for k, v in original.items()}
print("\nSquared values:", squared_values)
```

### Performance Comparison: Dictionary Comprehension vs. Traditional Methods

```python
import time

# Performance comparison
def using_for_loop(n):
    result = {}
    for i in range(n):
        result[i] = i**2
    return result

def using_dict_comp(n):
    return {i: i**2 for i in range(n)}

# Measure time for a large dictionary
n = 1000000

# Using for loop
start = time.time()
d1 = using_for_loop(n)
time_for_loop = time.time() - start
print(f"Time using for loop: {time_for_loop:.4f} seconds")

# Using dictionary comprehension
start = time.time()
d2 = using_dict_comp(n)
time_dict_comp = time.time() - start
print(f"Time using dict comprehension: {time_dict_comp:.4f} seconds")

print(f"Dictionary comprehension is {time_for_loop/time_dict_comp:.2f}x faster")
```

<a id="nested-dict"></a>
## 5. Nested Dictionaries

Nested dictionaries are dictionaries that contain other dictionaries as values. They are useful for representing hierarchical or complex data structures.

```python
# Creating a nested dictionary
university = {
    'Computer Science': {
        'Alice': {'age': 20, 'gpa': 3.8},
        'Bob': {'age': 22, 'gpa': 3.5}
    },
    'Physics': {
        'Charlie': {'age': 21, 'gpa': 3.9},
        'Diana': {'age': 23, 'gpa': 3.7}
    }
}

# Accessing elements in a nested dictionary
print(f"Bob's age: {university['Computer Science']['Bob']['age']}")

# Adding a new student to an existing department
university['Physics']['Evan'] = {'age': 19, 'gpa': 3.6}

# Adding a new department
university['Mathematics'] = {}
university['Mathematics']['Frank'] = {'age': 20, 'gpa': 3.4}

# Updating nested values
university['Computer Science']['Alice']['gpa'] = 3.9

# Iterating through a nested dictionary
print("\nAll students by department:")
for department, students in university.items():
    print(f"\n{department} Department:")
    for student, details in students.items():
        print(f"  - {student}: {details}")
```

### Practical Example: Representing JSON-like Data

Nested dictionaries are particularly useful for working with JSON data, which is a common format for web APIs and configuration files.

```python
import json

# A JSON-like structure represented as nested dictionaries
person = {
    "name": "John Smith",
    "age": 35,
    "contact": {
        "email": "john@example.com",
        "phone": "555-1234",
        "addresses": {
            "home": {
                "street": "123 Main St",
                "city": "Anytown",
                "state": "CA",
                "zip": "12345"
            },
            "work": {
                "street": "456 Business Ave",
                "city": "Workville",
                "state": "CA",
                "zip": "54321"
            }
        }
    },
    "skills": ["Python", "JavaScript", "SQL"],
    "experience": [
        {"company": "Tech Co", "years": 5, "position": "Developer"},
        {"company": "Code Inc", "years": 3, "position": "Analyst"}
    ]
}

# Convert dictionary to JSON string and pretty print
person_json = json.dumps(person, indent=2)
print("JSON representation:")
print(person_json)

# Parse a JSON string
parsed_person = json.loads(person_json)
print("\nAccessing parsed data:")
print(f"Work address: {parsed_person['contact']['addresses']['work']['street']}, " 
      f"{parsed_person['contact']['addresses']['work']['city']}")
print(f"Most recent job: {parsed_person['experience'][0]['company']}")
```

## 6. Introduction to Sets

Sets are another built-in data type in Python that represent an unordered collection of unique elements. They are useful for membership testing, removing duplicates, and mathematical operations like unions and intersections.

Key characteristics of sets:
- Unordered (elements have no defined order)
- Mutable (the set itself can be modified)
- No duplicates (all elements are unique)
- Can only contain immutable elements (strings, numbers, tuples, not lists or dictionaries)

### When to Use Sets

Sets are ideal when:
1. You need to ensure uniqueness of elements
2. You need to perform set operations (union, intersection, etc.)
3. You need fast membership testing
4. You need to eliminate duplicates from a collection

```python
# Comparing lists and sets for uniqueness and membership testing

# List example
numbers_list = [1, 2, 3, 2, 1, 4, 5, 4, 3, 6]
print(f"Original list: {numbers_list}")

# Removing duplicates from a list using a set
unique_numbers = list(set(numbers_list))
print(f"List with duplicates removed: {unique_numbers}")

# Checking membership
print("\nMembership checking:")

test_number = 5
# Using a list
start_time = time.time()
result = test_number in numbers_list
list_time = time.time() - start_time
print(f"List membership check: {result}, Time: {list_time:.10f} seconds")

# Using a set
numbers_set = set(numbers_list)
start_time = time.time()
result = test_number in numbers_set
set_time = time.time() - start_time
print(f"Set membership check: {result}, Time: {set_time:.10f} seconds")

# For a larger dataset, the difference becomes more apparent
large_list = list(range(1000000))
test_number = 999999

# List membership check
start_time = time.time()
result = test_number in large_list
large_list_time = time.time() - start_time
print(f"\nLarge list membership check: {result}, Time: {large_list_time:.6f} seconds")

# Set membership check
large_set = set(large_list)
start_time = time.time()
result = test_number in large_set
large_set_time = time.time() - start_time
print(f"Large set membership check: {result}, Time: {large_set_time:.6f} seconds")
print(f"Set is approximately {large_list_time/large_set_time:.0f}x faster for membership testing")
```

## 7. Set Creation and Operations

There are several ways to create sets in Python:
1. Using curly braces `{}`
2. Using the `set()` constructor
3. Using set comprehensions

```python
# 1. Using curly braces
fruits = {'apple', 'banana', 'orange', 'apple'}  # Note: duplicates are automatically removed
print(f"Fruits set: {fruits}")

# 2. Using the set() constructor
colors = set(['red', 'blue', 'green', 'blue'])
print(f"Colors set: {colors}")

# 3. Creating an empty set
# {} creates an empty dictionary, not a set
empty_dict = {}
print(f"Type of {{}}: {type(empty_dict)}")

# To create an empty set, use the set() constructor
empty_set = set()
print(f"Type of set(): {type(empty_set)}")

# 4. Creating a set from a string
letters = set('hello')
print(f"Set from 'hello': {letters}")  # Each character becomes an element

# 5. Set comprehension
even_squares = {x**2 for x in range(10) if x % 2 == 0}
print(f"Even squares: {even_squares}")
```

### Basic Set Operations

Sets support several operations for adding and removing elements:

```python
# Create a set
s = {1, 2, 3}
print(f"Initial set: {s}")

# Add an element
s.add(4)
print(f"After add(4): {s}")

# Add multiple elements using update()
s.update([5, 6, 7])
print(f"After update([5, 6, 7]): {s}")

# update() works with any iterable
s.update("abc")  # Adds each character as an element
print(f"After update('abc'): {s}")

# Remove an element with remove()
s.remove('a')
print(f"After remove('a'): {s}")

# Try to remove an element that doesn't exist
try:
    s.remove('z')
except KeyError as e:
    print(f"KeyError: {e}")

# Remove an element with discard() - no error if not present
s.discard('b')
print(f"After discard('b'): {s}")
s.discard('z')  # No error
print(f"After discard('z'): {s}")  # Set remains unchanged

# Pop an arbitrary element (usually the first one)
popped = s.pop()
print(f"Popped element: {popped}")
print(f"After pop(): {s}")

# Clear all elements
s.clear()
print(f"After clear(): {s}")
```

<a id="set-methods"></a>
## 8. Set Methods

Python sets come with several built-in methods for operations and comparisons.

```python
# Creating sets for demonstration
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(f"Set A: {A}")
print(f"Set B: {B}")

# Copying a set
C = A.copy()
print(f"C = A.copy(): {C}")

# Checking if a set is a subset of another
print(f"\nSubset and superset relations:")
D = {1, 2}
print(f"D = {D}")
print(f"D is subset of A (D <= A): {D <= A}")
print(f"D is subset of A (D.issubset(A)): {D.issubset(A)}")

# Checking if a set is a proper subset
print(f"D is proper subset of A (D < A): {D < A}")

# Checking if a set is a superset of another
print(f"A is superset of D (A >= D): {A >= D}")
print(f"A is superset of D (A.issuperset(D)): {A.issuperset(D)}")

# Checking if a set is a proper superset
print(f"A is proper superset of D (A > D): {A > D}")

# Checking if sets are disjoint (have no elements in common)
print(f"\nDisjoint check:")
E = {9, 10}
print(f"E = {E}")
print(f"A and B are disjoint: {A.isdisjoint(B)}")  # False because they share elements
print(f"A and E are disjoint: {A.isdisjoint(E)}")  # True because they don't share elements

# Membership testing
print(f"\nMembership testing:")
print(f"2 in A: {2 in A}")
print(f"9 not in A: {9 not in A}")
```

<a id="set-math"></a>
## 9. Mathematical Set Operations

Sets in Python support all the standard mathematical set operations, which can be performed using operators or methods.

```python
# Creating sets for demonstration
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(f"Set A: {A}")
print(f"Set B: {B}")

# Union (elements in either A or B)
print(f"\nUnion:")
print(f"A | B: {A | B}")
print(f"A.union(B): {A.union(B)}")

# Intersection (elements in both A and B)
print(f"\nIntersection:")
print(f"A & B: {A & B}")
print(f"A.intersection(B): {A.intersection(B)}")

# Difference (elements in A but not in B)
print(f"\nDifference:")
print(f"A - B: {A - B}")
print(f"A.difference(B): {A.difference(B)}")
print(f"B - A: {B - A}")
print(f"B.difference(A): {B.difference(A)}")

# Symmetric difference (elements in either A or B but not in both)
print(f"\nSymmetric Difference:")
print(f"A ^ B: {A ^ B}")
print(f"A.symmetric_difference(B): {A.symmetric_difference(B)}")

# Operations with multiple sets
C = {7, 8, 9, 10}
print(f"\nSet C: {C}")
print(f"A | B | C: {A | B | C}")
print(f"A.union(B, C): {A.union(B, C)}")
print(f"A & B & C: {A & B & C}")
print(f"A.intersection(B, C): {A.intersection(B, C)}")
```

### Update Operations

Sets also have in-place update operations that modify the set directly:

```python
# Creating sets for demonstration
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(f"Original set A: {A}")
print(f"Original set B: {B}")

# Update (union in-place)
A_copy = A.copy()
A_copy |= B  # Equivalent to A_copy.update(B)
print(f"\nA |= B: {A_copy}")

# Intersection update
A_copy = A.copy()
A_copy &= B  # Equivalent to A_copy.intersection_update(B)
print(f"A &= B: {A_copy}")

# Difference update
A_copy = A.copy()
A_copy -= B  # Equivalent to A_copy.difference_update(B)
print(f"A -= B: {A_copy}")

# Symmetric difference update
A_copy = A.copy()
A_copy ^= B  # Equivalent to A_copy.symmetric_difference_update(B)
print(f"A ^= B: {A_copy}")

# Using the method forms with multiple sets
A_copy = A.copy()
C = {7, 8, 9, 10}
A_copy.update(B, C)
print(f"\nA.update(B, C): {A_copy}")
```

### Example 1: Word Frequency Counter

```python
def word_frequency(text):
    """Count the frequency of each word in a text."""
    # Remove punctuation and convert to lowercase
    text = text.lower()
    for char in '.,!?;:"()[]{}':
        text = text.replace(char, '')
        
    # Split into words and count frequencies
    words = text.split()
    frequency = {}
    
    # Method 1: Using a dictionary directly
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    
    return frequency

# Alternative implementation using Counter from collections module
from collections import Counter

def word_frequency_counter(text):
    """Count the frequency of each word in a text using Counter."""
    # Remove punctuation and convert to lowercase
    text = text.lower()
    for char in '.,!?;:"()[]{}':
        text = text.replace(char, '')
        
    # Split into words and count frequencies
    words = text.split()
    return Counter(words)

# Sample text
sample_text = """
Python is an interpreted high-level general-purpose programming language. 
Python's design philosophy emphasizes code readability with its notable use of significant indentation. 
Its language constructs and object-oriented approach aim to help programmers write clear, logical code 
for small and large-scale projects. Python is dynamically-typed and garbage-collected.
"""

# Get word frequencies
freq_dict = word_frequency(sample_text)
freq_counter = word_frequency_counter(sample_text)

# Display top 10 most common words
print("Top 10 most common words (using custom function):")
sorted_words = sorted(freq_dict.items(), key=lambda x: x[1], reverse=True)[:10]
for word, count in sorted_words:
    print(f"  {word}: {count}")

print("\nTop 10 most common words (using Counter):")
for word, count in freq_counter.most_common(10):
    print(f"  {word}: {count}")
```

### Example 2: Removing Duplicates from a List

```python
# Different ways to remove duplicates

# Data with duplicates
data = [1, 2, 3, 1, 4, 2, 5, 6, 3, 7, 8, 9, 5, 10]

# Method 1: Using a set
unique_data1 = list(set(data))
# Note: This method doesn't preserve the original order

# Method 2: Using a dictionary to preserve order (Python 3.6+ behavior)
unique_data2 = list(dict.fromkeys(data))

# Method 3: Using a comprehension with a set for "seen" values
def remove_duplicates(items):
    seen = set()
    return [item for item in items if not (item in seen or seen.add(item))]

unique_data3 = remove_duplicates(data)

print(f"Original data: {data}")
print(f"Method 1 (set): {unique_data1}")
print(f"Method 2 (dict.fromkeys): {unique_data2}")
print(f"Method 3 (comprehension): {unique_data3}")

# Let's compare performance for a large list
import random

# Generate a list with many duplicate values
large_data = [random.randint(0, 1000) for _ in range(100000)]

# Timing the methods
import time

start = time.time()
unique1 = list(set(large_data))
time_set = time.time() - start

start = time.time()
unique2 = list(dict.fromkeys(large_data))
time_dict = time.time() - start

start = time.time()
unique3 = remove_duplicates(large_data)
time_comp = time.time() - start

print(f"\nPerformance comparison for large list:")
print(f"Method 1 (set): {time_set:.6f} seconds")
print(f"Method 2 (dict.fromkeys): {time_dict:.6f} seconds")
print(f"Method 3 (comprehension): {time_comp:.6f} seconds")
```

### Example 3: Simple Cache Implementation with Dictionary

```python
# Using a dictionary as a cache for expensive function calls
def fibonacci_with_cache(n, cache=None):
    """Calculate the Fibonacci number recursively using a cache."""
    if cache is None:
        cache = {}
    
    if n in cache:
        return cache[n]
    
    if n <= 1:
        result = n
    else:
        result = fibonacci_with_cache(n-1, cache) + fibonacci_with_cache(n-2, cache)
    
    cache[n] = result
    return result

# Let's compare with a version without caching
def fibonacci_without_cache(n):
    """Calculate the Fibonacci number recursively without caching."""
    if n <= 1:
        return n
    else:
        return fibonacci_without_cache(n-1) + fibonacci_without_cache(n-2)

# Test with a moderate value
n = 30

start = time.time()
result_with_cache = fibonacci_with_cache(n)
time_with_cache = time.time() - start

print(f"Fibonacci({n}) = {result_with_cache}")
print(f"Time with cache: {time_with_cache:.6f} seconds")

# The non-cached version would take too long for n=30, so let's use a smaller value
n_small = 20
start = time.time()
result_without_cache = fibonacci_without_cache(n_small)
time_without_cache = time.time() - start

print(f"\nFibonacci({n_small}) = {result_without_cache}")
print(f"Time without cache: {time_without_cache:.6f} seconds")
```

### Example 4: Efficient Membership Testing with Sets

```python
# Scenario: Check which words in a document are considered stop words
import time

# Define a list of common English stop words
stop_words = {"a", "an", "the", "in", "on", "at", "for", "with", "by", "to", 
              "and", "or", "but", "is", "are", "was", "were", "be", "been", 
              "being", "have", "has", "had", "do", "does", "did", "can", 
              "could", "will", "would", "should", "may", "might", "must", 
              "shall", "this", "that", "these", "those", "of", "from", "as"}

# Sample text
sample_text = """
Python is a programming language that lets you work quickly and integrate systems 
more effectively. Python is an interpreted, object-oriented, high-level 
programming language with dynamic semantics. Its high-level built-in data 
structures, combined with dynamic typing and dynamic binding, make it very 
attractive for Rapid Application Development, as well as for use as a scripting 
or glue language to connect existing components together.
"""

# Clean and split the text
words = sample_text.lower().replace('.', '').replace(',', '').split()

# Method 1: Using a set for stop word checking
def filter_with_set(words, stop_words):
    return [word for word in words if word not in stop_words]

# Method 2: Using a list for stop word checking
stop_words_list = list(stop_words)
def filter_with_list(words, stop_words_list):
    return [word for word in words if word not in stop_words_list]

# Compare performance
start_time = time.time()
filtered_set = filter_with_set(words, stop_words)
set_time = time.time() - start_time

start_time = time.time()
filtered_list = filter_with_list(words, stop_words_list)
list_time = time.time() - start_time

print(f"First 5 non-stop words: {filtered_set[:5]}")
print(f"\nPerformance comparison:")
print(f"Set method: {set_time:.8f} seconds")
print(f"List method: {list_time:.8f} seconds")
print(f"Set is approximately {list_time/set_time:.1f}x faster")

# For a more significant demonstration, let's do it with a larger text
# Generate a bigger text sample by repeating
big_text = sample_text * 1000  # 1000 times larger
big_words = big_text.lower().replace('.', '').replace(',', '').split()

start_time = time.time()
filtered_big_set = filter_with_set(big_words, stop_words)
big_set_time = time.time() - start_time

start_time = time.time()
filtered_big_list = filter_with_list(big_words, stop_words_list)
big_list_time = time.time() - start_time

print(f"\nPerformance comparison with large text:")
print(f"Set method: {big_set_time:.6f} seconds")
print(f"List method: {big_list_time:.6f} seconds")
print(f"Set is approximately {big_list_time/big_set_time:.1f}x faster with large text")
```

### Example 5: Finding Common Elements and Differences in Data

Sets are particularly useful when analyzing different groups of data to find commonalities and differences.

```python
# Scenario: Analyzing survey responses from different groups

# Survey data: Favorite programming languages by department
engineering = {"Python", "Java", "C++", "JavaScript", "Go", "Rust"}
data_science = {"Python", "R", "SQL", "Julia", "MATLAB", "Java"}
web_development = {"JavaScript", "TypeScript", "HTML", "CSS", "Python", "PHP"}

print(f"Engineering team favorites: {engineering}")
print(f"Data Science team favorites: {data_science}")
print(f"Web Development team favorites: {web_development}")

# Finding languages common to all departments
common_to_all = engineering & data_science & web_development
print(f"\nLanguages common to all departments: {common_to_all}")

# Languages common between pairs of departments
eng_ds_common = engineering & data_science
eng_web_common = engineering & web_development
ds_web_common = data_science & web_development

print(f"\nLanguages common to Engineering and Data Science: {eng_ds_common}")
print(f"Languages common to Engineering and Web Development: {eng_web_common}")
print(f"Languages common to Data Science and Web Development: {ds_web_common}")

# Languages unique to each department
eng_unique = engineering - (data_science | web_development)
ds_unique = data_science - (engineering | web_development)
web_unique = web_development - (engineering | data_science)

print(f"\nLanguages unique to Engineering: {eng_unique}")
print(f"Languages unique to Data Science: {ds_unique}")
print(f"Languages unique to Web Development: {web_unique}")

# Union of all languages
all_languages = engineering | data_science | web_development
print(f"\nAll languages mentioned: {all_languages}")
print(f"Total number of distinct languages: {len(all_languages)}")

# Languages in exactly one department
in_exactly_one = eng_unique | ds_unique | web_unique
print(f"\nLanguages used in exactly one department: {in_exactly_one}")

# Languages in at least two departments
in_at_least_two = all_languages - in_exactly_one
print(f"Languages used in at least two departments: {in_at_least_two}")
```

## Summary and Best Practices

### Dictionaries
- Use dictionaries when you need to associate values with keys
- Keys must be immutable (strings, numbers, tuples)
- Use the `.get()` method for safe access with default values
- Use dictionary comprehensions for concise transformations
- Since Python 3.7, dictionaries maintain insertion order

### Dictionaries Best Practices:
1. Use descriptive key names for readability
2. Prefer `.get()` over direct access to avoid KeyErrors
3. Use `.items()`, `.keys()`, and `.values()` for clean iteration
4. Consider `defaultdict` or `Counter` from collections module for special cases
5. Use nested dictionaries to represent hierarchical data

### Sets
- Use sets when you need unique values
- Use sets for membership testing (much faster than lists)
- Use sets for mathematical operations (union, intersection, difference)
- All set elements must be immutable

### Sets Best Practices:
1. Use sets over lists when order doesn't matter and uniqueness is important
2. Use set operations for comparing data collections
3. Convert collections to sets temporarily for fast duplicate removal
4. Remember that sets themselves are mutable but their elements must be immutable
5. Consider `frozenset` when you need an immutable set (e.g., as a dictionary key)

Both dictionaries and sets provide O(1) average case lookup time, making them extremely efficient for many common operations in Python.
