## 5. Date Arithmetic and Timedeltas

````python
# Working with timedeltas (duration)
today = datetime.date.today()
yesterday = today - datetime.timedelta(days=1)
tomorrow = today + datetime.timedelta(days=1)
next_week = today + datetime.timedelta(weeks=1)

print(f"Today: {today}")
print(f"Yesterday: {yesterday}")
print(f"Tomorrow: {tomorrow}")
print(f"Next week: {next_week}")
````

````python
# Create timedeltas with different units
one_day = datetime.timedelta(days=1)
one_hour = datetime.timedelta(hours=1)
one_minute = datetime.timedelta(minutes=1)
one_second = datetime.timedelta(seconds=1)
ten_milliseconds = datetime.timedelta(milliseconds=10)
complex_duration = datetime.timedelta(days=2, hours=5, minutes=30, seconds=15)

print(f"One day: {one_day}")
print(f"One hour: {one_hour}")
print(f"One minute: {one_minute}")
print(f"One second: {one_second}")
print(f"Ten milliseconds: {ten_milliseconds}")
print(f"Complex duration: {complex_duration}")
````

````python
# Calculate time differences
now = datetime.datetime.now()
future_time = now + datetime.timedelta(hours=3, minutes=45)
time_diff = future_time - now

print(f"Current time: {now}")
print(f"Future time: {future_time}")
print(f"Difference: {time_diff}")
print(f"Total seconds: {time_diff.total_seconds()}")

# Calculate age
birth_date = datetime.date(1990, 5, 15)
age_days = (today - birth_date).days
age_years = age_days // 365

print(f"\nBirth date: {birth_date}")
print(f"Age in days: {age_days}")
print(f"Approximate age in years: {age_years}")
````

````python
# Calculating business days
def is_business_day(date):
    return date.weekday() < 5  # 0-4 are Monday to Friday

def add_business_days(date, days):
    current_date = date
    remaining_days = days
    
    while remaining_days > 0:
        current_date += datetime.timedelta(days=1)
        if is_business_day(current_date):
            remaining_days -= 1
            
    return current_date

start_date = datetime.date.today()
end_date = add_business_days(start_date, 10)

print(f"Start date: {start_date} ({start_date.strftime('%A')})")
print(f"After 10 business days: {end_date} ({end_date.strftime('%A')})")
````

````python
# PyTZ is needed for timezone handling
# Let's explore available time zones
print(f"Total available time zones: {len(pytz.all_timezones)}")
print("\nSample time zones:")
for tz in pytz.all_timezones[:10]:  # Show first 10
    print(f"- {tz}")

print("\nCommon time zones:")
for tz in ['US/Eastern', 'US/Pacific', 'Europe/London', 'Asia/Tokyo', 'Australia/Sydney']:
    print(f"- {tz}")
````

````python
# Working with time zones
naive_now = datetime.datetime.now()  # No timezone info
utc_now = datetime.datetime.now(pytz.UTC)  # UTC timezone

print(f"Naive datetime: {naive_now}")
print(f"UTC datetime: {utc_now}")

# Create a timezone-aware datetime
eastern = pytz.timezone('US/Eastern')
eastern_time = eastern.localize(datetime.datetime.now().replace(microsecond=0))
print(f"Eastern time: {eastern_time}")

# Convert between time zones
pacific = pytz.timezone('US/Pacific')
tokyo = pytz.timezone('Asia/Tokyo')
london = pytz.timezone('Europe/London')

pacific_time = eastern_time.astimezone(pacific)
tokyo_time = eastern_time.astimezone(tokyo)
london_time = eastern_time.astimezone(london)

print(f"\nSame moment in different time zones:")
print(f"Eastern: {eastern_time}")
print(f"Pacific: {pacific_time}")
print(f"Tokyo: {tokyo_time}")
print(f"London: {london_time}")
````

````python
# Handling Daylight Saving Time
eastern = pytz.timezone('US/Eastern')

# Create two dates - one in winter, one in summer
winter_date = eastern.localize(datetime.datetime(2023, 1, 15, 12, 0))
summer_date = eastern.localize(datetime.datetime(2023, 7, 15, 12, 0))

print(f"Winter date: {winter_date}")
print(f"Summer date: {summer_date}")

# Compare with UTC
winter_utc = winter_date.astimezone(pytz.UTC)
summer_utc = summer_date.astimezone(pytz.UTC)

print(f"\nWinter in UTC: {winter_utc}")
print(f"Summer in UTC: {summer_utc}")
print(f"UTC offset in winter: {winter_date.strftime('%z')}")
print(f"UTC offset in summer: {summer_date.strftime('%z')}")
````

## 7. Parsing Dates from Strings

````python
# Parse dates from strings using strptime (string parse time)
date_string1 = "2023-05-15"
date_string2 = "15/08/2023"
date_string3 = "May 20, 2023"
date_string4 = "2023-05-15 14:30:45"

# Parse using different format strings
date1 = datetime.datetime.strptime(date_string1, "%Y-%m-%d")
date2 = datetime.datetime.strptime(date_string2, "%d/%m/%Y")
date3 = datetime.datetime.strptime(date_string3, "%B %d, %Y")
date4 = datetime.datetime.strptime(date_string4, "%Y-%m-%d %H:%M:%S")

print(f"Original string: {date_string1} -> Parsed: {date1}")
print(f"Original string: {date_string2} -> Parsed: {date2}")
print(f"Original string: {date_string3} -> Parsed: {date3}")
print(f"Original string: {date_string4} -> Parsed: {date4}")
````

````python
# Parsing dates with error handling
def parse_date(date_string, formats):
    """Try to parse a date string using multiple formats."""
    for fmt in formats:
        try:
            return datetime.datetime.strptime(date_string, fmt)
        except ValueError:
            continue
    raise ValueError(f"Cannot parse date string '{date_string}' with given formats")

# List of potential date formats
date_formats = [
    "%Y-%m-%d",           # 2023-05-15
    "%d/%m/%Y",           # 15/05/2023
    "%m/%d/%Y",           # 05/15/2023
    "%B %d, %Y",          # May 15, 2023
    "%d-%b-%Y",           # 15-May-2023
    "%Y-%m-%d %H:%M:%S",  # 2023-05-15 14:30:45
    "%m/%d/%y"            # 05/15/23
]

# Test with various strings
test_strings = [
    "2023-05-15",
    "15/05/2023",
    "05/15/2023",
    "May 15, 2023",
    "15-May-2023",
    "05/15/23"
]

for test_string in test_strings:
    try:
        parsed_date = parse_date(test_string, date_formats)
        print(f"Parsed '{test_string}' as {parsed_date}")
    except ValueError as e:
        print(f"Error: {e}")
````

````python
# Performance measurements with timestamps
def measure_execution_time(func, *args, **kwargs):
    """Measure execution time of a function."""
    start_time = time.time()
    result = func(*args, **kwargs)
    end_time = time.time()
    execution_time = end_time - start_time
    return result, execution_time

# Example function to measure
def fibonacci(n):
    """Calculate the nth Fibonacci number."""
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Measure execution time
n_values = [10, 15, 20, 25, 30]
results = []

for n in n_values:
    result, exec_time = measure_execution_time(fibonacci, n)
    results.append((n, result, exec_time))

print("Fibonacci execution time measurements:")
for n, result, exec_time in results:
    print(f"fibonacci({n}) = {result}, Time: {exec_time:.6f} seconds")
````

````python
# Time series analysis techniques

# --- Create a longer time series (3 years of monthly data) ---
dates = pd.date_range('2020-01-01', periods=36, freq='ME')  # <-- changed 'M' to 'ME'

# --- Create synthetic data with trend and seasonality ---
t = np.arange(len(dates))
trend = 0.1 * t  # Upward trend
seasonality = 5 * np.sin(2 * np.pi * t / 12)  # 12-month seasonality
noise = np.random.randn(len(dates)) * 2
data = trend + seasonality + noise

ts = pd.Series(data, index=dates)

# --- Plot the time series ---
plt.figure(figsize=(12, 6))
ts.plot()
plt.title('Time Series with Trend and Seasonality')
plt.xlabel('Date')
plt.ylabel('Value')
plt.grid(True)
plt.show()

# --- Calculate moving averages ---
ma3 = ts.rolling(window=3).mean()   # 3-month moving average
ma12 = ts.rolling(window=12).mean() # 12-month moving average

# --- Plot with moving averages ---
plt.figure(figsize=(12, 6))
ts.plot(label='Original')
ma3.plot(label='3-Month MA')
ma12.plot(label='12-Month MA')
plt.title('Time Series with Moving Averages')
plt.legend()
plt.grid(True)
plt.show()
````

````python
# Example 1: Date calculator
def date_calculator(start_date=None, days=0, months=0, years=0, business_days_only=False):
    """Calculate a future or past date based on the given parameters."""
    if start_date is None:
        start_date = datetime.date.today()
    elif isinstance(start_date, str):
        # Try to parse the date from string
        try:
            start_date = datetime.datetime.strptime(start_date, "%Y-%m-%d").date()
        except ValueError:
            raise ValueError("Date string must be in format YYYY-MM-DD")
    
    # Calculate year and month changes
    year = start_date.year + years
    month = start_date.month + months
    
    # Adjust year if month overflows
    year += (month - 1) // 12
    month = ((month - 1) % 12) + 1
    
    # Create intermediate date with year and month changes
    # Handle cases where the day might be invalid in the new month
    day = min(start_date.day, calendar.monthrange(year, month)[1])
    intermediate_date = datetime.date(year, month, day)
    
    # Add days
    if business_days_only:
        result_date = add_business_days(intermediate_date, days)
    else:
        result_date = intermediate_date + datetime.timedelta(days=days)
    
    return result_date

# Test the date calculator
print("Date Calculator Examples:")
print(f"Today: {datetime.date.today()}")
print(f"30 days from today: {date_calculator(days=30)}")
print(f"3 months from today: {date_calculator(months=3)}")
print(f"1 year ago: {date_calculator(years=-1)}")
print(f"10 business days from today: {date_calculator(days=10, business_days_only=True)}")
print(f"From 2023-01-31, add 1 month: {date_calculator('2023-01-31', months=1)}")
````

````python
# Example 2: Age calculator
def calculate_age(birth_date, reference_date=None):
    """Calculate age based on birth date and reference date."""
    if reference_date is None:
        reference_date = datetime.date.today()
    
    # Calculate years
    years = reference_date.year - birth_date.year
    
    # Adjust if birth date hasn't occurred yet this year
    if (reference_date.month, reference_date.day) < (birth_date.month, birth_date.day):
        years -= 1
        
    # Calculate months since last birthday
    if reference_date.day >= birth_date.day:
        months = reference_date.month - birth_date.month
    else:
        months = reference_date.month - birth_date.month - 1
        
    # Handle negative months (when birth month > reference month)
    if months < 0:
        months += 12
        
    # Calculate days
    if reference_date.day >= birth_date.day:
        days = reference_date.day - birth_date.day
    else:
        # Get last day of previous month
        last_month = reference_date.replace(day=1) - datetime.timedelta(days=1)
        days = reference_date.day + (last_month.day - birth_date.day)
        
    return {"years": years, "months": months, "days": days}

# Test the age calculator
birth_dates = [
    datetime.date(1990, 5, 15),  # Regular case
    datetime.date(2000, 1, 1),   # New Year's Day
    datetime.date(1985, 2, 29) if calendar.isleap(1985) else datetime.date(1984, 2, 29)  # Leap year
]

reference_date = datetime.date.today()
print("Age Calculator Examples:")
print(f"Reference date: {reference_date}")

for birth_date in birth_dates:
    age_data = calculate_age(birth_date, reference_date)
    print(f"Birth date: {birth_date} -> Age: {age_data['years']} years, {age_data['months']} months, {age_data['days']} days")
````

````python
# Example 3: Calendar generator
def generate_calendar(year, month):
    """Generate a calendar for the specified month and year."""
    # Get the calendar for the specified month
    cal = calendar.monthcalendar(year, month)
    
    # Get the month name
    month_name = calendar.month_name[month]
    
    # Print the calendar
    print(f"{month_name} {year}".center(20))
    print("Mo Tu We Th Fr Sa Su")
    
    for week in cal:
        week_str = ""
        for day in week:
            if day == 0:
                week_str += "   "
            else:
                week_str += f"{day:2d} "
        print(week_str)
        
# Generate a calendar for current month
today = datetime.date.today()
generate_calendar(today.year, today.month)

# Generate a calendar for December 2023
print("\n")
generate_calendar(2023, 12)
````

````python
# Example 4: Schedule creator
def create_schedule(start_date, days, start_time, duration_minutes, skip_weekends=True):
    """Create a schedule of dates and times."""
    if isinstance(start_date, str):
        start_date = datetime.datetime.strptime(start_date, "%Y-%m-%d").date()
        
    if isinstance(start_time, str):
        hours, minutes = map(int, start_time.split(':'))
        start_time = datetime.time(hours, minutes)
        
    current_date = start_date
    schedule = []
    
    while len(schedule) < days:
        # Skip weekends if requested
        if skip_weekends and current_date.weekday() >= 5:  # 5=Saturday, 6=Sunday
            current_date += datetime.timedelta(days=1)
            continue
            
        # Create datetime combining current date and start time
        start_datetime = datetime.datetime.combine(current_date, start_time)
        end_datetime = start_datetime + datetime.timedelta(minutes=duration_minutes)
        
        schedule.append({
            'date': current_date,
            'day_of_week': current_date.strftime('%A'),
            'start_time': start_time.strftime('%H:%M'),
            'end_time': end_datetime.time().strftime('%H:%M'),
            'duration_minutes': duration_minutes
        })
        
        current_date += datetime.timedelta(days=1)
        
    return schedule

# Create a schedule for a 5-day class
start_date = datetime.date.today()
class_schedule = create_schedule(start_date, 5, "09:00", 90)

# Display the schedule
print("5-Day Class Schedule:")
for i, slot in enumerate(class_schedule, 1):
    print(f"Day {i}: {slot['date']} ({slot['day_of_week']}) - {slot['start_time']} to {slot['end_time']}")
````

````python
# Example 5: Countdown timer
def countdown_to(target_date):
    """Calculate the time remaining until the target date."""
    if isinstance(target_date, str):
        target_date = datetime.datetime.strptime(target_date, "%Y-%m-%d %H:%M:%S")
    
    now = datetime.datetime.now()
    remaining = target_date - now
    
    if remaining.total_seconds() <= 0:
        return "The event has already passed!"
    
    # Calculate days, hours, minutes, and seconds
    total_seconds = int(remaining.total_seconds())
    days, remainder = divmod(total_seconds, 86400)
    hours, remainder = divmod(remainder, 3600)
    minutes, seconds = divmod(remainder, 60)
    
    return f"{days} days, {hours} hours, {minutes} minutes, {seconds} seconds"

# Set target dates
new_years_eve = f"{datetime.date.today().year}-12-31 23:59:59"
one_week_from_now = (datetime.datetime.now() + datetime.timedelta(days=7)).strftime("%Y-%m-%d %H:%M:%S")

# Calculate countdowns
print(f"Countdown to New Year's Eve: {countdown_to(new_years_eve)}")
print(f"Countdown to one week from now: {countdown_to(one_week_from_now)}")
````
