# Using pip and Package Management in Python

This notebook provides a comprehensive overview of Python package management with pip, covering installation, dependency management, virtual environments, and best practices.

## 1. Understanding Package Management

### What are Python Packages?

Python packages are collections of modules that extend Python's functionality. They allow developers to:
- Reuse code across projects
- Access community-developed tools and libraries
- Organize code in a modular way

### The Python Package Index (PyPI)

[PyPI](https://pypi.org/) is the official repository for Python packages with over 400,000 projects available for installation.

### Package Management Benefits

Effective package management helps you:
- Track project dependencies
- Create reproducible environments
- Avoid version conflicts
- Share code and dependencies consistently

## 2. Introduction to pip

`pip` is Python's standard package manager and comes pre-installed with Python 3.4 and newer. It provides a command-line interface for installing and managing Python packages.

```python
# Check if pip is installed and verify its version
!python -m pip --version
```

### Updating pip

It's a good practice to keep pip updated to the latest version to ensure compatibility with newer packages and security fixes.

```python
# Update pip to the latest version
!python -m pip install --upgrade pip
```

## 3. Installing Packages with pip

### Basic Package Installation

The most common use of pip is to install packages from PyPI.

```python
# Install a package (requests is a popular HTTP library)
!pip install requests
```

```python
# Let's verify that the package installed correctly
import requests

response = requests.get('https://httpbin.org/get')
print(f"Status Code: {response.status_code}")
print(f"Response: {response.json()}")
```

### Installing Specific Versions

Sometimes you need a specific version of a package for compatibility reasons.

```python
# Install a specific version of a package
!pip install numpy==1.22.3
```

```python
# Verify the installed version
import numpy as np
print(f"Installed NumPy version: {np.__version__}")
```

pip supports various version specifiers:

```python
# Examples of version specifiers (shown as comments to prevent execution)

# Install version 2.x.x (minimum 2.0.0, less than 3.0.0)
# !pip install 'pandas>=2.0.0,<3.0.0'

# Install the latest version but at least 4.10.0
# !pip install 'matplotlib>=4.10.0'

# Install any version except 1.2.3
# !pip install 'requests!=1.2.3'
```

## 8. Creating and Publishing Packages

### Basic Package Structure

A minimal Python package structure looks like this:

```python
# Visualization of a package directory structure
package_structure = """
my_package/
├── my_package/
│   ├── __init__.py
│   ├── module1.py
│   └── module2.py
├── tests/
│   └── test_my_package.py
├── README.md
├── setup.py
└── LICENSE
"""

print(package_structure)
```

### Creating a setup.py File

The `setup.py` file defines metadata about your package.

```python
%%writefile sample_setup.py
from setuptools import setup, find_packages

setup(
    name="my_package",
    version="0.1.0",
    author="Your Name",
    author_email="your.email@example.com",
    description="A short description of your package",
    long_description=open("README.md").read(),
    long_description_content_type="text/markdown",
    url="https://github.com/yourusername/my_package",
    packages=find_packages(),
    classifiers=[
        "Programming Language :: Python :: 3",
        "License :: OSI Approved :: MIT License",
        "Operating System :: OS Independent",
    ],
    python_requires=">=3.6",
    install_requires=[
        "numpy>=1.18.0",
        "pandas>=1.0.0",
    ],
)
```

### Building and Publishing a Package

These are the steps to build and publish a package to PyPI.

```python
# Commands for building and publishing a package (for reference)
build_publish_commands = """
# Install required tools
pip install setuptools wheel twine

# Build the package
python setup.py sdist bdist_wheel

# Upload to PyPI test server (optional)
twine upload --repository-url https://test.pypi.org/legacy/ dist/*

# Upload to PyPI
twine upload dist/*
"""

print(build_publish_commands)
```
