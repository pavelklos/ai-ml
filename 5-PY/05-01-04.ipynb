# Python Control Flow: If, Elif, and Else Statements

Control flow structures allow us to control the order in which statements are executed in our programs. Conditional statements, particularly the `if`, `elif`, and `else` statements, are fundamental tools that enable us to make decisions in our code based on certain conditions.

In this notebook, we'll explore Python's control flow mechanisms, focusing on conditional statements.

## Python Control Flow Basics

Control flow refers to the order in which individual statements, instructions, or function calls are executed in a program. In Python, we have several control flow tools:

1. **Conditional statements** (`if`, `elif`, `else`)
2. **Loops** (`for`, `while`)
3. **Function calls**
4. **Exception handling** (`try`, `except`, `finally`)

In this notebook, we'll focus on conditional statements, which let our programs make decisions based on whether certain conditions are true or false.

### Boolean Expressions

Before diving into conditional statements, let's understand boolean expressions. Boolean expressions evaluate to either `True` or `False`. These are the foundation of conditional statements.

```python
# Examples of boolean expressions
print(5 > 3)       # Greater than
print(10 == 10)    # Equal to
print(7 < 4)       # Less than
print(5 >= 5)      # Greater than or equal to
print(8 <= 4)      # Less than or equal to
print(6 != 7)      # Not equal to
```

## Simple Conditional Statements

The simplest form of a conditional statement in Python is the `if` statement. It executes a block of code only if a specified condition is true.

```python
# Basic if statement
age = 18

if age >= 18:
    print("You are an adult.")
    print("You can vote.")

print("This will always be executed regardless of the condition.")
```

### If-Else Statement

The `if-else` statement provides an alternative path of execution when the condition is false.

```python
# if-else statement
temperature = 15

if temperature > 25:
    print("It's a warm day!")
else:
    print("It's not very warm today.")

# Another example
is_raining = True

if is_raining:
    print("Remember to take an umbrella!")
else:
    print("No need for an umbrella today.")
```

## Multiple Conditions with elif

When you need to check multiple conditions, you can use `elif` (short for "else if"). This allows you to check additional conditions if the previous conditions are not met.

```python
# if-elif-else statement
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f"Your grade is: {grade}")
```

```python
# Another example with time of day
import datetime

current_hour = datetime.datetime.now().hour

if current_hour < 12:
    greeting = "Good morning!"
elif current_hour < 17:
    greeting = "Good afternoon!"
elif current_hour < 21:
    greeting = "Good evening!"
else:
    greeting = "Good night!"

print(greeting)
print(f"Current hour: {current_hour}")
```

## Nested Conditional Statements

You can place conditional statements inside other conditional statements. These are called nested conditionals.

```python
# Nested if statements
age = 25
has_license = True

if age >= 18:
    print("You are old enough to drive.")
    
    if has_license:
        print("You can drive legally.")
    else:
        print("You need to get a license first.")
else:
    print("You are too young to drive.")
```

```python
# A more complex example with nested conditions
user_type = "admin"
is_logged_in = True
access_level = 3

if is_logged_in:
    print("User is logged in.")
    
    if user_type == "admin":
        print("Welcome, Administrator!")
        
        if access_level >= 3:
            print("You have full access to the system.")
        else:
            print("You have limited administrative access.")
    
    elif user_type == "manager":
        print("Welcome, Manager!")
        print("You have access to management functions.")
    
    else:
        print("Welcome, User!")
        print("You have standard user access.")
else:
    print("Please log in to access the system.")
```

## Logical Operators in Conditions

Logical operators allow you to combine multiple conditions:

- `and`: True if both conditions are true
- `or`: True if at least one condition is true
- `not`: Inverts the truth value

```python
# Using logical AND
age = 25
income = 50000

if age > 21 and income > 30000:
    print("You qualify for the gold credit card.")
else:
    print("You do not qualify for the gold credit card.")
```

```python
# Using logical OR
is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("You can sleep in today!")
else:
    print("Set your alarm clock!")
```

```python
# Using logical NOT
is_working_day = True

if not is_working_day:
    print("Relax and enjoy your day off!")
else:
    print("Time to work!")
```

```python
# Complex logical expressions
age = 22
is_student = True
has_discount_card = False

if (age < 25 and is_student) or has_discount_card:
    print("You qualify for the discount!")
else:
    print("No discount available.")
```

## Conditional Expressions (Ternary Operator)

Python provides a concise way to write simple if-else statements using conditional expressions, also known as the ternary operator.

```python
# Traditional if-else
age = 20
if age >= 18:
    status = "adult"
else:
    status = "minor"
    
print(status)
```

```python
# Using ternary operator
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
```

```python
# More examples of ternary operators
score = 75
result = "Pass" if score >= 60 else "Fail"
print(f"Score: {score}, Result: {result}")

temperature = 28
weather = "Hot" if temperature > 25 else "Cold"
print(f"Temperature: {temperature}°C, Weather: {weather}")
```

## Common Patterns and Best Practices

Let's explore some common patterns and best practices when working with conditional statements in Python.

### 1. Avoiding Nested If Statements (Flattening Conditionals)

```python
# Deeply nested (harder to read)
def get_ticket_price(age, is_student, is_senior):
    if age < 12:
        return 5
    else:
        if is_student:
            return 10
        else:
            if is_senior:
                return 8
            else:
                return 15

# Flattened (easier to read)
def get_ticket_price_improved(age, is_student, is_senior):
    if age < 12:
        return 5
    if is_student:
        return 10
    if is_senior:
        return 8
    return 15

print(f"Original function: ${get_ticket_price(25, False, True)}")
print(f"Improved function: ${get_ticket_price_improved(25, False, True)}")
```

### 2. Using In Operator with Collections

```python
# Multiple if statements (verbose)
fruit = "apple"

if fruit == "apple" or fruit == "banana" or fruit == "orange":
    print(f"{fruit} is in our fruit basket.")
else:
    print(f"We don't have {fruit} in our fruit basket.")

# Using 'in' operator (cleaner)
fruits = ["apple", "banana", "orange"]

if fruit in fruits:
    print(f"{fruit} is in our fruit basket.")
else:
    print(f"We don't have {fruit} in our fruit basket.")
```

### 3. Truth Value Testing

```python
# Python objects have inherent truth values
# The following are considered False:
# - None
# - False
# - Zero (0, 0.0, 0j)
# - Empty sequences ('', [], (), {})
# Everything else is considered True

# Explicit check (more verbose)
names = []
if len(names) == 0:
    print("The list is empty.")
else:
    print("The list has some elements.")

# Truth value testing (cleaner)
if not names:  # Empty list evaluates to False
    print("The list is empty.")
else:
    print("The list has some elements.")
```

### 4. Early Returns in Functions

```python
# Using early returns for cleaner code
def check_password(password):
    # Early returns for invalid cases
    if len(password) < 8:
        return "Password too short"
    
    if not any(char.isdigit() for char in password):
        return "Password should contain at least one digit"
    
    if not any(char.isupper() for char in password):
        return "Password should contain at least one uppercase letter"
    
    # If we get here, all checks passed
    return "Password is valid"

print(check_password("password"))  # Too short
print(check_password("longpassword"))  # No digits or uppercase
print(check_password("Longpassword"))  # No digits
print(check_password("Longpassword123"))  # Valid
```

## Practical Examples

Let's look at some practical examples of using conditional statements in real-world scenarios.

### Example 1: Simple Calculator

```python
def simple_calculator(num1, num2, operation):
    """A simple calculator function."""
    if operation == '+':
        return num1 + num2
    elif operation == '-':
        return num1 - num2
    elif operation == '*':
        return num1 * num2
    elif operation == '/':
        if num2 == 0:  # Check for division by zero
            return "Error: Cannot divide by zero"
        return num1 / num2
    else:
        return "Error: Invalid operation"

# Test the calculator
print(simple_calculator(10, 5, '+'))  # Addition
print(simple_calculator(10, 5, '-'))  # Subtraction
print(simple_calculator(10, 5, '*'))  # Multiplication
print(simple_calculator(10, 5, '/'))  # Division
print(simple_calculator(10, 0, '/'))  # Division by zero
print(simple_calculator(10, 5, '%'))  # Invalid operation
```

### Example 2: BMI Calculator

```python
def calculate_bmi(weight_kg, height_m):
    """Calculate BMI and return the category."""
    if height_m <= 0:
        return "Error: Height must be positive"
    
    bmi = weight_kg / (height_m ** 2)
    
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal weight"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    
    return f"BMI: {bmi:.2f}, Category: {category}"

# Test with different values
print(calculate_bmi(70, 1.75))  # Normal weight
print(calculate_bmi(50, 1.80))  # Underweight
print(calculate_bmi(90, 1.70))  # Overweight
print(calculate_bmi(110, 1.65))  # Obese
print(calculate_bmi(70, 0))     # Error
```

### Example 3: Weather Clothing Advisor

```python
def clothing_advisor(temperature, is_raining, is_windy):
    """Suggest clothing based on weather conditions."""
    clothing = []
    
    # Temperature-based suggestions
    if temperature < 0:
        clothing.append("heavy winter coat")
        clothing.append("gloves")
        clothing.append("scarf")
        clothing.append("hat")
    elif temperature < 10:
        clothing.append("winter coat")
        clothing.append("light gloves")
    elif temperature < 20:
        clothing.append("light jacket or sweater")
    elif temperature < 30:
        clothing.append("t-shirt")
    else:
        clothing.append("light clothing")
        clothing.append("hat for sun protection")
    
    # Rain-based suggestions
    if is_raining:
        clothing.append("umbrella")
        clothing.append("waterproof shoes")
        if temperature > 20:  # It's warm but raining
            clothing.append("light raincoat")
        else:
            clothing.append("raincoat")
    
    # Wind-based suggestions
    if is_windy and temperature < 15:
        clothing.append("windbreaker")
    
    return "Suggested clothing: " + ", ".join(clothing)

# Test different weather conditions
print(clothing_advisor(5, False, True))    # Cold and windy
print(clothing_advisor(25, True, False))   # Warm and raining
print(clothing_advisor(-5, True, True))    # Very cold, raining, and windy
print(clothing_advisor(30, False, False))  # Hot day
```

### Example 4: Movie Ticket Pricing System

```python
def calculate_ticket_price(age, day_of_week, is_3d, is_imax):
    """Calculate movie ticket price based on various factors."""
    base_price = 12.00  # Standard adult ticket price
    
    # Age discounts
    if age < 5:
        return "Free for children under 5"
    elif age < 13:
        base_price *= 0.7  # 30% off for children
    elif age >= 65:
        base_price *= 0.8  # 20% off for seniors
    
    # Day of week adjustments
    if day_of_week.lower() == 'tuesday':  # Discount Tuesday
        base_price *= 0.75  # 25% off on Tuesdays
    elif day_of_week.lower() in ['friday', 'saturday', 'sunday']:  # Weekend premium
        base_price *= 1.15  # 15% premium on weekends
    
    # Format adjustments
    if is_3d:
        base_price += 3.50  # 3D surcharge
    if is_imax:
        base_price += 5.00  # IMAX surcharge
    
    return f"${base_price:.2f}"

# Test different scenarios
print(f"Child on Tuesday (regular): {calculate_ticket_price(10, 'Tuesday', False, False)}")
print(f"Adult on Weekend (3D): {calculate_ticket_price(35, 'Saturday', True, False)}")
print(f"Senior on Monday (IMAX): {calculate_ticket_price(70, 'Monday', False, True)}")
print(f"Adult on Weekend (3D IMAX): {calculate_ticket_price(40, 'Sunday', True, True)}")
print(f"Toddler any day: {calculate_ticket_price(3, 'Wednesday', False, False)}")
```

### Example 5: Password Strength Checker

```python
def check_password_strength(password):
    """Check the strength of a password and provide feedback."""
    strength = 0
    feedback = []
    
    # Check length
    if len(password) < 8:
        feedback.append("Password is too short. Use at least 8 characters.")
    elif len(password) >= 12:
        strength += 2
        feedback.append("Good length!")
    else:
        strength += 1
    
    # Check for lowercase letters
    if any(char.islower() for char in password):
        strength += 1
    else:
        feedback.append("Include lowercase letters.")
    
    # Check for uppercase letters
    if any(char.isupper() for char in password):
        strength += 1
    else:
        feedback.append("Include uppercase letters.")
    
    # Check for digits
    if any(char.isdigit() for char in password):
        strength += 1
    else:
        feedback.append("Include numbers.")
    
    # Check for special characters
    if any(not char.isalnum() for char in password):
        strength += 1
    else:
        feedback.append("Include special characters (!@#$%^&*).")
    
    # Check for common patterns
    common_passwords = ["password", "123456", "qwerty", "admin"]
    if password.lower() in common_passwords:
        strength = 0
        feedback = ["This is a commonly used password. Please choose something more secure."]
    
    # Determine strength category
    if strength < 2:
        category = "Very Weak"
    elif strength < 3:
        category = "Weak"
    elif strength < 4:
        category = "Moderate"
    elif strength < 5:
        category = "Strong"
    else:
        category = "Very Strong"
    
    # If password is good, but we have no feedback yet, add positive feedback
    if not feedback and strength >= 4:
        feedback.append("Excellent password!")
    
    return {
        "strength": strength,
        "category": category,
        "feedback": feedback
    }

# Test different passwords
passwords = [
    "pass",  # Too short
    "password",  # Common
    "Password123",  # Missing special characters
    "p@ssw0rd",  # Missing uppercase
    "P@ssw0rd123!"  # Strong password
]

for password in passwords:
    result = check_password_strength(password)
    print(f"Password: {password}")
    print(f"Strength: {result['strength']}/6")
    print(f"Category: {result['category']}")
    print(f"Feedback: {', '.join(result['feedback'])}")
    print("-----------------------------")
```
