## String Basics

In Python, a string is a sequence of characters enclosed within quotes. Strings are immutable, which means once created, they cannot be modified. Any operation that appears to modify a string actually creates a new string.

```python
# Creating strings with different types of quotes
single_quotes = 'Hello, Python!'
double_quotes = "Hello, Python!"
triple_quotes = '''This is a multi-line
string using triple single quotes.'''
triple_double_quotes = """This is also a multi-line
string using triple double quotes."""

print(single_quotes)
print(double_quotes)
print(triple_quotes)
print(triple_double_quotes)
```

### String Immutability

Strings in Python are immutable, which means you cannot change individual characters of a string after it's created.

```python
# Demonstrating string immutability
word = "Python"
print(f"Original string: {word}")

# This will raise a TypeError
try:
    word[0] = 'J'
except TypeError as e:
    print(f"Error: {e}")

# Instead, create a new string
new_word = 'J' + word[1:]
print(f"New string: {new_word}")
```

### Strings vs Other Data Types

Strings differ from other data types in several ways:

```python
# Comparing strings with other data types
string_value = "42"
integer_value = 42
float_value = 42.0

# Type checking
print(f"string_value is of type: {type(string_value)}")
print(f"integer_value is of type: {type(integer_value)}")
print(f"float_value is of type: {type(float_value)}")

# Conversions between types
converted_to_int = int(string_value)
converted_to_string = str(integer_value)
converted_to_float = float(string_value)

print(f"\nAfter conversion:")
print(f"int(string_value) = {converted_to_int}, type: {type(converted_to_int)}")
print(f"str(integer_value) = {converted_to_string}, type: {type(converted_to_string)}")
print(f"float(string_value) = {converted_to_float}, type: {type(converted_to_float)}")

# Operations differ across types
print("\nOperation examples:")
print(f"string_value + string_value = {string_value + string_value}")  # Concatenation for strings
print(f"integer_value + integer_value = {integer_value + integer_value}")  # Addition for integers
print(f"string_value * 3 = {string_value * 3}")  # Repetition for strings
print(f"integer_value * 3 = {integer_value * 3}")  # Multiplication for integers
```

## String Creation and Formatting

Python provides multiple ways to create and format strings to suit different needs.

```python
# Literal strings
simple_string = "This is a simple string"

# Multiline strings
multiline_string = """This is a multiline string.
It spans multiple lines.
You can write paragraphs this way."""

# Raw strings (ignore escape sequences)
raw_string = r"C:\Users\Documents\new_folder"  # Backslashes are not treated as escape characters

# F-strings (formatted string literals, available in Python 3.6+)
name = "Alice"
age = 30
f_string = f"My name is {name} and I am {age} years old."

# Expressions in f-strings
calculation = f"5 * 7 = {5 * 7}"

print(simple_string)
print("\n" + multiline_string)
print("\n" + raw_string)
print("\n" + f_string)
print(calculation)
```

### More Advanced F-string Features

```python
# Advanced f-string formatting (Python 3.6+)

# Number formatting
pi = 3.14159265359
print(f"Pi rounded to 2 decimal places: {pi:.2f}")

# Date formatting
from datetime import datetime
now = datetime.now()
print(f"Current date: {now:%Y-%m-%d}")
print(f"Current time: {now:%H:%M:%S}")

# Alignment
print(f"{'Left':<10}|{'Center':^10}|{'Right':>10}")
print(f"{'-'*10}+{'-'*10}+{'-'*10}")

# Using dictionaries with f-strings
person = {'name': 'Bob', 'age': 25}
print(f"Person's name is {person['name']} and age is {person['age']}")

# Using objects with f-strings
class User:
    def __init__(self, name, email):
        self.name = name
        self.email = email

user = User("Charlie", "charlie@example.com")
print(f"User {user.name} has email {user.email}")
```

## String Methods

Python provides a rich set of built-in methods for string manipulation. Here are some of the most commonly used string methods:

```python
# Let's define a sample string to work with
sample = "  Hello, World! Python is amazing.  "

# Basic information
print(f"Original string: '{sample}'")
print(f"Length: {len(sample)}")

# Trimming whitespace
print(f"\nstrip(): '{sample.strip()}'")
print(f"lstrip(): '{sample.lstrip()}'")
print(f"rstrip(): '{sample.rstrip()}'")

# Case conversion
print(f"\nuppercase: '{sample.upper()}'")
print(f"lowercase: '{sample.lower()}'")
print(f"title case: '{sample.title()}'")
print(f"capitalize: '{sample.capitalize()}'")
print(f"swapcase: '{sample.swapcase()}'")

# Searching methods
print(f"\nContains 'Python': {'Python' in sample}")
print(f"Index of 'Python': {sample.find('Python')}")  # Returns -1 if not found
try:
    print(f"Index of 'Python' (index()): {sample.index('Python')}")  # Raises ValueError if not found
except ValueError as e:
    print(f"Error with index(): {e}")

print(f"Count of 'o': {sample.count('o')}")
print(f"Starts with 'Hello': {sample.strip().startswith('Hello')}")
print(f"Ends with 'amazing.': {sample.strip().endswith('amazing.')}")

# Splitting and joining
words = sample.strip().split()
print(f"\nSplit into words: {words}")

comma_split = sample.split(',')
print(f"Split by comma: {comma_split}")

join_result = "-".join(words)
print(f"Joined with hyphens: {join_result}")

# Replace
print(f"\nReplace 'Python' with 'Programming': '{sample.replace('Python', 'Programming')}'")

# Check string content
print(f"\nIs alphanumeric: {sample.isalnum()}")
print(f"Is alphabetic: {sample.isalpha()}")
print(f"Is numeric: {sample.isnumeric()}")
print(f"Is decimal: {sample.isdecimal()}")
print(f"Is digit: {sample.isdigit()}")
print(f"Is lower: {sample.islower()}")
print(f"Is upper: {sample.isupper()}")
print(f"Is title: {sample.istitle()}")
print(f"Is space: {sample.isspace()}")

# For a cleaner check, let's use specific strings
print(f"\n'abc123'.isalnum(): {'abc123'.isalnum()}")
print(f"'abc'.isalpha(): {'abc'.isalpha()}")
print(f"'123'.isnumeric(): {'123'.isnumeric()}")
print(f"'123'.isdecimal(): {'123'.isdecimal()}")
print(f"'123'.isdigit(): {'123'.isdigit()}")
print(f"'hello'.islower(): {'hello'.islower()}")
print(f"'HELLO'.isupper(): {'HELLO'.isupper()}")
print(f"'Hello World'.istitle(): {'Hello World'.istitle()}")
print(f"'   '.isspace(): {'   '.isspace()}")
```

### Additional String Methods

Let's explore some more string methods that are useful for specific tasks:

```python
# Additional string methods

# Center, ljust, rjust - for alignment
text = "Python"
print(f"'{text.center(20)}'")
print(f"'{text.ljust(20)}'")
print(f"'{text.rjust(20)}'")

# Filling with characters
print(f"'{text.center(20, '*')}'")
print(f"'{text.ljust(20, '-')}'")
print(f"'{text.rjust(20, '=')}'")

# zfill - pad with zeros (useful for numbers)
number = "42"
print(f"'{number.zfill(5)}'")

# expandtabs - adjust tab size
tabbed_text = "Name\tAge\tCity"
print("Default tabs:")
print(tabbed_text)
print("\nAdjusted tabs (expandtabs(16)):")
print(tabbed_text.expandtabs(16))

# partition and rpartition
email = "user@example.com"
print(f"\npartition('@'): {email.partition('@')}")
print(f"rpartition('@'): {email.rpartition('@')}")

# splitlines - split by newline characters
multiline = "Line 1\nLine 2\nLine 3"
print(f"\nsplitlines(): {multiline.splitlines()}")

# translate - map characters using translation table
translation_table = str.maketrans({"a": "4", "e": "3", "i": "1", "o": "0"})
leet_speak = "elite hacker code"
print(f"\ntranslate: '{leet_speak.translate(translation_table)}'")

# encode and decode
original = "Pythön is great"
encoded = original.encode("utf-8")
print(f"\nencoded: {encoded}")
decoded = encoded.decode("utf-8")
print(f"decoded: {decoded}")
```

## String Manipulation Techniques

Now let's explore common techniques for string manipulation.

```python
# 1. String concatenation
first_name = "John"
last_name = "Doe"

# Using + operator
full_name = first_name + " " + last_name
print(f"Concatenation using + operator: {full_name}")

# Using join method
full_name_join = " ".join([first_name, last_name])
print(f"Concatenation using join: {full_name_join}")

# 2. String repetition
separator = "-" * 20
print(f"Repetition: {separator}")

# 3. Substring extraction
sentence = "The quick brown fox jumps over the lazy dog"
word = sentence[4:9]  # Extract 'quick'
print(f"\nExtracted substring: '{word}'")

# 4. String traversal using loops
print("\nTraversing a string:")
for char in "Python":
    print(f"Character: {char}")

# 5. Character counting
text = "banana"
char_count = {}
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1
print(f"\nCharacter count in '{text}': {char_count}")

# 6. Removing specific characters
dirty_string = "H!e@l#l$o%^&*"
clean_string = ''.join(char for char in dirty_string if char.isalnum())
print(f"\nCleaned string: '{clean_string}'")

# 7. Reversing a string
original = "Python"
reversed_string = original[::-1]
print(f"\nReversed string: '{reversed_string}'")

# Another way to reverse
reversed_manual = ''.join(reversed(original))
print(f"Reversed using reversed(): '{reversed_manual}'")

# 8. Check if a string is palindrome
def is_palindrome(s):
    # Remove spaces and convert to lowercase for accurate comparison
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print(f"\n'radar' is palindrome: {is_palindrome('radar')}")
print(f"'hello' is palindrome: {is_palindrome('hello')}")
print(f"'A man a plan a canal Panama' is palindrome: {is_palindrome('A man a plan a canal Panama')}")

# 9. Converting case
mixed_case = "This Is Mixed Case"
print(f"\nSwapping case: '{mixed_case.swapcase()}'")

# 10. Word count
sentence = "This is a simple sentence to count words in a string"
word_count = len(sentence.split())
print(f"\nNumber of words: {word_count}")
```

## Regular Expressions for Advanced String Operations

Regular expressions (regex) provide powerful pattern matching and manipulation capabilities for strings. Python's `re` module offers comprehensive support for regular expressions.

```python
import re

# Sample text for demonstration
text = """
Contact information:
John Doe, john.doe@example.com, (555) 123-4567
Jane Smith, jane_smith@company.co.uk, +1-987-654-3210
Bob Johnson, bob123@gmail.com, 555.777.8888
"""

print("Original text:")
print(text)
```

```python
# 1. Basic searching with re.search()
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'

first_email_match = re.search(email_pattern, text)
if first_email_match:
    print(f"First email found: {first_email_match.group()}")
else:
    print("No email found.")
```

```python
# 2. Finding all instances with re.findall()
all_emails = re.findall(email_pattern, text)
print(f"All emails found: {all_emails}")

# Find all phone numbers (various formats)
phone_pattern = r'(\(\d{3}\)\s*\d{3}-\d{4}|\+\d{1}-\d{3}-\d{3}-\d{4}|\d{3}\.\d{3}\.\d{4})'
all_phones = re.findall(phone_pattern, text)
print(f"All phone numbers found: {all_phones}")
```

```python
# 3. Regular expression substitution with re.sub()
# Format all phone numbers consistently
standardized_text = re.sub(
    r'(\(\d{3}\)\s*\d{3}-\d{4}|\+\d{1}-\d{3}-\d{3}-\d{4}|\d{3}\.\d{3}\.\d{4})',
    '[PHONE NUMBER]',
    text
)
print("Text with standardized phone numbers:")
print(standardized_text)

# Replace emails with redacted version
redacted_text = re.sub(email_pattern, '[EMAIL REDACTED]', text)
print("\nText with redacted emails:")
print(redacted_text)
```

```python
# 4. Advanced regex patterns with groups
name_email_pattern = r'([A-Z][a-z]+ [A-Z][a-z]+), ([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,})'

# Find all name-email pairs
name_email_pairs = re.findall(name_email_pattern, text)
print("Name and email pairs:")
for name, email in name_email_pairs:
    print(f"Name: {name}, Email: {email}")

# 5. Using re.match() - matches only at the beginning of the string
first_line_match = re.match(r'(.+)', text.strip())
if first_line_match:
    print(f"\nFirst line: {first_line_match.group(1)}")
```

```python
# 6. Using re.compile() for performance when using the same pattern multiple times
compiled_email_pattern = re.compile(email_pattern)

# Using the compiled pattern
first_email = compiled_email_pattern.search(text)
all_emails = compiled_email_pattern.findall(text)

print(f"\nUsing compiled pattern:")
print(f"First email: {first_email.group() if first_email else 'None'}")
print(f"All emails: {all_emails}")

# 7. Splitting text with re.split()
split_pattern = r',\s*'
first_contact_line = text.strip().split('\n')[1]
contact_parts = re.split(split_pattern, first_contact_line)

print(f"\nSplit result for '{first_contact_line}':")
for i, part in enumerate(contact_parts):
    print(f"Part {i+1}: {part}")
```

### Regular Expression Flags and Special Sequences

```python
# Regex flags
text_with_newlines = """This is line 1.
This is line 2.
THIS IS LINE 3 (uppercase).
This is line 4."""

# Case-insensitive matching
case_insensitive_search = re.findall(r'this', text_with_newlines, re.IGNORECASE)
print(f"Case-insensitive search for 'this': {case_insensitive_search}")

# Multi-line mode - ^ and $ match start/end of each line
multiline_search = re.findall(r'^This', text_with_newlines, re.MULTILINE)
print(f"Multi-line search for lines starting with 'This': {multiline_search}")

# Dot matches all (including newline)
dotall_search = re.search(r'line 1.*line 3', text_with_newlines, re.DOTALL)
print(f"DOTALL search result: {dotall_search.group() if dotall_search else 'No match'}")

# Special sequences
sample_text = "Call 555-123-4567 or visit room A-123 on 12/25/2023"

print("\nSpecial sequence examples:")
print(f"\\d (digits): {re.findall(r'\d+', sample_text)}")
print(f"\\w (word chars): {re.findall(r'\w+', sample_text)}")
print(f"\\s (whitespace): {re.findall(r'\s', sample_text)}")
print(f"\\b (word boundary): {re.findall(r'\b\w{4}\b', sample_text)}")  # 4-letter words
```

## String Case Handling

Python provides numerous methods for changing the case of strings. Let's explore them with examples.

```python
# String case conversion methods demonstration

sample_text = "Python programming LANGUAGE is Amazing"
print(f"Original text: '{sample_text}'")

# Convert to uppercase
uppercase = sample_text.upper()
print(f"\nupper(): '{uppercase}'")

# Convert to lowercase
lowercase = sample_text.lower()
print(f"lower(): '{lowercase}'")

# Capitalize first character, rest lowercase
capitalized = sample_text.capitalize()
print(f"capitalize(): '{capitalized}'")

# Title case (capitalize first letter of each word)
title_case = sample_text.title()
print(f"title(): '{title_case}'")

# Swap case (lowercase becomes uppercase and vice versa)
swapped_case = sample_text.swapcase()
print(f"swapcase(): '{swapped_case}'")

# Demonstration of when to use each method
print("\nPractical examples for when to use each method:")

# upper() - useful for case-insensitive comparisons
user_input = "yes"
if user_input.upper() == "YES":
    print("upper() example: User confirmed with 'yes'")

# lower() - normalizing data
emails = ["User@Example.com", "admin@COMPANY.com", "Support@service.COM"]
normalized_emails = [email.lower() for email in emails]
print(f"lower() example: Normalized emails: {normalized_emails}")

# capitalize() - for sentence formatting
sentence = "welcome to python programming."
print(f"capitalize() example: {sentence.capitalize()}")

# title() - for names, titles, and headings
name = "john doe"
print(f"title() example: {name.title()}")

# swapcase() - less common, but useful for special formatting
code = "camelCaseExample"
print(f"swapcase() example: {code.swapcase()}")

# casefold() - for case-insensitive comparisons (more aggressive than lower())
german_word1 = "straße"  # German word for "street"
german_word2 = "STRASSE"  # Uppercase version with standard "ss"
print(f"\nCasefold comparison:")
print(f"lower(): '{german_word1.lower()}' vs '{german_word2.lower()}'")
print(f"casefold(): '{german_word1.casefold()}' vs '{german_word2.casefold()}'")
print(f"Are equal with lower()? {german_word1.lower() == german_word2.lower()}")
print(f"Are equal with casefold()? {german_word1.casefold() == german_word2.casefold()}")
```

## String Slicing and Indexing

Python provides powerful string indexing and slicing capabilities that allow you to extract specific portions of strings.

```python
# Sample string for demonstration
text = "Python Programming"
print(f"Original string: '{text}'")
print(f"Length: {len(text)}")

# Basic indexing (accessing individual characters)
print("\n--- Basic Indexing ---")
print(f"First character: {text[0]}")
print(f"Fifth character: {text[4]}")
print(f"Last character: {text[-1]}")
print(f"Second to last character: {text[-2]}")

# Basic slicing (extracting substrings)
print("\n--- Basic Slicing ---")
print(f"First 6 characters: '{text[0:6]}'")  # From index 0 up to (but not including) index 6
print(f"Alternative first 6 characters: '{text[:6]}'")  # Omitting start index defaults to 0

print(f"From 7th character to end: '{text[7:]}'")  # Omitting end index goes to the end
print(f"From 2nd to 9th character: '{text[2:10]}'")  # From index 2 up to (but not including) index 10

# Negative indexing in slices
print("\n--- Slicing with Negative Indices ---")
print(f"Last 11 characters: '{text[-11:]}'")
print(f"Everything except last 11 characters: '{text[:-11]}'")
print(f"From 3rd character to 3rd from last: '{text[3:-3]}'")

# Step values in slices
print("\n--- Slicing with Step Values ---")
print(f"Every 2nd character: '{text[::2]}'")
print(f"Every 3rd character from index 1 to 12: '{text[1:13:3]}'")

# Reverse the string using slicing
print(f"Reversed string: '{text[::-1]}'")
print(f"Every 2nd character, reversed: '{text[::-2]}'")

# Examples of more complex slicing
print("\n--- Complex Slicing Examples ---")
print(f"Last 5 characters, reversed: '{text[-1:-6:-1]}'")
print(f"From middle to beginning, every 2nd character: '{text[len(text)//2::-2]}'")

# Practical examples
print("\n--- Practical Examples ---")
url = "https://www.example.com/path/to/resource.html"
print(f"URL: {url}")
print(f"Protocol: {url[:url.find('://')]}") 
print(f"Domain: {url[url.find('://')+3:url.find('/', 8)]}")
print(f"Path: {url[url.find('/', 8):]}")
print(f"File extension: {url[url.rfind('.')+1:]}")

# Another example with email
email = "user.name@example.com"
print(f"\nEmail: {email}")
print(f"Username: {email[:email.find('@')]}")
print(f"Domain: {email[email.find('@')+1:]}")

# Handling out-of-range indices
print("\n--- Handling Out-of-Range ---")
# This would cause IndexError
try:
    print(text[100])
except IndexError as e:
    print(f"Error accessing index 100: {e}")

# But slices handle out-of-range gracefully
print(f"Slice beyond end: '{text[10:100]}'")  # Works fine, just stops at the end
print(f"Completely out-of-range slice: '{text[100:200]}'")  # Returns empty string
```

## String Formatting Options

Python offers multiple ways to format strings. Let's compare the different approaches.

```python
# Sample variables for demonstration
name = "Alice"
age = 30
height = 1.75
languages = ["Python", "JavaScript", "Go"]
data = {"city": "New York", "country": "USA"}

print("--- Different String Formatting Methods ---")

# 1. Old-style % formatting (still works but less recommended)
old_style = "My name is %s. I am %d years old and %.2f meters tall." % (name, age, height)
print(f"% Operator: {old_style}")

# With dictionary
old_style_dict = "I live in %(city)s, %(country)s." % data
print(f"% Operator with dict: {old_style_dict}")

# 2. str.format() method
format_method = "My name is {}. I am {} years old and {:.2f} meters tall.".format(name, age, height)
print(f"str.format(): {format_method}")

# With positional indices
format_indexed = "My name is {0}. I am {1} years old. My name is still {0}.".format(name, age)
print(f"str.format() with indices: {format_indexed}")

# With named parameters
format_named = "My name is {name}. I am {age} years old and {height:.2f} meters tall.".format(
    name=name, age=age, height=height
)
print(f"str.format() with named params: {format_named}")

# Using dictionaries with ** unpacking
format_dict = "I live in {city}, {country}.".format(**data)
print(f"str.format() with dict unpacking: {format_dict}")

# 3. f-strings (available in Python 3.6+)
f_string = f"My name is {name}. I am {age} years old and {height:.2f} meters tall."
print(f"f-string: {f_string}")

# Expressions in f-strings
f_string_expr = f"In 5 years, I will be {age + 5} years old."
print(f"f-string with expression: {f_string_expr}")

# f-strings with dictionaries
f_string_dict = f"I live in {data['city']}, {data['country']}."
print(f"f-string with dictionary: {f_string_dict}")

print("\n--- Formatting Options ---")

# Number formatting
pi = 3.14159265359
print(f"Decimal places: {pi:.3f}")
print(f"Percentage: {0.25:.1%}")
print(f"Scientific notation: {1000000:.2e}")
print(f"With commas: {1234567:.2f}")
print(f"With commas (integer): {1234567:,}")

# Width and alignment
print(f"\nDefault: '{name}'")
print(f"Right aligned (10): '{name:>10}'")
print(f"Left aligned (10): '{name:<10}'")
print(f"Center aligned (10): '{name:^10}'")

# Fill character
print(f"Fill with dashes: '{name:-^10}'")
print(f"Fill with stars: '{name:*>10}'")

# Combining with number formatting
print(f"\nFormatted number: '{pi:10.2f}'")
print(f"Right-aligned number: '{pi:>10.2f}'")
print(f"Left-aligned number: '{pi:<10.2f}'")

# Format integers
print(f"\nDecimal: {42:d}")
print(f"Binary: {42:b}")
print(f"Octal: {42:o}")
print(f"Hexadecimal (lowercase): {42:x}")
print(f"Hexadecimal (uppercase): {42:X}")
print(f"With prefix: {42:#x}")

# Using format() function separately
print("\nUsing format() function:")
print(format(pi, ".2f"))
print(format(name, "^10"))
print(format(age, "04d"))  # Zero-padded to 4 digits
```

## Working with Special Characters

Python strings can contain special characters, escape sequences, Unicode characters, and even emojis. Let's explore how to work with them.

```python
# 1. Escape sequences
print("--- Escape Sequences ---")
print("Line 1\nLine 2")  # Newline
print("Name:\tAlice")    # Tab
print("This is a \"quoted\" text")  # Quotes
print("Path: C:\\Users\\Documents")  # Backslash
print("Bell sound: \a")  # Bell/alert (may not produce sound in all environments)

# Raw string to avoid escape sequence processing
print("\nRaw string:")
print(r"Path: C:\Users\Documents\file.txt")

# 2. Unicode characters
print("\n--- Unicode Characters ---")
print("\u03B1")  # Greek alpha character
print("\u03B2")  # Greek beta character
print("\u03B3")  # Greek gamma character
print("\u2665")  # Heart symbol
print("\u2764")  # Heavy heart symbol
print("\u00A9")  # Copyright symbol
print("\u00AE")  # Registered trademark symbol
print("\u221E")  # Infinity symbol

# Named Unicode characters
print("\nNamed Unicode:")
print("\N{GREEK CAPITAL LETTER DELTA}")
print("\N{SMILING FACE WITH SUNGLASSES}")

# 3. Emojis (which are actually Unicode characters)
print("\n--- Emojis ---")
print("\U0001F600")  # Grinning Face
print("\U0001F604")  # Smile with Open Mouth
print("\U0001F44D")  # Thumbs Up
print("\U0001F4BB")  # Laptop
print("\U0001F40D")  # Snake
print("\U0001F981")  # Lion

# 4. String encoding and decoding
print("\n--- Encoding and Decoding ---")
text = "Python is 👍"
print(f"Original: {text}")

# UTF-8 encoding (common for web and files)
utf8_bytes = text.encode('utf-8')
print(f"UTF-8 encoded bytes: {utf8_bytes}")
print(f"Decoded back from UTF-8: {utf8_bytes.decode('utf-8')}")

# ASCII encoding (will fail with non-ASCII characters without error handling)
try:
    ascii_bytes = text.encode('ascii')
    print(f"ASCII encoded: {ascii_bytes}")
except UnicodeEncodeError as e:
    print(f"ASCII encoding error: {e}")

# With error handling
ascii_bytes_ignore = text.encode('ascii', errors='ignore')
print(f"ASCII encoded (ignore errors): {ascii_bytes_ignore}")
print(f"Decoded back from ASCII (ignore): {ascii_bytes_ignore.decode('ascii')}")

ascii_bytes_replace = text.encode('ascii', errors='replace')
print(f"ASCII encoded (replace errors): {ascii_bytes_replace}")
print(f"Decoded back from ASCII (replace): {ascii_bytes_replace.decode('ascii')}")

# 5. Character code points
print("\n--- Character Code Points ---")
for char in "Python👍":
    print(f"Character: '{char}', Unicode: {ord(char)}, Hex: {hex(ord(char))}")

# Converting from code point to character
print(f"\nCharacter from code point 80: {chr(80)}")
print(f"Character from code point 128077 (👍): {chr(128077)}")

# 6. Special whitespace and control characters
print("\n--- Special Whitespace ---")
print("Non-breaking space: 'A\u00A0B'")
print("Zero-width space: 'A\u200BB'")
print("Carriage return + Line feed: 'Line 1\r\nLine 2'")
```

## Practical String Manipulation Examples

Let's explore some real-world examples of string manipulation in Python.

```python
# Example 1: Text Processing - Simple Word Frequency Counter
import re
from collections import Counter

def word_frequency(text):
    """Count frequency of words in text."""
    # Convert to lowercase and split into words (removing punctuation)
    words = re.findall(r'\b\w+\b', text.lower())
    return Counter(words)

sample_text = """
Python is an interpreted, high-level, general-purpose programming language.
Python's design philosophy emphasizes code readability with its notable use of significant whitespace.
Python is dynamically typed and garbage-collected. It supports multiple programming paradigms,
including structured, object-oriented, and functional programming.
"""

word_counts = word_frequency(sample_text)
print("--- Word Frequency Analysis ---")
print("Top 5 most common words:")
for word, count in word_counts.most_common(5):
    print(f"{word}: {count}")
```

```python
# Example 2: Data Cleaning - Standardizing Data
def clean_phone_number(phone):
    """Extract digits from a phone number and format it consistently."""
    # Remove all non-digit characters
    digits = ''.join(filter(str.isdigit, phone))
    
    # Format as (XXX) XXX-XXXX if it has 10 digits
    if len(digits) == 10:
        return f"({digits[0:3]}) {digits[3:6]}-{digits[6:]}"
    # Format as +X (XXX) XXX-XXXX if it has 11 digits and starts with 1 (country code)
    elif len(digits) == 11 and digits[0] == '1':
        return f"+{digits[0]} ({digits[1:4]}) {digits[4:7]}-{digits[7:]}"
    else:
        return digits  # Return as is if it doesn't match expected patterns

# Test with various phone number formats
phone_numbers = [
    "(555) 123-4567",
    "555.123.4567",
    "+1-555-123-4567",
    "5551234567",
    "1-555-123-4567",
    "555-123-4567 ext 890"  # With extension
]

print("\n--- Phone Number Standardization ---")
for number in phone_numbers:
    print(f"Original: {number}")
    print(f"Standardized: {clean_phone_number(number)}")
    print()
```

```python
# Example 3: Parsing CSV Data
import csv
from io import StringIO

# Sample CSV data as a string (normally you'd read from a file)
csv_data = """Name,Age,Email,Country
John Doe,34,john.doe@example.com,USA
Jane Smith,28,jane.smith@example.com,Canada
Bob Johnson,45,bob.johnson@example.com,UK
Alice Brown,31,alice.brown@example.com,Australia"""

# Parse CSV string into usable data
def parse_csv_string(csv_string):
    """Parse CSV string into a list of dictionaries."""
    reader = csv.DictReader(StringIO(csv_string))
    return list(reader)

parsed_data = parse_csv_string(csv_data)

print("--- CSV Parsing Example ---")
print("Parsed data:")
for row in parsed_data:
    print(row)

# Extract specific information
print("\nEmails by country:")
for person in parsed_data:
    print(f"{person['Name']} from {person['Country']}: {person['Email']}")

# Find people over 30
over_30 = [person for person in parsed_data if int(person['Age']) > 30]
print("\nPeople over 30:")
for person in over_30:
    print(f"{person['Name']}, {person['Age']}")
```

```python
# Example 4: Extracting Information from Structured Text
def parse_log_entry(log_line):
    """Parse a log entry into its components."""
    # Example log format: [TIMESTAMP] LEVEL: MESSAGE
    timestamp_match = re.search(r'\[(.*?)\]', log_line)
    timestamp = timestamp_match.group(1) if timestamp_match else ''
    
    level_match = re.search(r'\] (\w+):', log_line)
    level = level_match.group(1) if level_match else ''
    
    message_match = re.search(r': (.*)', log_line)
    message = message_match.group(1) if message_match else ''
    
    return {
        'timestamp': timestamp,
        'level': level,
        'message': message
    }

log_entries = [
    "[2023-05-10 08:30:45] INFO: Application started successfully",
    "[2023-05-10 08:31:23] WARNING: High memory usage detected",
    "[2023-05-10 08:32:15] ERROR: Database connection failed",
    "[2023-05-10 08:35:42] INFO: User 'admin' logged in"
]

print("\n--- Log Parsing Example ---")
parsed_logs = [parse_log_entry(entry) for entry in log_entries]
for entry in parsed_logs:
    print(f"Time: {entry['timestamp']}, Level: {entry['level']}, Message: {entry['message']}")

# Filter for warnings and errors
high_priority = [entry for entry in parsed_logs if entry['level'] in ('WARNING', 'ERROR')]
print("\nHigh priority entries:")
for entry in high_priority:
    print(f"[{entry['timestamp']}] {entry['level']}: {entry['message']}")
```

```python
# Example 5: Simple Text Analyzer
class TextAnalyzer:
    def __init__(self, text):
        self.text = text
        self.words = re.findall(r'\b\w+\b', text.lower())
    
    def word_count(self):
        return len(self.words)
    
    def character_count(self, include_spaces=True):
        if include_spaces:
            return len(self.text)
        else:
            return len(re.sub(r'\s', '', self.text))
    
    def sentence_count(self):
        # Simple sentence splitting (not perfect but works for basic text)
        return len(re.split(r'[.!?]+', self.text)) - 1
    
    def average_word_length(self):
        if not self.words:
            return 0
        return sum(len(word) for word in self.words) / len(self.words)
    
    def most_common_words(self, n=5):
        return Counter(self.words).most_common(n)
    
    def vocabulary_richness(self):
        """Calculate ratio of unique words to total words."""
        if not self.words:
            return 0
        return len(set(self.words)) / len(self.words)
    
    def generate_summary(self):
        """Generate a summary analysis of the text."""
        return {
            "total_characters": self.character_count(),
            "characters_no_spaces": self.character_count(include_spaces=False),
            "word_count": self.word_count(),
            "sentence_count": self.sentence_count(),
            "avg_word_length": round(self.average_word_length(), 2),
            "vocabulary_richness": round(self.vocabulary_richness(), 2),
            "most_common_words": self.most_common_words()
        }

sample_article = """
Python is a high-level, interpreted programming language created by Guido van Rossum and released in 1991. 
Python's design philosophy emphasizes code readability with its notable use of significant whitespace. 
Its language constructs and object-oriented approach aim to help programmers write clear, logical code for 
small and large-scale projects. Python is dynamically typed and garbage-collected. It supports multiple 
programming paradigms, including structured, object-oriented, and functional programming. Python is often 
described as a "batteries included" language due to its comprehensive standard library.
"""

analyzer = TextAnalyzer(sample_article)
summary = analyzer.generate_summary()

print("\n--- Text Analysis Example ---")
print(f"Text length: {summary['total_characters']} characters ({summary['characters_no_spaces']} excluding spaces)")
print(f"Word count: {summary['word_count']}")
print(f"Sentence count: {summary['sentence_count']}")
print(f"Average word length: {summary['avg_word_length']} characters")
print(f"Vocabulary richness: {summary['vocabulary_richness']}")
print("Most common words:")
for word, count in summary["most_common_words"]:
    print(f"  {word}: {count}")
```
