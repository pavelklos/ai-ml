## 1. Introduction to Python Functions

Functions are reusable blocks of code designed to perform a specific task. They help with:

- **Code Reusability**: Write once, use multiple times
- **Modularity**: Break down complex problems into smaller, manageable pieces
- **Abstraction**: Hide implementation details behind a simple interface
- **Maintainability**: Easier to update or fix code in one location
- **Testing**: Test individual functions in isolation

Python has built-in functions like `print()`, `len()`, and `max()`, but you can also define your own functions.

````python
# Examples of built-in functions
print(len("Hello"))  # Output: 5
print(max(5, 10, 15))  # Output: 15
print(round(3.14159, 2))  # Output: 3.14
print(type(42))  # Output: <class 'int'>
````

## 2. Defining and Calling Functions

In Python, you define a function using the `def` keyword followed by the function name and parentheses containing any parameters. The function body is indented beneath the definition.

### Basic Function Syntax
```python
def function_name(parameters):
    """Optional docstring describing the function"""
    # Function body - code that executes when the function is called
    return value  # Optional return statement
```

````python
# Defining a simple function with no parameters
def greet():
    print("Hello, world!")

# Calling the function
greet()
````

````python
# Function with a parameter
def greet_person(name):
    print(f"Hello, {name}!")

# Calling the function with an argument
greet_person("Alice")
````

### Function Naming Conventions

- Use lowercase letters with underscores (snake_case)
- Use descriptive names that indicate what the function does
- Avoid using Python keywords or built-in function names
- Start with a verb for functions that perform actions

````python
# Good function names
def calculate_area(radius):
    return 3.14159 * radius ** 2

def convert_celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True

# Testing the functions
print(f"Area of circle with radius 5: {calculate_area(5):.2f}")
print(f"25°C in Fahrenheit: {convert_celsius_to_fahrenheit(25):.1f}°F")
print(f"Is 17 a prime number? {is_prime(17)}")
````

## 3. Function Parameters and Arguments

Parameters are variables listed in the function definition that specify what information the function needs to work.
Arguments are the actual values passed to the function when it is called.

- **Parameters**: Variables in the function definition
- **Arguments**: Values passed to the function when called

````python
# Function with multiple parameters
def calculate_rectangle_area(length, width):
    return length * width

# Calling with positional arguments
area = calculate_rectangle_area(5, 3)
print(f"Rectangle area: {area} square units")
````

### Parameter Types

Python functions can have different types of parameters:

1. **Positional parameters**: Regular parameters matched by position
2. **Default parameters**: Parameters with predefined values
3. **Keyword arguments**: Arguments specified by parameter name
4. **Variable-length parameters**: Parameters that accept any number of values

````python
# Example showing a function with multiple parameters
def describe_person(name, age, occupation):
    print(f"{name} is {age} years old and works as a {occupation}.")

# Call with positional arguments
describe_person("John", 30, "software developer")
````

## 4. Default Parameter Values

You can assign default values to parameters, which will be used if no argument is provided for that parameter. This makes functions more flexible and allows you to specify common values that don't need to be provided every time.

````python
# Function with default parameters
def greet_with_message(name, message="Hello", punctuation="!"):
    print(f"{message}, {name}{punctuation}")

# Different ways to call this function
greet_with_message("Alice")  # Using default message and punctuation
greet_with_message("Bob", "Hi")  # Overriding default message
greet_with_message("Charlie", "Welcome", ".")  # Overriding both defaults
````

### Important Note About Default Values

Default parameters are evaluated only once when the function is defined, not each time it's called. This can lead to unexpected behavior with mutable default values like lists or dictionaries.

````python
# Problematic function with mutable default value
def add_item(item, item_list=[]):
    item_list.append(item)
    return item_list

# First call
print(add_item("apple"))  # Output: ['apple']

# Second call - the list is not empty anymore!
print(add_item("banana"))  # Output: ['apple', 'banana']

# Third call
print(add_item("cherry"))  # Output: ['apple', 'banana', 'cherry']

# The proper way to handle this case:
def add_item_properly(item, item_list=None):
    if item_list is None:
        item_list = []
    item_list.append(item)
    return item_list

print("\nUsing the proper approach:")
print(add_item_properly("apple"))  # Output: ['apple']
print(add_item_properly("banana"))  # Output: ['banana']
````

## 5. Keyword Arguments

Keyword arguments (kwargs) allow you to pass arguments by explicitly specifying the parameter name. This makes function calls more readable and allows you to specify parameters in any order.

````python
# Function with multiple parameters
def create_profile(name, age, city, occupation):
    return f"{name}, {age}, lives in {city} and works as a {occupation}."

# Using positional arguments (order matters)
profile1 = create_profile("John", 30, "New York", "engineer")
print(profile1)

# Using keyword arguments (order doesn't matter)
profile2 = create_profile(
    city="San Francisco",
    occupation="designer", 
    name="Sarah", 
    age=28
)
print(profile2)

# Mixing positional and keyword arguments
# Positional arguments must come before keyword arguments
profile3 = create_profile("Michael", 35, occupation="teacher", city="Chicago")
print(profile3)
````

## 6. Variable-Length Arguments

Python allows you to define functions that can accept a variable number of arguments:

- `*args`: Collects extra positional arguments as a tuple
- `**kwargs`: Collects extra keyword arguments as a dictionary

````python
# Function with variable number of positional arguments
def calculate_sum(*numbers):
    """Calculate the sum of any number of values"""
    return sum(numbers)

# Call with different numbers of arguments
print(calculate_sum(1, 2))  # 3
print(calculate_sum(1, 2, 3, 4, 5))  # 15
print(calculate_sum())  # 0
````

````python
# Function with variable keyword arguments
def build_person(**attributes):
    """Create a dictionary of person attributes"""
    print("Person attributes:")
    for key, value in attributes.items():
        print(f"  {key}: {value}")
    return attributes

# Call with different keyword arguments
person1 = build_person(name="John", age=30, occupation="developer")
print("\nPerson2:")
person2 = build_person(name="Lisa", age=28, city="Boston", hobby="painting", education="Master's")
````

````python
# Combining all parameter types
def process_data(required, *args, default_param="default", **kwargs):
    print(f"Required parameter: {required}")
    print(f"Variable positional arguments (args): {args}")
    print(f"Default parameter: {default_param}")
    print(f"Variable keyword arguments (kwargs): {kwargs}")

# Example call
process_data(
    "main data",  # required
    1, 2, 3,  # args
    default_param="custom value",  # override default
    extra1="value1",  # kwargs
    extra2="value2"   # kwargs
)
````

## 7. Return Values

Functions can send data back to the caller using the `return` statement. A function can return:
- A single value
- Multiple values (as a tuple)
- Nothing (returns `None` implicitly)
- Different values based on conditions

````python
# Function that returns a single value
def square(x):
    return x ** 2

result = square(4)
print(f"The square of 4 is {result}")
````

````python
# Function that returns multiple values
def get_min_max(numbers):
    return min(numbers), max(numbers)

# The returned values are actually packed into a tuple
min_val, max_val = get_min_max([5, 3, 9, 1, 7])
print(f"Min: {min_val}, Max: {max_val}")
````

````python
# Function with conditional returns
def get_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

print(f"Score 95: {get_grade(95)}")
print(f"Score 82: {get_grade(82)}")
print(f"Score 45: {get_grade(45)}")
````

````python
# Function with no return statement
def print_info(text):
    print(f"Info: {text}")
    # No return statement - implicitly returns None

# Capture the return value
result = print_info("This is some information")
print(f"Return value: {result}")
````

## 8. Function Documentation

Python functions should be documented using docstrings (multi-line string literals enclosed by triple quotes). Good documentation includes:

- A brief description of what the function does
- Parameter descriptions and types
- Return value description and type
- Examples or usage notes
- Exceptions that might be raised

````python
def calculate_discount(price, discount_percentage, max_discount=100):
    """
    Calculate the discounted price for an item.
    
    Parameters:
        price (float): The original price of the item
        discount_percentage (float): The discount percentage (0-100)
        max_discount (float, optional): The maximum discount amount. Defaults to 100.
        
    Returns:
        float: The price after applying the discount
        
    Examples:
        >>> calculate_discount(100, 20)
        80.0
        >>> calculate_discount(500, 10, max_discount=40)
        460.0
        
    Raises:
        ValueError: If price or discount percentage is negative
    """
    if price < 0 or discount_percentage < 0:
        raise ValueError("Price and discount percentage must be non-negative")
        
    discount_amount = price * (discount_percentage / 100)
    discount_amount = min(discount_amount, max_discount)
    return price - discount_amount
````

````python
# Access the docstring
print(calculate_discount.__doc__)
````

````python
# Example calls based on the documentation
try:
    print(f"$100 with 20% discount: ${calculate_discount(100, 20)}")
    print(f"$500 with 10% discount (max $40): ${calculate_discount(500, 10, max_discount=40)}")
    print(f"$200 with -5% discount: ${calculate_discount(200, -5)}")
except ValueError as e:
    print(f"Error: {e}")
````

## 9. Scope and Lifetime of Variables

Variable scope refers to which parts of a program can access a variable. Python follows LEGB rule for variable lookup:

- **L**ocal: Variables defined inside the current function
- **E**nclosing: Variables defined in enclosing functions
- **G**lobal: Variables defined at the top level of the module
- **B**uilt-in: Python's built-in functions and variables

````python
# Global variable
message = "Global message"

def outer_function():
    # Enclosing variable
    message = "Enclosing message"
    
    def inner_function():
        # Local variable
        message = "Local message"
        print(f"Inner function: {message}")  # Uses local message
    
    inner_function()
    print(f"Outer function: {message}")  # Uses enclosing message

# Call the function
outer_function()
print(f"Global scope: {message}")  # Uses global message
````

````python
# The global keyword allows modification of global variables from within functions
counter = 0

def increment_counter():
    global counter  # Declare that we want to use the global variable
    counter += 1
    print(f"Counter inside function: {counter}")

print(f"Initial counter: {counter}")
increment_counter()
print(f"Final counter: {counter}")
````

````python
# The nonlocal keyword is used to modify variables in an enclosing (but not global) scope
def outer():
    x = 10
    
    def inner():
        nonlocal x  # Refers to x in the enclosing function's scope
        x = 20
        print(f"x in inner(): {x}")
    
    print(f"x in outer() before inner(): {x}")
    inner()
    print(f"x in outer() after inner(): {x}")

outer()
````

## 10. Lambda Functions

Lambda functions (also called anonymous functions) are small, one-line functions defined with the `lambda` keyword. They are useful when you need a simple function for a short period, especially as an argument to higher-order functions.

````python
# Regular function vs lambda function
def square(x):
    return x ** 2

# Equivalent lambda function
square_lambda = lambda x: x ** 2

print(f"Regular function: {square(5)}")
print(f"Lambda function: {square_lambda(5)}")
````

````python
# Lambda functions with multiple parameters
add = lambda x, y: x + y
print(f"5 + 3 = {add(5, 3)}")

# Conditional expression in lambda
is_even = lambda x: True if x % 2 == 0 else False
print(f"Is 4 even? {is_even(4)}")
print(f"Is 7 even? {is_even(7)}")
````

````python
# Common use case: Sorting with custom key
people = [
    {'name': 'Alice', 'age': 25},
    {'name': 'Bob', 'age': 38},
    {'name': 'Charlie', 'age': 19},
    {'name': 'Diana', 'age': 42}
]

# Sort by age
people_by_age = sorted(people, key=lambda person: person['age'])
print("Sorted by age:")
for person in people_by_age:
    print(f"{person['name']}: {person['age']}")

# Sort by name length
people_by_name_length = sorted(people, key=lambda person: len(person['name']))
print("\nSorted by name length:")
for person in people_by_name_length:
    print(f"{person['name']}: {len(person['name'])} letters")
````

````python
# Lambda with map function
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(f"Original: {numbers}")
print(f"Squared: {squared}")

# Lambda with filter function
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even numbers: {even_numbers}")
````

## 11. Functions as First-Class Objects

In Python, functions are "first-class citizens" or "first-class objects", which means they can be:

1. Assigned to variables
2. Passed as arguments to other functions
3. Returned from functions
4. Stored in data structures like lists or dictionaries

````python
# Assigning a function to a variable
def greet(name):
    return f"Hello, {name}!"

# Assign the function (not the result of calling it) to a variable
greeting_function = greet

# Now we can call it through the new variable
print(greeting_function("Alex"))
````

````python
# Passing functions as arguments
def apply_function(func, value):
    """Apply the given function to a value and return the result"""
    return func(value)

def double(x):
    return x * 2

def square(x):
    return x ** 2

def negative(x):
    return -x

# Apply different functions to the same value
value = 5
print(f"Original value: {value}")
print(f"After doubling: {apply_function(double, value)}")
print(f"After squaring: {apply_function(square, value)}")
print(f"After negating: {apply_function(negative, value)}")
print(f"After custom lambda: {apply_function(lambda x: x + 10, value)}")
````

````python
# Returning functions from other functions (function factory)
def create_multiplier(factor):
    """Return a function that multiplies its input by the given factor"""
    def multiply(x):
        return x * factor
    return multiply

# Create specific multiplier functions
double = create_multiplier(2)
triple = create_multiplier(3)

print(f"Double 7: {double(7)}")
print(f"Triple 7: {triple(7)}")
````

````python
# Storing functions in data structures
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Error: Division by zero"
    return x / y

# Store functions in a dictionary
operations = {
    '+': add,
    '-': subtract,
    '*': multiply,
    '/': divide
}

# Create a simple calculator function
def calculate(x, operator, y):
    if operator in operations:
        return operations[operator](x, y)
    return "Error: Unknown operator"

# Test the calculator
print(f"10 + 5 = {calculate(10, '+', 5)}")
print(f"10 - 5 = {calculate(10, '-', 5)}")
print(f"10 * 5 = {calculate(10, '*', 5)}")
print(f"10 / 5 = {calculate(10, '/', 5)}")
print(f"10 / 0 = {calculate(10, '/', 0)}")
print(f"10 % 5 = {calculate(10, '%', 5)}")
````

## 12. Practical Examples

Let's look at some practical examples of using functions to solve real-world problems.

### Example 1: Text Analysis Tool

````python
def analyze_text(text):
    """Analyze a text and return various statistics."""
    # Ensure we're working with a string
    if not isinstance(text, str):
        raise TypeError("Input must be a string")
        
    # Remove leading/trailing whitespace
    text = text.strip()
    
    # Count characters
    char_count = len(text)
    
    # Count letters only
    letter_count = sum(c.isalpha() for c in text)
    
    # Count words
    words = text.split()
    word_count = len(words)
    
    # Count sentences (simple approximation)
    sentence_terminators = ['.', '!', '?']
    sentence_count = sum(text.count(term) for term in sentence_terminators)
    
    # Calculate average word length
    if word_count > 0:
        avg_word_length = sum(len(word) for word in words) / word_count
    else:
        avg_word_length = 0
        
    # Count unique words (case-insensitive)
    unique_words = len(set(word.lower().strip('.,!?;:"') for word in words))
    
    return {
        'character_count': char_count,
        'letter_count': letter_count,
        'word_count': word_count,
        'sentence_count': max(1, sentence_count),  # At least 1 sentence
        'avg_word_length': avg_word_length,
        'unique_words': unique_words
    }

# Example usage
sample_text = """
Python functions are an essential part of the language. They help in code reusability 
and make the code more organized. Functions accept parameters and return results. 
They can be simple or complex!
"""

result = analyze_text(sample_text)

print("Text Analysis Results:")
for key, value in result.items():
    if isinstance(value, float):
        print(f"  {key.replace('_', ' ').title()}: {value:.2f}")
    else:
        print(f"  {key.replace('_', ' ').title()}: {value}")
````

### Example 2: Data Transformation Pipeline

````python
# Let's simulate a data pipeline with multiple processing functions

def load_data(data_source):
    """Simulate loading data from a source"""
    print(f"Loading data from: {data_source}")
    # In a real app, this would load from a file or database
    return [
        {"id": 1, "name": " john  smith ", "age": "34", "email": "john@example.com"},
        {"id": 2, "name": "JANE DOE", "age": None, "email": "jane.doe@example.com"},
        {"id": 3, "name": "Bob  Johnson", "age": "42", "email": ""},
    ]

def clean_names(data):
    """Clean up the name field by stripping whitespace and standardizing case"""
    for item in data:
        if isinstance(item["name"], str):
            item["name"] = " ".join(item["name"].split()).title()
    return data

def convert_age_to_int(data):
    """Convert age values to integers"""
    for item in data:
        try:
            item["age"] = int(item["age"]) if item["age"] else None
        except (ValueError, TypeError):
            item["age"] = None
    return data

def validate_emails(data):
    """Add a field indicating if the email appears valid"""
    import re
    email_pattern = re.compile(r'^[\w.-]+@[\w.-]+\.[a-zA-Z]{2,}$')
    for item in data:
        if item["email"] and isinstance(item["email"], str):
            item["valid_email"] = bool(email_pattern.match(item["email"]))
        else:
            item["valid_email"] = False
    return data

def filter_valid_records(data):
    """Filter records that have a name and valid email"""
    return [item for item in data if item["name"] and item["valid_email"]]

def process_pipeline(data_source, pipeline_steps):
    """Process data through a pipeline of functions"""
    data = load_data(data_source)
    print(f"\nRaw data:")
    for item in data:
        print(f"  {item}")
    
    for i, step in enumerate(pipeline_steps, 1):
        print(f"\nStep {i}: {step.__name__}")
        data = step(data)
        for item in data:
            print(f"  {item}")
    
    return data

# Define the pipeline steps
pipeline = [
    clean_names,
    convert_age_to_int,
    validate_emails,
    filter_valid_records
]

# Execute the pipeline
final_data = process_pipeline("customer_database", pipeline)

print("\nFinal processed data:")
for item in final_data:
    print(f"  {item}")
````

### Example 3: Custom Decorator for Timing Functions

````python
import time
import functools

def timing_decorator(func):
    """A decorator that times how long a function takes to execute"""
    @functools.wraps(func)  # This preserves the original function metadata
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        execution_time = end_time - start_time
        print(f"Function '{func.__name__}' took {execution_time:.4f} seconds to execute")
        return result
    return wrapper

# Apply the decorator to some test functions
@timing_decorator
def slow_function():
    """A deliberately slow function"""
    time.sleep(1)  # Simulate a time-consuming task
    return "Slow function complete"

@timing_decorator
def calculate_fibonacci(n):
    """Calculate the nth Fibonacci number recursively"""
    if n <= 1:
        return n
    return calculate_fibonacci(n-1) + calculate_fibonacci(n-2)

# Test the functions
print(slow_function())
print(f"Fibonacci(20) = {calculate_fibonacci(20)}")
````
