# Regular Expressions in Python with the `re` Module

Regular expressions (regex) are powerful sequences of characters that define search patterns. Python's `re` module provides excellent support for working with regular expressions, enabling advanced pattern matching, text processing, and data extraction operations.

This notebook covers regular expressions in Python from the basics to advanced techniques, with practical examples.

## Import Required Libraries

To begin working with regular expressions in Python, we need to import the `re` module from Python's standard library. We'll also import other useful libraries for our examples.

```python
# Import the re module for regular expression operations
import re

# Optional libraries for examples
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Introduction to Regular Expressions

Regular expressions are specialized text strings used for describing search patterns. They are particularly useful for:

- Validating text input (email addresses, phone numbers, etc.)
- Searching for patterns in text data
- Extracting specific information from text
- Text replacement operations
- Parsing and transforming text data

Regular expressions have their own syntax, which might seem complex at first but offers tremendous power and flexibility once you understand the basics.

```python
# A simple regex example to find all digits in a string
text = "I have 2 apples and 3 oranges, that's 5 fruits total."
digits = re.findall(r'\d', text)
print(f"Found digits: {digits}")

# Finding words that contain 'a'
words_with_a = re.findall(r'\w*a\w*', text)
print(f"Words containing 'a': {words_with_a}")
```

## Basic Pattern Matching

The `re` module provides several functions for pattern matching:

- `re.search()`: Searches for a pattern anywhere in the string
- `re.match()`: Checks if the pattern matches at the beginning of the string
- `re.findall()`: Returns all non-overlapping matches as a list
- `re.finditer()`: Returns an iterator yielding match objects

Let's see these functions in action:

```python
text = "Python is a great programming language. Python is versatile."

# Using re.search() - finds the first occurrence
search_result = re.search(r'Python', text)
print(f"re.search() found: {search_result.group()} at position {search_result.start()}")

# Using re.match() - only matches at the beginning
match_result = re.match(r'Python', text)
print(f"re.match() result: {match_result.group() if match_result else 'No match'}")

match_result2 = re.match(r'Java', text)
print(f"re.match() for 'Java': {match_result2.group() if match_result2 else 'No match'}")

# Using re.findall() - finds all occurrences
findall_result = re.findall(r'Python', text)
print(f"re.findall() found: {findall_result}, count: {len(findall_result)}")

# Using re.finditer() - iterator of match objects
print("re.finditer() results:")
for match in re.finditer(r'Python', text):
    print(f"  Found '{match.group()}' at position {match.start()}-{match.end()}")
```

## Special Characters and Metacharacters

Regular expressions use special characters (metacharacters) with specific meanings:

| Character | Meaning |
|-----------|---------|
| `.` | Matches any character (except newline) |
| `\d` | Matches any digit (0-9) |
| `\D` | Matches any non-digit |
| `\w` | Matches any alphanumeric character (a-z, A-Z, 0-9, _) |
| `\W` | Matches any non-alphanumeric character |
| `\s` | Matches any whitespace character (space, tab, newline) |
| `\S` | Matches any non-whitespace character |
| `\b` | Matches word boundary |
| `\B` | Matches non-word boundary |
| `\` | Escapes special characters |

Let's see examples of these metacharacters:

```python
text = "User123 has $50.00 in account #ACCT-456"

# Finding digits with \d
digits = re.findall(r'\d+', text)
print(f"Digits: {digits}")

# Finding words with \w+
words = re.findall(r'\w+', text)
print(f"Words: {words}")

# Finding whitespace with \s
spaces = re.findall(r'\s', text)
print(f"Whitespace count: {len(spaces)}")

# Using dot (.) to match any character
any_char = re.findall(r'h.s', text)
print(f"'h.s' matches: {any_char}")

# Escaping special characters
price = re.search(r'\$\d+\.\d+', text)
print(f"Price: {price.group() if price else 'Not found'}")
```

## Character Classes

Character classes allow you to specify a set of characters to match. They are defined using square brackets `[]`:

- `[abc]` - Matches any single character that is either 'a', 'b', or 'c'
- `[^abc]` - Matches any character except 'a', 'b', or 'c'
- `[a-z]` - Matches any lowercase letter from 'a' to 'z'
- `[A-Z]` - Matches any uppercase letter from 'A' to 'Z'
- `[0-9]` - Matches any digit from 0 to 9
- `[a-zA-Z0-9]` - Matches any alphanumeric character

Let's explore character classes:

```python
text = "The price of the item is $29.99, item code: ABC-123"

# Finding lowercase letters
lowercase = re.findall(r'[a-z]+', text)
print(f"Lowercase words: {lowercase}")

# Finding uppercase letters
uppercase = re.findall(r'[A-Z]+', text)
print(f"Uppercase sequences: {uppercase}")

# Finding digits
digits = re.findall(r'[0-9]+', text)
print(f"Numeric sequences: {digits}")

# Using negated character class
not_alphanumeric = re.findall(r'[^a-zA-Z0-9]+', text)
print(f"Non-alphanumeric sequences: {not_alphanumeric}")

# Combining character classes
code = re.search(r'[A-Z]+-[0-9]+', text)
print(f"Product code: {code.group() if code else 'Not found'}")
```

## Quantifiers

Quantifiers specify how many times a character or group should be matched:

- `*` - Matches 0 or more occurrences
- `+` - Matches 1 or more occurrences
- `?` - Matches 0 or 1 occurrence
- `{n}` - Matches exactly n occurrences
- `{n,}` - Matches n or more occurrences
- `{n,m}` - Matches between n and m occurrences (inclusive)

Let's see quantifiers in action:

```python
text = "color colour flavor flavour"

# Using * quantifier (0 or more)
star_pattern = re.findall(r'colou*r', text)
print(f"'colou*r' matches: {star_pattern}")

# Using + quantifier (1 or more)
plus_pattern = re.findall(r'flavo+r', text)
print(f"'flavo+r' matches: {plus_pattern}")

# Using ? quantifier (0 or 1)
question_pattern = re.findall(r'colou?r', text)
print(f"'colou?r' matches: {question_pattern}")

# Using {n} - exactly n occurrences
text2 = "42 142 1442 14442"
exact_pattern = re.findall(r'1{1}4{2}2', text2)
print(f"'1{{1}}4{{2}}2' matches: {exact_pattern}")

# Using {n,} - n or more occurrences
min_pattern = re.findall(r'14{2,}2', text2)
print(f"'14{{2,}}2' matches: {min_pattern}")

# Using {n,m} - between n and m occurrences
range_pattern = re.findall(r'14{1,3}2', text2)
print(f"'14{{1,3}}2' matches: {range_pattern}")
```

## Grouping and Capturing

Parentheses `()` are used to group parts of a regex pattern and capture the matched text. Captured groups can be accessed using the `.group()` method or backreferences.

- `(...)` - Creates a capturing group
- `(?:...)` - Creates a non-capturing group

Let's explore grouping and capturing:

```python
# Basic grouping
text = "John Smith and Jane Doe"
name_pattern = r'(\w+) (\w+)'

matches = re.findall(name_pattern, text)
print(f"Names found (first, last): {matches}")

# Using match objects and accessing groups
for match in re.finditer(name_pattern, text):
    print(f"Full name: {match.group(0)}")
    print(f"First name: {match.group(1)}")
    print(f"Last name: {match.group(2)}")
    
# Named groups with (?P<name>...)
pattern_with_names = r'(?P<first>\w+) (?P<last>\w+)'
for match in re.finditer(pattern_with_names, text):
    print(f"Using named groups - {match.group('first')} {match.group('last')}")
    
# Non-capturing groups
phone = "123-456-7890"
area_code = re.search(r'((?:\d{3})-)\d{3}-\d{4}', phone)
print(f"Area code with dash: {area_code.group(1) if area_code else 'Not found'}")
```

## Anchors and Boundaries

Anchors and boundaries let you specify positions in the text where matches should occur:

- `^` - Matches the start of a string
- `$` - Matches the end of a string
- `\b` - Matches a word boundary
- `\B` - Matches a non-word boundary

Let's see these in action:

```python
# Start of string with ^
text = "Python is great. I love Python!"
starts_with = re.match(r'^Python', text)
print(f"String starts with 'Python': {bool(starts_with)}")

# End of string with $
ends_with = re.search(r'Python!$', text)
print(f"String ends with 'Python!': {bool(ends_with)}")

# Word boundaries with \b
text2 = "Cats are cute, I have a cat."
cats = re.findall(r'\bcat\b', text2, re.IGNORECASE)
print(f"Complete 'cat' words: {cats}")

# Without word boundaries
cat_partial = re.findall(r'cat', text2, re.IGNORECASE)
print(f"All 'cat' occurrences (even within words): {cat_partial}")
```

## Regular Expression Functions

Let's explore additional regex functions provided by the `re` module:

- `re.split()` - Splits a string by occurrences of the pattern
- `re.sub()` - Replaces occurrences of the pattern
- `re.subn()` - Like `re.sub()` but returns a tuple of (new_string, number_of_replacements)
- `re.escape()` - Escapes special characters in a string

```python
# re.split() example
text = "apple,banana;orange,grape;pear"
fruits = re.split(r'[,;]', text)
print(f"Fruits after splitting: {fruits}")

# re.sub() example
text2 = "Phone: 123-456-7890"
formatted_phone = re.sub(r'(\d{3})-(\d{3})-(\d{4})', r'(\1) \2-\3', text2)
print(f"Formatted phone: {formatted_phone}")

# re.subn() example
text3 = "I like cats. Cats are great."
new_text, count = re.subn(r'[Cc]ats', 'dogs', text3)
print(f"New text: '{new_text}', Replacements: {count}")

# re.escape() example
pattern_text = "Hello (world)? * + [123]"
escaped_pattern = re.escape(pattern_text)
print(f"Original: '{pattern_text}'")
print(f"Escaped: '{escaped_pattern}'")
```

## Compilation and Flags

For better performance, especially when using the same pattern multiple times, you can compile regular expressions using `re.compile()`. You can also use flags to modify regex behavior.

Common flags include:
- `re.IGNORECASE` or `re.I` - Case-insensitive matching
- `re.MULTILINE` or `re.M` - Makes `^` and `$` match at line breaks
- `re.DOTALL` or `re.S` - Makes `.` match newlines too
- `re.VERBOSE` or `re.X` - Allows you to write more readable patterns with comments

```python
# Compiling a regex pattern
email_pattern = re.compile(r'[\w.-]+@[\w.-]+\.\w+')

text = "Contact us at info@example.com or support@company.org"
emails = email_pattern.findall(text)
print(f"Emails: {emails}")

# Using flags with compiled regex
text2 = """Python
python
PYTHON"""

# Case-insensitive matching
pattern_i = re.compile(r'^python', re.IGNORECASE)
matches_i = pattern_i.findall(text2)
print(f"Case-insensitive matches: {matches_i}")

# Multiline matching
pattern_m = re.compile(r'^python', re.IGNORECASE | re.MULTILINE)
matches_m = pattern_m.findall(text2)
print(f"Multiline matches: {matches_m}")

# Verbose pattern for better readability
phone_regex = re.compile(r'''
    \((\d{3})\)     # Area code
    \s              # Space
    (\d{3})         # First 3 digits
    -               # Dash
    (\d{4})         # Last 4 digits
''', re.VERBOSE)

text3 = "Call (123) 456-7890 for help"
phone = phone_regex.search(text3)
if phone:
    print(f"Phone: ({phone.group(1)}) {phone.group(2)}-{phone.group(3)}")
```

## Substitution with re.sub

The `re.sub()` function is powerful for text replacement. It can:

1. Replace matches with a fixed string
2. Use backreferences to parts of the match
3. Use a function to dynamically determine the replacement

Let's explore these capabilities:

```python
# Basic replacement
text = "Hello world! Hello universe!"
new_text = re.sub(r'Hello', 'Hi', text)
print(f"Basic replacement: {new_text}")

# Using backreferences
text2 = "John Smith, Jane Doe, Robert Johnson"
swapped = re.sub(r'(\w+) (\w+)', r'\2, \1', text2)
print(f"Swapped names: {swapped}")

# Using a function for replacement
def capitalize_match(match):
    return match.group(0).upper()

text3 = "apple banana orange"
capitalized = re.sub(r'\b\w+\b', capitalize_match, text3)
print(f"Capitalized words: {capitalized}")

# Using named groups in substitution
text4 = "John Smith"
formatted = re.sub(r'(?P<first>\w+) (?P<last>\w+)', r'\g<last>, \g<first>', text4)
print(f"Formatted name: {formatted}")
```

## Common Regex Patterns

Here are some commonly used regex patterns for various types of data validation:

```python
# Email validation
email_pattern = r'^[\w.-]+@[\w.-]+\.\w+$'

test_emails = ['user@example.com', 'invalid@email', 'user.name@domain.co.uk']
for email in test_emails:
    is_valid = bool(re.match(email_pattern, email))
    print(f"Email '{email}' is {'valid' if is_valid else 'invalid'}")

# URL validation (simplified)
url_pattern = r'^https?:\/\/(?:www\.)?[-a-zA-Z0-9@:%._\+~#=]{1,256}\.[a-zA-Z0-9()]{1,6}\b(?:[-a-zA-Z0-9()@:%_\+.~#?&\/=]*)$'

test_urls = ['https://www.example.com', 'http://sub.domain.co.uk/page?param=1', 'invalid-url']
for url in test_urls:
    is_valid = bool(re.match(url_pattern, url))
    print(f"URL '{url}' is {'valid' if is_valid else 'invalid'}")

# Phone number validation (US format)
phone_pattern = r'^\(?(\d{3})\)?[-.\s]?(\d{3})[-.\s]?(\d{4})$'

test_phones = ['(123) 456-7890', '123-456-7890', '1234567890', '12-345-6789']
for phone in test_phones:
    is_valid = bool(re.match(phone_pattern, phone))
    print(f"Phone '{phone}' is {'valid' if is_valid else 'invalid'}")

# Date validation (MM/DD/YYYY)
date_pattern = r'^(0?[1-9]|1[0-2])\/(0?[1-9]|[12]\d|3[01])\/(\d{4})$'

test_dates = ['01/01/2023', '1/1/2023', '13/01/2023', '12/32/2023']
for date in test_dates:
    is_valid = bool(re.match(date_pattern, date))
    print(f"Date '{date}' is {'valid' if is_valid else 'invalid'}")
```

## Practical Examples

Let's solve some real-world problems using regular expressions:

```python
# Example 1: Extract all hashtags from a tweet
tweet = "Just attended #PyCon2023! Great talks on #Python #DataScience and #MachineLearning"
hashtags = re.findall(r'#(\w+)', tweet)
print(f"Hashtags: {hashtags}")

# Example 2: Extract data from log entries
log_entries = [
    "INFO 2023-05-15 12:34:56 - User login successful for user123",
    "ERROR 2023-05-15 12:35:30 - Database connection failed",
    "WARNING 2023-05-15 12:36:15 - Disk space below 20%"
]

log_pattern = re.compile(r'(INFO|WARNING|ERROR) (\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}) - (.+)')

print("\nParsed log entries:")
for log in log_entries:
    match = log_pattern.match(log)
    if match:
        level, timestamp, message = match.groups()
        print(f"Level: {level}, Time: {timestamp}, Message: {message}")

# Example 3: Cleaning and standardizing data
addresses = [
    "123 Main St, Apt 4, New York, NY 10001",
    "456 Elm Avenue Suite 200, Los Angeles CA 90001",
    "789 Oak Dr. Chicago IL 60601"
]

# Standardize addresses
def standardize_address(address):
    # Replace various separators with commas
    address = re.sub(r'(?<!\d)[,.]?\s+(?=(Suite|Apt|Unit|#))', ', ', address)
    # Ensure state and zip are separated by a space
    address = re.sub(r'([A-Z]{2})(\d{5})', r'\1 \2', address)
    # Add comma before state
    address = re.sub(r'([^,])\s+([A-Z]{2}\s+\d{5})', r'\1, \2', address)
    return address

print("\nStandardized addresses:")
for address in addresses:
    print(f"Original: {address}")
    print(f"Cleaned:  {standardize_address(address)}")
    print()

# Example 4: Extract data from HTML
html = """<div class="product">
    <h2>Smartphone X</h2>
    <p class="price">$799.99</p>
    <p class="desc">Latest model with advanced features</p>
</div>
<div class="product">
    <h2>Laptop Pro</h2>
    <p class="price">$1299.99</p>
    <p class="desc">Powerful performance for professionals</p>
</div>"""

# Extract product information
product_pattern = re.compile(r'<h2>(.+?)</h2>.*?<p class="price">\$(.+?)</p>.*?<p class="desc">(.+?)</p>', re.DOTALL)
products = product_pattern.findall(html)

print("Extracted products:")
for name, price, description in products:
    print(f"Product: {name}")
    print(f"Price: ${price}")
    print(f"Description: {description}")
    print()
```
