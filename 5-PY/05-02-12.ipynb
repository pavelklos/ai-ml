## 2. Understanding defaultdict

The `defaultdict` is a subclass of the built-in `dict` class. The main difference is that `defaultdict` never raises a `KeyError`. Instead, it provides a default value for a nonexistent key by calling a factory function that you provide when creating the `defaultdict`.

```python
# Creating a defaultdict with int as the default factory
word_count = defaultdict(int)

# No need to initialize keys before incrementing
for word in ["apple", "banana", "apple", "orange", "banana", "apple"]:
    word_count[word] += 1
    
print("Word counts:", dict(word_count))
```

```python
# Compare to regular dictionary approach
regular_dict = {}
words = ["apple", "banana", "apple", "orange", "banana", "apple"]

# Have to check if key exists first
for word in words:
    if word in regular_dict:
        regular_dict[word] += 1
    else:
        regular_dict[word] = 1
        
print("Regular dict approach:", regular_dict)
```

### Using different factory functions

```python
# defaultdict with list as factory
grouped_words = defaultdict(list)

words = ["apple", "bat", "car", "apple", "banana", "bat"]
for i, word in enumerate(words):
    grouped_words[word].append(i)  # Automatically creates a list if key doesn't exist
    
print("Word positions:", dict(grouped_words))
```

```python
# Using a custom factory function
def default_value():
    return "Not Found"

d = defaultdict(default_value)
d['a'] = 'exists'

print(d['a'])  # Key exists
print(d['b'])  # Key doesn't exist, returns the default value
```

### Nested defaultdict

```python
# Creating nested defaultdicts
nested_dict = defaultdict(lambda: defaultdict(list))

# Add data without checking if keys exist
nested_dict['fruits']['red'].append('apple')
nested_dict['fruits']['yellow'].append('banana')
nested_dict['fruits']['red'].append('cherry')
nested_dict['vegetables']['green'].append('spinach')

# Print the nested structure
for category, colors in nested_dict.items():
    print(f"{category}:")
    for color, items in colors.items():
        print(f"  {color}: {items}")
```

## 3. Working with Counter

The `Counter` is a specialized dictionary for counting hashable objects. It's a subclass of `dict` that helps count objects.

```python
# Basic Counter usage
c = Counter(['apple', 'orange', 'apple', 'banana', 'apple', 'orange'])
print("Counter:", c)
```

```python
# Initialize with a dictionary
c = Counter({'apple': 3, 'orange': 2, 'banana': 1})
print("Counter from dict:", c)
```

```python
# Initialize with keyword arguments
c = Counter(apple=3, orange=2, banana=1)
print("Counter from kwargs:", c)
```

### Working with Counter methods

```python
# Counting characters in a text
text = "the quick brown fox jumps over the lazy dog"
c = Counter(text)

# Get the most common elements
print("Most common characters:")
for char, count in c.most_common(5):
    print(f"'{char}': {count}")

# Counting words in a text
word_counter = Counter(text.split())
print("\nWord frequency:", word_counter)
```

```python
# Mathematical operations with Counter
c1 = Counter(['a', 'b', 'c', 'a', 'b', 'a'])
c2 = Counter(['a', 'c', 'd', 'c', 'd'])

print("c1:", c1)
print("c2:", c2)

# Add two counters together: c1[x] + c2[x]
print("\nc1 + c2:", c1 + c2)

# Subtract (keeping only positive counts): c1[x] - c2[x]
print("c1 - c2:", c1 - c2)

# Intersection: min(c1[x], c2[x]) 
print("c1 & c2:", c1 & c2)

# Union: max(c1[x], c2[x])
print("c1 | c2:", c1 | c2)
```

```python
# Updating a Counter
c = Counter(['apple', 'orange', 'apple'])
print("Initial counter:", c)

# Add more elements
c.update(['apple', 'banana', 'banana'])
print("After update:", c)

# Subtract elements (removes only if result is positive)
c.subtract(['apple', 'apple', 'orange', 'orange'])
print("After subtract:", c)

# Elements method - returns each element repeated count times
print("\nElements:")
print(list(c.elements()))
```

```python
# Visualizing Counter data using a bar chart
text = "To be or not to be, that is the question."
word_counter = Counter(text.lower().replace(',', '').replace('.', '').split())

# Create a simple bar chart
plt.figure(figsize=(10, 5))
plt.bar(word_counter.keys(), word_counter.values())
plt.title('Word Frequency')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Using namedtuple

The `namedtuple` is a factory function for creating tuple subclasses with named fields. Named tuples are immutable and can be accessed by index like regular tuples, but also allow accessing fields by name.

```python
# Creating a namedtuple type
Person = namedtuple('Person', ['name', 'age', 'job'])

# Creating namedtuple instances
alice = Person('Alice', 30, 'Engineer')
bob = Person('Bob', 25, 'Designer')

print(f"{alice.name} is a {alice.age}-year-old {alice.job}")
print(f"{bob.name} is a {bob.age}-year-old {bob.job}")

# Access by index (like a regular tuple)
print(f"\nAlice's job (by index): {alice[2]}")

# Converting to a dictionary
alice_dict = alice._asdict()
print(f"\nAs dictionary: {alice_dict}")
```

```python
# Creating a namedtuple with defaults
Employee = namedtuple('Employee', ['name', 'department', 'salary'], defaults=[None, 0])

# Creating instances with and without defaults
emp1 = Employee('John')  # Uses defaults for department and salary
emp2 = Employee('Lisa', 'Engineering')  # Uses default for salary
emp3 = Employee('Mike', 'Marketing', 50000)  # Specifies all values

employees = [emp1, emp2, emp3]
for emp in employees:
    print(f"{emp.name}, {emp.department}, ${emp.salary}")
```

```python
# Creating a new instance with _replace
Point = namedtuple('Point', ['x', 'y'])
p1 = Point(10, 20)
print(f"Original point: {p1}")

# Create a new point with x=15
p2 = p1._replace(x=15)
print(f"New point: {p2}")

# Original is unchanged (immutable)
print(f"Original still: {p1}")
```

```python
# Using namedtuple for data analysis
Sale = namedtuple('Sale', ['product', 'price', 'quantity', 'date'])

# Creating a list of sales records
sales = [
    Sale('Laptop', 1200, 1, '2023-01-15'),
    Sale('Mouse', 25, 5, '2023-01-15'),
    Sale('Monitor', 300, 2, '2023-01-16'),
    Sale('Keyboard', 50, 3, '2023-01-17'),
    Sale('Laptop', 1200, 2, '2023-01-17'),
    Sale('Mouse', 25, 10, '2023-01-18')
]

# Calculate total sales value
total_sales = sum(sale.price * sale.quantity for sale in sales)
print(f"Total sales value: ${total_sales}")

# Group sales by product
product_sales = defaultdict(int)
for sale in sales:
    product_sales[sale.product] += sale.price * sale.quantity

print("\nSales by product:")
for product, total in product_sales.items():
    print(f"{product}: ${total}")

# Convert to DataFrame for analysis
sales_df = pd.DataFrame([s._asdict() for s in sales])
print("\nSales DataFrame:")
display(sales_df)
```

### When to use namedtuple

Named tuples are particularly useful:

1. When you need lightweight object types with immutable attributes
2. When memory usage is a concern (they're more memory efficient than classes)
3. When you want to return multiple values from a function with descriptive field names
4. When you want to make your code more readable by using field names instead of indices

## 5. Other Useful Collections

The collections module provides several other useful collection types that we'll explore briefly.

### deque - Double-Ended Queue

```python
# Create a new deque with some items
d = deque(['a', 'b', 'c'])
print("Original deque:", d)

# Add to both ends
d.append('d')  # Add to right end
d.appendleft('z')  # Add to left end
print("After appends:", d)

# Pop from both ends
right = d.pop()  # Remove from right end
left = d.popleft()  # Remove from left end
print(f"Popped right: {right}, left: {left}")
print("After pops:", d)
```

```python
# Deque rotation
d = deque([1, 2, 3, 4, 5])
print("Original:", d)

# Rotate right by 2 positions
d.rotate(2)
print("Rotated right by 2:", d)

# Rotate left by 3 positions
d.rotate(-3)
print("Rotated left by 3:", d)

# Creating a bounded deque with maxlen
d = deque(maxlen=3)
for i in range(5):
    d.append(i)
    print(f"After adding {i}: {d}")
```

### OrderedDict

```python
# Create an OrderedDict
od = OrderedDict()
od['apple'] = 5
od['banana'] = 3
od['cherry'] = 7
od['date'] = 2

print("OrderedDict:")
for key, value in od.items():
    print(f"{key}: {value}")

# Move an item to the end
od.move_to_end('banana')
print("\nAfter moving 'banana' to the end:")
for key, value in od.items():
    print(f"{key}: {value}")

# Move an item to the beginning
od.move_to_end('date', last=False)
print("\nAfter moving 'date' to the beginning:")
for key, value in od.items():
    print(f"{key}: {value}")

# Note: In Python 3.7+, regular dictionaries preserve insertion order,
# but OrderedDict still provides extra features like move_to_end()
```

### ChainMap

```python
# Create multiple dictionaries
defaults = {'theme': 'default', 'language': 'en', 'font_size': 12}
user_settings = {'theme': 'dark'}
session_settings = {'font_size': 14}

# Combine them with ChainMap - later mappings have priority
settings = ChainMap(session_settings, user_settings, defaults)

print(f"Theme: {settings['theme']}")
print(f"Language: {settings['language']}")
print(f"Font size: {settings['font_size']}")

# Adding a new mapping with highest priority
temporary_settings = {'language': 'fr'}
new_settings = settings.new_child(temporary_settings)

print("\nAfter adding temporary settings:")
print(f"Theme: {new_settings['theme']}")
print(f"Language: {new_settings['language']}")
print(f"Font size: {new_settings['font_size']}")
```

## 6. Performance Considerations

Let's compare the performance of some of these collections against their standard library alternatives.

```python
# Performance test: defaultdict vs dict for counting
import random
import string

# Generate random words
def random_word(length=5):
    return ''.join(random.choice(string.ascii_lowercase) for _ in range(length))

words = [random_word() for _ in range(100000)]

# Time using regular dict
start = time.time()
d = {}
for word in words:
    if word in d:
        d[word] += 1
    else:
        d[word] = 1
regular_dict_time = time.time() - start

# Time using defaultdict
start = time.time()
dd = defaultdict(int)
for word in words:
    dd[word] += 1
defaultdict_time = time.time() - start

# Time using Counter
start = time.time()
c = Counter(words)
counter_time = time.time() - start

print(f"Regular dict: {regular_dict_time:.6f} seconds")
print(f"defaultdict: {defaultdict_time:.6f} seconds")
print(f"Counter: {counter_time:.6f} seconds")
```

```python
# Performance: deque vs list for queue operations
n = 100000

# Using list as queue (inefficient for popleft)
start = time.time()
queue_list = []
for i in range(n):
    queue_list.append(i)
for i in range(n):
    queue_list.pop(0)  # inefficient O(n) operation
list_time = time.time() - start

# Using deque as queue (efficient)
start = time.time()
queue_deque = deque()
for i in range(n):
    queue_deque.append(i)
for i in range(n):
    queue_deque.popleft()  # efficient O(1) operation
deque_time = time.time() - start

print(f"List as queue: {list_time:.6f} seconds")
print(f"Deque as queue: {deque_time:.6f} seconds")
print(f"Deque is {list_time/deque_time:.1f}x faster")

# Visualize the performance comparison
plt.figure(figsize=(10, 5))
plt.bar(['Regular dict', 'defaultdict', 'Counter', 'List as queue', 'Deque as queue'], 
        [regular_dict_time, defaultdict_time, counter_time, list_time, deque_time])
plt.title('Performance Comparison')
plt.ylabel('Time (seconds)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Building a simple cache with OrderedDict
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity):
        self.cache = OrderedDict()
        self.capacity = capacity
        
    def get(self, key):
        if key not in self.cache:
            return -1
        
        # Move the accessed key to the end (most recently used)
        self.cache.move_to_end(key)
        return self.cache[key]
    
    def put(self, key, value):
        # If key exists, update and move to end
        if key in self.cache:
            self.cache[key] = value
            self.cache.move_to_end(key)
            return
        
        # If cache is full, remove the least recently used item (first item)
        if len(self.cache) >= self.capacity:
            self.cache.popitem(last=False)  # Remove first item (least recently used)
            
        # Add new key-value pair
        self.cache[key] = value
    
    def __str__(self):
        return str(dict(self.cache))

# Test the LRU cache
lru = LRUCache(3)  # Cache with capacity 3
lru.put('a', 1)
print(f"After adding 'a': {lru}")
lru.put('b', 2)
print(f"After adding 'b': {lru}")
lru.put('c', 3)
print(f"After adding 'c': {lru}")

# Access 'a' (moves it to the end)
lru.get('a')
print(f"After accessing 'a': {lru}")

# Add 'd', should evict 'b' (least recently used)
lru.put('d', 4)
print(f"After adding 'd': {lru}")
```

```python
# Using deque for moving average calculation
def moving_average(data, window_size=3):
    window = deque(maxlen=window_size)
    averages = []
    
    for value in data:
        window.append(value)
        if len(window) == window_size:  # Only calculate when window is full
            averages.append(sum(window) / window_size)
            
    return averages

# Test with stock price data
prices = [100, 102, 99, 101, 105, 108, 109, 106, 107, 104, 103, 105, 108, 110]

# Calculate 3-day moving average
ma3 = moving_average(prices, 3)

# Calculate 5-day moving average
ma5 = moving_average(prices, 5)

# Plot the results
plt.figure(figsize=(12, 6))
plt.plot(prices, label='Daily Price', marker='o')
plt.plot(range(2, len(prices)), ma3, label='3-day MA', linestyle='--')
plt.plot(range(4, len(prices)), ma5, label='5-day MA', linestyle='-.')
plt.legend()
plt.title('Stock Price with Moving Averages')
plt.xlabel('Day')
plt.ylabel('Price')
plt.grid(True)
plt.tight_layout()
plt.show()

print("Original prices:", prices)
print("3-day moving average:", ma3)
print("5-day moving average:", ma5)
```

## 8. Best Practices

Here are some best practices for working with the collections module:

### 1. Choosing the Right Collection

| Collection | Use when you need... |
|------------|------------------------|
| `defaultdict` | A dictionary that provides default values for missing keys |
| `Counter` | To count occurrences of hashable objects |
| `namedtuple` | A lightweight, immutable data structure with named fields |
| `deque` | Fast appends and pops from both ends of a collection |
| `OrderedDict` | A dictionary that remembers insertion order (less important in Python 3.7+) |
| `ChainMap` | To search through multiple dictionaries as a single mapping |

### 2. Performance Considerations

- Use `defaultdict` when you find yourself repeatedly checking if keys exist
- Use `Counter` for counting instead of building your own solution
- Use `deque` over lists when you need frequent operations at both ends
- Use `namedtuple` instead of dictionaries for small data records to save memory

### 3. Common Patterns

#### Group items with defaultdict

```python
# Grouping items by a specific attribute
items = [('apple', 'fruit'), ('carrot', 'vegetable'), 
         ('banana', 'fruit'), ('celery', 'vegetable')]

# Group by category
grouped = defaultdict(list)
for name, category in items:
    grouped[category].append(name)
    
print(dict(grouped))
```

#### Use Counter for frequency analysis

```python
# Finding most common elements
data = [1, 2, 3, 1, 2, 1, 4, 5, 3, 1, 2, 6, 2, 3, 1]
c = Counter(data)

# Most common
print(f"Three most common elements: {c.most_common(3)}")

# Elements with count > 2
frequent = [item for item, count in c.items() if count > 2]
print(f"Elements appearing more than twice: {frequent}")
```

#### Use namedtuple for self-documenting code

```python
# Without namedtuple
def get_rectangle_data():
    return (100, 50, (255, 0, 0))  # Width, height, color

rect_data = get_rectangle_data()
print(f"Rectangle: width={rect_data[0]}, height={rect_data[1]}, color={rect_data[2]}")

# With namedtuple (more self-documenting)
Rectangle = namedtuple('Rectangle', ['width', 'height', 'color'])

def get_rectangle_data_named():
    return Rectangle(100, 50, (255, 0, 0))

rect = get_rectangle_data_named()
print(f"Rectangle: width={rect.width}, height={rect.height}, color={rect.color}")
```

### 4. Debugging Tips

- Remember that `defaultdict` returns the default value for missing keys, so check your default factory if results are unexpected
- `namedtuples` are immutable - you can't change values after creation
- When using `Counter`, remember that it only counts the occurrences that are specified - it doesn't auto-account for missing items

### Exercise 1: Word Frequency Counter

Create a function that reads a text file and returns the top N most frequent words, ignoring case and punctuation.

```python
# Exercise 1: First, let's create a sample text file
sample_text = """
Alice was beginning to get very tired of sitting by her sister on the bank, 
and of having nothing to do: once or twice she had peeped into the book her sister 
was reading, but it had no pictures or conversations in it, 'and what is the use 
of a book,' thought Alice 'without pictures or conversations?'

So she was considering in her own mind (as well as she could, for the hot day 
made her feel very sleepy and stupid), whether the pleasure of making a daisy-chain 
would be worth the trouble of getting up and picking the daisies, when suddenly a 
White Rabbit with pink eyes ran close by her.
"""

# Write to a temporary file
import tempfile
with tempfile.NamedTemporaryFile(mode='w+', delete=False, suffix='.txt') as f:
    f.write(sample_text)
    temp_filename = f.name

print(f"Created sample text file: {temp_filename}")
```

```python
# Your solution here (Exercise 1)

# Expected to see a function that:
# 1. Opens and reads the file
# 2. Processes text to remove punctuation and standardize case
# 3. Uses Counter to count word frequencies
# 4. Returns the N most common words
```

```python
# Solution for Exercise 1
def top_words(filename, n=10):
    import re
    from collections import Counter
    
    with open(filename, 'r') as f:
        text = f.read()
    
    # Remove punctuation and convert to lowercase
    words = re.findall(r'\b\w+\b', text.lower())
    
    # Count the words
    word_counts = Counter(words)
    
    # Return the top N words
    return word_counts.most_common(n)

# Test the function
print("Top 10 most common words:")
for word, count in top_words(temp_filename, 10):
    print(f"{word}: {count}")
```

### Exercise 2: Student Records with namedtuple

Create a system that manages student records using namedtuple and provides functions to calculate average grades, find the top student, etc.

```python
# Your solution here (Exercise 2)

# Expected to see:
# 1. Definition of Student namedtuple 
# 2. Creation of student records
# 3. Functions to calculate stats like average grade
# 4. Function to find the top student
```

```python
# Solution for Exercise 2
from collections import namedtuple

# Define a Student record
Student = namedtuple('Student', ['id', 'name', 'age', 'grades'])

# Create some student records
students = [
    Student(1, 'Alice', 20, [85, 90, 92, 88]),
    Student(2, 'Bob', 19, [75, 80, 85, 70]),
    Student(3, 'Charlie', 21, [95, 92, 88, 94]),
    Student(4, 'Diana', 20, [70, 75, 80, 72]),
    Student(5, 'Eva', 22, [88, 84, 82, 90])
]

# Function to calculate average grade for a student
def average_grade(student):
    return sum(student.grades) / len(student.grades)

# Function to find the student with the highest average
def top_student(students):
    return max(students, key=average_grade)

# Function to get students sorted by average grade
def students_by_performance(students):
    return sorted(students, key=average_grade, reverse=True)

# Display all student averages
print("Student Grade Averages:")
for student in students:
    avg = average_grade(student)
    print(f"{student.name}: {avg:.2f}")

# Find the top student
best = top_student(students)
print(f"\nTop student: {best.name} with average {average_grade(best):.2f}")

# Display students in order of performance
print("\nStudents by performance:")
for i, student in enumerate(students_by_performance(students), 1):
    print(f"{i}. {student.name}: {average_grade(student):.2f}")
```

Create a product inventory system that tracks products by category, allowing easy lookup of all products in a category.

```python
# Your solution here (Exercise 3)

# Expected to see:
# 1. Use of defaultdict to organize products by category
# 2. Functions to add products, look up products by category
# 3. Demo of the system
```

```python
# Solution for Exercise 3
from collections import defaultdict

class Inventory:
    def __init__(self):
        self.products_by_category = defaultdict(list)
        self.products_by_id = {}
    
    def add_product(self, product_id, name, category, price, quantity):
        product = {
            'id': product_id,
            'name': name,
            'category': category,
            'price': price,
            'quantity': quantity
        }
        self.products_by_category[category].append(product)
        self.products_by_id[product_id] = product
        return product
    
    def get_by_category(self, category):
        return self.products_by_category[category]  # Returns empty list if category doesn't exist
    
    def get_by_id(self, product_id):
        return self.products_by_id.get(product_id)
    
    def update_quantity(self, product_id, new_quantity):
        if product_id in self.products_by_id:
            self.products_by_id[product_id]['quantity'] = new_quantity
            return True
        return False
    
    def get_all_categories(self):
        return list(self.products_by_category.keys())

# Create inventory and add products
inventory = Inventory()

inventory.add_product(1, 'Laptop', 'Electronics', 999.99, 10)
inventory.add_product(2, 'Smartphone', 'Electronics', 699.99, 25)
inventory.add_product(3, 'Headphones', 'Electronics', 149.99, 30)
inventory.add_product(4, 'T-shirt', 'Clothing', 19.99, 100)
inventory.add_product(5, 'Jeans', 'Clothing', 49.99, 50)
inventory.add_product(6, 'Novel', 'Books', 12.99, 200)
inventory.add_product(7, 'Cookbook', 'Books', 24.99, 80)

# Demo the system
print("All categories:", inventory.get_all_categories())

print("\nElectronics products:")
for product in inventory.get_by_category('Electronics'):
    print(f"  {product['name']} - ${product['price']} ({product['quantity']} in stock)")

print("\nLooking up product by ID:")
laptop = inventory.get_by_id(1)
print(f"  Product ID 1: {laptop['name']} - ${laptop['price']}")

# Update quantity
inventory.update_quantity(1, 5)
print(f"\nUpdated laptop quantity: {inventory.get_by_id(1)['quantity']}")

# Looking for non-existent category (returns empty list)
print(f"\nToys category has {len(inventory.get_by_category('Toys'))} products")
```

In this notebook, we've explored the Python `collections` module and its specialized container datatypes:
