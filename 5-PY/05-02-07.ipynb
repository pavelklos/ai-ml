## 1. Classes and Objects Fundamentals

Classes are the blueprint for creating objects in object-oriented programming. They define attributes (data) and methods (functions) that objects of the class will have.

```python
# Basic class definition
class Person:
    """A simple class representing a person"""
    
    # Class attribute (shared by all instances)
    species = "Homo sapiens"
    
    # Initializer / Constructor
    def __init__(self, name, age):
        # Instance attributes (unique to each instance)
        self.name = name
        self.age = age
    
    # Instance method
    def introduce(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
    
    # Instance method
    def celebrate_birthday(self):
        self.age += 1
        return f"{self.name} is now {self.age} years old!"


# Creating objects (instances) of the class
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

# Accessing attributes and methods
print(person1.name)  # Output: Alice
print(person2.age)   # Output: 25
print(person1.introduce())  # Output: Hello, my name is Alice and I am 30 years old.
print(Person.species)  # Output: Homo sapiens (accessing class attribute)

# Modifying object state
print(person2.celebrate_birthday())  # Output: Bob is now 26 years old!
print(person2.age)  # Output: 26
```

### Key Concepts of Classes and Objects:

1. **Class**: A blueprint for creating objects.
2. **Object/Instance**: A specific realization of a class.
3. **Attributes**: Data stored inside a class or instance.
   - Class attributes: Shared by all instances
   - Instance attributes: Unique to each instance
4. **Methods**: Functions defined inside a class that describe behaviors.
5. **Self**: The first parameter of instance methods, referring to the instance itself.
6. **Constructor**: The `__init__` method that initializes new objects.

Now that we understand the basics, let's explore inheritance.

<a id="inheritance"></a>
## 2. Understanding Inheritance

Inheritance allows a class (child/derived class) to inherit attributes and methods from another class (parent/base class). This promotes code reuse and establishes a relationship between classes.

```python
# Base/Parent class
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
        
    def make_sound(self):
        print("Some generic animal sound")
        
    def info(self):
        return f"{self.name} is a {self.species}"

# Derived/Child class
class Dog(Animal):  # Dog inherits from Animal
    def __init__(self, name, breed):
        # Call the parent class's __init__ method
        super().__init__(name, species="Dog")
        self.breed = breed
        
    def make_sound(self):
        print("Woof!")
        
    def info(self):
        basic_info = super().info()
        return f"{basic_info} of breed {self.breed}"

# Create instances
generic_animal = Animal("Generic", "Unknown")
my_dog = Dog("Rex", "German Shepherd")

# Test inherited and overridden methods
print(generic_animal.info())  # Output: Generic is a Unknown
generic_animal.make_sound()   # Output: Some generic animal sound

print(my_dog.info())  # Output: Rex is a Dog of breed German Shepherd
my_dog.make_sound()   # Output: Woof!

# Check inheritance relationship
print(isinstance(my_dog, Dog))      # True
print(isinstance(my_dog, Animal))   # True - Dog is a subclass of Animal
print(issubclass(Dog, Animal))      # True
```

### Benefits of Inheritance:

1. **Code Reusability**: Child classes inherit attributes and methods from parent classes, reducing code duplication.
2. **Extensibility**: Child classes can extend parent functionality by adding new methods or overriding existing ones.
3. **Hierarchical Models**: Inheritance allows for building hierarchical models that mirror real-world relationships.
4. **Modularity**: Classes in an inheritance hierarchy can be modified independently.

### Inheritance Terminology:

- **Parent/Base/Super class**: The class being inherited from
- **Child/Derived/Sub class**: The class that inherits from the base class
- **Overriding**: Redefining a method in a subclass that was already defined in the parent class
- **Extension**: Adding new functionality in the derived class

<a id="types-of-inheritance"></a>
## 3. Types of Inheritance

Python supports various types of inheritance patterns:

1. **Single Inheritance**: A class inherits from one parent class
2. **Multi-level Inheritance**: A class inherits from a child class
3. **Multiple Inheritance**: A class inherits from multiple parent classes
4. **Hierarchical Inheritance**: Multiple classes inherit from a single base class

Let's explore each type with examples:

```python
# 1. Single Inheritance (already demonstrated above with Animal -> Dog)

# 2. Multi-level Inheritance
class Animal:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        return f"{self.name} is eating"

class Mammal(Animal):
    def walk(self):
        return f"{self.name} is walking"

class Dog(Mammal):
    def bark(self):
        return f"{self.name} is barking"

# Create instance of multi-level inheritance
my_dog = Dog("Buddy")
print(my_dog.eat())    # Inherited from Animal
print(my_dog.walk())   # Inherited from Mammal
print(my_dog.bark())   # Defined in Dog

# 3. Multiple Inheritance
class Swimmer:
    def swim(self):
        return f"{self.name} is swimming"

class Flyer:
    def fly(self):
        return f"{self.name} is flying"

class Duck(Animal, Swimmer, Flyer):
    def quack(self):
        return f"{self.name} is quacking"

# Create instance with multiple inheritance
my_duck = Duck("Donald")
print(my_duck.eat())   # From Animal
print(my_duck.swim())  # From Swimmer
print(my_duck.fly())   # From Flyer
print(my_duck.quack()) # From Duck

# 4. Hierarchical Inheritance
class Cat(Animal):
    def meow(self):
        return f"{self.name} is meowing"

# Create instances of different child classes from the same parent
animal = Animal("Generic Animal")
dog = Dog("Rex")
cat = Cat("Whiskers")

print(animal.eat())
print(dog.eat())  # Both dog and cat inherited eat() from Animal
print(cat.eat())
```

### Method Resolution Order (MRO)

When dealing with multiple inheritance, Python needs to determine the order in which it looks for methods and attributes in the parent classes. This is called the Method Resolution Order (MRO).

Python uses the C3 Linearization algorithm to determine the MRO, which ensures:
1. A child class comes before its parent classes
2. The order of parent classes in the inheritance declaration is preserved

```python
# Demonstrating Method Resolution Order (MRO)
class A:
    def method(self):
        return "Method from class A"

class B(A):
    def method(self):
        return "Method from class B"

class C(A):
    def method(self):
        return "Method from class C"

class D(B, C):
    pass

# Let's check the MRO
print(D.__mro__)

# Create an instance and call method()
d = D()
print(d.method())  # Which method will be called?

# The output shows that D.method() resolves to B.method() because:
# 1. D is checked first (but it doesn't have method())
# 2. B is checked next because it appears first in D's inheritance list
# 3. Since B has method(), it is called
```

<a id="method-overriding"></a>
## 4. Method Overriding and super()

**Method Overriding** occurs when a child class provides a specific implementation of a method that is already defined in its parent class. This allows child classes to change behavior inherited from parent classes.

The `super()` function is used to call methods from the parent class in the child class, allowing for extension rather than complete replacement.

```python
class Vehicle:
    def __init__(self, name, color):
        self.name = name
        self.color = color
        
    def get_info(self):
        return f"{self.color} {self.name}"
    
    def move(self):
        return f"{self.name} is moving"

class Car(Vehicle):
    def __init__(self, name, color, model):
        # Call parent's __init__ using super()
        super().__init__(name, color)
        self.model = model
        
    def get_info(self):
        # Extend the parent's get_info method
        basic_info = super().get_info()
        return f"{basic_info}, Model: {self.model}"
    
    def move(self):
        return f"{self.name} is driving on the road"

# Create instances
vehicle = Vehicle("Generic Vehicle", "Gray")
car = Car("Toyota", "Red", "Camry")

# Test methods
print(vehicle.get_info())  # Gray Generic Vehicle
print(vehicle.move())      # Generic Vehicle is moving

print(car.get_info())      # Red Toyota, Model: Camry
print(car.move())          # Toyota is driving on the road
```

### Key Points About Method Overriding:

1. **Method Signature**: The overridden method in the child class should have the same name, parameters, and return type as the parent class method.

2. **Using `super()`**: The `super()` function allows you to call the parent class method from within the child class, enabling you to extend rather than completely replace the parent's functionality.

3. **Constructor Overriding**: A common use case is overriding the `__init__` method to initialize additional attributes while still calling the parent's constructor.

4. **Dynamic Method Resolution**: Python automatically resolves which method to call based on the object's type at runtime.

<a id="polymorphism"></a>
## 5. Polymorphism in Python

Polymorphism means "many forms" and allows objects of different classes to be treated as objects of a common superclass. The key benefit is that it enables a single interface to represent different underlying forms (data types).

In Python, polymorphism is achieved through method overriding (as we've seen) and through duck typing.

```python
# Polymorphism through inheritance
class Shape:
    def area(self):
        raise NotImplementedError("Subclass must implement abstract method")
    
    def perimeter(self):
        raise NotImplementedError("Subclass must implement abstract method")

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length * self.width
    
    def perimeter(self):
        return 2 * (self.length + self.width)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        import math
        return math.pi * self.radius ** 2
    
    def perimeter(self):
        import math
        return 2 * math.pi * self.radius

# Polymorphic function
def print_shape_info(shape):
    print(f"Area: {shape.area()}")
    print(f"Perimeter: {shape.perimeter()}")

# Create objects
rectangle = Rectangle(5, 10)
circle = Circle(7)

# The same function works for different shapes
print("Rectangle:")
print_shape_info(rectangle)  

print("\nCircle:")
print_shape_info(circle)
```

### Benefits of Polymorphism:

1. **Interface Uniformity**: Objects of different types can be accessed through the same interface.
2. **Code Reusability**: The same code works with different data types.
3. **Extensibility**: New classes can be added that implement the interface, without modifying existing code.
4. **Flexibility**: Functions can be designed to operate on objects of various types as long as they support the required methods.

In the example above, `print_shape_info()` works with any object that has `area()` and `perimeter()` methods, regardless of the actual class. This is polymorphism in action.

<a id="duck-typing"></a>
## 6. Duck Typing

Python employs a concept called "duck typing" - if an object walks like a duck and quacks like a duck, it is a duck. This means that Python doesn't care about the object's type, only that it has the methods and properties needed for the operation.

Unlike statically typed languages that determine compatibility at compile time, Python checks for compatibility at runtime, making it more flexible.

```python
# Duck typing example

class Duck:
    def sound(self):
        return "Quack"
    
    def walk(self):
        return "Walking like a duck"

class Person:
    def sound(self):
        return "Hello"
    
    def walk(self):
        return "Walking like a human"

class Dog:
    def sound(self):
        return "Woof"
    
    def walk(self):
        return "Walking like a dog"

# Function that doesn't care about the type of object, only that it has sound() and walk() methods
def interact_with(animal):
    print(animal.sound())
    print(animal.walk())
    print()

# Create instances
donald = Duck()
john = Person()
fido = Dog()

# Each works with interact_with() because they implement the required methods
interact_with(donald)
interact_with(john)
interact_with(fido)
```

### Duck Typing vs. Traditional Polymorphism:

1. **No Inheritance Requirement**: With duck typing, classes don't need to share a common ancestor to be used polymorphically.

2. **Runtime Checking**: Compatibility is checked at runtime, not compile time.

3. **Emphasis on Behavior**: The focus is on what an object can do (its methods) rather than what it is (its class).

4. **Flexibility**: Duck typing allows for more flexible code designs but with less formal interfaces.

Duck typing is a powerful feature in Python that enables more flexible and generic code, but it can also lead to runtime errors if objects don't have the expected methods.

## 7. Abstract Base Classes

Abstract Base Classes (ABCs) define a common API for a set of subclasses. They can't be instantiated directly but serve as a template for other classes.

Python provides the `abc` module for defining ABCs:

```python
from abc import ABC, abstractmethod

# Abstract base class
class Animal(ABC):
    
    def __init__(self, name):
        self.name = name
    
    @abstractmethod
    def sound(self):
        """Make the animal's sound"""
        pass
    
    @abstractmethod
    def move(self):
        """Describe how the animal moves"""
        pass
    
    # Concrete method (not abstract)
    def introduce(self):
        return f"I am {self.name}, and I say {self.sound()}"

# Concrete implementations
class Dog(Animal):
    def sound(self):
        return "Woof!"
    
    def move(self):
        return "runs on four legs"

class Bird(Animal):
    def sound(self):
        return "Tweet!"
    
    def move(self):
        return "flies through the air"

# This would raise an error: TypeError: Can't instantiate abstract class Animal with abstract methods move, sound
# generic_animal = Animal("Generic")

# These work fine because they implement all abstract methods
dog = Dog("Rex")
bird = Bird("Tweety")

print(dog.introduce())  # I am Rex, and I say Woof!
print(bird.introduce()) # I am Tweety, and I say Tweet!

# We can see the move method works correctly
print(f"{dog.name} {dog.move()}")  # Rex runs on four legs
print(f"{bird.name} {bird.move()}") # Tweety flies through the air
```

### Benefits of Abstract Base Classes:

1. **Contract Enforcement**: ABCs enforce that derived classes implement specific methods.

2. **Documentation**: They clearly document what methods subclasses need to implement.

3. **Consistency**: They help ensure a consistent interface across related classes.

4. **Type Checking**: Python's `isinstance()` and `issubclass()` work with ABCs.

### ABC Components:

- The `ABC` class from the `abc` module is used as a metaclass for abstract base classes.
- The `@abstractmethod` decorator marks methods that must be implemented by subclasses.
- Abstract classes can include concrete (implemented) methods as well as abstract ones.

<a id="special-methods"></a>
## 8. Special Methods for Polymorphic Behavior

Python's special methods (also called dunder methods for "double underscore") allow classes to integrate with Python's built-in functionality. These methods enable polymorphic behavior with Python's operators and built-in functions.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # String representation
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    # Formal string representation
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    # Addition
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    # Subtraction
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    # Length/magnitude
    def __len__(self):
        import math
        return int(math.sqrt(self.x**2 + self.y**2))
    
    # Equals comparison
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        return self.x == other.x and self.y == other.y
    
    # Less than comparison (for sorting)
    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return self.x**2 + self.y**2 < other.x**2 + other.y**2

# Create vectors
v1 = Vector(3, 4)
v2 = Vector(5, 6)

# Use special methods through built-in operations
print(v1)               # __str__: Vector(3, 4)
print(v1 + v2)          # __add__: Vector(8, 10)
print(v1 - v2)          # __sub__: Vector(-2, -2)
print(len(v1))          # __len__: 5
print(v1 == Vector(3, 4)) # __eq__: True
print(v1 < v2)          # __lt__: True

# Sorting using __lt__
vectors = [Vector(5, 2), Vector(3, 4), Vector(1, 7)]
sorted_vectors = sorted(vectors)
print([str(v) for v in sorted_vectors])  # Sorted by magnitude
```

### Common Special Methods:

1. **String representation**:
   - `__str__`: For informal string representation (str())
   - `__repr__`: For formal string representation (repr())

2. **Operators**:
   - `__add__`, `__sub__`, `__mul__`, `__truediv__`: Arithmetic operations (+, -, *, /)
   - `__eq__`, `__ne__`, `__lt__`, `__gt__`, `__le__`, `__ge__`: Comparisons (==, !=, <, >, <=, >=)

3. **Container behavior**:
   - `__len__`: Length (len())
   - `__getitem__`, `__setitem__`: Indexing (obj[index])
   - `__contains__`: Membership test (in)
   - `__iter__`: Iteration (for x in obj)

4. **Function behavior**:
   - `__call__`: Make instance callable like a function

By implementing these special methods, you can create classes that behave similarly to built-in types and seamlessly integrate with Python's syntax.

## 9. Practical OOP Design

Let's build a practical example that combines inheritance and polymorphism: a simple banking system with different account types.

```python
from abc import ABC, abstractmethod
from datetime import datetime

# Abstract base class
class BankAccount(ABC):
    def __init__(self, account_number, owner_name, balance=0):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance
        self.transaction_history = []
    
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self.balance += amount
        self.log_transaction("Deposit", amount)
        return f"Deposited ${amount}. New balance: ${self.balance}"
    
    @abstractmethod
    def withdraw(self, amount):
        """Each account type must implement its own withdrawal logic"""
        pass
    
    def get_balance(self):
        return f"Current balance: ${self.balance}"
    
    def log_transaction(self, transaction_type, amount):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        transaction = {
            "timestamp": timestamp,
            "type": transaction_type,
            "amount": amount,
            "balance": self.balance
        }
        self.transaction_history.append(transaction)
    
    def get_transaction_history(self):
        return self.transaction_history
    
    def __str__(self):
        return f"{self.__class__.__name__} - Account #{self.account_number} - Owner: {self.owner_name} - Balance: ${self.balance}"

# Concrete implementation: Checking Account
class CheckingAccount(BankAccount):
    def __init__(self, account_number, owner_name, balance=0, overdraft_limit=100):
        super().__init__(account_number, owner_name, balance)
        self.overdraft_limit = overdraft_limit
    
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
            
        if self.balance - amount >= -self.overdraft_limit:
            self.balance -= amount
            self.log_transaction("Withdrawal", amount)
            
            if self.balance < 0:
                return f"Withdrew ${amount} (${-self.balance} in overdraft). New balance: ${self.balance}"
            else:
                return f"Withdrew ${amount}. New balance: ${self.balance}"
        else:
            raise ValueError(f"Cannot withdraw ${amount}. It would exceed your overdraft limit.")

# Concrete implementation: Savings Account
class SavingsAccount(BankAccount):
    def __init__(self, account_number, owner_name, balance=0, interest_rate=0.01):
        super().__init__(account_number, owner_name, balance)
        self.interest_rate = interest_rate
    
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
            
        if amount <= self.balance:
            self.balance -= amount
            self.log_transaction("Withdrawal", amount)
            return f"Withdrew ${amount}. New balance: ${self.balance}"
        else:
            raise ValueError("Insufficient funds for withdrawal")
    
    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        self.log_transaction("Interest", interest)
        return f"Added interest: ${interest:.2f}. New balance: ${self.balance}"

# Concrete implementation: Fixed Deposit Account
class FixedDepositAccount(BankAccount):
    def __init__(self, account_number, owner_name, balance=0, term_months=12, interest_rate=0.05, maturity_date=None):
        super().__init__(account_number, owner_name, balance)
        self.term_months = term_months
        self.interest_rate = interest_rate
        
        # If no maturity date provided, set it to term_months from now
        if maturity_date is None:
            from datetime import datetime, timedelta
            self.maturity_date = datetime.now() + timedelta(days=term_months * 30)
        else:
            self.maturity_date = maturity_date
    
    def withdraw(self, amount):
        if datetime.now() < self.maturity_date:
            raise ValueError(f"Cannot withdraw before maturity date ({self.maturity_date.strftime('%Y-%m-%d')})")
        
        if amount <= self.balance:
            self.balance -= amount
            self.log_transaction("Withdrawal", amount)
            return f"Withdrew ${amount}. New balance: ${self.balance}"
        else:
            raise ValueError("Insufficient funds for withdrawal")
    
    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        self.log_transaction("Interest", interest)
        return f"Added interest: ${interest:.2f}. New balance: ${self.balance}"
    
    def __str__(self):
        basic_info = super().__str__()
        return f"{basic_info} - Matures on: {self.maturity_date.strftime('%Y-%m-%d')}"

# Demonstration
def demonstrate_bank_accounts():
    # Create accounts
    checking = CheckingAccount("CH001", "John Doe", 1000, 200)
    savings = SavingsAccount("SV001", "Jane Smith", 5000, 0.02)
    fd = FixedDepositAccount("FD001", "Robert Johnson", 10000, 6, 0.06)
    
    print("--- Initial Account Information ---")
    print(checking)
    print(savings)
    print(fd)
    print()
    
    # Perform operations on checking account
    print("--- Checking Account Operations ---")
    print(checking.deposit(500))
    print(checking.withdraw(1200))  # This will use overdraft
    try:
        print(checking.withdraw(600))  # This should exceed overdraft limit
    except ValueError as e:
        print(f"Error: {e}")
    print()
    
    # Perform operations on savings account
    print("--- Savings Account Operations ---")
    print(savings.deposit(1000))
    print(savings.withdraw(2000))
    print(savings.add_interest())
    try:
        print(savings.withdraw(10000))  # This should exceed balance
    except ValueError as e:
        print(f"Error: {e}")
    print()
    
    # Perform operations on fixed deposit account
    print("--- Fixed Deposit Account Operations ---")
    print(fd.deposit(2000))
    print(fd.add_interest())
    try:
        print(fd.withdraw(1000))  # This should fail before maturity date
    except ValueError as e:
        print(f"Error: {e}")
    print()
    
    # Show polymorphism with a list of different account types
    accounts = [checking, savings, fd]
    
    print("--- Polymorphic Operations ---")
    for i, account in enumerate(accounts):
        print(f"Account {i+1}: {account.get_balance()}")
    
    print("\n--- Transaction History for Checking Account ---")
    for transaction in checking.get_transaction_history():
        print(f"{transaction['timestamp']} - {transaction['type']} - ${transaction['amount']} - Balance: ${transaction['balance']}")

# Run the demonstration
demonstrate_bank_accounts()
```

### Design Patterns Demonstrated:

1. **Inheritance Hierarchy**: Base abstract class with specialized subclasses.
2. **Polymorphism**: Common interface through the base class, with specialized behavior in subclasses.
3. **Encapsulation**: Data and methods are bundled together.
4. **Abstraction**: Abstract methods define a contract for subclasses.

This practical example shows how inheritance and polymorphism work together to create a flexible, maintainable system:

1. **Flexibility**: New account types can be added by creating new subclasses.
2. **Code Reuse**: Common functionality is in the base class.
3. **Type-Specific Behavior**: Each account type implements its own rules for withdrawals.
4. **Consistent Interface**: All accounts share the same method names and parameters.
