# 9 Fundamental Machine Learning Paradigms with Examples

## 1. Regression
**Description**: Predicts continuous values. Used for forecasting, trend analysis, and modeling relationships between variables.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# Generate data
np.random.seed(42)
X = np.linspace(0, 10, 100)
y = 2*X + 1 + np.random.normal(0, 1, 100)
df = pd.DataFrame({'Hours_Studied': X, 'Exam_Score': y})

# Visualize
plt.scatter(X, y)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')

# Model
X_train, X_test, y_train, y_test = train_test_split(X.reshape(-1,1), y, test_size=0.2)
model = LinearRegression()
model.fit(X_train, y_train)

# Evaluate
print(f"R²: {model.score(X_test, y_test):.2f}")
plt.plot(X, model.predict(X.reshape(-1,1)), color='red')
plt.show()
```

## 2. Classification
**Description**: Predicts discrete class labels. Used for spam detection, image recognition, medical diagnosis.

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Generate data
X, y = make_classification(n_samples=200, n_features=2, n_redundant=0, 
                           n_classes=2, random_state=42)
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1], 'Class': y})

# Visualize
plt.scatter(X[:,0], X[:,1], c=y)
plt.title('Classification Problem')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')

# Model
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = LogisticRegression()
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
```

## 3. Clustering
**Description**: Groups similar data points. Used for customer segmentation, image compression, anomaly detection.

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Generate data
X, _ = make_blobs(n_samples=150, centers=3, random_state=42)
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1]})

# Visualize
plt.scatter(X[:,0], X[:,1])
plt.title('Unlabeled Data')

# Model
model = KMeans(n_clusters=3)
model.fit(X)
labels = model.predict(X)

# Visualize clusters
plt.scatter(X[:,0], X[:,1], c=labels)
plt.title('Clustered Data')
plt.show()
```

## 4. Dimensionality Reduction
**Description**: Reduces number of features while preserving structure. Used for visualization, noise reduction, efficiency.

```python
from sklearn.decomposition import PCA

# Generate data
np.random.seed(42)
X = np.random.randn(100, 3)
X[:,2] = X[:,0] + X[:,1] + 0.1*np.random.randn(100)
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1], 'Feature3': X[:,2]})

# Model
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Visualize
plt.scatter(X_pca[:,0], X_pca[:,1])
plt.title('PCA Reduced Data (3D→2D)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

## 5. Ensemble Learning
**Description**: Combines multiple models for better performance. Used for competitions, robust systems.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification

# Generate data
X, y = make_classification(n_samples=200, n_features=5, random_state=42)
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1], 'Class': y})

# Model
model = RandomForestClassifier(n_estimators=100)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model.fit(X_train, y_train)

# Evaluate
print(f"Accuracy: {model.score(X_test, y_test):.2f}")
```

## 6. Anomaly Detection
**Description**: Identifies rare items/events. Used for fraud detection, system health monitoring.

```python
from sklearn.ensemble import IsolationForest

# Generate data
X = 0.3 * np.random.randn(100, 2)
X = np.r_[X + 2, X - 2, [[0, 0], [2.5, 2.5]]]  # Add outliers
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1]})

# Model
model = IsolationForest(contamination=0.05)
model.fit(X)
anomalies = model.predict(X)

# Visualize
plt.scatter(X[:,0], X[:,1], c=anomalies)
plt.title('Anomaly Detection')
plt.show()
```

## 7. Time Series Analysis
**Description**: Analyzes time-dependent data. Used for stock prediction, weather forecasting.

```python
from statsmodels.tsa.arima.model import ARIMA

# Generate data
dates = pd.date_range('2023-01-01', periods=100)
values = np.cumsum(np.random.randn(100)) + 50
df = pd.DataFrame({'Date': dates, 'Value': values}).set_index('Date')

# Visualize
df.plot()
plt.title('Time Series Data')

# Model
model = ARIMA(df, order=(1,1,1))
results = model.fit()

# Forecast
forecast = results.forecast(steps=5)
print(f"Forecast: {forecast}")
```

## 8. Semi-Supervised Learning
**Description**: Uses both labeled and unlabeled data. Used when labeling is expensive.

```python
from sklearn.semi_supervised import LabelSpreading

# Generate data
X, y = make_classification(n_samples=200, n_features=2, random_state=42)
y[50:] = -1  # Unlabel most of the data
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1], 'Label': y})

# Model
model = LabelSpreading(kernel='knn', alpha=0.8)
model.fit(X, y)

# Evaluate
pred_labels = model.transduction_
plt.scatter(X[:,0], X[:,1], c=pred_labels)
plt.title('Semi-Supervised Learning')
plt.show()
```

## 9. Supervised Learning
**Description**: Learns from labeled training data. Most common paradigm for predictive modeling.

```python
from sklearn.neighbors import KNeighborsClassifier

# Generate data
X, y = make_classification(n_samples=200, n_features=2, n_classes=3,
                           n_clusters_per_class=1, random_state=42)
df = pd.DataFrame({'Feature1': X[:,0], 'Feature2': X[:,1], 'Class': y})

# Visualize
plt.scatter(X[:,0], X[:,1], c=y)
plt.title('Supervised Learning Problem')

# Model
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Evaluate
print(f"Accuracy: {model.score(X_test, y_test):.2f}")
```

## Final Decision: Most Used Paradigm in Real ML
After examining all paradigms, Supervised Learning (particularly classification and regression) is the most widely used in real-world machine learning applications because:

1. Most business problems have clear input-output pairs (labeled data)
2. Easier to evaluate and explain to stakeholders
3. Well-established algorithms and frameworks
4. Directly solves common predictive tasks

However, the best paradigm depends on the specific problem - unsupervised methods like clustering are growing rapidly, and ensemble methods often win competitions.
