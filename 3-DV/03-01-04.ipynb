# Pie Charts and When to Use Them

This notebook explores pie charts in data visualization, including their creation, customization, appropriate use cases, and alternatives. We'll learn when pie charts are effective and when other visualization types might be better choices.

## 1. Import Required Libraries

First, let's import the necessary libraries to create and customize pie charts.

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for our plots
plt.style.use('seaborn-v0_8')
sns.set_palette("colorblind")

# For displaying plots inline in Jupyter
%matplotlib inline
```

## 2. Introduction to Pie Charts

Pie charts are circular statistical graphics divided into slices to illustrate numerical proportions. The arc length of each slice (and consequently its area and central angle) is proportional to the quantity it represents.

### Key characteristics:
- Represent parts of a whole
- Sum of all slices equals 100%
- Best used for displaying relative proportions when you have a small number of categories
- Each slice represents a category's proportion of the total

## 3. Creating Basic Pie Charts

Let's create a simple pie chart showing market share of different smartphone brands.

```python
# Sample data for smartphone market share (Q2 2023)
brands = ['Samsung', 'Apple', 'Xiaomi', 'Oppo', 'Others']
market_share = [20, 16, 13, 10, 41]

# Create a simple pie chart
plt.figure(figsize=(10, 7))
plt.pie(market_share, labels=brands)
plt.title('Smartphone Market Share (Q2 2023)', fontsize=14)
plt.show()
```

### Creating a Pie Chart from a Pandas DataFrame

We can also create pie charts directly from pandas DataFrames.

```python
# Create a DataFrame with the market share data
df_market = pd.DataFrame({
    'Brand': brands,
    'Market_Share': market_share
})

# Display the DataFrame
print(df_market)

# Create a pie chart from the DataFrame
plt.figure(figsize=(10, 7))
df_market.plot.pie(y='Market_Share', labels=brands, autopct='%1.1f%%', figsize=(10, 7))
plt.title('Smartphone Market Share (Q2 2023)', fontsize=14)
plt.ylabel('')  # Hide the y-label
plt.show()
```

## 4. Customizing Pie Charts

Pie charts can be customized in various ways to enhance their visual appeal and readability.

```python
# Custom colors and styling
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
explode = (0.1, 0, 0, 0, 0)  # Explode the 1st slice (Samsung)

plt.figure(figsize=(12, 8))
plt.pie(market_share, 
        labels=brands, 
        colors=colors, 
        autopct='%1.1f%%',  # Show percentages
        startangle=90,      # Rotate the chart
        shadow=True,        # Add shadow
        explode=explode,    # Explode a slice
        textprops={'fontsize': 14})  # Text properties

plt.title('Customized Smartphone Market Share Pie Chart', fontsize=16)
plt.axis('equal')  # Equal aspect ratio ensures the pie is circular
plt.tight_layout()
plt.show()
```

## 5. Exploding Slices

Exploding slices is a technique to emphasize specific segments by pulling them out from the main pie chart.

```python
# Exploding multiple slices with different distances
explode_multiple = (0.1, 0.2, 0, 0, 0)  # Explode Samsung and Apple slices

plt.figure(figsize=(10, 7))
plt.pie(market_share, 
        labels=brands, 
        explode=explode_multiple,
        autopct='%1.1f%%',
        shadow=True,
        startangle=140)
plt.title('Pie Chart with Multiple Exploded Slices', fontsize=14)
plt.axis('equal')
plt.show()
```

## 6. Adding Labels and Percentages

Well-labeled pie charts communicate information more effectively. Let's explore different ways to add informative labels to pie charts.

```python
# Adding custom labels with percentages and absolute values
def autopct_format(values):
    def my_format(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{p:.1f}%\n({v:d})'.format(p=pct, v=val)
    return my_format

plt.figure(figsize=(12, 8))
pie_wedge_collection, texts, autotexts = plt.pie(market_share, 
                                               labels=brands, 
                                               autopct=autopct_format(market_share),
                                               shadow=True,
                                               startangle=90,
                                               textprops={'fontsize': 12})

# Styling the text labels
for text in texts:
    text.set_color('darkblue')
    
# Styling the percentage labels
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')
    
plt.title('Smartphone Market Share with Detailed Labels', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

### Adding a Legend

Adding a legend can sometimes improve readability, especially when labels might overlap or be hard to read.

```python
plt.figure(figsize=(12, 8))
plt.pie(market_share, 
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        shadow=True)

# Add a legend outside the pie chart
plt.legend(brands, title="Smartphone Brands", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

plt.title('Smartphone Market Share with External Legend', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

## 7. When to Use Pie Charts

Pie charts are most effective in the following scenarios:

1. **Showing parts of a whole**: When you want to show how different parts contribute to a total.
2. **Small number of categories**: Ideally 5-7 categories at most.
3. **Significant differences between slices**: When values have notable differences.
4. **Presenting simple proportions**: For high-level, non-technical audiences.
5. **Categorical data**: For nominal (categorical) data rather than continuous data.

Let's see an example of an effective pie chart use case:

```python
# Energy sources - a good use case for pie charts
energy_sources = ['Oil', 'Coal', 'Natural Gas', 'Nuclear', 'Renewables']
percentages = [33, 27, 24, 6, 10]

plt.figure(figsize=(10, 7))
plt.pie(percentages, 
        labels=energy_sources, 
        autopct='%1.1f%%',
        startangle=90,
        colors=sns.color_palette("Set3", len(energy_sources)),
        shadow=False)
plt.title('Global Energy Consumption by Source', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

## 8. When Not to Use Pie Charts

Pie charts have several limitations and should be avoided in these situations:

1. **Too many categories**: Pie charts with many slices become cluttered and hard to read
2. **Similar-sized slices**: When slices are similar in size, differences are hard to perceive
3. **When precise comparisons are needed**: Humans are not good at comparing angles or areas
4. **Time series data**: Showing changes over time is better with line or bar charts
5. **Multiple variables**: Pie charts only show one variable (the proportion)
6. **Small differences**: When differences between slices are small

Let's demonstrate a bad example of using pie charts:

```python
# Too many categories - bad use case for pie charts
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
values = [8.1, 8.0, 8.3, 8.4, 8.2, 8.5, 8.6, 8.4, 8.3, 8.1, 8.0, 8.1]

plt.figure(figsize=(10, 7))
plt.pie(values, labels=months)
plt.title('Monthly Sales Distribution (%) - A Poor Pie Chart Example', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()

print("This pie chart fails because:")
print("1. Too many categories (12 months)")
print("2. All slices are nearly identical in size (8.0% - 8.6%)")
print("3. It's difficult to compare the months or see patterns")
```

## 9. Alternatives to Pie Charts

When pie charts aren't appropriate, consider these alternatives:

### 1. Bar Charts
Bar charts are excellent for comparing values across categories.

```python
# Bar chart as an alternative to pie chart
plt.figure(figsize=(12, 7))
bars = plt.bar(brands, market_share, color=colors)

# Adding the values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.5,
            f'{height}%', ha='center', fontsize=12)

plt.title('Smartphone Market Share (Q2 2023)', fontsize=16)
plt.xlabel('Brand', fontsize=12)
plt.ylabel('Market Share (%)', fontsize=12)
plt.ylim(0, max(market_share) * 1.15)  # Add some space for labels
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 2. Stacked Bar Charts
Stacked bar charts are useful for showing part-to-whole relationships while also allowing comparisons across different groups.

```python
# Creating a stacked bar chart for market share across regions
regions = ['North America', 'Europe', 'Asia', 'Rest of World']

# Market share by brand and region (fictional data)
data = {
    'Samsung': [15, 20, 22, 19],
    'Apple': [35, 25, 10, 12],
    'Xiaomi': [5, 10, 18, 12],
    'Oppo': [3, 5, 15, 8],
    'Others': [42, 40, 35, 49]
}

df = pd.DataFrame(data, index=regions)

# Plot stacked bar chart
ax = df.plot(kind='bar', stacked=True, figsize=(12, 7), colormap='tab10')
ax.set_xlabel('Region', fontsize=12)
ax.set_ylabel('Market Share (%)', fontsize=12)
ax.set_title('Smartphone Market Share by Region and Brand', fontsize=16)
ax.legend(title='Brands', bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Display percentages on the bars
for container in ax.containers:
    ax.bar_label(container, label_type='center', fmt='%.0f%%')

plt.tight_layout()
plt.show()
```

### 3. Donut Charts
Donut charts are a variation of pie charts with a hole in the middle, which can be more aesthetically pleasing and provide space for additional information.

```python
# Create a donut chart
plt.figure(figsize=(10, 7))
plt.pie(market_share, 
        labels=brands, 
        autopct='%1.1f%%',
        startangle=90,
        colors=colors,
        wedgeprops={'edgecolor': 'white', 'linewidth': 2})

# Add a circle at the center to create a donut chart
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

# Add text in the center
plt.text(0, 0, 'Smartphone\nMarket Share', 
         horizontalalignment='center', 
         verticalalignment='center', 
         fontsize=16)

plt.title('Donut Chart: Smartphone Market Share', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

### 4. Treemaps
Treemaps display hierarchical data as nested rectangles, with the size of each rectangle proportional to its value.

```python
# Create a treemap using plotly
import plotly.express as px

# Prepare the data
df_treemap = pd.DataFrame({
    'Brand': brands,
    'Market_Share': market_share
})

fig = px.treemap(df_treemap, 
                path=['Brand'], 
                values='Market_Share',
                title='Smartphone Market Share as a Treemap')

fig.update_layout(
    title_font_size=22,
    font_size=14,
    margin=dict(t=50, l=25, r=25, b=25)
)

fig.show()
```

## 10. Practical Examples

Let's explore some real-world examples of effective pie chart usage.

### Example 1: Website Traffic by Source

```python
# Website traffic by source - a good use case for pie charts
traffic_sources = ['Organic Search', 'Paid Search', 'Social Media', 'Direct', 'Referral', 'Email']
traffic_values = [45, 20, 15, 10, 7, 3]

plt.figure(figsize=(12, 8))
plt.pie(traffic_values, 
        labels=traffic_sources, 
        autopct='%1.1f%%',
        startangle=140,
        explode=(0.1, 0, 0, 0, 0, 0),
        colors=sns.color_palette("pastel", len(traffic_sources)),
        wedgeprops={'edgecolor': 'white', 'linewidth': 2})
plt.title('Website Traffic by Source', fontsize=16)
plt.axis('equal')

# Add a shadow effect
plt.gca().add_artist(plt.Circle((0, 0), 0.7, fc='white'))
plt.tight_layout()
plt.show()
```

### Example 2: Expenditure Breakdown

A pie chart is appropriate for showing how a budget is allocated across different categories.

```python
# Monthly budget allocation - another good use case
budget_categories = ['Housing', 'Food', 'Transportation', 'Entertainment', 'Savings', 'Other']
budget_percentages = [35, 20, 15, 10, 15, 5]

# Create a DataFrame
df_budget = pd.DataFrame({
    'Category': budget_categories,
    'Percentage': budget_percentages
})

# Sort values to make the pie chart more readable
df_budget_sorted = df_budget.sort_values('Percentage', ascending=False)

# Create two subplots: a pie chart and a table
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8), gridspec_kw={'width_ratios': [1.5, 1]})

# Pie chart
wedges, texts, autotexts = ax1.pie(df_budget_sorted['Percentage'], 
                                  autopct='%1.1f%%',
                                  startangle=90,
                                  explode=[0.05]*len(budget_categories),
                                  colors=sns.color_palette("bright", len(budget_categories)))

# Add a legend with percentages
legend_labels = [f"{cat} ({pct}%)" for cat, pct in zip(df_budget_sorted['Category'], df_budget_sorted['Percentage'])]
ax1.legend(wedges, legend_labels, title="Budget Categories", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

ax1.set_title('Monthly Budget Allocation', fontsize=16)
ax1.axis('equal')

# Table showing the actual numbers
table_data = [['Category', 'Percentage (%)']]
for index, row in df_budget_sorted.iterrows():
    table_data.append([row['Category'], row['Percentage']])

ax2.axis('tight')
ax2.axis('off')
table = ax2.table(cellText=table_data, 
                 colWidths=[0.5, 0.3], 
                 loc='center',
                 cellLoc='center')
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.2)

# Style the header row
for i in range(len(table_data[0])):
    table[(0, i)].set_facecolor('#4472C4')
    table[(0, i)].set_text_props(color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```

### Example 3: Comparing Pie Charts with Bar Charts

Let's compare a pie chart and a bar chart for the same dataset to see their strengths and weaknesses.

```python
# Create sample data - survey responses about favorite programming languages
languages = ['Python', 'JavaScript', 'Java', 'C++', 'C#', 'Other']
responses = [35, 25, 15, 10, 10, 5]

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# Pie chart
ax1.pie(responses, 
       labels=languages,
       autopct='%1.1f%%',
       startangle=90,
       colors=sns.color_palette("Dark2", len(languages)))
ax1.set_title('Favorite Programming Languages (Pie Chart)', fontsize=16)
ax1.axis('equal')

# Bar chart
bars = ax2.bar(languages, responses, color=sns.color_palette("Dark2", len(languages)))
ax2.set_title('Favorite Programming Languages (Bar Chart)', fontsize=16)
ax2.set_xlabel('Language')
ax2.set_ylabel('Number of Responses')

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.5,
            f'{height}', ha='center', fontsize=12)

ax2.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

print("Comparison:")
print("- Pie chart shows proportion of each language relative to the whole (percentage)")
print("- Bar chart makes it easier to compare exact values between categories")
print("- Bar chart would scale better if there were more categories or similar-sized values")
```

## Summary

### Key Takeaways About Pie Charts:

1. **Effective Use Cases**:
   - Showing parts of a whole
   - Displaying data with clear differences between slices
   - When you have a small number of categories (ideally 5-7)
   - For high-level, non-technical presentations

2. **Avoid Using Pie Charts When**:
   - You have many categories
   - Values are similar in size
   - Precise comparisons are needed
   - Showing trends over time

3. **Best Practices**:
   - Use clear, contrasting colors
   - Label directly or use a legend
   - Consider exploding important slices
   - Show percentages for context
   - Use alternatives like bar charts when appropriate

4. **Alternatives**:
   - Bar charts for comparing values
   - Stacked bar charts for part-to-whole relationships with multiple groups
   - Donut charts for a more modern look
   - Treemaps for hierarchical data
