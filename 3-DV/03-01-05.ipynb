## 1. Import Required Libraries <a id="import"></a>

Let's start by importing the necessary libraries for data manipulation and visualization.

```python
# Import the necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns

# Set the style for seaborn
sns.set_style("whitegrid")

# For inline plotting in Jupyter notebooks
%matplotlib inline

# For better resolution plots
%config InlineBackend.figure_format = 'retina'

print("Libraries imported successfully!")
```

## 2. Basic Plot with Default Settings <a id="basic"></a>

Before we dive into customization, let's create some basic plots with default settings to establish a baseline.

```python
# Generate some sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.exp(-x/5)

# Create a simple figure with three subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# Line plot
ax1.plot(x, y1)
ax1.set_title("Default Line Plot")

# Scatter plot
ax2.scatter(x, y2)
ax2.set_title("Default Scatter Plot")

# Bar plot
categories = ['A', 'B', 'C', 'D', 'E']
values = [25, 40, 30, 55, 15]
ax3.bar(categories, values)
ax3.set_title("Default Bar Plot")

plt.tight_layout()
plt.show()
```

## 3. Adding and Customizing Titles <a id="titles"></a>

A well-formatted title can significantly enhance the readability and professionalism of your plot. Let's explore various title customization options.

```python
# Create a figure
plt.figure(figsize=(12, 8))

# Basic plot
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.plot(x, y3, label='sin(x)*exp(-x/5)')

# Basic title
plt.title('Basic Title')

# Display the plot
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Title customization options
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# 1. Title with custom font size and weight
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Custom Font Size and Weight', fontsize=16, fontweight='bold')

# 2. Title with custom color and font family
axs[0, 1].plot(x, y2)
axs[0, 1].set_title('Custom Color and Font', fontsize=14, 
                   color='navy', fontfamily='serif', fontstyle='italic')

# 3. Title with custom position and alignment
axs[1, 0].plot(x, y3)
axs[1, 0].set_title('Custom Position and Alignment', 
                   fontsize=14, y=1.05, loc='right', pad=10)

# 4. Main title and subtitle
axs[1, 1].plot(x, y1)
axs[1, 1].set_title('Main Title', fontsize=16, fontweight='bold')
axs[1, 1].text(5, 0, 'This is a subtitle with additional information', 
             ha='center', fontsize=12, style='italic', color='gray')

# Add a title for the entire figure
fig.suptitle('Different Title Customization Options', fontsize=20, y=0.98)

plt.tight_layout()
plt.subplots_adjust(top=0.90)  # Make room for suptitle
plt.show()
```

## 4. Working with Axis Labels <a id="axis-labels"></a>

Proper axis labels help viewers understand what data is being presented. Let's explore how to customize axis labels.

```python
# Create a figure with 2x2 subplots for demonstrating axis label customization
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. Basic axis labels
axs[0, 0].plot(x, y1)
axs[0, 0].set_xlabel('X-axis')
axs[0, 0].set_ylabel('Y-axis')
axs[0, 0].set_title('Basic Axis Labels')

# 2. Custom font, size and color for axis labels
axs[0, 1].plot(x, y2)
axs[0, 1].set_xlabel('Time (seconds)', fontsize=12, fontfamily='serif', color='darkblue')
axs[0, 1].set_ylabel('Amplitude', fontsize=12, fontfamily='serif', color='darkred')
axs[0, 1].set_title('Custom Font, Size and Color')

# 3. Custom rotation and alignment
axs[1, 0].plot(x, y3)
axs[1, 0].set_xlabel('Frequency (Hz)', fontsize=12, rotation=15)
axs[1, 0].set_ylabel('Power (dB)', fontsize=12, rotation=75, labelpad=15)
axs[1, 0].set_title('Custom Rotation and Padding')

# 4. Mathematical expressions using LaTeX
axs[1, 1].plot(x, np.sin(x) * np.exp(-x/10))
axs[1, 1].set_xlabel(r'$\theta$ (radians)', fontsize=14)
axs[1, 1].set_ylabel(r'$\sin(\theta) \cdot e^{-\theta/10}$', fontsize=14)
axs[1, 1].set_title('Mathematical Expressions in Labels')

plt.tight_layout()
plt.show()
```

## 5. Legend Creation and Customization <a id="legends"></a>

Legends help identify different data series in your plots. Let's explore various legend customization options.

```python
# Create a figure to demonstrate legend customization
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. Basic legend
axs[0, 0].plot(x, y1, label='sin(x)')
axs[0, 0].plot(x, y2, label='cos(x)')
axs[0, 0].plot(x, y3, label='sin(x)*exp(-x/5)')
axs[0, 0].legend()
axs[0, 0].set_title('Default Legend')

# 2. Custom legend location
axs[0, 1].plot(x, y1, label='sin(x)')
axs[0, 1].plot(x, y2, label='cos(x)')
axs[0, 1].plot(x, y3, label='sin(x)*exp(-x/5)')
axs[0, 1].legend(loc='lower center')  # Try: 'best', 'upper right', 'center left', etc.
axs[0, 1].set_title('Custom Legend Location')

# 3. Custom legend appearance
axs[1, 0].plot(x, y1, '-', label='sin(x)')
axs[1, 0].plot(x, y2, '--', label='cos(x)')
axs[1, 0].plot(x, y3, '-.', label='sin(x)*exp(-x/5)')
axs[1, 0].legend(
    frameon=True,
    fancybox=True,
    shadow=True,
    framealpha=0.7,
    borderpad=1,
    title='Functions',
    title_fontsize=12
)
axs[1, 0].set_title('Custom Legend Appearance')

# 4. Legend outside the plot
axs[1, 1].plot(x, y1, label='sin(x)')
axs[1, 1].plot(x, y2, label='cos(x)')
axs[1, 1].plot(x, y3, label='sin(x)*exp(-x/5)')
axs[1, 1].legend(
    bbox_to_anchor=(1.05, 1),  # Outside the right edge
    loc='upper left',
    borderaxespad=0,
    ncol=1  # Number of columns
)
axs[1, 1].set_title('Legend Outside Plot')

plt.tight_layout()
plt.show()
```

```python
# More advanced legend customization

# Generate some additional data
categories = ['A', 'B', 'C', 'D', 'E']
values1 = [25, 40, 30, 55, 15]
values2 = [15, 35, 40, 30, 25]
values3 = [35, 25, 20, 40, 35]

plt.figure(figsize=(12, 6))

# Create custom legend entries
from matplotlib.lines import Line2D

# Plot data
plt.bar(categories, values1, alpha=0.7, label='Group 1')
plt.bar(categories, values2, bottom=values1, alpha=0.7, label='Group 2')
plt.bar(categories, values3, bottom=np.array(values1)+np.array(values2), alpha=0.7, label='Group 3')

# Create a custom legend
legend_elements = [
    Line2D([0], [0], color='cornflowerblue', lw=4, label='Line Data'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='forestgreen', 
           markersize=10, label='Scatter Points'),
    Line2D([0], [0], marker='s', color='w', markerfacecolor='crimson', 
           markersize=10, label='Square Markers'),
]

# Add a second legend for the custom elements
first_legend = plt.legend()
plt.gca().add_artist(first_legend)  # Add the first legend (original)
plt.legend(handles=legend_elements, loc='lower right', title='Custom Legend')

plt.title('Stacked Bar Chart with Multiple Legends')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.tight_layout()
plt.show()
```

## 6. Text Annotations and Formatting <a id="annotations"></a>

Text annotations help highlight specific data points or features in your plots. Let's explore various annotation techniques.

```python
# Create a figure for annotations demo
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. Simple text annotations
x_data = np.array([1, 2, 3, 4, 5])
y_data = np.array([5, 7, 2, 8, 4])

axs[0, 0].scatter(x_data, y_data, s=100)
for i, txt in enumerate(y_data):
    axs[0, 0].annotate(f"{txt}", (x_data[i], y_data[i]), 
                      xytext=(5, 5), textcoords='offset points')
    
axs[0, 0].set_xlim(0, 6)
axs[0, 0].set_ylim(0, 10)
axs[0, 0].set_title('Simple Text Annotations')

# 2. Annotations with arrows
x_data = np.linspace(0, 10, 100)
y_data = np.sin(x_data)

axs[0, 1].plot(x_data, y_data)
axs[0, 1].annotate('Maximum', xy=(π/2, 1), xytext=(π/2, 0.5),
                 arrowprops=dict(facecolor='black', shrink=0.05, width=1.5),
                 horizontalalignment='center')

axs[0, 1].annotate('Minimum', xy=(3*π/2, -1), xytext=(3*π/2, -0.5),
                 arrowprops=dict(facecolor='red', shrink=0.05, width=1.5),
                 horizontalalignment='center')

axs[0, 1].set_title('Annotations with Arrows')
axs[0, 1].set_xlim(0, 10)
axs[0, 1].set_ylim(-1.5, 1.5)

# 3. Text box annotations
# Create a sample plot
x_points = np.linspace(0, 10, 50)
y_points = np.sin(x_points) + np.random.normal(0, 0.2, 50)

axs[1, 0].scatter(x_points, y_points, alpha=0.7)
axs[1, 0].set_title('Text Box Annotation')

# Add a textbox with statistics
stats_text = f"Statistics:\nMean: {y_points.mean():.2f}\nStd: {y_points.std():.2f}\nMax: {y_points.max():.2f}\nMin: {y_points.min():.2f}"

text_box = {
    'boxstyle': 'round',
    'facecolor': 'wheat',
    'alpha': 0.8
}

axs[1, 0].text(0.05, 0.95, stats_text, transform=axs[1, 0].transAxes,
             fontsize=10, verticalalignment='top', bbox=text_box)

# 4. Fancy annotations with custom styles
axs[1, 1].plot(x, np.sin(x/2)*np.exp(-x/10))
axs[1, 1].set_title('Fancy Annotations')

# Add fancy annotations
axs[1, 1].annotate('Peak', xy=(3, 0.3), xytext=(4, 0.5),
                  arrowprops=dict(facecolor='green', shrink=0.05,
                                  width=2, headwidth=10),
                  fontsize=12, color='darkgreen',
                  bbox=dict(boxstyle="round,pad=0.5", facecolor='lightgreen', alpha=0.7))

axs[1, 1].annotate('Decay', xy=(7, 0), xytext=(8, -0.3),
                  arrowprops=dict(arrowstyle="fancy",
                                 fc="0.6", ec="none",
                                 connectionstyle="arc3,rad=-0.3"),
                  fontsize=12, color='navy',
                  bbox=dict(boxstyle="square,pad=0.5", facecolor='lightblue', alpha=0.7))

plt.tight_layout()
plt.show()
```

## 7. Modifying Tick Labels and Positions <a id="ticks"></a>

Customizing tick marks and their labels can significantly improve the readability of your plots. Let's explore various tick customization options.

```python
# Create a figure for tick customization
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. Custom tick positions and labels
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Custom Tick Positions and Labels')

# Set custom tick positions and labels for x-axis
axs[0, 0].set_xticks([0, π/2, π, 3*π/2, 2*π, 5*π/2, 3*π])
axs[0, 0].set_xticklabels(['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$', r'$5\pi/2$', r'$3\pi$'])

# Set custom tick positions and labels for y-axis
axs[0, 0].set_yticks([-1, -0.5, 0, 0.5, 1])
axs[0, 0].set_yticklabels(['-1.0', '-0.5', '0.0', '0.5', '1.0'])

# 2. Tick formatting for different data types
import matplotlib.ticker as ticker
from matplotlib.dates import DateFormatter
import datetime as dt

# Create time series data
dates = pd.date_range(start='2023-01-01', periods=100, freq='D')
values = np.cumsum(np.random.randn(100))

axs[0, 1].plot(dates, values)
axs[0, 1].set_title('Date Formatting for Tick Labels')

# Format the date ticks
date_formatter = DateFormatter('%b %d')
axs[0, 1].xaxis.set_major_formatter(date_formatter)
axs[0, 1].xaxis.set_major_locator(ticker.MaxNLocator(6))  # Set max number of ticks

# Rotate the tick labels for better readability
plt.setp(axs[0, 1].get_xticklabels(), rotation=45, ha='right')

# 3. Logarithmic scales and custom tick formatting
# Generate data for log scale
x_log = np.logspace(0, 3, 100)
y_log = x_log**0.5

axs[1, 0].loglog(x_log, y_log)
axs[1, 0].set_title('Logarithmic Scale with Custom Formatting')

# Set log-scale formatter
axs[1, 0].xaxis.set_major_formatter(ticker.LogFormatter())
axs[1, 0].yaxis.set_major_formatter(ticker.LogFormatter())

# Add grid for better readability in log scale
axs[1, 0].grid(True, which="both", ls="-")

# 4. Minor ticks and scientific notation
x_sci = np.linspace(0, 5, 100)
y_sci = np.exp(x_sci) * 1000

axs[1, 1].plot(x_sci, y_sci)
axs[1, 1].set_title('Scientific Notation and Minor Ticks')

# Enable minor ticks
axs[1, 1].minorticks_on()

# Use scientific notation for y-axis
axs[1, 1].ticklabel_format(axis='y', style='sci', scilimits=(0, 0))

# Customize the scientific notation
formatter = ticker.ScalarFormatter(useMathText=True)
formatter.set_scientific(True)
formatter.set_powerlimits((0, 0))
axs[1, 1].yaxis.set_major_formatter(formatter)

plt.tight_layout()
plt.show()
```

## 8. Creating Custom Color Schemes <a id="colors"></a>

Color is a powerful tool for data visualization. Let's explore how to create custom color schemes for your plots.

```python
# Create a figure for color scheme demonstration
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# 1. Using predefined color maps
x_grid = np.linspace(0, 10, 20)
y_grid = np.linspace(0, 10, 20)
X, Y = np.meshgrid(x_grid, y_grid)
Z = np.sin(X) * np.cos(Y)

contour = axs[0, 0].contourf(X, Y, Z, 20, cmap='viridis')
fig.colorbar(contour, ax=axs[0, 0])
axs[0, 0].set_title('Predefined Colormap (viridis)')

# 2. Custom color palettes with Seaborn
categories = ['A', 'B', 'C', 'D', 'E']
values = [25, 40, 30, 55, 15]

# Use a custom color palette
custom_palette = sns.color_palette("Set2")
sns.barplot(x=categories, y=values, palette=custom_palette, ax=axs[0, 1])
axs[0, 1].set_title('Custom Seaborn Color Palette')

# 3. Creating a custom colormap
from matplotlib.colors import LinearSegmentedColormap

# Create a custom colormap
colors = ["darkblue", "royalblue", "lightblue", "lightgreen", "yellowgreen", "gold", "orange", "red"]
custom_cmap = LinearSegmentedColormap.from_list("custom_cmap", colors)

# Create a heatmap with the custom colormap
data = np.random.rand(10, 10)
im = axs[1, 0].imshow(data, cmap=custom_cmap)
fig.colorbar(im, ax=axs[1, 0])
axs[1, 0].set_title('Custom Colormap')

# 4. Consistent colors across multiple plot elements
# Create data with multiple series
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.exp(-x/5)

# Define a color scheme
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # These are default matplotlib colors

# Plot multiple series with consistent colors
axs[1, 1].plot(x, y1, color=colors[0], label='sin(x)')
axs[1, 1].plot(x, y2, color=colors[1], label='cos(x)')
axs[1, 1].plot(x, y3, color=colors[2], label='sin(x)*exp(-x/5)')

# Add markers with the same colors
axs[1, 1].scatter([π/2, π, 3*π/2], [1, -1, 0], color=colors[0], s=100)
axs[1, 1].scatter([0, π, 2*π], [1, -1, 1], color=colors[1], s=100)
axs[1, 1].scatter([0, 5], [0, 0], color=colors[2], s=100)

# Add a legend
axs[1, 1].legend()
axs[1, 1].set_title('Consistent Color Scheme Across Plot Elements')

plt.tight_layout()
plt.show()
```

## 9. Multiple Plots with Consistent Styling <a id="multiple"></a>

When creating multiple plots, it's important to maintain visual consistency. Let's explore how to create multiple plots with consistent styling.

```python
# Set a consistent style for all plots
plt.style.use('seaborn-v0_8-whitegrid')

# Create a figure with multiple subplots
fig = plt.figure(figsize=(14, 10))

# Define a consistent color palette
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Create a GridSpec layout
gs = fig.add_gridspec(3, 3)

# 1. Line plot
ax1 = fig.add_subplot(gs[0, :2])
x = np.linspace(0, 10, 100)
for i, a in enumerate([0.5, 1, 1.5]):
    ax1.plot(x, np.sin(a*x), color=colors[i], label=f'sin({a}x)')
ax1.set_title('Sine Functions', fontsize=14)
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('sin(ax)', fontsize=12)
ax1.legend()

# 2. Bar plot
ax2 = fig.add_subplot(gs[0, 2])
categories = ['A', 'B', 'C', 'D']
values = [15, 30, 45, 20]
ax2.bar(categories, values, color=colors[0])
ax2.set_title('Bar Chart', fontsize=14)
ax2.set_xlabel('Category', fontsize=12)
ax2.set_ylabel('Value', fontsize=12)

# 3. Scatter plot
ax3 = fig.add_subplot(gs[1, :2])
n_points = 50
x = np.random.rand(n_points)
y = 2*x + np.random.normal(0, 0.2, n_points)
ax3.scatter(x, y, color=colors[1], alpha=0.7)
ax3.set_title('Scatter Plot', fontsize=14)
ax3.set_xlabel('x', fontsize=12)
ax3.set_ylabel('y', fontsize=12)

# 4. Pie chart
ax4 = fig.add_subplot(gs[1, 2])
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
ax4.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors)
ax4.set_title('Pie Chart', fontsize=14)

# 5. Histogram
ax5 = fig.add_subplot(gs[2, :])
data = np.random.normal(0, 1, 1000)
ax5.hist(data, bins=30, color=colors[2], alpha=0.7)
ax5.set_title('Histogram', fontsize=14)
ax5.set_xlabel('Value', fontsize=12)
ax5.set_ylabel('Frequency', fontsize=12)

# Add a title for the entire figure
fig.suptitle('Multiple Plots with Consistent Styling', fontsize=20, y=0.98)

# Adjust layout
plt.tight_layout()
plt.subplots_adjust(top=0.9)  # Make room for suptitle

plt.show()
```

## 10. Saving Customized Plots <a id="saving"></a>

Once you've created a beautifully customized plot, you'll want to save it. Let's explore options for saving plots in different formats.

```python
# Create a simple plot to demonstrate saving options
plt.figure(figsize=(10, 6))

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), linewidth=2, label='sin(x)')
plt.plot(x, np.cos(x), linewidth=2, label='cos(x)')

plt.title('Trigonometric Functions', fontsize=16)
plt.xlabel('x', fontsize=14)
plt.ylabel('y', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)

# Display the plot
plt.tight_layout()
plt.show()

# Save the plot in different formats with different settings
# PNG format (raster) - good for web display
plt.savefig('trig_functions.png', dpi=100)
print("Saved as PNG with default DPI (100)")

# PNG with higher resolution for better quality
plt.savefig('trig_functions_hires.png', dpi=300)
print("Saved as PNG with high DPI (300)")

# SVG format (vector) - good for publications and further editing
plt.savefig('trig_functions.svg')
print("Saved as SVG (vector format)")

# PDF format (vector) - good for publications
plt.savefig('trig_functions.pdf')
print("Saved as PDF")

# Save with transparent background
plt.savefig('trig_functions_transparent.png', transparent=True)
print("Saved with transparent background")

# Save with custom dimensions (in inches)
plt.savefig('trig_functions_custom_size.png', figsize=(8, 4))
print("Saved with custom dimensions")

print("\nNote: In a real notebook, you would typically save to specific directories.")
print("The files above would be saved in your working directory.")
```
