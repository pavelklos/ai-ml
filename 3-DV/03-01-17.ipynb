# Introduction to Visualization Ethics

This notebook explores the ethical dimensions of data visualization, covering principles, common issues, and best practices for creating honest and effective visualizations.

```python
# Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for consistent visualization
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("colorblind")

# Configure plot settings
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12
```

## Understanding Visualization Ethics

### What is Visualization Ethics?

Visualization ethics refers to the moral principles and standards that guide the creation and presentation of data visualizations. It's about creating visualizations that:

- Tell the truth about the data
- Don't mislead or manipulate the audience
- Respect the audience's context and needs
- Present information responsibly and accurately

### Core Principles of Ethical Data Visualization

1. **Truth**: Represent data accurately and completely
2. **Transparency**: Be open about methods, sources, and limitations
3. **Fairness**: Present information without bias or manipulation
4. **Responsibility**: Consider the impact of your visualization choices
5. **Context**: Provide appropriate context for proper interpretation

Let's explore how the same data can be presented in different ways - some ethical, some not.

```python
# Example: Creating a dataset for demonstration
np.random.seed(42)  # For reproducibility
years = range(2013, 2023)
sales = [100, 105, 103, 107, 110, 112, 115, 113, 117, 120]

# Create a DataFrame
df = pd.DataFrame({'Year': years, 'Sales': sales})
print(df)
```

### Ethical vs. Unethical Visualization Examples

Let's visualize this sales data in both ethical and unethical ways.

```python
# Example 1: Ethical visualization - Shows complete context with axis starting at 0
plt.figure(figsize=(10, 5))
plt.plot(df['Year'], df['Sales'], marker='o', linewidth=2)
plt.title('Annual Sales (2013-2022)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Sales (in millions $)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.ylim(0, max(df['Sales']) * 1.1)  # Start from 0 to show true scale
plt.tight_layout()
plt.show()
```

```python
# Example 2: Unethical visualization - Truncated y-axis exaggerates changes
plt.figure(figsize=(10, 5))
plt.plot(df['Year'], df['Sales'], marker='o', linewidth=2, color='red')
plt.title('Annual Sales Showing DRAMATIC Growth! (2013-2022)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Sales (in millions $)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.ylim(95, max(df['Sales']) * 1.05)  # Truncated y-axis to exaggerate differences
plt.tight_layout()
plt.show()
```

### Commentary on the Ethical Difference

The two visualizations above show the same data, but convey very different impressions:

1. **Ethical Visualization**:
   - Shows the full context by starting the y-axis at 0
   - Allows viewer to accurately assess the magnitude of changes
   - Uses neutral language in the title
   - Presents a fair representation of the data

2. **Unethical Visualization**:
   - Truncates the y-axis to make small changes appear dramatic
   - Exaggerates the growth pattern visually
   - Uses emotionally charged language in the title
   - Creates a misleading impression about the data trend

This simple example demonstrates how visualization choices can dramatically alter the story told by the data.

## Common Ethical Issues in Data Visualization

Let's explore some common ethical issues that arise in data visualization:

1. Truncated axes
2. Manipulated scales
3. Cherry-picking data
4. Correlation-causation confusion
5. Misleading color choices
6. Omitting uncertainty

```python
# Demonstrating cherry-picking data - Creating dataset with seasonal fluctuations
months = np.tile(["Jan", "Feb", "Mar", "Apr", "May", "Jun", 
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"], 3)
years = np.repeat([2020, 2021, 2022], 12)

# Creating seasonal data with an overall growth trend
np.random.seed(42)
base = 100
seasonal_pattern = [0, 5, 10, 15, 25, 40, 60, 55, 35, 20, 10, 5]  # Seasonal variation
trend = np.linspace(0, 30, 36)  # Overall upward trend
noise = np.random.normal(0, 5, 36)  # Random variation

visitors = base + np.tile(seasonal_pattern, 3) + trend + noise

# Create dataframe
seasonal_df = pd.DataFrame({
    'Month': months,
    'Year': years,
    'Visitors': visitors.astype(int)
})

print(seasonal_df.head(15))
```

```python
# Unethical: Cherry-picking data by selecting only specific months
plt.figure(figsize=(10, 5))
cherry_picked = seasonal_df[(seasonal_df['Month'].isin(['Jan', 'Jul'])) & (seasonal_df['Year'] >= 2020)]

# Group by month and year
jan_data = cherry_picked[cherry_picked['Month'] == 'Jan']
jul_data = cherry_picked[cherry_picked['Month'] == 'Jul']

plt.scatter(jan_data['Year'], jan_data['Visitors'], color='blue', label='January', s=100)
plt.scatter(jul_data['Year'], jul_data['Visitors'], color='red', label='July', s=100)
plt.plot(jan_data['Year'], jan_data['Visitors'], 'b--')
plt.plot(jul_data['Year'], jul_data['Visitors'], 'r--')

plt.title('Visitor Growth Comparison (Jan vs. Jul)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Visitors', fontsize=12)
plt.xticks(sorted(seasonal_df['Year'].unique()))
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Commentary on Cherry-Picking

The visualization above is misleading because:

1. It only shows January and July data, ignoring all other months
2. It creates a false impression of extreme growth by comparing the lowest month (January) with one of the highest months (July)
3. It implies that growth is much more dramatic than the full data would suggest

Let's create an ethical version that shows the complete seasonal pattern:

```python
# Ethical: Showing complete seasonal patterns across years
plt.figure(figsize=(12, 6))

# Create a proper time series index
seasonal_df['Date'] = pd.to_datetime([f"{y}-{m}-01" for m, y in zip(seasonal_df['Month'], seasonal_df['Year'])])
seasonal_df = seasonal_df.sort_values('Date')

# Plot complete data
for year in sorted(seasonal_df['Year'].unique()):
    year_data = seasonal_df[seasonal_df['Year'] == year]
    plt.plot(year_data['Month'], year_data['Visitors'], marker='o', label=str(year))

plt.title('Monthly Visitors (2020-2022) - Complete Seasonal Pattern', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Visitors', fontsize=12)
plt.legend(title='Year')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Manipulated Scales Example

Another common ethical issue is manipulating the scale to make differences appear larger or smaller than they actually are.

```python
# Create data for different product sales
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales_2021 = [1020, 1040, 1015, 1025]
sales_2022 = [1060, 1065, 1055, 1045]

# Calculate the percentage change
pct_change = [(new - old) / old * 100 for old, new in zip(sales_2021, sales_2022)]

# Unethical: Manipulated scale that makes small differences look dramatic
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(products, pct_change, color='purple')

# Add data labels on top of bars
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points",
                ha='center', va='bottom', fontsize=12)

plt.title('Product Sales Growth (2021-2022)', fontsize=14)
plt.ylabel('Growth Percentage (%)', fontsize=12)
plt.ylim(0, 6)  # Truncated y-axis makes differences look bigger
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

```python
# Ethical: Using appropriate scale to show the true magnitude of differences
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(products, pct_change, color='green')

# Add data labels on top of bars
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha='center', va='bottom', fontsize=12)

plt.title('Product Sales Growth (2021-2022)', fontsize=14)
plt.ylabel('Growth Percentage (%)', fontsize=12)
plt.ylim(0, 100)  # More appropriate scale showing context (percentage is out of 100)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

### Commentary on Scale Manipulation

The first chart with a truncated scale (0-6%) makes small differences between products appear dramatic. The second chart, with a more appropriate scale (0-100%), provides proper context that these are actually modest differences - all products grew by around 3-4%. 

A middle-ground approach would be to use a scale from 0 to 10% or 0 to 20%, which still provides context while allowing viewers to see the differences. The key ethical principle is to choose scales that provide appropriate context without artificially exaggerating or minimizing differences.

## Misleading Visualizations: Examples and Analysis

Let's look at some specific techniques that can create misleading visualizations and how to correct them.

```python
# Creating a dataset showing minimal change
years = range(2013, 2023)
values = [98.2, 98.3, 98.5, 98.7, 98.6, 98.8, 99.0, 99.1, 99.2, 99.3]
metric_df = pd.DataFrame({'Year': years, 'Satisfaction': values})

# Unethical: Dual axis manipulation
fig, ax1 = plt.subplots(figsize=(10, 6))

# Primary axis - Company satisfaction
color = 'tab:blue'
ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('Customer Satisfaction (%)', color=color, fontsize=12)
ax1.plot(metric_df['Year'], metric_df['Satisfaction'], color=color, marker='o', linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)
ax1.set_ylim(98, 100)  # Truncated axis

# Secondary axis - Industry average (fictional data, deliberately misleading)
ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Industry Average (%)', color=color, fontsize=12)
ax2.plot(metric_df['Year'], [85, 85, 86, 86, 87, 87, 88, 88, 89, 90], color=color, marker='s', linewidth=2)
ax2.tick_params(axis='y', labelcolor=color)
ax2.set_ylim(0, 100)  # Full scale on this axis

plt.title('Our Customer Satisfaction vs. Industry Average (2013-2022)', fontsize=14)
plt.grid(False)
plt.tight_layout()
plt.show()
```

### Commentary on Dual Axis Manipulation

The visualization above uses a misleading technique:

1. It shows the company's customer satisfaction on a truncated scale (98-100%), making tiny changes look significant
2. It shows a fictional "industry average" on a full scale (0-100%)
3. The visual comparison between the two lines is invalid because they use different scales
4. This creates a false impression that the company vastly outperforms the industry

Let's create an ethical version with both datasets on the same scale:

```python
# Ethical: Using consistent scales for comparison
fig, ax = plt.subplots(figsize=(10, 6))

# Create fictional industry average data
industry_avg = [85, 85, 86, 86, 87, 87, 88, 88, 89, 90]

# Plot both on the same scale
ax.plot(metric_df['Year'], metric_df['Satisfaction'], color='blue', marker='o', 
         linewidth=2, label='Our Company')
ax.plot(metric_df['Year'], industry_avg, color='red', marker='s', 
         linewidth=2, label='Industry Average')

ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Customer Satisfaction (%)', fontsize=12)
ax.set_ylim(0, 100)  # Full scale for honest comparison
ax.grid(True, alpha=0.3)
ax.legend()

plt.title('Customer Satisfaction Comparison (2013-2022)', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Another ethical option: Use same scale but zoom in and acknowledge it
fig, ax = plt.subplots(1, 2, figsize=(15, 6))

# First plot: Full scale (0-100%)
ax[0].plot(metric_df['Year'], metric_df['Satisfaction'], color='blue', marker='o', 
         linewidth=2, label='Our Company')
ax[0].plot(metric_df['Year'], industry_avg, color='red', marker='s', 
         linewidth=2, label='Industry Average')

ax[0].set_xlabel('Year', fontsize=12)
ax[0].set_ylabel('Customer Satisfaction (%)', fontsize=12)
ax[0].set_ylim(0, 100)
ax[0].grid(True, alpha=0.3)
ax[0].legend()
ax[0].set_title('Full Scale (0-100%)', fontsize=14)

# Second plot: Zoomed in to show detail (with clear labeling)
ax[1].plot(metric_df['Year'], metric_df['Satisfaction'], color='blue', marker='o', 
         linewidth=2, label='Our Company')
ax[1].plot(metric_df['Year'], industry_avg, color='red', marker='s', 
         linewidth=2, label='Industry Average')

ax[1].set_xlabel('Year', fontsize=12)
ax[1].set_ylabel('Customer Satisfaction (%)', fontsize=12)
ax[1].set_ylim(80, 100)  # Zoomed in scale, but still includes both datasets
ax[1].grid(True, alpha=0.3)
ax[1].legend()
ax[1].set_title('Zoomed View (80-100%)', fontsize=14)

plt.suptitle('Customer Satisfaction Comparison (2013-2022)', fontsize=16)
plt.tight_layout()
plt.subplots_adjust(top=0.88)
plt.show()
```

## Ethical Frameworks for Data Visualization

When creating visualizations, we can apply several ethical frameworks to guide our decisions:

### 1. Transparency Framework
- Clearly state data sources
- Acknowledge limitations and uncertainties
- Explain methodological choices
- Provide access to underlying data when possible

### 2. Fairness Framework
- Present multiple perspectives
- Avoid cherry-picking or selective representation
- Consider diverse audiences and needs
- Provide appropriate context

### 3. Harm Minimization Framework
- Consider potential misinterpretations
- Evaluate possible negative impacts
- Avoid reinforcing harmful stereotypes
- Protect privacy and sensitive information

Let's see how to apply these frameworks in practice:

```python
# Creating data to demonstrate frameworks
np.random.seed(42)
experiment_data = pd.DataFrame({
    'Group': np.repeat(['Treatment', 'Control'], 50),
    'Result': np.concatenate([
        np.random.normal(65, 15, 50),  # Treatment group
        np.random.normal(60, 15, 50)   # Control group
    ])
})

# Add outcome classification
experiment_data['Outcome'] = ['Success' if x >= 70 else 'Failure' for x in experiment_data['Result']]

print(experiment_data.head(10))
print("\nSummary statistics by group:")
print(experiment_data.groupby('Group')['Result'].describe())
```

```python
# Applying ethical frameworks to visualization

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 1. Transparency Framework: Show full distributions with uncertainty
sns.boxplot(x='Group', y='Result', data=experiment_data, ax=axes[0])
axes[0].set_title('Transparent Representation\nShowing Full Distribution', fontsize=14)
axes[0].set_ylabel('Result Score', fontsize=12)
axes[0].grid(True, axis='y', alpha=0.3)

# 2. Fairness Framework: Show same data with multiple perspectives
treatment = experiment_data[experiment_data['Group'] == 'Treatment']
control = experiment_data[experiment_data['Group'] == 'Control']
axes[1].bar(['Control', 'Treatment'], 
          [control['Result'].mean(), treatment['Result'].mean()],
          yerr=[control['Result'].std() / np.sqrt(len(control)), 
                treatment['Result'].std() / np.sqrt(len(treatment))],
          capsize=10, color=['#3498db', '#e74c3c'])
axes[1].set_ylabel('Mean Score (with std error)', fontsize=12)
axes[1].set_title('Fair Representation\nShowing Uncertainty with Error Bars', fontsize=14)
axes[1].grid(True, axis='y', alpha=0.3)

# 3. Harm Minimization: Avoid misinterpretation by showing actual effect size context
# Calculate Cohen's d effect size
mean_diff = treatment['Result'].mean() - control['Result'].mean()
pooled_std = np.sqrt((treatment['Result'].std()**2 + control['Result'].std()**2) / 2)
effect_size = mean_diff / pooled_std

# Create plot with effect size context
axes[2].bar(['Mean Difference'], [mean_diff], color='purple')
axes[2].set_title(f'Contextual Representation\nEffect Size (Cohen\'s d): {effect_size:.2f}', fontsize=14)
axes[2].set_ylabel('Mean Difference (Treatment - Control)', fontsize=12)
axes[2].axhline(y=0, color='red', linestyle='-', alpha=0.3)
axes[2].grid(True, axis='y', alpha=0.3)

# Add annotation for effect size interpretation
effect_interp = "Small" if effect_size < 0.5 else "Medium" if effect_size < 0.8 else "Large"
axes[2].annotate(f"Effect Size Interpretation: {effect_interp}", 
                xy=(0, mean_diff), 
                xytext=(0, mean_diff/2 if mean_diff > 0 else mean_diff*1.5),
                ha='center', fontsize=12,
                bbox=dict(boxstyle='round,pad=0.5', fc='yellow', alpha=0.3))

plt.tight_layout()
plt.show()
```

### Commentary on Ethical Framework Application

The three visualizations above demonstrate different ethical frameworks applied to the same data:

1. **Transparency Framework**: Shows the full distribution of both groups using boxplots, clearly revealing overlap and outliers. This gives viewers a complete picture of the data.

2. **Fairness Framework**: Presents means with error bars, acknowledging uncertainty in the results. This approach ensures viewers understand that there's variability in the data.

3. **Harm Minimization Framework**: Contextualizes the difference by showing the effect size and its interpretation, helping to prevent overinterpretation of small differences.

Each approach has ethical merit, and combining these perspectives provides the most complete and honest representation of the data.

## Accessibility and Inclusivity

Ethical data visualization also means creating visualizations that are accessible to all audiences, including those with visual impairments or color blindness. This section explores how to make visualizations more accessible.

```python
# Create data for category comparison
categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
values = [23, 45, 56, 78, 32]

# Example 1: Poor accessibility - Red-green color scheme (problematic for color blindness)
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.bar(categories, values, color=['red', 'green', 'red', 'green', 'red'])
plt.title('Inaccessible: Red-Green Color Scheme', fontsize=14)
plt.ylabel('Values', fontsize=12)
plt.xticks(rotation=45)

# Example 2: Better accessibility - Colorblind-friendly palette with patterns
plt.subplot(1, 2, 2)
colorblind_palette = sns.color_palette("colorblind", len(categories))
bars = plt.bar(categories, values, color=colorblind_palette)

# Add patterns for additional differentiation
patterns = ['/', '\\', '.', 'o', '*']
for bar, pattern in zip(bars, patterns):
    bar.set_hatch(pattern)

plt.title('Accessible: Colorblind-Friendly with Patterns', fontsize=14)
plt.ylabel('Values', fontsize=12)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Text readability examples
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Poor text readability
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

axes[0].plot(x, y1, label='Sin(x)', color='lightblue')
axes[0].plot(x, y2, label='Cos(x)', color='pink')
axes[0].set_title('Poor Text Readability Example', fontsize=10, color='lightgray')
axes[0].set_xlabel('x values', fontsize=8)
axes[0].set_ylabel('y values', fontsize=8)
axes[0].legend(fontsize=8, loc='lower right', framealpha=0.3)
axes[0].grid(True, alpha=0.1)
axes[0].set_facecolor('lightgrey')  # Low contrast background

# Good text readability
axes[1].plot(x, y1, label='Sin(x)', linewidth=2.5, color='#0077BB')
axes[1].plot(x, y2, label='Cos(x)', linewidth=2.5, color='#EE7733')
axes[1].set_title('Good Text Readability Example', fontsize=14, fontweight='bold')
axes[1].set_xlabel('X Values', fontsize=12)
axes[1].set_ylabel('Y Values', fontsize=12)
axes[1].legend(fontsize=12, loc='lower right', framealpha=0.7)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Accessibility Best Practices

Based on the examples above, here are key accessibility best practices for ethical data visualization:

1. **Color choices**:
   - Use colorblind-friendly palettes
   - Don't rely solely on color to convey information
   - Ensure sufficient contrast between colors
   - Add patterns or textures for additional differentiation

2. **Text readability**:
   - Use sufficient font sizes (12pt minimum for important information)
   - Choose high-contrast color combinations
   - Use clear, standard fonts
   - Make titles and labels descriptive and legible

3. **Alternative representations**:
   - Consider multiple ways to represent the same information
   - Provide text alternatives or descriptions when possible
   - Use direct labeling instead of relying on legends when space permits

4. **Simplify complexity**:
   - Break complex visualizations into simpler components
   - Use clear hierarchy of information
   - Direct attention to key insights

By applying these accessibility principles, we make our visualizations usable by a wider audience, which is an important ethical consideration.

## Practical Guidelines for Ethical Visualization

Let's implement some practical guidelines for creating ethical visualizations:

1. Use appropriate scales
2. Label axes clearly and completely
3. Provide sufficient context
4. Use color responsibly
5. Acknowledge uncertainty
6. Maintain proportional representation
7. Be transparent about data sources and processing

```python
# Create sample time series data with uncertainty
np.random.seed(42)
dates = pd.date_range(start='2022-01-01', periods=12, freq='M')
baseline = np.array([10, 12, 15, 18, 22, 25, 23, 20, 17, 15, 13, 11])
trend = baseline + np.random.normal(0, 2, 12)  # Add some noise

# Calculate upper and lower uncertainty bounds
lower_bound = trend - 1.5
upper_bound = trend + 1.5

# Create ethical time series visualization
plt.figure(figsize=(12, 6))

# Plot the main trend line
plt.plot(dates, trend, 'b-', linewidth=2, label='Observed Values')

# Add uncertainty area
plt.fill_between(dates, lower_bound, upper_bound, color='b', alpha=0.2, 
                 label='Uncertainty Range')

# Add proper labels and context
plt.title('Monthly Measurements with Uncertainty (2022)', fontsize=14)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Measurement Value', fontsize=12)
plt.grid(True, alpha=0.3)

# Start y-axis at 0 for appropriate scale
plt.ylim(0, max(upper_bound) * 1.1)

# Add annotations for context
plt.annotate('Seasonal Peak', xy=(dates[5], trend[5]), xytext=(dates[5], trend[5]+3),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5),
            ha='center')

# Add text box with data source information
props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
plt.text(dates[0], max(upper_bound) * 1.05, 
         "Data Source: Monthly measurements\nCollection Method: Automated sensors\nMargin of Error: ±1.5 units", 
         fontsize=10, bbox=props)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Example of ethical proportional representation with pie charts
# Create data
categories = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [15, 30, 45, 10]

# Create two visualizations side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 7))

# Left: Unethical 3D pie chart with distorted proportions
ax1.pie(sizes, labels=categories, autopct='%1.1f%%', startangle=90, 
       shadow=True, explode=(0.1, 0, 0, 0))
ax1.set_title('Unethical: 3D Pie Chart with Distortion', fontsize=14)

# Right: Ethical horizontal bar chart with clear proportions
categories_with_values = [f"{cat} ({size}%)" for cat, size in zip(categories, sizes)]
ax2.barh(categories_with_values, sizes, color=plt.cm.Paired(np.arange(len(categories))/len(categories)))
ax2.set_title('Ethical: Horizontal Bar Chart with Clear Proportions', fontsize=14)
ax2.set_xlabel('Percentage (%)', fontsize=12)
ax2.grid(True, alpha=0.3)

# Add values at end of bars
for i, v in enumerate(sizes):
    ax2.text(v + 0.5, i, f"{v}%", va='center')

plt.tight_layout()
plt.show()
```

### Key Ethical Guidelines Demonstrated

The examples above showcase several important ethical guidelines:

1. **Acknowledge uncertainty**: The time series chart includes confidence intervals, showing the uncertainty in the measurements.

2. **Provide context**: The annotation and text box provide important context about data collection and interpretation.

3. **Use appropriate representations**: The comparison between 3D pie charts and horizontal bar charts demonstrates how choice of chart type affects perception of proportions.

4. **Clear labeling**: All axes are clearly labeled, and data values are explicitly shown where appropriate.

5. **Start axes at zero**: The time series y-axis starts at zero, providing proper context for the magnitude of values.

6. **Transparency about sources**: The data source and methodology are explicitly stated in the visualization.

These guidelines help ensure that our visualizations present data in an honest and transparent manner, enabling viewers to accurately interpret the information.

## Case Study Analysis

Let's analyze a case study of a real-world scenario where visualization ethics are important.

### Case Study: COVID-19 Data Visualization

COVID-19 data visualization has been particularly challenging from an ethical standpoint. We'll create a simplified example to demonstrate ethical vs. unethical approaches.

```python
# Simulate COVID-19 case data
np.random.seed(42)
dates = pd.date_range('2020-03-01', '2020-12-31', freq='D')
base_trend = np.linspace(10, 1000, len(dates))  # Overall increasing trend
seasonal = 300 * np.sin(np.linspace(0, 4*np.pi, len(dates)))  # Oscillations
noise = np.random.normal(0, 50, len(dates))  # Random noise
cases = np.maximum(0, base_trend + seasonal + noise).astype(int)  # Ensure no negative values

# Create a dataframe
covid_df = pd.DataFrame({
    'Date': dates,
    'Cases': cases
})

# Calculate 7-day moving average
covid_df['7-Day Average'] = covid_df['Cases'].rolling(window=7).mean()

# Create testing data (simplistic model)
tests_base = np.linspace(100, 10000, len(dates))
testing_noise = np.random.normal(0, 500, len(dates))
tests = np.maximum(0, tests_base + testing_noise).astype(int)
covid_df['Tests'] = tests
covid_df['Positivity Rate'] = (covid_df['Cases'] / covid_df['Tests'] * 100).round(2)

print(covid_df.head())
```

```python
# Unethical visualization: Cherry-picking the timeframe and not showing testing context
plt.figure(figsize=(12, 6))
# Select only a specific timeframe that shows the steepest increase
start_date = '2020-07-01'
end_date = '2020-10-15'
selected_data = covid_df[(covid_df['Date'] >= start_date) & (covid_df['Date'] <= end_date)]

plt.plot(selected_data['Date'], selected_data['Cases'], color='red', linewidth=2)
plt.fill_between(selected_data['Date'], 0, selected_data['Cases'], color='red', alpha=0.3)
plt.title('ALARMING RISE IN COVID-19 CASES!', fontsize=16, fontweight='bold')
plt.ylabel('Daily Cases', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Ethical visualization: Showing complete context with multiple metrics
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(14, 12), sharex=True)

# Plot 1: Daily cases and 7-day average for the full time period
ax1.bar(covid_df['Date'], covid_df['Cases'], alpha=0.3, color='blue', label='Daily Cases')
ax1.plot(covid_df['Date'], covid_df['7-Day Average'], color='blue', linewidth=2.5, 
        label='7-Day Average')
ax1.set_title('COVID-19 Daily Cases and 7-Day Moving Average', fontsize=14)
ax1.set_ylabel('Number of Cases', fontsize=12)
ax1.legend()
ax1.grid(True, alpha=0.3)

# Highlight the period shown in the unethical visualization
ax1.axvspan(pd.Timestamp(start_date), pd.Timestamp(end_date), alpha=0.2, color='red')
ax1.text(pd.Timestamp('2020-08-15'), max(covid_df['Cases'])*0.8, 'Period shown in\nselective visualization', 
        ha='center', fontsize=12, color='red')

# Plot 2: Testing volumes
ax2.bar(covid_df['Date'], covid_df['Tests'], alpha=0.3, color='green')
ax2.set_title('COVID-19 Testing Volume', fontsize=14)
ax2.set_ylabel('Number of Tests', fontsize=12)
ax2.grid(True, alpha=0.3)

# Plot 3: Positivity rate
ax3.plot(covid_df['Date'], covid_df['Positivity Rate'], color='purple', linewidth=2.5)
ax3.set_title('COVID-19 Test Positivity Rate', fontsize=14)
ax3.set_ylabel('Positivity Rate (%)', fontsize=12)
ax3.set_xlabel('Date', fontsize=12)
ax3.grid(True, alpha=0.3)

# Add annotation about the data context
props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
ax1.text(pd.Timestamp('2020-04-01'), max(covid_df['Cases'])*0.9, 
         "Data Source: Simulated COVID-19 Data\nNote: Case numbers should be interpreted\nin the context of testing volume and positivity rate.", 
         fontsize=12, bbox=props)

plt.tight_layout()
plt.show()
```

### Case Study Analysis

The two visualizations of the COVID-19 data demonstrate dramatically different approaches:

#### Unethical Visualization:
1. **Cherry-picking timeframe**: Shows only the period with the steepest increase
2. **Missing context**: Doesn't show testing data or positivity rates
3. **Alarmist language**: Uses emotionally charged language in the title
4. **No smoothing**: Uses daily fluctuations rather than showing the trend
5. **No source information**: Provides no information about data sources or limitations

#### Ethical Visualization:
1. **Complete timeframe**: Shows the entire data period
2. **Multiple metrics**: Includes cases, testing volume, and positivity rate for context
3. **Trend smoothing**: Uses 7-day moving average to show meaningful trends
4. **Context notes**: Includes notes about data interpretation
5. **Highlights selective framing**: Shows which portion was cherry-picked in the unethical visualization
6. **Neutral language**: Uses descriptive rather than emotional language

This case study demonstrates how visualization choices can dramatically alter the impression given by the same underlying data. Ethical visualization requires showing complete context and avoiding selective framing that might mislead viewers.

## Summary: Principles of Ethical Data Visualization

Throughout this notebook, we've explored various aspects of visualization ethics. Here are the key principles to remember:

1. **Truth and Accuracy**
   - Represent data accurately and completely
   - Avoid distortions and manipulations
   - Use appropriate scales and axes

2. **Context and Completeness**
   - Provide necessary context for proper interpretation
   - Show relevant related metrics
   - Acknowledge limitations and uncertainty

3. **Fairness and Balance**
   - Present multiple perspectives when appropriate
   - Avoid cherry-picking data or timeframes
   - Use neutral, descriptive language

4. **Transparency**
   - Clearly state data sources and methodologies
   - Explain unusual visualization choices
   - Make underlying data available when possible

5. **Accessibility**
   - Design for diverse audiences including those with visual impairments
   - Use colorblind-friendly color schemes
   - Ensure readability with appropriate fonts and contrast

6. **Responsibility**
   - Consider the potential impact of your visualization
   - Be aware of potential misinterpretations
   - Prioritize clarity over aesthetics when they conflict

By following these principles, we can create visualizations that are not only effective but also ethical and trustworthy.
