# Creating Histograms and Distribution Plots

Histograms and distribution plots are essential tools in data visualization that help us understand the underlying distribution of our data. This notebook explores how to create and customize these plots using popular Python libraries.

## Import Required Libraries

Let's start by importing the essential libraries we'll need for creating histograms and distribution plots.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for our visualizations
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("notebook")

# For displaying plots inline in the notebook
%matplotlib inline
```

## Understanding Histogram Basics

A histogram is a graphical representation of the distribution of data that groups data into bins (intervals) and shows the frequency of data points within each bin.

Key concepts for histograms:
- **Bins**: The intervals into which the data range is divided
- **Frequency**: The count of observations falling into each bin
- **Density**: The normalized frequency (area under the histogram equals 1)
- **Shape**: Describes the overall pattern of the distribution (symmetric, skewed, bimodal, etc.)

Let's generate some sample data to work with throughout this notebook:

```python
# Generate sample data sets with different distributions
np.random.seed(42)  # For reproducibility

# Normal distribution
normal_data = np.random.normal(loc=0, scale=1, size=1000)

# Skewed distribution (log-normal)
skewed_data = np.random.lognormal(mean=0, sigma=0.5, size=1000)

# Bimodal distribution (mixture of two normal distributions)
bimodal_data = np.concatenate([
    np.random.normal(loc=-2, scale=0.5, size=500),
    np.random.normal(loc=2, scale=0.5, size=500)
])

# Uniform distribution
uniform_data = np.random.uniform(low=-3, high=3, size=1000)

# Create a DataFrame to store our data
df = pd.DataFrame({
    'Normal': normal_data,
    'Skewed': skewed_data,
    'Bimodal': bimodal_data,
    'Uniform': uniform_data
})

# Display basic statistics of our data
df.describe()
```

## Creating Histograms with Matplotlib

Matplotlib provides the `plt.hist()` function for creating histograms. Let's explore how to create basic histograms and adjust key parameters like the number of bins.

```python
# Create a basic histogram with matplotlib
plt.figure(figsize=(10, 6))
plt.hist(normal_data, bins=30)
plt.title('Basic Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Compare the effect of different bin sizes
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Effect of Bin Size on Histogram Appearance', fontsize=16)

bin_sizes = [5, 10, 30, 100]
plot_positions = [(0, 0), (0, 1), (1, 0), (1, 1)]

for bins, pos in zip(bin_sizes, plot_positions):
    i, j = pos
    axes[i, j].hist(normal_data, bins=bins, color='skyblue', edgecolor='black')
    axes[i, j].set_title(f'{bins} Bins')
    axes[i, j].set_xlabel('Value')
    axes[i, j].set_ylabel('Frequency')
    axes[i, j].grid(True, alpha=0.3)

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

```python
# Compare different distributions
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Comparing Different Distributions', fontsize=16)

distributions = ['Normal', 'Skewed', 'Bimodal', 'Uniform']
plot_positions = [(0, 0), (0, 1), (1, 0), (1, 1)]
colors = ['skyblue', 'lightgreen', 'salmon', 'plum']

for dist, pos, color in zip(distributions, plot_positions, colors):
    i, j = pos
    axes[i, j].hist(df[dist], bins=30, color=color, edgecolor='black')
    axes[i, j].set_title(f'{dist} Distribution')
    axes[i, j].set_xlabel('Value')
    axes[i, j].set_ylabel('Frequency')
    axes[i, j].grid(True, alpha=0.3)

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

## Customizing Histogram Appearance

Customizing the appearance of histograms can make them more informative and visually appealing. Let's explore various customization options in Matplotlib.

```python
# Create a more customized histogram
plt.figure(figsize=(12, 6))

# Main histogram with customizations
plt.hist(normal_data, bins=30, 
         color='skyblue', 
         edgecolor='black', 
         alpha=0.7,
         density=True,  # Normalize to create a density plot
         label='Histogram')

# Add a vertical line for the mean
plt.axvline(x=np.mean(normal_data), color='red', linestyle='--', linewidth=2, 
           label=f'Mean: {np.mean(normal_data):.2f}')

# Add a vertical line for the median
plt.axvline(x=np.median(normal_data), color='green', linestyle='-.', linewidth=2,
           label=f'Median: {np.median(normal_data):.2f}')

# Add +/- 1 standard deviation markers
std = np.std(normal_data)
mean = np.mean(normal_data)
plt.axvline(x=mean + std, color='purple', linestyle=':', linewidth=2,
           label=f'+1 Std Dev: {mean + std:.2f}')
plt.axvline(x=mean - std, color='purple', linestyle=':', linewidth=2,
           label=f'-1 Std Dev: {mean - std:.2f}')

# Add titles and labels
plt.title('Customized Histogram with Statistical Annotations', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)

# Add grid and legend
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```

## Distribution Plots with Seaborn

Seaborn provides enhanced visualization capabilities for distribution plots with functions like `histplot`, `kdeplot`, and others. Let's explore these functions and their features.

```python
# Create a basic Seaborn histplot
plt.figure(figsize=(12, 6))
sns.histplot(data=normal_data, kde=True, bins=30, color='skyblue')
plt.title('Seaborn Histplot with KDE', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Count', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Exploring histplot options
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Exploring Seaborn Histplot Options', fontsize=16)

# Standard histogram
sns.histplot(data=normal_data, bins=30, ax=axes[0, 0])
axes[0, 0].set_title('Standard Histogram')

# Histogram with KDE overlay
sns.histplot(data=normal_data, bins=30, kde=True, ax=axes[0, 1])
axes[0, 1].set_title('Histogram with KDE')

# Histogram with different stat
sns.histplot(data=normal_data, bins=30, stat='density', ax=axes[1, 0])
axes[1, 0].set_title('Density Histogram')

# Histogram with cumulative distribution
sns.histplot(data=normal_data, bins=30, cumulative=True, ax=axes[1, 1])
axes[1, 1].set_title('Cumulative Histogram')

for ax in axes.flat:
    ax.set_xlabel('Value')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

## Kernel Density Estimation (KDE) Plots

KDE plots provide a smooth estimation of the probability density function of a random variable. They're like a smoothed version of the histogram and can reveal the shape of the distribution more clearly.

```python
# Create KDE plots for different distributions
plt.figure(figsize=(12, 6))

for col, color in zip(['Normal', 'Skewed', 'Bimodal', 'Uniform'], 
                      ['blue', 'green', 'red', 'purple']):
    sns.kdeplot(data=df[col], label=col, color=color, linewidth=2)

plt.title('KDE Plots of Different Distributions', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Exploring KDE bandwidth parameter
plt.figure(figsize=(12, 6))

bandwidths = [0.05, 0.2, 0.5, 1.0, 2.0]
for bw, color in zip(bandwidths, ['blue', 'green', 'red', 'purple', 'orange']):
    sns.kdeplot(data=normal_data, bw_adjust=bw, label=f'Bandwidth = {bw}', color=color)

plt.title('Effect of Bandwidth on KDE Plots', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## Combining Histograms and KDE

Combining histograms with KDE plots provides both the actual data distribution (histogram) and the estimated probability density function (KDE). This gives a more complete picture of the data.

```python
# Create a figure with multiple subplots showing different combinations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Combining Histograms and KDE for Different Distributions', fontsize=16)

# Normal distribution
sns.histplot(df['Normal'], kde=True, bins=30, color='skyblue', ax=axes[0, 0])
axes[0, 0].set_title('Normal Distribution')

# Skewed distribution
sns.histplot(df['Skewed'], kde=True, bins=30, color='lightgreen', ax=axes[0, 1])
axes[0, 1].set_title('Skewed Distribution')

# Bimodal distribution
sns.histplot(df['Bimodal'], kde=True, bins=30, color='salmon', ax=axes[1, 0])
axes[1, 0].set_title('Bimodal Distribution')

# Uniform distribution
sns.histplot(df['Uniform'], kde=True, bins=30, color='plum', ax=axes[1, 1])
axes[1, 1].set_title('Uniform Distribution')

for ax in axes.flat:
    ax.set_xlabel('Value')
    ax.set_ylabel('Count')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

```python
# Customize a combined histogram and KDE plot
plt.figure(figsize=(12, 6))

# Create histogram with semi-transparency
plt.hist(normal_data, bins=30, alpha=0.5, color='skyblue', edgecolor='black', density=True, label='Histogram')

# Add KDE plot on top
kde_x = np.linspace(min(normal_data), max(normal_data), 1000)
kde = sns.kdeplot(normal_data, color='darkblue', linewidth=2, label='KDE')

# Add annotations
plt.axvline(x=np.mean(normal_data), color='red', linestyle='--', 
           linewidth=2, label=f'Mean: {np.mean(normal_data):.2f}')
plt.axvline(x=np.median(normal_data), color='green', linestyle='-.', 
           linewidth=2, label=f'Median: {np.median(normal_data):.2f}')

plt.title('Custom Histogram with KDE Overlay and Statistical Annotations', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## Comparing Multiple Distributions

When working with multiple datasets or groups, comparing their distributions can provide valuable insights. Let's explore methods for comparing distributions visually.

```python
# Create a figure for comparing distributions
plt.figure(figsize=(12, 6))

# Plot KDEs for each distribution
for col, color in zip(['Normal', 'Skewed', 'Bimodal', 'Uniform'],
                     ['blue', 'green', 'red', 'purple']):
    sns.kdeplot(data=df[col], label=col, color=color, linewidth=2)

plt.title('Comparing Multiple Distributions with KDE', fontsize=16)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Using violin plots to compare distributions
plt.figure(figsize=(10, 6))
sns.violinplot(data=df)
plt.title('Comparing Distributions with Violin Plots', fontsize=16)
plt.xlabel('Distribution Type', fontsize=14)
plt.ylabel('Value', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()
```

## Visualization Best Practices

Creating effective histogram and distribution plots involves more than just calling the right functions. Let's explore some best practices and techniques for creating more informative visualizations.

```python
# Create a well-formatted, publication-ready histogram
plt.figure(figsize=(12, 8))

# Create the main histogram
n, bins, patches = plt.hist(normal_data, bins=30, density=True, 
                          alpha=0.7, color='skyblue', edgecolor='black',
                          label='Data Distribution')

# Add a smoothed KDE curve
sns.kdeplot(normal_data, color='navy', linewidth=2, label='Density Estimate')

# Add statistical annotations
mean = np.mean(normal_data)
median = np.median(normal_data)
std = np.std(normal_data)

# Add vertical lines for statistics
plt.axvline(x=mean, color='red', linestyle='--', linewidth=2, 
           label=f'Mean: {mean:.2f}')
plt.axvline(x=median, color='green', linestyle=':', linewidth=2,
           label=f'Median: {median:.2f}')

# Add annotation text for standard deviation
plt.text(0.95, 0.95, f'Standard Deviation: {std:.2f}',
        transform=plt.gca().transAxes, fontsize=12,
        verticalalignment='top', horizontalalignment='right',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

# Add title and labels with appropriate font sizes
plt.title('Distribution of Normal Data', fontsize=18, pad=20)
plt.xlabel('Value', fontsize=14)
plt.ylabel('Density', fontsize=14)

# Customize the tick parameters
plt.tick_params(axis='both', which='major', labelsize=12)

# Add grid with appropriate styling
plt.grid(True, alpha=0.3, linestyle='--')

# Add legend with good positioning
plt.legend(fontsize=12, loc='upper left')

# Adjust layout
plt.tight_layout()

plt.show()
```

```python
# Function to calculate the optimal number of bins using Freedman-Diaconis rule
def freedman_diaconis_bins(data):
    """Calculate the optimal bin width using Freedman-Diaconis rule"""
    data = np.asarray(data)
    IQR = np.subtract(*np.percentile(data, [75, 25]))
    n = len(data)
    bin_width = 2 * IQR * n**(-1/3)
    
    # Calculate the number of bins
    data_range = np.max(data) - np.min(data)
    if bin_width > 0:
        bins = np.ceil(data_range / bin_width)
    else:
        bins = 30  # Default if IQR is too small
    
    return int(bins)

# Compare different bin selection methods
plt.figure(figsize=(15, 10))

# Raw data as rug plot
plt.subplot(3, 2, 1)
sns.rugplot(normal_data, color='navy')
plt.title('Raw Data (Rug Plot)')
plt.grid(True, alpha=0.3)

# Fixed number of bins
plt.subplot(3, 2, 2)
plt.hist(normal_data, bins=10, color='skyblue', edgecolor='black')
plt.title('Fixed: 10 Bins')
plt.grid(True, alpha=0.3)

# Freedman-Diaconis rule
fd_bins = freedman_diaconis_bins(normal_data)
plt.subplot(3, 2, 3)
plt.hist(normal_data, bins=fd_bins, color='lightgreen', edgecolor='black')
plt.title(f'Freedman-Diaconis: {fd_bins} Bins')
plt.grid(True, alpha=0.3)

# Sturges' formula
sturges_bins = int(np.ceil(np.log2(len(normal_data))) + 1)
plt.subplot(3, 2, 4)
plt.hist(normal_data, bins=sturges_bins, color='salmon', edgecolor='black')
plt.title(f'Sturges: {sturges_bins} Bins')
plt.grid(True, alpha=0.3)

# Square-root rule
sqrt_bins = int(np.sqrt(len(normal_data)))
plt.subplot(3, 2, 5)
plt.hist(normal_data, bins=sqrt_bins, color='plum', edgecolor='black')
plt.title(f'Square Root: {sqrt_bins} Bins')
plt.grid(True, alpha=0.3)

# Scott's normal reference rule (using numpy's auto bin selection)
plt.subplot(3, 2, 6)
plt.hist(normal_data, bins='auto', color='lightyellow', edgecolor='black')
plt.title('NumPy Auto Bins')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.suptitle('Comparison of Bin Selection Methods', fontsize=16, y=1.02)
plt.show()
```
