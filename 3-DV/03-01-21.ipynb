## Basic Plot Types
Basic plots are the foundation of data visualization. These include line plots, bar charts, scatter plots, and pie charts. These plots are simple but effective in communicating different types of information.

### Line Plots
**When to use:** Line plots are great for displaying trends over time or continuous data where the order matters, such as time series data.

```python
# Matplotlib line plot
plt.figure(figsize=(12, 6))
plt.plot(x, y, 'b-', label='sin(x)')
plt.plot(x, np.cos(x), 'r-', label='cos(x)')
plt.title('Line Plot of Trigonometric Functions')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()

# Seaborn line plot with real data
plt.figure(figsize=(12, 6))
sns.lineplot(x='year', y='passengers', data=sns.load_dataset('flights'))
plt.title('Number of Passengers Over Time')
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.show()

# Plotly interactive line plot
fig = px.line(flights, x='year', y='passengers', title='Monthly Passenger Numbers',
              color='month', line_group='month', hover_name='month')
fig.update_layout(xaxis_title='Year', yaxis_title='Passengers')
fig.show()
```

### Bar Charts
**When to use:** Bar charts are ideal for comparing quantities across different categories.

```python
# Matplotlib basic bar chart
plt.figure(figsize=(12, 6))
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = [15, 34, 23, 47]
plt.bar(categories, values, color='skyblue')
plt.title('Simple Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()

# Seaborn bar chart with real data
plt.figure(figsize=(12, 6))
sns.barplot(x='class', y='fare', data=titanic)
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare ($)')
plt.show()

# Plotly interactive bar chart with grouped bars
survival_class = titanic.groupby(['class', 'survived']).size().reset_index(name='count')
fig = px.bar(survival_class, x='class', y='count', color='survived',
             title='Survival Count by Passenger Class',
             labels={'survived': 'Survived', 'count': 'Number of Passengers', 'class': 'Passenger Class'},
             barmode='group')
fig.show()
```

### Scatter Plots
**When to use:** Scatter plots are excellent for showing the relationship between two continuous variables and identifying patterns, clusters, or outliers.

```python
# Matplotlib scatter plot
plt.figure(figsize=(10, 8))
plt.scatter(iris['sepal_length'], iris['sepal_width'], alpha=0.7)
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.grid(True)
plt.show()

# Seaborn scatter plot with hue for categories
plt.figure(figsize=(10, 8))
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', data=iris, s=100)
plt.title('Sepal Length vs Sepal Width by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()

# Plotly interactive scatter plot
fig = px.scatter(iris, x='sepal_length', y='sepal_width', color='species',
                 size='petal_width', hover_data=['petal_length'],
                 title='Sepal Length vs Sepal Width')
fig.update_layout(xaxis_title='Sepal Length (cm)',
                  yaxis_title='Sepal Width (cm)')
fig.show()
```

### Pie Charts
**When to use:** Pie charts work well for showing proportions and percentages of a whole, particularly when you have a small number of categories.

```python
# Matplotlib pie chart
plt.figure(figsize=(10, 8))
class_counts = titanic['class'].value_counts()
plt.pie(class_counts, labels=class_counts.index, autopct='%1.1f%%',
        startangle=90, shadow=True, explode=(0.1, 0, 0))
plt.title('Distribution of Passenger Classes')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.show()

# Plotly interactive pie chart
sex_survived = titanic.groupby('sex')['survived'].mean().reset_index()
sex_survived['survived'] = sex_survived['survived'] * 100  # Convert to percentage
fig = px.pie(sex_survived, values='survived', names='sex',
             title='Survival Rate by Gender',
             hover_data=['survived'], labels={'survived':'Survival Rate (%)'})
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```

## Statistical Plot Types
Statistical plots help visualize distributions, summary statistics, and uncertainty in data.

### Box Plots
**When to use:** Box plots are ideal for showing the distribution of data based on a five-number summary (minimum, first quartile, median, third quartile, and maximum). They're great for spotting outliers and comparing distributions across groups.

```python
# Seaborn box plot
plt.figure(figsize=(12, 6))
sns.boxplot(x='class', y='age', data=titanic)
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()

# Plotly interactive box plot
fig = px.box(titanic, x='class', y='age', color='sex',
             title='Age Distribution by Passenger Class and Sex')
fig.update_layout(xaxis_title='Passenger Class',
                  yaxis_title='Age')
fig.show()
```

### Violin Plots
**When to use:** Violin plots are similar to box plots but also show the probability density of the data at different values. They're useful when you want to see the distribution shape along with summary statistics.

```python
# Seaborn violin plot
plt.figure(figsize=(12, 6))
sns.violinplot(x='class', y='fare', hue='survived', split=True, data=titanic)
plt.title('Fare Distribution by Class and Survival')
plt.xlabel('Passenger Class')
plt.ylabel('Fare ($)')
plt.show()

# Plotly interactive violin plot
fig = px.violin(titanic, x='class', y='fare', color='survived', box=True,
                title='Fare Distribution by Class and Survival')
fig.update_layout(xaxis_title='Passenger Class',
                  yaxis_title='Fare ($)')
fig.show()
```

### Error Bars
**When to use:** Error bars help visualize the uncertainty or variability in data, often showing confidence intervals or standard error.

```python
# Matplotlib error bars
plt.figure(figsize=(12, 6))

# Calculate mean and standard error for age by class
class_names = []
age_means = []
age_errors = []

for class_name, group in titanic.groupby('class'):
    class_names.append(class_name)
    age_means.append(group['age'].mean())
    age_errors.append(group['age'].std() / np.sqrt(len(group)))

plt.bar(class_names, age_means, yerr=age_errors, capsize=10, color='skyblue', alpha=0.7)
plt.title('Average Age by Passenger Class (with Standard Error)')
plt.xlabel('Passenger Class')
plt.ylabel('Average Age')
plt.show()

# Seaborn error bars using pointplot
plt.figure(figsize=(12, 6))
sns.pointplot(x='class', y='age', data=titanic, ci=95, join=False, capsize=0.2)
plt.title('Average Age by Passenger Class (with 95% CI)')
plt.xlabel('Passenger Class')
plt.ylabel('Average Age')
plt.show()
```

## Categorical Plot Types
These plots are designed specifically for categorical data and showing relationships between categorical variables.

### Bar Plots
**When to use:** Bar plots for categorical data show the relationship between a numerical and a categorical variable by displaying a summary statistic (like count, mean, etc.) for each category.

```python
# Seaborn count plot (show counts for each category)
plt.figure(figsize=(10, 6))
sns.countplot(x='class', hue='survived', data=titanic)
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

# Seaborn categorical bar plot (showing means)
plt.figure(figsize=(10, 6))
sns.barplot(x='day', y='total_bill', hue='sex', data=tips)
plt.title('Average Bill by Day and Gender')
plt.xlabel('Day of the Week')
plt.ylabel('Average Total Bill ($)')
plt.show()
```

### Heatmaps
**When to use:** Heatmaps are excellent for displaying the relationship between two categorical variables, showing frequency counts or another metric using color intensity.

```python
# Create a cross-tabulation of two categorical variables
class_sex_counts = pd.crosstab(titanic['class'], titanic['sex'])

# Seaborn heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(class_sex_counts, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Number of Passengers by Class and Gender')
plt.show()

# Correlation heatmap
plt.figure(figsize=(12, 10))
correlation_matrix = titanic.select_dtypes(include=[np.number]).corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numerical Variables in Titanic Dataset')
plt.tight_layout()
plt.show()

# Plotly interactive heatmap
fig = px.imshow(correlation_matrix, text_auto=True, 
                aspect="auto", color_continuous_scale='RdBu_r')
fig.update_layout(title='Correlation Matrix of Numerical Variables')
fig.show()
```

## Distribution Plot Types
Distribution plots help visualize the distribution of data, which is essential for understanding data patterns and checking assumptions.

### Histograms
**When to use:** Histograms show the distribution of a single numerical variable by dividing it into bins and showing the frequency of observations in each bin.

```python
# Matplotlib histogram
plt.figure(figsize=(12, 6))
plt.hist(titanic['age'].dropna(), bins=30, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()

# Seaborn histogram with KDE
plt.figure(figsize=(12, 6))
sns.histplot(data=titanic, x='age', hue='sex', kde=True, bins=30, alpha=0.6)
plt.title('Age Distribution by Gender')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()

# Plotly interactive histogram
fig = px.histogram(titanic, x='age', color='class', marginal='box',
                  nbins=30, opacity=0.7,
                  title='Age Distribution by Passenger Class')
fig.update_layout(xaxis_title='Age',
                  yaxis_title='Count',
                  barmode='overlay')
fig.show()
```

### Kernel Density Plots (KDE)
**When to use:** KDE plots show the estimated probability density function of a continuous variable, giving a smooth curve that represents the distribution.

```python
# Seaborn KDE plot
plt.figure(figsize=(12, 6))
sns.kdeplot(data=titanic, x='fare', hue='class', fill=True, alpha=0.5, multiple='stack')
plt.title('Kernel Density Estimate of Fare by Passenger Class')
plt.xlabel('Fare ($)')
plt.ylabel('Density')
plt.grid(True, alpha=0.3)
plt.show()

# Plotly interactive KDE plot
fig = px.histogram(titanic, x='fare', color='class', 
                   marginal='kde', histnorm='probability density',
                   title='Fare Distribution by Passenger Class')
fig.update_layout(xaxis_title='Fare ($)',
                  yaxis_title='Probability Density')
fig.show()
```

### Cumulative Distribution Functions (CDF)
**When to use:** CDFs show the cumulative probability of a variable being less than or equal to a certain value, which is useful for understanding the entire distribution at a glance.

```python
# Seaborn ECDF (empirical cumulative distribution function) plot
plt.figure(figsize=(12, 6))
sns.ecdfplot(data=titanic, x='age', hue='class')
plt.title('Empirical Cumulative Distribution of Age by Passenger Class')
plt.xlabel('Age')
plt.ylabel('Cumulative Probability')
plt.grid(True, alpha=0.3)
plt.show()

# Computing and plotting CDF manually
plt.figure(figsize=(12, 6))
for class_name in ['First', 'Second', 'Third']:
    # Get age data for this class
    ages = titanic[titanic['class'] == class_name]['age'].dropna()
    
    # Sort the ages
    ages_sorted = np.sort(ages)
    
    # Calculate the proportional rank of each age
    cumulative_prob = np.arange(1, len(ages_sorted) + 1) / len(ages_sorted)
    
    # Plot the CDF
    plt.step(ages_sorted, cumulative_prob, label=f'{class_name} Class', where='post')

plt.title('Cumulative Distribution Function of Age by Passenger Class')
plt.xlabel('Age')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Relational Plot Types
Relational plots show the relationship between two or more variables in the data.

### Scatter Plots
We've already covered basic scatter plots in the "Basic Plot Types" section. Here we'll look at more advanced relational scatter plots.

```python
# Seaborn relational plot with regression line
plt.figure(figsize=(10, 8))
sns.regplot(x='total_bill', y='tip', data=tips, scatter_kws={'alpha':0.5})
plt.title('Relationship Between Bill Amount and Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.grid(True, alpha=0.3)
plt.show()

# Adding a third dimension with color and size
plt.figure(figsize=(10, 8))
sns.scatterplot(x='total_bill', y='tip', hue='day', size='size', sizes=(20, 200), data=tips)
plt.title('Relationship Between Bill, Tip, Day, and Party Size')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.grid(True, alpha=0.3)
plt.show()
```

### Pair Plots
**When to use:** Pair plots create a grid of scatter plots for each pair of variables in a dataset, allowing you to see all pairwise relationships at once.

```python
# Seaborn pair plot
plt.figure(figsize=(12, 10))
sns.pairplot(iris, hue='species', height=2.5)
plt.suptitle('Pair Plot of Iris Dataset', y=1.02)
plt.show()

# Customized pair plot with different plot types
plt.figure(figsize=(12, 10))
g = sns.pairplot(tips, hue='time', 
                 diag_kind='kde',
                 plot_kws={'alpha': 0.6},
                 diag_kws={'alpha': 0.5})
g.fig.suptitle('Pair Plot of Tips Dataset by Time of Day', y=1.02)
plt.show()
```

### Joint Plots
**When to use:** Joint plots show the relationship between two variables with bivariate and univariate graphs combined, typically with a scatter plot in the middle and histograms or KDE plots on the axes.

```python
# Seaborn joint plot with scatter and histograms
plt.figure(figsize=(10, 10))
sns.jointplot(x='total_bill', y='tip', data=tips, height=8)
plt.suptitle('Joint Distribution of Bill and Tip', y=1.02)
plt.show()

# Joint plot with regression line
plt.figure(figsize=(10, 10))
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg', height=8)
plt.suptitle('Joint Distribution of Bill and Tip with Regression', y=1.02)
plt.show()

# Joint plot with KDE
plt.figure(figsize=(10, 10))
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde', height=8)
plt.suptitle('Joint Density of Bill and Tip', y=1.02)
plt.show()
```

## Specialized Plot Types

### Radar Charts (Spider Plots)
**When to use:** Radar charts are useful for displaying multivariate data in the form of a two-dimensional chart with three or more quantitative variables represented on axes starting from the same point.

```python
# Create a sample dataset for player statistics
players = ['Player A', 'Player B', 'Player C']
attributes = ['Speed', 'Power', 'Technique', 'Stamina', 'Accuracy']

# Random scores for each player on each attribute (scale of 0-10)
np.random.seed(42)
scores = np.random.randint(5, 11, size=(len(players), len(attributes)))

# Create a radar chart with Plotly
fig = go.Figure()

for i, player in enumerate(players):
    fig.add_trace(go.Scatterpolar(
        r=scores[i],
        theta=attributes,
        fill='toself',
        name=player
    ))

fig.update_layout(
    polar=dict(
        radialaxis=dict(
            visible=True,
            range=[0, 10]
        )
    ),
    showlegend=True,
    title='Player Attributes Comparison'
)

fig.show()
```

### Network Graphs
**When to use:** Network graphs are ideal for showing relationships or connections between entities, such as social networks, system architectures, or organizational structures.

```python
# Create a simple network graph
G = nx.karate_club_graph()

# Get node positions
pos = nx.spring_layout(G, seed=42)

# Create a Matplotlib figure
plt.figure(figsize=(12, 8))

# Draw the graph
nx.draw(G, pos, with_labels=True, node_color='skyblue', 
        node_size=700, edge_color='gray', linewidths=0.5,
        font_size=10)

plt.title('Zachary\'s Karate Club Network')
plt.axis('off')
plt.tight_layout()
plt.show()

# Create an interactive network graph with Plotly
edge_x = []
edge_y = []
for edge in G.edges():
    x0, y0 = pos[edge[0]]
    x1, y1 = pos[edge[1]]
    edge_x.extend([x0, x1, None])
    edge_y.extend([y0, y1, None])

edge_trace = go.Scatter(
    x=edge_x, y=edge_y,
    line=dict(width=0.5, color='#888'),
    hoverinfo='none',
    mode='lines')

node_x = []
node_y = []
for node in G.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)

node_trace = go.Scatter(
    x=node_x, y=node_y,
    mode='markers+text',
    hoverinfo='text',
    text=[str(i) for i in G.nodes()],
    marker=dict(
        showscale=True,
        colorscale='YlGnBu',
        size=10,
        colorbar=dict(
            thickness=15,
            title='Node Connections',
            xanchor='left',
            titleside='right'
        ),
        line_width=2))

# Color node points by the number of connections
node_adjacencies = []
for node, adjacencies in enumerate(G.adjacency()):
    node_adjacencies.append(len(adjacencies[1]))

node_trace.marker.color = node_adjacencies
node_trace.marker.size = [v * 2 + 5 for v in node_adjacencies]

fig = go.Figure(data=[edge_trace, node_trace],
             layout=go.Layout(
                title='Zachary\'s Karate Club Network',
                titlefont_size=16,
                showlegend=False,
                hovermode='closest',
                margin=dict(b=20,l=5,r=5,t=40),
                xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
                yaxis=dict(showgrid=False, zeroline=False, showticklabels=False))
                )

fig.show()
```

## 3D Plot Types
Three-dimensional plots add an extra dimension to your visualizations, which can be useful for visualizing complex relationships.

### 3D Scatter Plots
**When to use:** 3D scatter plots are excellent for visualizing relationships among three continuous variables.

```python
# Create a 3D scatter plot with Matplotlib
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Get iris data
x = iris['sepal_length']
y = iris['sepal_width']
z = iris['petal_length']
species = iris['species']

# Create a color map
colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}
species_colors = [colors[s] for s in species]

# Plot the scatter points
scatter = ax.scatter(x, y, z, c=species_colors, s=50, alpha=0.8)

# Add labels
ax.set_xlabel('Sepal Length (cm)')
ax.set_ylabel('Sepal Width (cm)')
ax.set_zlabel('Petal Length (cm)')
ax.set_title('3D Scatter Plot of Iris Dataset')

# Add a legend
from matplotlib.lines import Line2D
legend_elements = [Line2D([0], [0], marker='o', color='w', 
                          markerfacecolor=color, label=species, markersize=10)
                   for species, color in colors.items()]
ax.legend(handles=legend_elements)

plt.show()

# Create an interactive 3D scatter plot with Plotly
fig = px.scatter_3d(iris, x='sepal_length', y='sepal_width', z='petal_length',
                   color='species', size='petal_width',
                   title='3D Scatter Plot of Iris Dataset')
fig.update_layout(scene=dict(
                    xaxis_title='Sepal Length (cm)',
                    yaxis_title='Sepal Width (cm)',
                    zaxis_title='Petal Length (cm)'))
fig.show()
```

### Surface Plots
**When to use:** Surface plots are ideal for visualizing a function of two variables or displaying three-dimensional data on a regular grid.

```python
# Create a surface plot with Matplotlib
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Create data for the surface plot
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create the surface plot
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.8)

# Add a color bar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)

# Add labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Surface Plot: z = sin(sqrt(x² + y²))')

plt.show()

# Create an interactive surface plot with Plotly
fig = go.Figure(data=[go.Surface(z=Z, x=x, y=y, colorscale='Viridis')])
fig.update_layout(title='Surface Plot: z = sin(sqrt(x² + y²))',
                 scene=dict(
                     xaxis_title='X',
                     yaxis_title='Y',
                     zaxis_title='Z'))
fig.show()
```

### 3D Wireframes
**When to use:** Wireframe plots show a 3D surface by plotting grid lines, which can be useful for seeing the structure of a surface without the solid fill that might obscure details.

```python
# Create a wireframe plot with Matplotlib
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Create data for the wireframe plot
u = np.linspace(0, 2 * np.pi, 30)
v = np.linspace(0, np.pi, 20)
x = 3 * np.outer(np.cos(u), np.sin(v))
y = 3 * np.outer(np.sin(u), np.sin(v))
z = 3 * np.outer(np.ones(np.size(u)), np.cos(v))

# Create the wireframe plot
wire = ax.plot_wireframe(x, y, z, color='blue', linewidth=0.5)

# Add labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Wireframe of a Sphere')

plt.show()

# Create an interactive wireframe-like plot with Plotly
fig = go.Figure()

# Add wireframe using a mesh grid
for i, _u in enumerate(u):
    fig.add_trace(go.Scatter3d(
        x=x[i], y=y[i], z=z[i],
        mode='lines',
        line=dict(color='blue', width=2),
        showlegend=False
    ))
    
for j in range(len(v)):
    xj = [x[i][j] for i in range(len(u))]
    yj = [y[i][j] for i in range(len(u))]
    zj = [z[i][j] for i in range(len(u))]
    fig.add_trace(go.Scatter3d(
        x=xj, y=yj, z=zj,
        mode='lines',
        line=dict(color='blue', width=2),
        showlegend=False
    ))

fig.update_layout(
    title='3D Wireframe of a Sphere',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    )
)

fig.show()
```

## Comparison of Plot Types
Let's compare different plot types for the same dataset to demonstrate which visual works best for different insights.

```python
# Let's use the Titanic dataset to compare different plot types for survival analysis

# Create a figure with multiple subplots
fig, axes = plt.subplots(2, 2, figsize=(16, 14))

# Plot 1: Bar chart showing survival by class
sns.countplot(x='class', hue='survived', data=titanic, ax=axes[0, 0])
axes[0, 0].set_title('Bar Chart: Survival by Class')
axes[0, 0].set_xlabel('Passenger Class')
axes[0, 0].set_ylabel('Count')
axes[0, 0].legend(title='Survived', labels=['No', 'Yes'])

# Plot 2: Stacked percentage bar chart
class_survival = pd.crosstab(titanic['class'], titanic['survived'], normalize='index') * 100
class_survival.plot(kind='bar', stacked=True, ax=axes[0, 1], colormap='viridis')
axes[0, 1].set_title('Stacked Bar Chart: Survival Rate by Class')
axes[0, 1].set_xlabel('Passenger Class')
axes[0, 1].set_ylabel('Percentage (%)')
axes[0, 1].legend(title='Survived', labels=['No', 'Yes'])

# Plot 3: Pie charts for each class
classes = titanic['class'].unique()
survival_by_class = {}
for cls in classes:
    survival_by_class[cls] = titanic[titanic['class'] == cls]['survived'].value_counts()

# Create pie charts
for i, (cls, data) in enumerate(survival_by_class.items()):
    if i >= 3:  # Just to make sure we don't go out of bounds
        break
    ax = axes[1, i % 2]
    data.plot(kind='pie', autopct='%1.1f%%', startangle=90, ax=ax, 
              colors=['salmon', 'lightgreen'], labels=['Did not survive', 'Survived'])
    ax.set_ylabel('')
    ax.set_title(f'Pie Chart: Survival in {cls} Class')

plt.tight_layout()
plt.suptitle('Comparing Different Plot Types for Visualizing Titanic Survival Data', 
             fontsize=16, y=1.02)
plt.show()

# Use Plotly to create an interactive comparison
fig = make_subplots(rows=2, cols=2, 
                   specs=[[{'type': 'bar'}, {'type': 'bar'}],
                          [{'type': 'domain'}, {'type': 'domain'}]],
                   subplot_titles=('Survival Count by Class', 
                                  'Survival Rate by Class', 
                                  'Survival in First Class',
                                  'Age Distribution by Survival'))

# Plot 1: Bar chart
survival_counts = titanic.groupby(['class', 'survived']).size().reset_index(name='count')
for i, surv in enumerate(['No', 'Yes']):
    fig.add_trace(
        go.Bar(x=survival_counts[survival_counts['survived'] == i]['class'], 
               y=survival_counts[survival_counts['survived'] == i]['count'],
               name=f'Survived: {surv}'),
        row=1, col=1
    )

# Plot 2: Stacked percentage bar chart
for i, cls in enumerate(['First', 'Second', 'Third']):
    class_data = titanic[titanic['class'] == cls]
    survived = class_data['survived'].mean() * 100
    not_survived = 100 - survived
    fig.add_trace(
        go.Bar(x=[cls], y=[not_survived], name='Did not survive', 
               legendgroup='Not Survived', showlegend=i==0),
        row=1, col=2
    )
    fig.add_trace(
        go.Bar(x=[cls], y=[survived], name='Survived',
               legendgroup='Survived', showlegend=i==0),
        row=1, col=2
    )

# Plot 3: Pie chart for first class
first_class = titanic[titanic['class'] == 'First']['survived'].value_counts()
fig.add_trace(
    go.Pie(labels=['Did not survive', 'Survived'], 
           values=first_class.values, 
           hole=.3),
    row=2, col=1
)

# Plot 4: Box plot of age by survival
fig.add_trace(
    go.Box(x=titanic[titanic['survived'] == 0]['age'].dropna(), 
           name='Did not survive',
           marker_color='salmon'),
    row=2, col=2
)
fig.add_trace(
    go.Box(x=titanic[titanic['survived'] == 1]['age'].dropna(), 
           name='Survived',
           marker_color='lightgreen'),
    row=2, col=2
)

# Update layout
fig.update_layout(height=800, width=1000, title_text="Interactive Comparison of Plot Types",
                 barmode='stack')
fig.update_xaxes(title_text="Passenger Class", row=1, col=1)
fig.update_yaxes(title_text="Count", row=1, col=1)
fig.update_xaxes(title_text="Passenger Class", row=1, col=2)
fig.update_yaxes(title_text="Percentage (%)", row=1, col=2)
fig.update_xaxes(title_text="Age", row=2, col=2)

fig.show()
```

## Decision Framework for Plot Selection
Choosing the right visualization is crucial for effective data communication. Here's a methodical approach to selecting the right plot based on your data and analysis goals.

### Step 1: Identify Your Data Type

1. **Numerical Data**
   - Continuous (e.g., height, weight, time)
   - Discrete (e.g., count of items, number of children)

2. **Categorical Data**
   - Nominal (e.g., gender, country, product type)
   - Ordinal (e.g., education level, satisfaction rating)

3. **Time Series Data**
   - Regular intervals (e.g., daily temperature, monthly sales)
   - Irregular intervals (e.g., event timestamps)

4. **Geospatial Data**
   - Point locations (e.g., store locations, incident reports)
   - Regions (e.g., countries, states, zip codes)

### Step 2: Determine Your Analysis Goal

1. **Distribution Analysis**
   - Understanding the shape, center, and spread of data

2. **Comparison Analysis**
   - Comparing values across different categories or groups

3. **Relationship Analysis**
   - Examining correlations or patterns between variables

4. **Composition Analysis**
   - Understanding parts of a whole

5. **Trend Analysis**
   - Tracking changes over time

### Step 3: Consider the Number of Variables

1. **Univariate Analysis** (one variable)
2. **Bivariate Analysis** (two variables)
3. **Multivariate Analysis** (three or more variables)

### Step 4: Select the Appropriate Plot Type

Below is a decision framework to help you select the appropriate plot type based on the above considerations:

#### Plot Selection Matrix

| Analysis Goal | Data Types | Recommended Plot Types |
|---------------|------------|------------------------|
| **Distribution** | Single Numerical | Histogram, Density Plot, Box Plot, Violin Plot |
| **Distribution** | Single Categorical | Bar Chart, Pie Chart (if few categories) |
| **Comparison** | Numerical across Categories | Bar Chart, Box Plot, Violin Plot |
| **Comparison** | Multiple Categorical | Grouped Bar Chart, Stacked Bar Chart, Heatmap |
| **Relationship** | Two Numerical | Scatter Plot, Hex Bin Plot, 2D Density Plot |
| **Relationship** | Numerical & Categorical | Box Plot, Violin Plot, Grouped Bar Chart |
| **Relationship** | Two Categorical | Heatmap, Mosaic Plot |
| **Composition** | Categorical Parts of Whole | Pie Chart, Stacked Bar Chart, Treemap |
| **Trend** | Numerical over Time | Line Chart, Area Chart |
| **Trend** | Categorical over Time | Stacked Area Chart, Stream Graph |
| **Geospatial** | Point Data | Scatter Map, Bubble Map |
| **Geospatial** | Regional Data | Choropleth Map, Cartogram |
| **Multivariate** | 3+ Variables | Scatter Plot Matrix, Parallel Coordinates, Radar Chart |

```python
# Let's create an interactive plot selection tool
def recommend_plots(data_type, analysis_goal, num_variables):
    """
    Recommend plot types based on data type, analysis goal, and number of variables.
    
    Parameters:
    data_type (str): Type of data ('numerical', 'categorical', 'time_series', 'geospatial', 'mixed')
    analysis_goal (str): Analysis goal ('distribution', 'comparison', 'relationship', 'composition', 'trend')
    num_variables (int): Number of variables (1, 2, 3+)
    
    Returns:
    list: List of recommended plot types
    """
    recommendations = []
    
    # Distribution Analysis
    if analysis_goal == 'distribution':
        if data_type == 'numerical':
            recommendations = ['Histogram', 'Density Plot', 'Box Plot', 'Violin Plot', 'CDF Plot']
        elif data_type == 'categorical':
            recommendations = ['Bar Chart', 'Pie Chart (if few categories)', 'Tree Map']
        elif data_type == 'time_series':
            recommendations = ['Line Chart', 'Histogram by Time Period', 'Box Plot by Time Period']
    
    # Comparison Analysis
    elif analysis_goal == 'comparison':
        if data_type == 'numerical':
            if num_variables == 1:
                recommendations = ['Box Plot', 'Violin Plot', 'Strip Plot', 'Swarm Plot']
            else:
                recommendations = ['Grouped Box Plot', 'Violin Plot', 'Scatter Plot', 'Bubble Chart']
        elif data_type == 'categorical':
            recommendations = ['Bar Chart', 'Grouped Bar Chart', 'Heatmap']
        elif data_type == 'time_series':
            recommendations = ['Multi-line Chart', 'Area Chart', 'Slope Chart']
        elif data_type == 'geospatial':
            recommendations = ['Choropleth Map', 'Graduated Symbol Map', 'Heat Map']
    
    # Relationship Analysis
    elif analysis_goal == 'relationship':
        if num_variables == 2:
            if data_type == 'numerical':
                recommendations = ['Scatter Plot', '2D Density Plot', 'Hexbin Plot', 'Contour Plot']
            elif data_type == 'categorical':
                recommendations = ['Heatmap', 'Mosaic Plot']
            elif data_type == 'mixed':
                recommendations = ['Box Plot', 'Violin Plot', 'Strip Plot', 'Swarm Plot']
        else:  # 3+ variables
            recommendations = ['Scatter Plot Matrix', 'Parallel Coordinates', 'Radar Chart', '3D Scatter Plot']
    
    # Composition Analysis
    elif analysis_goal == 'composition':
        if data_type == 'categorical':
            recommendations = ['Pie Chart', 'Stacked Bar Chart', 'Tree Map', 'Waffle Chart']
        elif data_type == 'time_series':
            recommendations = ['Stacked Area Chart', 'Stream Graph']
    
    # Trend Analysis
    elif analysis_goal == 'trend':
        if data_type == 'time_series':
            recommendations = ['Line Chart', 'Area Chart', 'Slope Chart']
        elif data_type == 'numerical':
            recommendations = ['Line Chart', 'Area Chart']
    
    return recommendations

# Example usage of our recommendation function
data_types = ['numerical', 'categorical', 'time_series', 'geospatial', 'mixed']
analysis_goals = ['distribution', 'comparison', 'relationship', 'composition', 'trend']
variable_counts = [1, 2, '3+']

# Let's test a few combinations
examples = [
    ('numerical', 'distribution', 1),
    ('categorical', 'comparison', 2),
    ('numerical', 'relationship', 2),
    ('time_series', 'trend', 2),
    ('mixed', 'relationship', 3)
]

for data_type, analysis_goal, num_variables in examples:
    print(f"Data Type: {data_type}, Goal: {analysis_goal}, Variables: {num_variables}")
    recommendations = recommend_plots(data_type, analysis_goal, num_variables)
    print("Recommended Plots:", ", ".join(recommendations))
    print()
```

### Additional Considerations for Plot Selection

When choosing a visualization, also consider:

1. **Audience**: Who will be viewing the visualization? Tailor the complexity based on their technical expertise.

2. **Medium**: Where will the visualization be presented (print, web, presentation)? Interactive visualizations are great for the web but not for print.

3. **Message**: What key insight are you trying to convey? Choose a plot type that emphasizes this message.

4. **Accessibility**: Consider color blindness and other accessibility concerns when designing visualizations.

5. **Data Volume**: Some plot types don't work well with very large datasets (e.g., scatter plots with millions of points).

6. **Interactivity Needs**: Determine if your audience needs to interact with the data to explore it further.

Remember, effective data visualization is both a science and an art. While these guidelines can help you make better choices, experimentation and feedback are key to creating impactful visualizations.
