# Basic Boxplots and Violin Plots

This notebook explores how to create and interpret boxplots and violin plots, which are powerful visualization tools for understanding data distributions.

## Import Required Libraries

First, we'll import the necessary Python libraries for data manipulation and visualization:
- **pandas**: for data handling and manipulation
- **numpy**: for numerical operations
- **matplotlib**: for creating basic plots
- **seaborn**: for enhanced statistical visualizations

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for better visualization
sns.set(style="whitegrid")

# Make plots larger and more readable
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# For reproducibility
np.random.seed(42)
```

## Understanding Boxplots

A boxplot (also known as a box-and-whisker plot) provides a visual summary of the distribution of data by displaying:

- **Median**: The middle line in the box
- **First quartile (Q1)**: The bottom of the box (25th percentile)
- **Third quartile (Q3)**: The top of the box (75th percentile)
- **Interquartile Range (IQR)**: The height of the box (Q3 - Q1)
- **Whiskers**: Lines extending from the box, typically to the smallest and largest non-outlier values
- **Outliers**: Individual points beyond the whiskers

Boxplots are particularly useful for:
1. Identifying the central tendency of data
2. Visualizing data spread and variability
3. Detecting outliers
4. Comparing distributions across different groups

```python
# Create some sample data for demonstration
np.random.seed(42)

# Create three different distributions for comparison
data1 = np.random.normal(0, 1, 100)  # Normal distribution
data2 = np.random.normal(2, 0.5, 100)  # Normal distribution with different mean and std
data3 = np.random.exponential(2, 100)  # Exponential distribution

# Create a DataFrame for easy manipulation
df = pd.DataFrame({
    'Normal (μ=0, σ=1)': data1,
    'Normal (μ=2, σ=0.5)': data2,
    'Exponential (λ=0.5)': data3
})

# Display the first few rows of the DataFrame
df.head()
```

## Creating Basic Boxplots

Now let's create some basic boxplots using both matplotlib and seaborn.

```python
# Create a basic boxplot using matplotlib
plt.figure(figsize=(12, 6))
plt.boxplot(df.values, labels=df.columns)
plt.title('Basic Boxplot Using Matplotlib')
plt.ylabel('Value')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Create a boxplot using seaborn
plt.figure(figsize=(12, 6))
sns.boxplot(data=df)
plt.title('Basic Boxplot Using Seaborn')
plt.ylabel('Value')
plt.show()
```

```python
# Horizontal boxplot using seaborn
# First melt the dataframe to long format
df_melted = df.melt(var_name='Distribution', value_name='Value')

plt.figure(figsize=(10, 6))
sns.boxplot(x='Value', y='Distribution', data=df_melted)
plt.title('Horizontal Boxplot Using Seaborn')
plt.xlabel('Value')
plt.tight_layout()
plt.show()
```

## Customizing Boxplots

Now let's explore how to customize boxplots by changing colors, adding labels, and handling outliers.

```python
# Create a customized boxplot with matplotlib
fig, ax = plt.subplots(figsize=(12, 6))

# Custom boxplot with different colors
boxplot = ax.boxplot(
    df.values,
    labels=df.columns,
    patch_artist=True,  # Fill boxes with color
    widths=0.6,  # Width of the boxes
    showfliers=True,  # Show outliers
    flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markersize': 8},  # Customize outliers
    whiskerprops={'linestyle': '--', 'linewidth': 1.5},  # Customize whiskers
    medianprops={'color': 'black', 'linewidth': 2}  # Customize median lines
)

# Fill boxes with different colors
colors = ['lightblue', 'lightgreen', 'lightpink']
for box, color in zip(boxplot['boxes'], colors):
    box.set(facecolor=color)

plt.title('Customized Boxplot Using Matplotlib')
plt.ylabel('Value')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Customized boxplot with seaborn
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df,
    palette="pastel",  # Color palette
    linewidth=1.5,  # Line width
    fliersize=8,  # Outlier size
    notch=True  # Add notch to indicate confidence interval around median
)

plt.title('Customized Boxplot Using Seaborn')
plt.ylabel('Value')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Understanding Violin Plots

Violin plots are a combination of boxplots and kernel density plots. They show:

- The full distribution of data through a kernel density estimation
- Statistical summary similar to boxplots (median, quartiles)
- The width of the "violin" indicates the density or frequency of data points

Violin plots are particularly useful for:
1. Visualizing the full distribution shape
2. Comparing distributions across multiple categories
3. Identifying multimodal distributions (multiple peaks)
4. Providing more detail about the distribution than a boxplot

The underlying technique for violin plots is **kernel density estimation (KDE)** - a non-parametric way to estimate the probability density function of a random variable.

```python
# Create a basic violin plot using seaborn
plt.figure(figsize=(12, 6))
sns.violinplot(data=df)
plt.title('Basic Violin Plot Using Seaborn')
plt.ylabel('Value')
plt.tight_layout()
plt.show()
```

```python
# Create horizontal violin plots
plt.figure(figsize=(10, 6))
sns.violinplot(x='Value', y='Distribution', data=df_melted)
plt.title('Horizontal Violin Plot')
plt.xlabel('Value')
plt.tight_layout()
plt.show()
```

## Creating Basic Violin Plots

Violin plots combine the features of boxplots with kernel density plots to provide a comprehensive view of data distribution. Unlike boxplots, violin plots show the full distribution shape through kernel density estimation.

Seaborn is the preferred library for creating violin plots due to its simple API and aesthetic defaults. Here are some ways to create basic violin plots:

1. **Vertical violin plots** - The most common orientation, with categories on the x-axis
2. **Horizontal violin plots** - Useful when you have long category names or many categories
3. **Split violin plots** - For comparing distributions across two groups

Let's create each type using our sample data:

```python
# Create a basic violin plot using seaborn
plt.figure(figsize=(12, 6))
sns.violinplot(data=df)
plt.title('Basic Violin Plot Using Seaborn')
plt.ylabel('Value')
plt.tight_layout()
plt.show()
```

The resulting violin plot shows:
- The width at each point represents the density of data at that value
- The central white dot represents the median
- The thick black bar in the center represents the interquartile range
- The thin black line extends to the upper and lower adjacent values

Violin plots excel at revealing:
- Multimodal distributions (multiple peaks)
- Skewness in the data
- The full shape of the distribution

## Customizing Violin Plots

Violin plots can be customized in many ways, such as changing colors, adjusting bandwidth, adding inner representations, and more.

```python
# Create custom violin plots with different inner representations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Different inner representation styles
inner_types = ['box', 'quartile', 'stick', 'point']
titles = ['Violin with Box', 'Violin with Quartile', 'Violin with Stick', 'Violin with Point']

for i, (ax, inner, title) in enumerate(zip(axes.flatten(), inner_types, titles)):
    sns.violinplot(data=df, ax=ax, inner=inner, palette='pastel')
    ax.set_title(title)
    ax.set_ylabel('Value')

plt.tight_layout()
plt.show()
```

```python
# Customize violin plot with various options
plt.figure(figsize=(12, 6))
sns.violinplot(
    data=df,
    palette="Set3",  # Color palette
    bw=0.2,  # Bandwidth adjustment (smoothing parameter)
    cut=0,  # Don't extend the density past the data
    linewidth=2,  # Border line width
    inner='box',  # Show box plot inside violin
    saturation=0.7  # Color saturation
)
plt.title('Customized Violin Plot')
plt.ylabel('Value')
plt.tight_layout()
plt.show()
```

## Combining Boxplots and Violin Plots

Sometimes it's useful to combine boxplots and violin plots for a more comprehensive view of the data distribution.

```python
# Create a hybrid plot with violins and boxplots overlay
plt.figure(figsize=(14, 7))

# Plot violin first
sns.violinplot(data=df, color="0.8", inner=None)

# Overlay with boxplot
sns.boxplot(data=df, width=0.2, boxprops={'zorder': 2}, palette="Set2")

plt.title('Combined Violin and Box Plots')
plt.ylabel('Value')
plt.tight_layout()
plt.show()
```

## Comparing Distributions with Boxplots and Violin Plots

One of the main use cases for boxplots and violin plots is comparing distributions across different categories or groups.

```python
# Generate categorical data for comparison
np.random.seed(42)
n = 100

# Create a DataFrame with measurements across different conditions and groups
categories = ['Group A', 'Group B', 'Group C']
conditions = ['Condition 1', 'Condition 2']

data = []

for condition in conditions:
    for category in categories:
        if condition == 'Condition 1':
            # Different mean for each group
            values = np.random.normal(loc=5 + categories.index(category), scale=1.0, size=n)
        else:
            # Different variances for each group
            values = np.random.normal(loc=7, scale=0.5 + categories.index(category)*0.5, size=n)
            
        for value in values:
            data.append({
                'Category': category,
                'Condition': condition,
                'Value': value
            })

# Create dataframe
df_categorical = pd.DataFrame(data)

# Display the first few rows
df_categorical.head()
```

```python
# Compare distributions with boxplots across categories and conditions
plt.figure(figsize=(14, 7))
sns.boxplot(x='Category', y='Value', hue='Condition', data=df_categorical)
plt.title('Comparison of Distributions with Boxplots')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Compare distributions with violin plots across categories and conditions
plt.figure(figsize=(14, 7))
sns.violinplot(x='Category', y='Value', hue='Condition', data=df_categorical, split=True, inner='quartile')
plt.title('Comparison of Distributions with Violin Plots (Split)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Real-World Examples

Let's apply boxplots and violin plots to real datasets from seaborn to demonstrate their practical applications.

```python
# Load some real-world datasets
iris = sns.load_dataset("iris")
tips = sns.load_dataset("tips")

# Display the first few rows of each
print("Iris Dataset:")
print(iris.head())
print("\nTips Dataset:")
print(tips.head())
```

```python
# Boxplot for Iris dataset comparing sepal length across species
plt.figure(figsize=(12, 6))
sns.boxplot(x='species', y='sepal_length', data=iris)
plt.title('Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Violin plot for Iris dataset comparing all measurements across species
plt.figure(figsize=(14, 8))

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for i, feature in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.violinplot(x='species', y=feature, data=iris, inner='quartile')
    plt.title(f'{feature.replace("_", " ").title()} by Species')
    plt.tight_layout()
    
plt.tight_layout()
plt.show()
```

```python
# Comparing tip amounts by day and time using boxplots
plt.figure(figsize=(12, 6))
sns.boxplot(x='day', y='tip', hue='time', data=tips)
plt.title('Tip Amount by Day and Time')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Comparing tip percentage by day and time using violin plots
# First calculate tip percentage
tips['tip_pct'] = 100 * tips['tip'] / tips['total_bill']

plt.figure(figsize=(14, 7))
sns.violinplot(x='day', y='tip_pct', hue='time', data=tips, split=True)
plt.title('Tip Percentage by Day and Time')
plt.ylabel('Tip Percentage (%)')
plt.tight_layout()
plt.show()
```

## Conclusion

Boxplots and violin plots are powerful visualization tools for understanding data distributions:

**Boxplots**:
- Provide a clear visual summary of central tendency and spread
- Highlight outliers effectively
- Work well for comparing distributions
- Are simpler and often easier to interpret

**Violin Plots**:
- Show the full density distribution
- Reveal multimodal distributions
- Provide more detailed shape information
- Better for complex distributions

**Which to use?**
- Use boxplots for straightforward statistical summaries and outlier detection
- Use violin plots when the shape of the distribution is important
- Consider combining them for comprehensive visualization

Both plot types are essential tools in the data visualization toolkit and can provide valuable insights about your data's characteristics.
