# Correlation Visualization Techniques

This notebook provides a comprehensive exploration of correlation visualization techniques using scatter plots and heatmaps. We'll learn how to effectively visualize relationships between variables and interpret correlation patterns.

## Learning Objectives
- Understand the concept of correlation and how to visualize it
- Create and interpret scatter plots for correlation analysis
- Generate and customize correlation heatmaps
- Apply correlation visualization techniques to real-world data analysis

## 1. Import Required Libraries

First, we'll import the necessary libraries for correlation visualization:
- NumPy for numerical operations
- Pandas for data manipulation
- Matplotlib for basic visualization
- Seaborn for enhanced statistical visualizations
- Plotly for interactive plots

```python
# Import basic libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Import plotly for interactive visualizations
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set visualization styles and options
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# Display plots inline
%matplotlib inline
```

## 2. Create Sample Datasets

To demonstrate correlation visualization techniques, we'll create several synthetic datasets with different correlation strengths:
- Strong positive correlation
- Strong negative correlation
- Weak correlation
- No correlation

We'll also load a real-world dataset for practical examples.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Sample size for our synthetic datasets
n = 100

# Create strongly positively correlated data
x_pos = np.linspace(0, 10, n)
y_pos = x_pos + np.random.normal(0, 1, n)
df_pos = pd.DataFrame({'x': x_pos, 'y': y_pos})
df_pos['correlation'] = 'Strong Positive'

# Create strongly negatively correlated data
x_neg = np.linspace(0, 10, n)
y_neg = 10 - x_neg + np.random.normal(0, 1, n)
df_neg = pd.DataFrame({'x': x_neg, 'y': y_neg})
df_neg['correlation'] = 'Strong Negative'

# Create weakly correlated data
x_weak = np.linspace(0, 10, n)
y_weak = x_weak/2 + np.random.normal(0, 3, n)
df_weak = pd.DataFrame({'x': x_weak, 'y': y_weak})
df_weak['correlation'] = 'Weak'

# Create uncorrelated data
x_none = np.linspace(0, 10, n)
y_none = np.random.normal(5, 2, n)
df_none = pd.DataFrame({'x': x_none, 'y': y_none})
df_none['correlation'] = 'None'

# Combine all datasets
df_combined = pd.concat([df_pos, df_neg, df_weak, df_none])

# Calculate correlation coefficients
corr_pos = np.corrcoef(x_pos, y_pos)[0, 1]
corr_neg = np.corrcoef(x_neg, y_neg)[0, 1]
corr_weak = np.corrcoef(x_weak, y_weak)[0, 1]
corr_none = np.corrcoef(x_none, y_none)[0, 1]

print(f"Positive correlation: {corr_pos:.3f}")
print(f"Negative correlation: {corr_neg:.3f}")
print(f"Weak correlation: {corr_weak:.3f}")
print(f"No correlation: {corr_none:.3f}")
```

```python
# Load a real-world dataset (Iris dataset)
from sklearn.datasets import load_iris

iris = load_iris()
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df_iris['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# Display the first few rows of the dataset
print("Iris dataset sample:")
df_iris.head()
```

```python
# Create a more complex synthetic dataset with multiple correlated features
n = 200
np.random.seed(42)

# Base features
feature1 = np.random.normal(0, 1, n)
feature2 = np.random.normal(0, 1, n)

# Derived features with known correlations
feature3 = 0.7 * feature1 + 0.3 * feature2 + np.random.normal(0, 0.5, n)  # Correlated with both 1 & 2
feature4 = -0.8 * feature1 + 0.1 * feature2 + np.random.normal(0, 0.5, n)  # Strongly neg. corr with 1
feature5 = 0.2 * feature1 + np.random.normal(0, 1.5, n)  # Weakly correlated with 1
feature6 = np.random.normal(0, 1, n)  # Independent

# Create DataFrame
df_multi = pd.DataFrame({
    'Feature1': feature1,
    'Feature2': feature2,
    'Feature3': feature3,
    'Feature4': feature4,
    'Feature5': feature5,
    'Feature6': feature6
})

# Display the first few rows and the shape
print("Multi-feature dataset shape:", df_multi.shape)
df_multi.head()
```

## 3. Basic Scatter Plots for Correlation

Scatter plots are one of the simplest and most effective ways to visualize the correlation between two variables. Let's create scatter plots for our different correlation examples and interpret them.

```python
# Create subplots for each correlation type
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Scatter Plots Showing Different Correlation Types', fontsize=16)

# Plot each correlation type
sns.scatterplot(x='x', y='y', data=df_pos, ax=axes[0, 0], color='blue')
axes[0, 0].set_title(f'Strong Positive Correlation (r = {corr_pos:.3f})')
axes[0, 0].text(1, 8, f'r = {corr_pos:.3f}', fontsize=14, bbox=dict(facecolor='white', alpha=0.8))

sns.scatterplot(x='x', y='y', data=df_neg, ax=axes[0, 1], color='red')
axes[0, 1].set_title(f'Strong Negative Correlation (r = {corr_neg:.3f})')
axes[0, 1].text(1, 2, f'r = {corr_neg:.3f}', fontsize=14, bbox=dict(facecolor='white', alpha=0.8))

sns.scatterplot(x='x', y='y', data=df_weak, ax=axes[1, 0], color='green')
axes[1, 0].set_title(f'Weak Correlation (r = {corr_weak:.3f})')
axes[1, 0].text(1, 8, f'r = {corr_weak:.3f}', fontsize=14, bbox=dict(facecolor='white', alpha=0.8))

sns.scatterplot(x='x', y='y', data=df_none, ax=axes[1, 1], color='purple')
axes[1, 1].set_title(f'No Correlation (r = {corr_none:.3f})')
axes[1, 1].text(1, 8, f'r = {corr_none:.3f}', fontsize=14, bbox=dict(facecolor='white', alpha=0.8))

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

### Interpreting Scatter Plots

- **Strong Positive Correlation**: Points follow a line with a positive slope. As x increases, y tends to increase as well.
- **Strong Negative Correlation**: Points follow a line with a negative slope. As x increases, y tends to decrease.
- **Weak Correlation**: Points show a general trend but with considerable scatter around that trend.
- **No Correlation**: Points appear randomly scattered with no discernible pattern.

The Pearson correlation coefficient (r) quantifies these relationships, ranging from -1 (perfect negative correlation) to +1 (perfect positive correlation), with 0 indicating no correlation.

```python
# Add regression lines to better visualize the correlation patterns
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Scatter Plots with Regression Lines', fontsize=16)

# Plot each correlation type with regression line
sns.regplot(x='x', y='y', data=df_pos, ax=axes[0, 0], color='blue', line_kws={"color":"red"})
axes[0, 0].set_title(f'Strong Positive Correlation (r = {corr_pos:.3f})')

sns.regplot(x='x', y='y', data=df_neg, ax=axes[0, 1], color='red', line_kws={"color":"blue"})
axes[0, 1].set_title(f'Strong Negative Correlation (r = {corr_neg:.3f})')

sns.regplot(x='x', y='y', data=df_weak, ax=axes[1, 0], color='green', line_kws={"color":"red"})
axes[1, 0].set_title(f'Weak Correlation (r = {corr_weak:.3f})')

sns.regplot(x='x', y='y', data=df_none, ax=axes[1, 1], color='purple', line_kws={"color":"blue"})
axes[1, 1].set_title(f'No Correlation (r = {corr_none:.3f})')

plt.tight_layout()
plt.subplots_adjust(top=0.9)
plt.show()
```

## 5. Understanding Correlation Matrices

A correlation matrix shows the correlation coefficients between all pairs of variables in a dataset. This is especially useful when dealing with multiple variables where examining individual scatter plots might become cumbersome.

Let's create correlation matrices for our datasets and explore different correlation methods:
- Pearson (linear relationships)
- Spearman (monotonic relationships)
- Kendall (concordance between pairs)

```python
# Calculate the correlation matrix for the Iris dataset (numeric columns only)
corr_iris = df_iris.select_dtypes('number').corr()
print("Pearson Correlation Matrix for Iris Dataset:")
display(corr_iris)
```

```python
# Calculate correlation matrices using different methods for our multi-feature dataset
corr_pearson = df_multi.corr(method='pearson')
corr_spearman = df_multi.corr(method='spearman')
corr_kendall = df_multi.corr(method='kendall')

# Display the matrices
print("Pearson Correlation (Linear Relationship):")
display(corr_pearson.round(3))

print("\nSpearman Correlation (Monotonic Relationship):")
display(corr_spearman.round(3))

print("\nKendall Correlation (Concordance between Pairs):")
display(corr_kendall.round(3))
```

### Comparison of Correlation Methods

- **Pearson**: Measures linear relationships between variables. Assumes data is normally distributed and sensitive to outliers.
- **Spearman**: Measures monotonic relationships (variables change together, but not necessarily at a constant rate). More robust to outliers.
- **Kendall**: Measures the proportion of concordant versus discordant pairs. More robust and often preferred for small sample sizes.

The choice of correlation method depends on your data characteristics and analysis goals. If you observe significant differences between these methods, your data might contain outliers or non-linear relationships.

## 6. Creating Basic Heatmaps

Heatmaps are powerful tools for visualizing correlation matrices. The color intensity represents the strength of correlation between variables, making patterns easy to identify.

Let's create heatmaps for our correlation matrices:

```python
# Create a basic heatmap for the Iris dataset correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(corr_iris, annot=True, cmap='coolwarm', vmin=-1, vmax=1, square=True)
plt.title('Correlation Heatmap of Iris Dataset Features', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Create a heatmap for the multi-feature dataset correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(
    corr_pearson, 
    annot=True, 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    square=True,
    fmt='.2f',
    linewidths=0.5
)
plt.title('Pearson Correlation Heatmap of Multi-Feature Dataset', fontsize=14)
plt.tight_layout()
plt.show()
```

### Interpreting Correlation Heatmaps

- **Color Scale**: In the 'coolwarm' color scheme:
  - Dark blue: Strong negative correlation (-1)
  - White: No correlation (0)
  - Dark red: Strong positive correlation (+1)
- **Diagonal**: The diagonal always shows 1.0 (perfect correlation of each variable with itself)
- **Symmetry**: The heatmap is symmetrical across the diagonal since the correlation between A and B is the same as B and A
- **Annotations**: The numbers inside each cell represent the actual correlation coefficient values

## 7. Advanced Heatmap Customization

Let's explore advanced heatmap customization techniques:
1. Masking to hide redundant information
2. Clustering variables by similarity
3. Applying different color schemes
4. Adjusting visual properties for better interpretation

```python
# Create a heatmap with a mask to hide the upper triangle (redundant information)
plt.figure(figsize=(10, 8))

# Create a mask for the upper triangle
mask = np.triu(np.ones_like(corr_pearson, dtype=bool))

# Create a custom colormap
cmap = sns.diverging_palette(240, 10, as_cmap=True)

# Create heatmap with mask
sns.heatmap(
    corr_pearson, 
    mask=mask,
    annot=True, 
    cmap=cmap, 
    vmin=-1, 
    vmax=1, 
    center=0,
    square=True,
    fmt='.2f',
    linewidths=0.5,
    cbar_kws={"shrink": .8}
)

plt.title('Correlation Heatmap (Lower Triangle Only)', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Create a clustered heatmap to group similar variables together
plt.figure(figsize=(12, 10))

# Use hierarchical clustering to reorder the rows/columns
sns.clustermap(
    corr_pearson, 
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    annot=True,
    fmt='.2f',
    linewidths=0.5,
    standard_scale=None,  # No standardization for correlation matrix
    method='complete',    # Linkage method for hierarchical clustering
    figsize=(12, 10)
)

plt.title('Clustered Correlation Heatmap', fontsize=14, pad=50)
plt.show()
```

```python
# Create a heatmap highlighting only strong correlations
plt.figure(figsize=(12, 10))

# Define a function to determine cell text color based on correlation value
def text_color(val):
    color = 'white' if abs(val) > 0.5 else 'black'
    return color

# Create a custom annotation array
corr_rounded = corr_pearson.round(2)
annot_array = corr_rounded.astype(str)

# Set weak correlations to empty string
for i in range(len(corr_rounded.index)):
    for j in range(len(corr_rounded.columns)):
        if abs(corr_rounded.iloc[i, j]) < 0.3 and i != j:
            annot_array.iloc[i, j] = ""

# Create the heatmap with custom annotations
sns.heatmap(
    corr_pearson, 
    annot=annot_array, 
    cmap='RdBu_r', 
    vmin=-1, 
    vmax=1, 
    square=True,
    linewidths=0.5,
    annot_kws={"size": 10},
    cbar_kws={"shrink": .8, "label": "Correlation Coefficient"}
)

plt.title('Correlation Heatmap (Only Strong Correlations Labeled)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 8. Interpreting Correlations

Understanding correlation patterns in your visualizations is critical for drawing meaningful insights:

1. **Correlation Strength**: The absolute value indicates strength
   - 0.0 to 0.3: Weak correlation
   - 0.3 to 0.7: Moderate correlation
   - 0.7 to 1.0: Strong correlation

2. **Correlation Direction**: Positive or negative
   - Positive: Variables increase or decrease together
   - Negative: As one variable increases, the other decreases

Let's examine some common patterns and pitfalls in correlation analysis.

```python
# Create a dataset with a non-linear relationship that has low Pearson correlation
n = 100
np.random.seed(42)

# Create a quadratic relationship
x_quad = np.linspace(-5, 5, n)
y_quad = x_quad**2 + np.random.normal(0, 2, n)

df_quad = pd.DataFrame({'x': x_quad, 'y': y_quad})

# Calculate correlations with different methods
pearson_corr = df_quad.corr().iloc[0, 1]
spearman_corr = df_quad.corr(method='spearman').iloc[0, 1]
kendall_corr = df_quad.corr(method='kendall').iloc[0, 1]

# Plot
plt.figure(figsize=(10, 6))
plt.scatter(x_quad, y_quad, alpha=0.7)
plt.title(f'Non-linear Relationship Example\nPearson: {pearson_corr:.3f}, Spearman: {spearman_corr:.3f}, Kendall: {kendall_corr:.3f}')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("This is an important example showing how Pearson correlation can miss non-linear relationships.")
print("The parabola shape has near-zero Pearson correlation even though there's a clear pattern.")
print("Spearman and Kendall capture the monotonic portions better but still don't fully represent the relationship.")
```

```python
# Demonstrate the impact of outliers on correlation
np.random.seed(42)

# Create data with a strong correlation
x_base = np.linspace(0, 10, 50)
y_base = x_base + np.random.normal(0, 1, 50)

# Add outliers
x_outlier = np.array([12, 0])
y_outlier = np.array([0, 12])  # These go against the correlation pattern

x_with_outliers = np.concatenate([x_base, x_outlier])
y_with_outliers = np.concatenate([y_base, y_outlier])

# Calculate correlations
corr_base = np.corrcoef(x_base, y_base)[0, 1]
corr_with_outliers = np.corrcoef(x_with_outliers, y_with_outliers)[0, 1]

# Create plots
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Plot without outliers
ax[0].scatter(x_base, y_base)
ax[0].set_title(f'Without Outliers (r = {corr_base:.3f})')
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')
ax[0].grid(True, alpha=0.3)

# Plot with outliers
ax[1].scatter(x_with_outliers, y_with_outliers)
ax[1].scatter(x_outlier, y_outlier, color='red', s=100, label='Outliers')
ax[1].set_title(f'With Outliers (r = {corr_with_outliers:.3f})')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
ax[1].legend()
ax[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Correlation without outliers: {corr_base:.3f}")
print(f"Correlation with outliers: {corr_with_outliers:.3f}")
print("Note how just two outliers dramatically reduce the correlation coefficient.")
```

### Correlation vs. Causation

One of the most important principles in data analysis is understanding that **correlation does not imply causation**. A strong correlation between two variables does not necessarily mean that one causes the other.

Common scenarios that can lead to misleading correlations:

1. **Confounding variables**: A third variable affects both observed variables
2. **Reverse causality**: The assumed direction of cause and effect is backward
3. **Coincidental correlation**: Random chance (especially with small datasets)
4. **Non-independence**: Data points are not independent of each other

When interpreting correlations, always consider these possibilities and look for supporting evidence before making causal claims.

## 10. Real-world Example: Correlation Analysis

Let's apply correlation visualization techniques to a real-world dataset. We'll use a Boston housing dataset, which is commonly used for regression problems.

```python
# Load the Boston Housing dataset
from sklearn.datasets import load_boston

boston = load_boston()
df_boston = pd.DataFrame(boston.data, columns=boston.feature_names)
df_boston['PRICE'] = boston.target

# Show a sample of the dataset
print("Boston Housing Dataset Sample:")
df_boston.head()
```

```python
# Check for missing values
print("Missing values in Boston Housing dataset:")
print(df_boston.isna().sum())

# Summary statistics
print("\nSummary statistics:")
display(df_boston.describe())
```

```python
# Calculate the correlation matrix
corr_boston = df_boston.corr()

# Create a heatmap with a mask to visualize the correlations
plt.figure(figsize=(14, 12))

# Create a mask for the upper triangle
mask = np.triu(np.ones_like(corr_boston, dtype=bool))

# Draw the heatmap
sns.heatmap(
    corr_boston,
    mask=mask,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    fmt='.2f',
    linewidths=0.5,
    cbar_kws={"shrink": .8}
)

plt.title('Boston Housing Dataset Correlation Matrix', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Identify features most strongly correlated with price
price_correlations = corr_boston['PRICE'].sort_values(ascending=False)
print("Features most correlated with housing prices:")
display(price_correlations)

# Create scatter plots for the most significant correlations with price
top_features = price_correlations[1:6].index  # Excluding PRICE itself
plt.figure(figsize=(16, 10))

for i, feature in enumerate(top_features):
    plt.subplot(2, 3, i+1)
    sns.regplot(x=feature, y='PRICE', data=df_boston)
    plt.title(f'{feature} vs PRICE (r = {corr_boston.loc[feature, "PRICE"]:.3f})')
    
plt.tight_layout()
plt.show()
```

```python
# Create a correlation scatterplot matrix (pairplot) for features strongly correlated with price
# Include the top 5 features and the target variable
selected_features = list(price_correlations[:6].index)
plt.figure(figsize=(12, 10))
sns.pairplot(df_boston[selected_features], height=2)
plt.suptitle('Pairplot of Most Important Features', y=1.02)
plt.show()
```

```python
# Create a clustered heatmap to identify groups of related features
plt.figure(figsize=(14, 12))
sns.clustermap(
    corr_boston, 
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    annot=True,
    fmt='.2f',
    linewidths=0.5,
    method='complete',
    figsize=(14, 12)
)
plt.title('Clustered Correlation Heatmap of Boston Housing Dataset', fontsize=16, pad=50)
plt.show()
```

```python
# Create an interactive scatter plot matrix with Plotly
import plotly.express as px

# Select the most important features including the target
selected_features = list(price_correlations[:6].index)
fig = px.scatter_matrix(
    df_boston[selected_features],
    dimensions=selected_features,
    color='PRICE',
    opacity=0.7,
    title="Interactive Scatter Matrix of Key Boston Housing Features",
    color_continuous_scale=px.colors.sequential.Viridis
)
fig.update_layout(width=1000, height=800)
fig.show()
```

## Conclusion

In this notebook, we've explored various techniques for visualizing correlations in data:

1. **Scatter plots** provide a simple way to visualize the relationship between two variables:
   - Basic scatter plots show patterns but need quantitative analysis
   - Adding regression lines helps visualize the trend direction
   - Confidence intervals show the uncertainty in the relationship

2. **Correlation matrices and heatmaps** allow us to visualize relationships between multiple variables simultaneously:
   - Different correlation methods (Pearson, Spearman, Kendall) capture different aspects of relationships
   - Heatmaps provide an intuitive color-coded view of correlation strength
   - Advanced customization techniques like masking and clustering improve interpretability

3. **Important considerations in correlation analysis**:
   - Correlation does not imply causation
   - Outliers can significantly impact correlation coefficients
   - Non-linear relationships may not be captured by Pearson correlation
   - The choice of correlation method should match your data characteristics

4. **Comprehensive visualization approaches**:
   - Combining multiple visualization types provides more complete insights
   - Interactive plots allow for exploration of relationships
   - Real-world datasets often contain complex patterns requiring multiple visualization techniques

By applying these visualization techniques, you can effectively explore relationships in your data, identify patterns, and communicate insights to stakeholders.
