# Introduction to Seaborn

Seaborn is a Python data visualization library based on matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. This notebook introduces you to the core features of Seaborn, demonstrates various plot types, and shows customization options.

## 1. Import Required Libraries
First, let's import the necessary libraries including Pandas for data manipulation, Matplotlib for basic plotting support, and Seaborn as the main visualization library. We'll also load some sample datasets.

```python
# Import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the plotting style and context
sns.set_theme()

# For displaying plots in the notebook
%matplotlib inline

# Load some sample datasets that come with seaborn
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")
planets = sns.load_dataset("planets")

# Let's look at the first few rows of one dataset
tips.head()
```

## 2. Understanding Seaborn and Its Relationship with Matplotlib

Seaborn is built on top of Matplotlib and provides a higher-level interface for creating statistical visualizations. Key advantages of Seaborn include:

- **Beautiful default styles**: Seaborn comes with attractive preset themes
- **Built-in statistical functions**: Automatically calculates and visualizes statistical relationships
- **Integration with Pandas**: Works seamlessly with Pandas DataFrames
- **Dataset-oriented API**: Functions operate on dataframes, making it easy to map variables to visual properties

While Matplotlib gives you fine-grained control over plots, Seaborn simplifies creating common statistical visualizations with fewer lines of code.

```python
# Let's compare a basic scatter plot in Matplotlib vs Seaborn

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Matplotlib plot
ax1.scatter(iris['sepal_length'], iris['sepal_width'], c=iris['species'].astype('category').cat.codes)
ax1.set_xlabel('Sepal Length')
ax1.set_ylabel('Sepal Width')
ax1.set_title('Scatter Plot with Matplotlib')

# Seaborn plot
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species', ax=ax2)
ax2.set_title('Scatter Plot with Seaborn')

plt.tight_layout()
plt.show()
```

As you can see, the Seaborn plot automatically includes a legend, uses a nice color palette, and generally looks more polished with less code.

## 3. Creating Basic Plots with Seaborn

Seaborn provides intuitive functions to create common visualizations. Let's explore some of the basic plot types:

1. Scatter plots
2. Line plots 
3. Bar plots
4. Distribution plots

```python
# Scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', size='size', sizes=(20, 200))
plt.title('Scatter Plot: Tips vs Total Bill')
plt.show()
```

```python
# Line plot
plt.figure(figsize=(12, 6))
sns.lineplot(data=flights, x='year', y='passengers', marker='o')
plt.title('Line Plot: Airline Passengers Over Time')
plt.show()
```

```python
# Bar plot
plt.figure(figsize=(10, 6))
sns.barplot(data=tips, x='day', y='total_bill')
plt.title('Bar Plot: Average Total Bill by Day')
plt.show()
```

```python
# Distribution plot (histogram and KDE)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(data=tips, x='total_bill', kde=True)
plt.title('Histogram with KDE')

plt.subplot(1, 2, 2)
sns.kdeplot(data=tips, x='total_bill', fill=True)
plt.title('KDE Plot')

plt.tight_layout()
plt.show()
```

## 4. Customizing Seaborn Plots

Seaborn plots can be extensively customized to suit specific visualization needs. Let's look at various customization options including:

- Colors and palettes
- Figure size and layout
- Titles, labels, and annotations
- Legend positioning
- Plot styles

```python
# Customizing a scatter plot with various options
plt.figure(figsize=(10, 6))

# Create a custom color palette
custom_palette = sns.color_palette("Set2")

# Create the plot with customized elements
sns.scatterplot(
    data=iris, 
    x='sepal_length', 
    y='petal_length', 
    hue='species',
    style='species',
    size='sepal_width',
    sizes=(20, 200),
    palette=custom_palette,
    alpha=0.7
)

# Customize the plot appearance
plt.title('Iris Dataset: Sepal Length vs Petal Length', fontsize=16, fontweight='bold')
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Species', title_fontsize=12, fontsize=10, loc='upper left')

# Add a text annotation
plt.annotate('Setosa cluster', xy=(5.0, 1.5), xytext=(5.5, 1.0),
             arrowprops=dict(arrowstyle='->', color='black'), fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Customizing color palettes demonstration
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

palettes = ['viridis', 'plasma', 'inferno', 'magma', 'crest', 'flare']

for i, palette in enumerate(palettes):
    sns.barplot(data=tips, x='day', y='total_bill', palette=palette, ax=axes[i])
    axes[i].set_title(f'Palette: {palette}')
    axes[i].set_ylim(0, 30)

plt.tight_layout()
plt.show()
```

## 5. Statistical Plots in Seaborn

Seaborn excels at creating statistical visualizations. Let's explore some of the most common ones:

1. Distribution plots (histograms and KDE plots)
2. Box plots
3. Violin plots
4. Swarm plots

These plots are useful for understanding data distributions and comparing groups.

```python
# Create a figure with distribution plots
plt.figure(figsize=(14, 10))

# Histogram showing distribution of total_bill
plt.subplot(2, 2, 1)
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)
plt.title('Histogram of Total Bill')

# Box plot showing distribution of total_bill by day
plt.subplot(2, 2, 2)
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Box Plot of Total Bill by Day')

# Violin plot showing distribution of total_bill by day
plt.subplot(2, 2, 3)
sns.violinplot(data=tips, x='day', y='total_bill', inner='quartile')
plt.title('Violin Plot of Total Bill by Day')

# Swarm plot showing all data points
plt.subplot(2, 2, 4)
sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex', palette='Set2')
plt.title('Swarm Plot of Total Bill by Day and Sex')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Combining box plot with swarm plot for better visualization
plt.figure(figsize=(12, 6))

# Create the box plot
sns.boxplot(data=tips, x='day', y='total_bill', hue='time', palette='pastel', dodge=True)

# Overlay with swarm plot
sns.swarmplot(data=tips, x='day', y='total_bill', hue='time', dodge=True, color='.25', size=5, alpha=0.5)

plt.title('Box Plot with Swarm Plot: Total Bill by Day and Time')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 6. Category Plots in Seaborn

Seaborn offers specialized plots for visualizing categorical data:

1. Bar plots (show point estimates and confidence intervals)
2. Count plots (show counts of observations)
3. Point plots (show point estimates and confidence intervals using markers)
4. Categorical scatter plots (show the scatter points on categorical axes)

```python
# Create a figure with various categorical plots
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Bar plot
sns.barplot(data=titanic, x='class', y='fare', hue='sex', ax=axes[0, 0])
axes[0, 0].set_title('Bar Plot: Average Fare by Class and Sex')
axes[0, 0].set_ylabel('Average Fare')

# Count plot
sns.countplot(data=titanic, x='class', hue='who', ax=axes[0, 1])
axes[0, 1].set_title('Count Plot: Passengers by Class and Passenger Type')
axes[0, 1].set_ylabel('Count')

# Point plot
sns.pointplot(data=titanic, x='class', y='survived', hue='sex', ax=axes[1, 0])
axes[1, 0].set_title('Point Plot: Survival Rate by Class and Sex')
axes[1, 0].set_ylabel('Survival Rate')

# Strip plot (categorical scatter)
sns.stripplot(data=titanic, x='class', y='age', hue='survived', dodge=True, ax=axes[1, 1], jitter=True, alpha=0.6)
axes[1, 1].set_title('Strip Plot: Age Distribution by Class and Survival')
axes[1, 1].set_ylabel('Age')

plt.tight_layout()
plt.show()
```

```python
# Categorical box plots with multiple grouped variables
plt.figure(figsize=(14, 6))

# Using the "catplot" function with kind="box"
g = sns.catplot(
    data=tips,
    kind="box",
    x="day", 
    y="total_bill", 
    hue="smoker",
    col="time",
    palette="Set3",
    height=6,
    aspect=0.8
)

g.fig.suptitle('Box Plots Grouped by Multiple Variables', y=1.05, fontsize=16)
g.set_axis_labels("Day", "Total Bill")
g.set_titles("{col_name}")

plt.tight_layout()
plt.show()
```

## 7. Regression Plots

Seaborn makes it easy to visualize relationships between variables and fit regression models:

1. `regplot`: Simple regression plot on a single set of variables
2. `lmplot`: Regression plot that allows faceting using FacetGrid

These plots automatically fit and visualize a linear regression model to the data, along with confidence intervals.

```python
# Basic regplot showing the relationship between two variables
plt.figure(figsize=(10, 6))
sns.regplot(data=tips, x='total_bill', y='tip')
plt.title('Regression Plot: Tip vs Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
# Advanced lmplot with faceting
sns.lmplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',
    col='time',
    row='sex',
    height=5,
    aspect=1,
    scatter_kws={'alpha': 0.6},
    line_kws={'linewidth': 2}
)

plt.suptitle('Regression Models Faceted by Multiple Variables', y=1.05, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Customizing the regression line and confidence interval
plt.figure(figsize=(12, 6))

sns.regplot(
    data=tips,
    x='total_bill',
    y='tip',
    scatter_kws={'color': 'skyblue', 'alpha': 0.6, 's': 60},
    line_kws={'color': 'red', 'linewidth': 2},
    order=2,  # polynomial regression of order 2
    ci=95     # 95% confidence interval
)

plt.title('Polynomial Regression with 95% Confidence Interval')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## 8. Matrix Plots: Heatmaps and Correlation Visualization

Heatmaps are excellent for visualizing matrix data such as correlation matrices. Seaborn makes creating and customizing heatmaps straightforward.

```python
# Creating a correlation matrix heatmap
# First calculate the correlation matrix
corr = iris.drop('species', axis=1).corr()

# Create the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr,
    annot=True,  # Show the correlation values
    cmap='coolwarm',  # Color map
    vmin=-1,  # Minimum value for color scaling
    vmax=1,   # Maximum value for color scaling
    linewidths=0.5,  # Width of lines between cells
    cbar_kws={'label': 'Correlation Coefficient'}  # Customize color bar
)
plt.title('Correlation Matrix of Iris Features', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Creating a heatmap with a pivot table
# First, create a pivot table from the flights dataset
flights_pivot = flights.pivot_table(
    values='passengers',
    index='month',
    columns='year'
)

# Plot the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    flights_pivot,
    cmap='YlGnBu',
    annot=True,
    fmt='d',  # Display as integers
    linewidths=0.5,
    cbar_kws={'label': 'Passengers'}
)
plt.title('Number of Airline Passengers by Month and Year', fontsize=16)
plt.tight_layout()
plt.show()
```

## 9. Pair Plots and Joint Plots

Pair plots and joint plots are powerful tools for exploring relationships between multiple variables:

1. `pairplot`: Shows pairwise relationships between variables in a dataset
2. `jointplot`: Shows a bivariate relationship with marginal distributions

These plots are particularly useful in exploratory data analysis.

```python
# Create a pairplot to visualize pairwise relationships
sns.pairplot(
    iris,
    hue='species',
    palette='viridis',
    diag_kind='kde',
    plot_kws={'alpha': 0.6},
    height=2.5
)
plt.suptitle('Pairwise Relationships in Iris Dataset', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Creating various joint plots
fig, axes = plt.subplots(2, 2, figsize=(14, 14))
plt.subplots_adjust(wspace=0.4, hspace=0.4)

# Default jointplot (scatter with histograms)
g1 = sns.jointplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    kind='scatter',
    color='blueviolet',
    height=6
)
g1.fig.suptitle('Scatter Joint Plot', y=1.02, fontsize=14)
g1.fig.tight_layout()

# Hex jointplot (useful for large datasets)
g2 = sns.jointplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    kind='hex',
    color='darkgreen',
    height=6
)
g2.fig.suptitle('Hex Joint Plot', y=1.02, fontsize=14)
g2.fig.tight_layout()

# KDE jointplot
g3 = sns.jointplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    kind='kde',
    fill=True,
    cmap='Blues',
    height=6
)
g3.fig.suptitle('KDE Joint Plot', y=1.02, fontsize=14)
g3.fig.tight_layout()

# Regression jointplot
g4 = sns.jointplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    kind='reg',
    color='crimson',
    height=6
)
g4.fig.suptitle('Regression Joint Plot', y=1.02, fontsize=14)
g4.fig.tight_layout()

plt.show()
```

## 10. Customizing Plot Styles and Themes

Seaborn provides several built-in themes to control the overall appearance of plots. You can easily switch between different styles and customize various aspects:

1. Setting the theme with `set_theme()`
2. Changing the context with `set_context()`
3. Setting the style with `set_style()`
4. Creating custom color palettes

```python
# Exploring different Seaborn styles
styles = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
fig, axes = plt.subplots(len(styles), 1, figsize=(12, 15))

for i, style in enumerate(styles):
    # Set the style
    sns.set_style(style)
    
    # Create a simple plot
    sns.lineplot(data=flights, x='year', y='passengers', ax=axes[i], marker='o')
    
    # Set title with style name
    axes[i].set_title(f"Style: '{style}'", fontsize=14)
    axes[i].set_ylim(0, 650)

plt.tight_layout()
plt.show()

# Reset to default style
sns.set_style("darkgrid")
```

```python
# Demonstrating different contexts (controls scaling)
contexts = ['paper', 'notebook', 'talk', 'poster']
fig, axes = plt.subplots(len(contexts), 1, figsize=(10, 15))

for i, context in enumerate(contexts):
    # Set the context
    sns.set_context(context)
    
    # Create a simple plot
    sns.barplot(data=tips, x='day', y='total_bill', ax=axes[i])
    
    # Set title with context name
    axes[i].set_title(f"Context: '{context}'", fontsize=14)

plt.tight_layout()
plt.show()

# Reset to default context
sns.set_context("notebook")
```

```python
# Creating and using custom color palettes
plt.figure(figsize=(14, 10))

# Create various custom palettes
palettes = {
    'Custom Sequential': sns.color_palette("Blues", n_colors=5),
    'Custom Diverging': sns.color_palette("RdBu_r", n_colors=5),
    'Custom Categorical': sns.color_palette("Set3", n_colors=5),
    'Custom Cubehelix': sns.cubehelix_palette(5, start=.5, rot=-.75),
    'Custom Light': sns.light_palette("seagreen", n_colors=5),
    'Custom Dark': sns.dark_palette("purple", n_colors=5)
}

# Plot all palettes
for i, (name, palette) in enumerate(palettes.items()):
    plt.subplot(3, 2, i+1)
    sns.barplot(x=list(range(5)), y=[3, 5, 2, 4, 6], palette=palette)
    plt.title(name)
    plt.xlabel('')
    plt.ylabel('')

plt.tight_layout()
plt.show()
```
