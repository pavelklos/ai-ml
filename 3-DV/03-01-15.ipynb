# Adding Annotations to Plots

This notebook explores various techniques for adding informative annotations to your data visualizations. Annotations help highlight key insights, explain patterns, and draw attention to important data points in your plots.

## 1. Import Required Libraries

```python
# Import the necessary visualization and data processing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patheffects import withStroke
import seaborn as sns

# Set the style for our plots
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context("notebook", font_scale=1.2)

# For reproducible results
np.random.seed(42)
```

## 2. Creating Basic Plots for Annotation

Let's create several different types of plots that we'll use to demonstrate annotation techniques.

```python
# Create some example data
x = np.linspace(0, 10, 100)
y = 4 + 2 * np.sin(2 * x) + np.random.normal(0, 0.5, 100)

# Create figure with multiple subplots to use for different examples
plt.figure(figsize=(15, 10))

# 1. Simple line plot
plt.subplot(2, 2, 1)
plt.plot(x, y, 'b-', linewidth=2)
plt.title('Line Plot')
plt.xlabel('x')
plt.ylabel('y')

# 2. Scatter plot
plt.subplot(2, 2, 2)
scatter_data = pd.DataFrame({
    'x': np.random.normal(0, 2, 30),
    'y': np.random.normal(0, 2, 30)
})
plt.scatter(scatter_data['x'], scatter_data['y'], alpha=0.7)
plt.title('Scatter Plot')
plt.xlabel('x')
plt.ylabel('y')

# 3. Bar chart
plt.subplot(2, 2, 3)
categories = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 4]
plt.bar(categories, values)
plt.title('Bar Chart')
plt.xlabel('Category')
plt.ylabel('Value')

# 4. Time series
plt.subplot(2, 2, 4)
dates = pd.date_range(start='2023-01-01', periods=20)
ts_data = pd.Series(np.cumsum(np.random.randn(20)), index=dates)
plt.plot(dates, ts_data)
plt.title('Time Series')
plt.xlabel('Date')
plt.ylabel('Value')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 3. Text Annotations

The simplest form of annotation is adding text to your plots. Matplotlib provides two main functions for this:
- `plt.text()`: Add text at specific coordinates
- `plt.annotate()`: Add text with additional features like arrows

Let's see how to use these functions to add informative text to your plots.

```python
# Create a line plot with text annotations
x = np.linspace(0, 10, 100)
y = 2 + 3 * np.sin(x)

plt.figure(figsize=(12, 6))
plt.plot(x, y, 'b-')

# Simple text annotation
plt.text(1.5, 4.5, "Maximum", fontsize=12, color='red')

# Text with different style
plt.text(4.5, 0, "Minimum", fontsize=12, color='blue', 
         bbox=dict(facecolor='white', alpha=0.7, boxstyle='round,pad=0.5'))

# Text with rotation
plt.text(8, 4, "Upward trend", fontsize=12, color='green', rotation=45)

# Text with mathematical expressions using LaTeX
plt.text(7.5, -0.5, r'$y = 2 + 3\sin(x)$', fontsize=14)

plt.title('Text Annotation Examples')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Positioning Text Annotations

You can position text in different coordinate systems:

1. **Data coordinates** (default): Position based on your data values
2. **Figure coordinates**: Position relative to the entire figure (0,0 is bottom-left, 1,1 is top-right)
3. **Axes coordinates**: Position relative to the axes (0,0 is bottom-left, 1,1 is top-right)

Let's see examples of each:

```python
plt.figure(figsize=(10, 6))

# Create a simple plot
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)

# 1. Data coordinates (default)
plt.text(5, 0.5, "Data coordinates (5, 0.5)", 
         bbox=dict(facecolor='yellow', alpha=0.5))

# 2. Axes coordinates
plt.text(0.5, 0.5, "Axes coordinates (0.5, 0.5)", 
         transform=plt.gca().transAxes,
         bbox=dict(facecolor='lightblue', alpha=0.5))

# 3. Figure coordinates
plt.text(0.1, 0.9, "Figure coordinates (0.1, 0.9)", 
         transform=plt.gcf().transFigure,
         bbox=dict(facecolor='lightgreen', alpha=0.5))

plt.title('Different Text Coordinate Systems')
plt.tight_layout()
plt.show()
```

## 4. Adding Arrows and Annotations

Annotating specific data points often requires arrows pointing to the feature of interest. The `plt.annotate()` function is perfect for this, providing both text and directional indicators.

```python
# Create a scatter plot with annotations and arrows
np.random.seed(42)
x = np.random.rand(20) * 10
y = np.random.rand(20) * 10

plt.figure(figsize=(10, 8))
plt.scatter(x, y, s=100, c=np.random.rand(20), cmap='viridis')

# Basic annotation with arrow
plt.annotate('Highest point', xy=(x[3], y[3]), xytext=(x[3]+1, y[3]+1),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5))

# Curved arrow with custom style
plt.annotate('Interesting cluster', xy=(x[5], y[5]), xytext=(x[5]-3, y[5]+2),
             arrowprops=dict(facecolor='red', shrink=0.05, width=2,
                            connectionstyle="arc3,rad=.3"))

# Arrow with different head style
plt.annotate('Outlier', xy=(x[10], y[10]), xytext=(x[10]+2, y[10]-2),
             arrowprops=dict(facecolor='green', shrink=0.05, 
                            headwidth=10, headlength=10))

# Fancy box with arrow
plt.annotate('Low value', xy=(x[15], y[15]), xytext=(x[15]-1, y[15]-3),
             bbox=dict(boxstyle="round4,pad=.5", fc="skyblue", ec="b", lw=2),
             arrowprops=dict(arrowstyle="->", color='blue',
                            connectionstyle="arc3,rad=-.2"))

plt.title('Annotations with Arrows', fontsize=16)
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(True, alpha=0.3)
plt.colorbar(label='Random Value')
plt.tight_layout()
plt.show()
```

### Arrow Styles

Matplotlib offers many arrow styles through the `arrowprops` parameter. Here's a showcase of different arrow styles:

```python
plt.figure(figsize=(12, 8))
plt.scatter([5]*6, [1, 2, 3, 4, 5, 6], s=100, c='blue')

arrow_styles = [
    '->', '-|>', '<-', '<|-', '<|-|>', 'fancy', 
    'simple', 'wedge'
]

for i, style in enumerate(arrow_styles[:6]):  # Show first 6 styles
    plt.annotate(f"Arrow style: '{style}'", 
                 xy=(5, i+1), 
                 xytext=(8, i+1),
                 arrowprops=dict(arrowstyle=style, lw=2))

plt.xlim(0, 12)
plt.ylim(0, 7)
plt.title('Different Arrow Styles', fontsize=16)
plt.grid(False)
plt.tight_layout()
plt.show()
```

## 5. Annotating Line Plots

Line plots often need annotations to highlight important features like peaks, valleys, trends, or specific data points.

```python
# Create data for line plot
x = np.linspace(0, 10, 100)
y = 3 + 2 * np.sin(2*x) * np.exp(-0.2*x)

# Find indices of local maxima and minima
from scipy.signal import find_peaks
peaks, _ = find_peaks(y)
valleys, _ = find_peaks(-y)

plt.figure(figsize=(12, 6))
plt.plot(x, y, 'b-', lw=2)

# Annotate all peaks
for i in peaks:
    plt.plot(x[i], y[i], 'ro', ms=8)  # Red dot at peak
    plt.annotate(f'Peak: {y[i]:.2f}', 
                xy=(x[i], y[i]),
                xytext=(x[i]+0.2, y[i]+0.3),
                arrowprops=dict(facecolor='red', shrink=0.05, width=1,
                                connectionstyle="arc3,rad=0.2"))
    
# Annotate all valleys
for i in valleys:
    plt.plot(x[i], y[i], 'go', ms=8)  # Green dot at valley
    plt.annotate(f'Valley: {y[i]:.2f}', 
                xy=(x[i], y[i]),
                xytext=(x[i]-0.5, y[i]-0.5),
                arrowprops=dict(facecolor='green', shrink=0.05, width=1,
                                connectionstyle="arc3,rad=-0.2"))

# Annotate trend
plt.annotate('Decaying amplitude', 
            xy=(7, 2), 
            xytext=(7, 4),
            ha='center',
            arrowprops=dict(arrowstyle="->", lw=1.5, color='purple'),
            bbox=dict(boxstyle="round", fc="white", ec="purple"))

plt.title('Annotating Features in Line Plots', fontsize=16)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Highlighting Regions of Interest

We can also highlight specific regions in line plots:

```python
# Creating a time series with annotations highlighting specific regions
dates = pd.date_range('2023-01-01', periods=90)
values = 100 + np.cumsum(np.random.randn(90) * 3)
ts = pd.Series(values, index=dates)

fig, ax = plt.subplots(figsize=(12, 6))
ts.plot(ax=ax, lw=2)

# 1. Highlight a time period with a rectangle
start_date = pd.Timestamp('2023-01-15')
end_date = pd.Timestamp('2023-02-10')
ax.axvspan(start_date, end_date, color='yellow', alpha=0.3)
ax.text(start_date + (end_date - start_date)/2, max(values)-10, 
        'Volatility period',
        ha='center', va='center', fontsize=12, 
        bbox=dict(facecolor='yellow', alpha=0.5))

# 2. Annotate the maximum value
max_date = ts.idxmax()
max_value = ts.max()
ax.annotate(f'Maximum: {max_value:.1f}', 
           xy=(max_date, max_value),
           xytext=(max_date + pd.Timedelta(days=5), max_value + 5),
           arrowprops=dict(facecolor='red', shrink=0.05))

# 3. Mark a specific threshold with a horizontal line
threshold = 120
ax.axhline(y=threshold, color='red', linestyle='--', alpha=0.7)
ax.text(dates[5], threshold+2, 'Threshold', color='red')

# 4. Add explanatory text for a trend
trend_start = pd.Timestamp('2023-02-25')
trend_end = pd.Timestamp('2023-03-20')
mid_point = trend_start + (trend_end - trend_start)/2
trend_y = ts.loc[trend_start:trend_end].mean()

# Draw arrow showing the trend direction
ax.annotate('Upward trend', 
           xy=(trend_end, ts.loc[trend_end]),
           xytext=(mid_point, trend_y - 15),
           arrowprops=dict(facecolor='green', width=2, headwidth=10),
           bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="green", lw=1))

plt.title('Time Series with Annotated Regions and Events', fontsize=16)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 6. Annotating Scatter Plots

Scatter plots often benefit from annotations that identify outliers, clusters, or specific data points of interest.

```python
# Create scatter data with a few outliers
np.random.seed(42)
x = np.random.randn(100)
y = x * 3 + np.random.randn(100) * 2

# Add a few outliers
x = np.append(x, [4, -5, -4])
y = np.append(y, [0, -2, 15])

# Create a DataFrame for easier handling
df = pd.DataFrame({'x': x, 'y': y})

# Identify outliers (using arbitrary criteria for demonstration)
outliers = df[(abs(df.x) > 3) | (abs(df.y) > 10)]

plt.figure(figsize=(10, 8))
plt.scatter(df.x, df.y, alpha=0.6)

# Highlight outliers with a different color
plt.scatter(outliers.x, outliers.y, color='red', s=100)

# Annotate each outlier
for idx, row in outliers.iterrows():
    # Determine text position based on quadrant
    if row.x > 0 and row.y > 0:  # top-right
        xytext = (10, -10)
    elif row.x < 0 and row.y > 0:  # top-left
        xytext = (-10, -10)
    elif row.x < 0 and row.y < 0:  # bottom-left
        xytext = (-10, 10)
    else:  # bottom-right
        xytext = (10, 10)
        
    plt.annotate(f'Outlier\n({row.x:.1f}, {row.y:.1f})', 
                xy=(row.x, row.y),
                xytext=xytext,
                textcoords='offset points',
                ha='center',
                arrowprops=dict(arrowstyle='->', color='red'))

# Annotate the overall trend
plt.annotate('Positive correlation', 
            xy=(0, 0), 
            xytext=(-4, 6),
            arrowprops=dict(facecolor='blue', shrink=0.05, width=2),
            bbox=dict(boxstyle="round", fc="lightblue", ec="blue"))

# Add regression line
from scipy import stats
slope, intercept, r_value, p_value, std_err = stats.linregress(df.x, df.y)
xline = np.linspace(min(df.x), max(df.x), 100)
yline = slope * xline + intercept
plt.plot(xline, yline, 'k--', alpha=0.5)

# Add correlation coefficient
plt.annotate(f'r = {r_value:.2f}', 
            xy=(2, -10),
            fontsize=12,
            bbox=dict(boxstyle="square", fc="white"))

plt.title('Annotating Outliers in Scatter Plots', fontsize=16)
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Labeling Points in a Scatter Plot

When working with labeled data, you might want to add labels to specific points:

```python
# Create a scatter plot with labeled points (e.g., countries data)
countries = ['USA', 'China', 'India', 'Japan', 'Germany', 
             'Brazil', 'UK', 'France', 'Italy', 'Canada']
gdp = [21.43, 14.34, 2.87, 5.06, 3.85, 1.87, 2.67, 2.63, 1.88, 1.64]
life_exp = [77.3, 76.9, 69.7, 84.6, 81.3, 75.9, 81.3, 82.7, 83.6, 82.3]
population = [331e6, 1412e6, 1380e6, 126e6, 83e6, 212e6, 67e6, 65e6, 60e6, 38e6]

# Normalize population for marker size
pop_sizes = [p/1e7 for p in population]  

plt.figure(figsize=(14, 8))
scatter = plt.scatter(gdp, life_exp, s=pop_sizes, c=range(len(countries)), 
                    cmap='viridis', alpha=0.7)

# Add labels to all points
for i, country in enumerate(countries):
    # Adjust text position based on which country it is
    if country == 'China':
        xytext = (-15, -25)
    elif country in ['USA', 'Japan']:
        xytext = (10, 0)
    else:
        xytext = (5, 5)
        
    plt.annotate(country, 
                xy=(gdp[i], life_exp[i]),
                xytext=xytext,
                textcoords='offset points',
                fontsize=10,
                bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.8))

# Add explanatory annotations
plt.annotate('High GDP,\nHigh Life Expectancy', 
            xy=(20, 83),
            xytext=(15, 75),
            arrowprops=dict(arrowstyle="->", color='blue'),
            bbox=dict(boxstyle="round", fc="lightblue", alpha=0.7),
            fontsize=12)

plt.title('GDP vs Life Expectancy by Country (2023)', fontsize=16)
plt.xlabel('GDP (Trillion USD)')
plt.ylabel('Life Expectancy (Years)')
plt.colorbar(scatter, label="Country Index")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7. Annotating with Coordinates

Understanding different coordinate systems in matplotlib is important for precise annotations. We'll explore:
- Data coordinates (default)
- Axes coordinates (`ax.transAxes`)
- Figure coordinates (`fig.transFigure`)

```python
fig = plt.figure(figsize=(12, 9))

# Create three different subplots
ax1 = plt.subplot2grid((2, 2), (0, 0))
ax2 = plt.subplot2grid((2, 2), (0, 1))
ax3 = plt.subplot2grid((2, 2), (1, 0), colspan=2)

# Generate some data for the plots
x = np.linspace(0, 10, 100)
ax1.plot(x, np.sin(x))
ax1.set_title('Sine Wave')

ax2.plot(x, np.cos(x), 'g-')
ax2.set_title('Cosine Wave')

ax3.plot(x, np.sin(x) * np.cos(x), 'r-')
ax3.set_title('Sine * Cosine')

# 1. Data coordinates (specific to each axes)
ax1.annotate('Local maximum', 
           xy=(np.pi/2, 1), 
           xytext=(np.pi/2-1, 0.5),
           arrowprops=dict(facecolor='black', shrink=0.05))

ax2.annotate('Local minimum', 
           xy=(np.pi, -1), 
           xytext=(np.pi-1, -0.5),
           arrowprops=dict(facecolor='green', shrink=0.05))

# 2. Axes coordinates (0-1 range within each subplot)
ax1.text(0.5, 0.1, 'Axes coordinates (0.5, 0.1)', 
        transform=ax1.transAxes,
        ha='center',
        bbox=dict(facecolor='yellow', alpha=0.5))

ax2.text(0.5, 0.1, 'Axes coordinates (0.5, 0.1)', 
        transform=ax2.transAxes,
        ha='center',
        bbox=dict(facecolor='lightgreen', alpha=0.5))

# 3. Figure coordinates (0-1 range of entire figure)
plt.figtext(0.5, 0.05, 'Figure coordinates (0.5, 0.05) - at bottom center', 
          ha='center', fontsize=12, 
          bbox=dict(facecolor='lightblue', alpha=0.5))

# 4. Arrow spanning from one axes to another using figure coordinates
# Create a custom arrow
fig.patches.extend([
    patches.FancyArrowPatch(
        (0.25, 0.4),  # starting point in figure coordinates
        (0.75, 0.4),  # ending point in figure coordinates
        transform=fig.transFigure,
        connectionstyle="arc3,rad=0.3",
        arrowstyle="fancy",
        color="purple",
        lw=2
    )
])

# Add annotation explaining the arrow
plt.figtext(0.5, 0.5, 'Custom arrow in figure coordinates\nconnecting subplots', 
          ha='center', va='center', fontsize=10,
          bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

plt.tight_layout(rect=[0, 0.1, 1, 0.95])  # Adjust layout to make room for the bottom annotation
plt.show()
```

## 8. Custom Annotation Styles

You can create custom annotation styles using various Matplotlib features like custom text boxes, fonts, and styling options.

```python
# Create a plot with custom annotation styles
plt.figure(figsize=(12, 8))

# Generate example data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, 'b-', label='sin(x)')
plt.plot(x, y2, 'r-', label='cos(x)')

# 1. Rounded box with shadow
plt.annotate('Rounded Box with Shadow', 
           xy=(2.5, 0.5), 
           xytext=(4, 0.8),
           arrowprops=dict(facecolor='black', shrink=0.05),
           bbox=dict(boxstyle="round,pad=0.5", fc="yellow", 
                    ec="orange", lw=2, alpha=0.8,
                    boxstyle="round,pad=0.5,rounding_size=0.2",
                    path_effects=[withStroke(linewidth=5, foreground="black")]
                   ))

# 2. Fancy sawtooth-shaped box
plt.annotate('Sawtooth Box', 
           xy=(5, -0.5), 
           xytext=(7, -0.8),
           arrowprops=dict(facecolor='green', shrink=0.05),
           bbox=dict(boxstyle="sawtooth,pad=0.6", fc="lightgreen", 
                    ec="darkgreen", lw=2))

# 3. Round4 box (super-rounded corners)
plt.annotate('Super-Rounded Box', 
           xy=(7, 0.5), 
           xytext=(8, 0.8),
           arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.2"),
           bbox=dict(boxstyle="round4,pad=0.6", fc="lightblue", 
                    ec="blue", lw=2, alpha=0.8))

# 4. Circle annotation
plt.annotate('Circle Annotation', 
           xy=(1, -0.5), 
           xytext=(1, -0.8),
           arrowprops=dict(arrowstyle="wedge,tail_width=0.7", fc="purple", ec="purple"),
           bbox=dict(boxstyle="circle,pad=0.5", fc="lavender", 
                   ec="purple", lw=2))

# 5. LaTeX math text with custom style
plt.annotate(r'$\int_{0}^{\pi} \sin(x) dx = 2$', 
           xy=(np.pi, 0), 
           xytext=(np.pi - 1, -0.3),
           fontsize=15,
           arrowprops=dict(arrowstyle="-", connectionstyle="angle,angleA=0,angleB=90"),
           bbox=dict(boxstyle="square", fc="white", ec="black", lw=1))

# 6. Multi-line text with custom font
plt.annotate("Multi-line\nCustom Font\nAnnotation", 
           xy=(8.5, 0), 
           xytext=(7, 0),
           fontsize=12,
           family='serif',
           style='italic',
           weight='bold',
           arrowprops=dict(facecolor='black'),
           bbox=dict(boxstyle="square", fc=(1.0, 0.9, 0.8), ec="brown", lw=1))

plt.title('Custom Annotation Styles', fontsize=16)
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Text Box Styles

Matplotlib offers various box styles for text annotations. Here's a showcase of different options:

```python
# Create a reference chart of box styles
fig, ax = plt.subplots(figsize=(12, 8))

# Hide axis ticks and spines
ax.set_xticks([])
ax.set_yticks([])
for spine in ax.spines.values():
    spine.set_visible(False)

# Available boxstyles
boxstyles = {
    'square': 'square box',
    'round': 'box with rounded corners',
    'round4': 'box with super rounded corners',
    'circle': 'circular box',
    'darrow': 'box with arrow pointing down',
    'larrow': 'box with arrow pointing left',
    'rarrow': 'box with arrow pointing right',
    'roundtooth': 'box with rounded sawtooth edge',
    'sawtooth': 'box with sawtooth edge'
}

# Position for text boxes
x, y = 0.5, 0.9
dy = 0.1

for boxstyle, description in boxstyles.items():
    ax.text(x, y, f'{boxstyle}: {description}',
            ha='center',
            bbox=dict(boxstyle=boxstyle, fc='lightyellow', ec='red'),
            fontsize=12)
    y -= dy

plt.title('Available Box Styles for Annotations', fontsize=16)
plt.tight_layout()
plt.show()
```

## 9. Adding Annotations to Subplots

When working with multiple subplots, you might need to add annotations that relate to individual subplots or span across multiple plots.

```python
# Create a figure with multiple subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
axs = axs.flatten()  # Flatten the 2x2 array for easier indexing

# Generate some example data
x = np.linspace(0, 10, 100)
trends = {
    0: {'data': 2*np.sin(x), 'title': 'Cyclical Trend', 'color': 'blue'},
    1: {'data': np.exp(0.1*x), 'title': 'Exponential Growth', 'color': 'red'},
    2: {'data': np.log(x+1), 'title': 'Logarithmic Trend', 'color': 'green'},
    3: {'data': 5 + 0.5*x - 0.05*x**2, 'title': 'Quadratic Trend', 'color': 'purple'}
}

# Create each subplot
for i, ax in enumerate(axs):
    ax.plot(x, trends[i]['data'], color=trends[i]['color'])
    ax.set_title(trends[i]['title'])
    ax.grid(True, alpha=0.3)
    
    # Add an annotation specific to each subplot
    midpoint_x = 5
    midpoint_y = trends[i]['data'][len(x)//2]
    
    ax.annotate(f'Mid-point: {midpoint_y:.2f}',
              xy=(midpoint_x, midpoint_y),
              xytext=(midpoint_x + 2, midpoint_y),
              arrowprops=dict(arrowstyle='->'),
              bbox=dict(boxstyle='round', fc='white', ec=trends[i]['color']))
              
# Add a global title
fig.suptitle('Various Trend Patterns', fontsize=16, y=0.98)

# Add annotations that span subplots using figure coordinates
# Arrow connecting the top-right and bottom-left plots
fig.patches.extend([
    patches.FancyArrowPatch(
        (0.75, 0.75),  # start point (top-right plot, figure coordinates)
        (0.25, 0.25),  # end point (bottom-left plot, figure coordinates)
        transform=fig.transFigure,
        connectionstyle="arc3,rad=-0.3",
        arrowstyle="fancy",
        color="gray",
        lw=2,
        alpha=0.6
    )
])

# Add text spanning multiple subplots
plt.figtext(0.5, 0.5, 'Compare Trends', 
          ha='center', va='center', fontsize=14,
          bbox=dict(boxstyle='round', fc='lightyellow', ec='orange'),
          transform=fig.transFigure)

# Add annotations to specific regions
plt.figtext(0.25, 0.9, 'Periodic patterns', ha='center', fontsize=10, 
           bbox=dict(fc='lightblue', ec='blue', alpha=0.7))

plt.figtext(0.75, 0.9, 'Unlimited growth', ha='center', fontsize=10,
           bbox=dict(fc='mistyrose', ec='red', alpha=0.7))

plt.figtext(0.25, 0.1, 'Diminishing returns', ha='center', fontsize=10,
           bbox=dict(fc='lightgreen', ec='green', alpha=0.7))

plt.figtext(0.75, 0.1, 'Rise and fall', ha='center', fontsize=10, 
           bbox=dict(fc='lavender', ec='purple', alpha=0.7))

plt.tight_layout(rect=[0.03, 0.03, 0.97, 0.95])  # Adjust layout to make room for annotations
plt.show()
```

## 10. Annotating Statistical Information

Adding statistical information to plots can make them more informative. Let's see how to add various statistical annotations.

```python
# Creating example data
np.random.seed(42)
group1 = np.random.normal(0, 1, 50)
group2 = np.random.normal(2, 1.5, 50)

plt.figure(figsize=(12, 8))

# Create boxplots
boxplot = plt.boxplot([group1, group2], patch_artist=True, labels=['Group 1', 'Group 2'])
for box, color in zip(boxplot['boxes'], ['lightblue', 'lightgreen']):
    box.set(facecolor=color)

# Add statistical annotations
from scipy import stats

# Calculate statistics
g1_mean = np.mean(group1)
g2_mean = np.mean(group2)
g1_median = np.median(group1)
g2_median = np.median(group2)
g1_std = np.std(group1)
g2_std = np.std(group2)

# Perform t-test
t_stat, p_value = stats.ttest_ind(group1, group2)

# Annotate means
plt.hlines(y=g1_mean, xmin=0.8, xmax=1.2, colors='red', linestyles='dashed')
plt.hlines(y=g2_mean, xmin=1.8, xmax=2.2, colors='red', linestyles='dashed')

plt.annotate(f'Mean = {g1_mean:.2f}', 
           xy=(1, g1_mean),
           xytext=(0.7, g1_mean + 0.5),
           arrowprops=dict(arrowstyle='->', color='red'),
           bbox=dict(boxstyle="round", fc="white", ec="red"))

plt.annotate(f'Mean = {g2_mean:.2f}', 
           xy=(2, g2_mean),
           xytext=(2.3, g2_mean + 0.5),
           arrowprops=dict(arrowstyle='->', color='red'),
           bbox=dict(boxstyle="round", fc="white", ec="red"))

# Annotate standard deviations
plt.annotate(f'Std Dev = {g1_std:.2f}', 
           xy=(1, g1_mean + g1_std),
           xytext=(0.5, g1_mean + g1_std + 1),
           arrowprops=dict(arrowstyle='->', color='blue'))

plt.annotate(f'Std Dev = {g2_std:.2f}', 
           xy=(2, g2_mean + g2_std),
           xytext=(2.3, g2_mean + g2_std + 1),
           arrowprops=dict(arrowstyle='->', color='blue'))

# Add p-value information
plt.annotate(f'p-value = {p_value:.4f}\nt-statistic = {t_stat:.2f}',
           xy=(1.5, np.max([np.max(group1), np.max(group2)]) + 1),
           ha='center',
           bbox=dict(boxstyle="round", fc="lightyellow", ec="orange"))

# Add comparison arrow
y_arrow = np.max([np.max(group1), np.max(group2)]) + 0.5
plt.annotate('', 
           xy=(1, y_arrow), 
           xytext=(2, y_arrow),
           arrowprops=dict(arrowstyle='<->', color='purple', lw=2))
           
plt.annotate('Difference', 
           xy=(1.5, y_arrow),
           xytext=(1.5, y_arrow + 0.5),
           ha='center',
           color='purple')

plt.title('Statistical Comparison Between Groups', fontsize=16)
plt.ylabel('Value')
plt.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### Correlation Plot with Statistical Annotations

Let's create a correlation plot with annotations showing the correlation coefficient and other statistics:

```python
# Create correlated data
np.random.seed(42)
x = np.random.normal(0, 1, 50)
y = x * 0.7 + np.random.normal(0, 0.5, 50)  # Correlated with x
z = np.random.normal(0, 1, 50)  # Uncorrelated with x

# Calculate statistics
corr_xy = np.corrcoef(x, y)[0, 1]  # Correlation between x and y
corr_xz = np.corrcoef(x, z)[0, 1]  # Correlation between x and z

fig, ax = plt.subplots(1, 2, figsize=(15, 6))

# First plot: strong correlation
ax[0].scatter(x, y, alpha=0.7)
ax[0].set_title('Strong Correlation')

# Add regression line
slope_xy, intercept_xy, r_value_xy, p_value_xy, std_err_xy = stats.linregress(x, y)
x_line = np.linspace(min(x), max(x), 100)
y_line = slope_xy * x_line + intercept_xy
ax[0].plot(x_line, y_line, 'r--')

# Add correlation annotation
ax[0].annotate(f'r = {corr_xy:.2f}\np-value = {p_value_xy:.4f}',
              xy=(np.mean(x), max(y)),
              xytext=(np.mean(x), max(y) + 0.5),
              ha='center',
              bbox=dict(boxstyle="round", fc="lightgreen", ec="green"))

# Add regression equation
ax[0].annotate(f'y = {slope_xy:.2f}x + {intercept_xy:.2f}', 
              xy=(min(x) + 0.5, max(y) - 0.5),
              bbox=dict(boxstyle="square", fc="white", ec="red"))

# Second plot: weak correlation
ax[1].scatter(x, z, alpha=0.7)
ax[1].set_title('Weak Correlation')

# Add regression line
slope_xz, intercept_xz, r_value_xz, p_value_xz, std_err_xz = stats.linregress(x, z)
z_line = slope_xz * x_line + intercept_xz
ax[1].plot(x_line, z_line, 'r--')

# Add correlation annotation
ax[1].annotate(f'r = {corr_xz:.2f}\np-value = {p_value_xz:.4f}',
              xy=(np.mean(x), max(z)),
              xytext=(np.mean(x), max(z) + 0.5),
              ha='center',
              bbox=dict(boxstyle="round", fc="lightcoral", ec="red"))

# Add regression equation
ax[1].annotate(f'y = {slope_xz:.2f}x + {intercept_xz:.2f}', 
              xy=(min(x) + 0.5, max(z) - 0.5),
              bbox=dict(boxstyle="square", fc="white", ec="red"))

# Global title
plt.suptitle('Correlation Analysis with Statistical Annotations', fontsize=16)

for a in ax:
    a.grid(True, alpha=0.3)
    a.set_xlabel('x')
    a.set_ylabel('y')

plt.tight_layout(rect=[0, 0, 1, 0.95])  # Leave space for suptitle
plt.show()
```
