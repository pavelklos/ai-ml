# Basic Plots with Matplotlib

This notebook covers fundamental plotting techniques using matplotlib, one of Python's most popular data visualization libraries. We'll explore how to create and customize line plots, bar charts, and scatter plots - the building blocks of data visualization.

## Import Required Libraries

```python
# Import necessary libraries for visualization
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Enable inline plotting in the notebook
%matplotlib inline

# Set a larger figure size as default for better visibility
plt.rcParams['figure.figsize'] = [10, 6]
# Use a more modern style
plt.style.use('seaborn-v0_8-whitegrid')
```

## Line Plots

Line plots are ideal for showing trends over time or continuous data. They can reveal patterns, seasonality, and relationships between variables.

```python
# Create basic data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create a simple line plot
plt.figure()
plt.plot(x, y1)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
# Multiple line plots with customization
plt.figure()
plt.plot(x, y1, label='sin(x)', color='blue', linestyle='-', linewidth=2)
plt.plot(x, y2, label='cos(x)', color='red', linestyle='--', linewidth=2)
plt.title('Multiple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Line plot with markers
plt.figure()
plt.plot(x[::10], y1[::10], 'bo-', label='sin(x) with markers', markersize=8)
plt.plot(x[::10], y2[::10], 'rs-', label='cos(x) with markers', markersize=8)
plt.title('Line Plot with Markers')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```

```python
# Real-world example: Monthly temperature data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temp_2022 = [5, 7, 10, 15, 20, 25, 28, 27, 22, 15, 10, 6]
temp_2023 = [4, 6, 11, 16, 22, 27, 30, 29, 23, 17, 11, 7]

plt.figure()
plt.plot(months, temp_2022, 'o-', label='2022')
plt.plot(months, temp_2023, 's-', label='2023')
plt.title('Monthly Average Temperature')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True)
plt.show()
```

## Bar Charts

Bar charts are perfect for comparing categorical data or displaying discrete quantities. They can be vertical (bar charts) or horizontal (barh charts).

```python
# Simple bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [22, 35, 14, 28, 19]

plt.figure()
plt.bar(categories, values)
plt.title('Simple Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

```python
# Customized bar chart
plt.figure()
plt.bar(categories, values, color='skyblue', edgecolor='black', linewidth=1.5, alpha=0.7)
plt.title('Customized Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')

# Adding value labels above each bar
for i, v in enumerate(values):
    plt.text(i, v + 0.5, str(v), ha='center')
    
plt.show()
```

```python
# Horizontal bar chart
plt.figure()
plt.barh(categories, values, color='lightgreen', edgecolor='darkgreen', height=0.5)
plt.title('Horizontal Bar Chart')
plt.xlabel('Values')
plt.ylabel('Categories')

# Adding value labels to the right of each bar
for i, v in enumerate(values):
    plt.text(v + 0.5, i, str(v), va='center')
    
plt.show()
```

```python
# Grouped bar chart
x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, values, width, label='Group 1', color='lightblue')
rects2 = ax.bar(x + width/2, [25, 32, 20, 35, 27], width, label='Group 2', color='lightgreen')

ax.set_title('Grouped Bar Chart')
ax.set_xlabel('Categories')
ax.set_ylabel('Values')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

plt.show()
```

```python
# Stacked bar chart
values1 = [10, 20, 15, 12, 8]
values2 = [12, 15, 8, 16, 11]

plt.figure()
plt.bar(categories, values1, color='skyblue', label='Series 1')
plt.bar(categories, values2, bottom=values1, color='lightgreen', label='Series 2')
plt.title('Stacked Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.legend()
plt.show()
```

```python
# Real-world example: Sales comparison by quarter
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
products = {
    'Product A': [120, 150, 135, 180],
    'Product B': [90, 115, 130, 105],
    'Product C': [60, 75, 80, 95]
}

x = np.arange(len(quarters))
width = 0.25
multiplier = 0

fig, ax = plt.subplots()

for product, sales in products.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, sales, width, label=product)
    ax.bar_label(rects, padding=3)
    multiplier += 1

ax.set_title('Quarterly Sales by Product')
ax.set_xlabel('Quarter')
ax.set_ylabel('Sales (thousands $)')
ax.set_xticks(x + width, quarters)
ax.legend(loc='upper left')
ax.set_ylim(0, 200)

plt.show()
```

## Scatter Plots

Scatter plots are excellent for showing the relationship between two continuous variables. They can reveal correlations, clusters, and outliers in your data.

```python
# Basic scatter plot
x = np.random.rand(50)
y = np.random.rand(50)

plt.figure()
plt.scatter(x, y)
plt.title('Basic Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
# Customized scatter plot with varying sizes and colors
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

plt.figure()
plt.scatter(x, y, c=colors, s=sizes, alpha=0.7, cmap='viridis')
plt.colorbar(label='Color Value')
plt.title('Scatter Plot with Varying Size and Color')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
# Scatter plot showing correlation
np.random.seed(42)
x = np.random.normal(0, 1, 100)
y = x + np.random.normal(0, 0.5, 100)  # y is correlated with x plus some noise

plt.figure()
plt.scatter(x, y, alpha=0.7)
plt.title('Scatter Plot with Correlation')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

# Calculate and display correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]
plt.annotate(f'Correlation: {correlation:.2f}', xy=(0.05, 0.95), xycoords='axes fraction')

plt.show()
```

```python
# Real-world example: Height vs Weight with gender distinction
np.random.seed(42)
men_heights = np.random.normal(175, 7, 50)  # in cm
men_weights = men_heights * 0.7 + np.random.normal(0, 5, 50)  # in kg

women_heights = np.random.normal(162, 5, 50)  # in cm
women_weights = women_heights * 0.65 + np.random.normal(0, 4, 50)  # in kg

plt.figure()
plt.scatter(men_heights, men_weights, color='blue', alpha=0.7, label='Men')
plt.scatter(women_heights, women_weights, color='red', alpha=0.7, label='Women')
plt.title('Height vs Weight by Gender')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.legend()
plt.grid(True)
plt.show()
```

## Combining Different Plot Types

Combining different plot types can help present multiple aspects of your data in a single visualization.

```python
# Scatter plot with a trend line
np.random.seed(42)
x = np.linspace(0, 10, 50)
y = 2 * x + 1 + np.random.normal(0, 2, 50)  # Linear relationship with noise

# Calculate the linear regression
slope, intercept = np.polyfit(x, y, 1)
trend_y = slope * x + intercept

plt.figure()
plt.scatter(x, y, color='blue', alpha=0.7, label='Data points')
plt.plot(x, trend_y, color='red', linestyle='-', linewidth=2, label=f'Trend line: y = {slope:.2f}x + {intercept:.2f}')
plt.title('Scatter Plot with Trend Line')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Bar chart with a line plot overlay
categories = ['A', 'B', 'C', 'D', 'E']
values1 = [22, 35, 14, 28, 19]  # For bars
values2 = [24, 32, 18, 30, 25]  # For line

fig, ax1 = plt.subplots()

# Bar chart
ax1.bar(categories, values1, color='skyblue', alpha=0.7, label='Actual')
ax1.set_xlabel('Categories')
ax1.set_ylabel('Actual Values', color='skyblue')
ax1.tick_params(axis='y', labelcolor='skyblue')

# Add a second y-axis for the line plot
ax2 = ax1.twinx()
ax2.plot(categories, values2, 'ro-', linewidth=2, markersize=8, label='Target')
ax2.set_ylabel('Target Values', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Add a title and legends
plt.title('Bar Chart with Line Plot Overlay')
fig.tight_layout()

# Add legends for both axes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.show()
```

## Customizing Plot Appearance

Matplotlib offers extensive customization options to make your visualizations more effective and attractive.

```python
# Customizing plot appearance
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(12, 7))

# Custom plot with extensive styling
plt.plot(x, y, 'b-', linewidth=2.5)
plt.title('Customized Plot Appearance', fontsize=18, fontweight='bold', pad=15)
plt.xlabel('X-axis', fontsize=14, labelpad=10)
plt.ylabel('Y-axis', fontsize=14, labelpad=10)

# Customize ticks
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# Add grid with custom appearance
plt.grid(True, linestyle='--', alpha=0.7)

# Add a text annotation
plt.text(5, 0.5, 'Important Point', fontsize=14, 
         bbox=dict(facecolor='yellow', alpha=0.3))

# Add custom background color to the plot area
plt.gca().set_facecolor('#f8f9fa')

# Customize spines (the box around the plot)
for spine in plt.gca().spines.values():
    spine.set_linewidth(1.5)
    
plt.show()
```

```python
# Using different matplotlib style sheets
style_options = ['default', 'seaborn-v0_8-darkgrid', 'ggplot', 'bmh', 'dark_background']

fig, axes = plt.subplots(len(style_options), 1, figsize=(10, 15))
fig.tight_layout(pad=3.0)

for ax, style in zip(axes, style_options):
    with plt.style.context(style):
        ax.plot(x, np.sin(x), 'b-')
        ax.set_title(f'Style: {style}', fontsize=12)
        ax.set_xlabel('X-axis')
        ax.set_ylabel('Y-axis')
        ax.grid(True)
        
plt.show()
```

## Saving Plots

Matplotlib allows you to save your plots in various formats for use in reports, presentations, or websites.

```python
# Create a plot to save
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-')
plt.title('Plot to Save')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

# Save in different formats
plt.savefig('sine_plot.png', dpi=100)  # PNG format
plt.savefig('sine_plot.jpg', dpi=200)  # JPG format
plt.savefig('sine_plot.pdf')  # PDF format (vector)
plt.savefig('sine_plot.svg')  # SVG format (vector)

print("Plots saved in multiple formats.")
plt.show()
```

```python
# Demonstrate different DPI settings
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, dpi in zip(axes, [72, 150, 300]):
    ax.plot(x, y, 'b-')
    ax.set_title(f'DPI: {dpi}')
    ax.set_xlabel('X-axis')
    ax.set_ylabel('Y-axis')
    ax.grid(True)

plt.tight_layout()

# Save with different DPI settings
plt.savefig('sine_plot_72dpi.png', dpi=72)
plt.savefig('sine_plot_150dpi.png', dpi=150)
plt.savefig('sine_plot_300dpi.png', dpi=300)

print("Plots saved with different DPI settings.")
plt.show()
```

## Multiple Subplots

Creating multiple subplots allows you to compare different visualizations side by side.

```python
# Creating basic subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.tight_layout(pad=3.0)

# Upper left: Line plot
axes[0, 0].plot(x, np.sin(x), 'b-')
axes[0, 0].set_title('Line Plot')
axes[0, 0].set_xlabel('X-axis')
axes[0, 0].set_ylabel('Y-axis')
axes[0, 0].grid(True)

# Upper right: Bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [22, 35, 14, 28, 19]
axes[0, 1].bar(categories, values, color='skyblue')
axes[0, 1].set_title('Bar Chart')
axes[0, 1].set_xlabel('Categories')
axes[0, 1].set_ylabel('Values')

# Lower left: Scatter plot
np.random.seed(42)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
axes[1, 0].scatter(x_scatter, y_scatter, alpha=0.7)
axes[1, 0].set_title('Scatter Plot')
axes[1, 0].set_xlabel('X-axis')
axes[1, 0].set_ylabel('Y-axis')
axes[1, 0].grid(True)

# Lower right: Histogram
np.random.seed(42)
data = np.random.normal(0, 1, 1000)
axes[1, 1].hist(data, bins=30, color='lightgreen', edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Histogram')
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Frequency')

plt.show()
```

```python
# Creating subplots with shared axes
x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
fig.tight_layout(pad=3.0)

# First subplot
axes[0].plot(x, np.sin(x), 'b-', label='sin(x)')
axes[0].set_title('Sine Function')
axes[0].set_ylabel('sin(x)')
axes[0].legend()
axes[0].grid(True)

# Second subplot
axes[1].plot(x, np.cos(x), 'r-', label='cos(x)')
axes[1].set_title('Cosine Function')
axes[1].set_ylabel('cos(x)')
axes[1].legend()
axes[1].grid(True)

# Third subplot
axes[2].plot(x, np.sin(x) * np.cos(x), 'g-', label='sin(x)×cos(x)')
axes[2].set_title('Sine × Cosine')
axes[2].set_xlabel('X-axis')
axes[2].set_ylabel('sin(x)×cos(x)')
axes[2].legend()
axes[2].grid(True)

plt.show()
```

```python
# Create a subplot with grid specification
plt.figure(figsize=(12, 8))

# Create a 2x3 grid
plt.subplot(2, 3, 1)  # row 1, col 1
plt.plot(x, np.sin(x))
plt.title('Plot 1: sin(x)')
plt.grid(True)

plt.subplot(2, 3, 2)  # row 1, col 2
plt.plot(x, np.cos(x), 'r-')
plt.title('Plot 2: cos(x)')
plt.grid(True)

plt.subplot(2, 3, 3)  # row 1, col 3
plt.plot(x, np.tan(x), 'g-')
plt.title('Plot 3: tan(x)')
plt.grid(True)
plt.ylim(-5, 5)  # Limit y range for tan

plt.subplot(2, 3, 4)  # row 2, col 1
plt.plot(x, x**2, 'm-')
plt.title('Plot 4: x²')
plt.grid(True)

plt.subplot(2, 3, 5)  # row 2, col 2
plt.plot(x, np.exp(x), 'c-')
plt.title('Plot 5: eˣ')
plt.grid(True)

plt.subplot(2, 3, 6)  # row 2, col 3
plt.plot(x, np.log(x + 1), 'y-')
plt.title('Plot 6: ln(x+1)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Complex layout with GridSpec
from matplotlib.gridspec import GridSpec

fig = plt.figure(figsize=(12, 8))
gs = GridSpec(3, 3, figure=fig)

# Create a large plot that spans the first two rows
ax1 = fig.add_subplot(gs[0:2, :])
ax1.plot(x, np.sin(x), 'b-')
ax1.set_title('Main Plot: sin(x)')
ax1.grid(True)

# Create smaller plots in the third row
ax2 = fig.add_subplot(gs[2, 0])
ax2.bar(['A', 'B', 'C', 'D'], [10, 15, 7, 12], color='skyblue')
ax2.set_title('Bar Chart')

ax3 = fig.add_subplot(gs[2, 1])
ax3.scatter(np.random.rand(20), np.random.rand(20), color='green')
ax3.set_title('Scatter Plot')

ax4 = fig.add_subplot(gs[2, 2])
ax4.hist(np.random.normal(0, 1, 100), bins=15, color='purple', alpha=0.7)
ax4.set_title('Histogram')

plt.tight_layout()
plt.show()
```
