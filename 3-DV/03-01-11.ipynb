## 3. Saving Matplotlib Plots to Files

The primary way to save matplotlib visualizations is using the `plt.savefig()` function. Let's explore its basic usage and various parameters.

```python
# Basic saving of a figure
plt.figure(figsize=(10, 6))
plt.plot(x, y, '-o', markersize=4, linewidth=2, color='royalblue', alpha=0.7)
plt.title('Line Plot Saved with savefig()', fontsize=16)
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)

# Save using savefig
save_path = os.path.join(save_dir, 'basic_line_plot.png')
plt.savefig(save_path)
plt.close()  # Close the figure to free memory

print(f"Plot saved to {save_path}")
```

### 3.1 Understanding savefig() Parameters

The `savefig()` function has many parameters to control how your visualization is saved:

- **filename**: Path where the file should be saved
- **dpi**: Resolution in dots per inch
- **format**: File format (e.g., 'png', 'pdf', 'svg', 'jpg')
- **bbox_inches**: Portion of the figure to save ('tight' removes extra whitespace)
- **transparent**: Whether to save with a transparent background
- **facecolor**: Background color of the figure
- **edgecolor**: Edge color of the figure
- **pad_inches**: Padding around the figure

```python
# Save with specific parameters
plt.figure(figsize=(10, 6))
plt.plot(x, y, '-o', markersize=4, linewidth=2, color='royalblue', alpha=0.7)
plt.title('Line Plot with Custom Save Parameters', fontsize=16)
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)

# Save with custom parameters
save_path = os.path.join(save_dir, 'custom_params_line_plot.png')
plt.savefig(save_path, 
           dpi=300,  # Higher resolution
           bbox_inches='tight',  # Remove extra whitespace
           pad_inches=0.5,  # Add some padding
           transparent=False,  # Opaque background
           facecolor='#f0f0f0',  # Light gray background
           edgecolor='black')  # Black edge
plt.close()

print(f"Plot saved to {save_path} with custom parameters")
```

## 4. File Formats for Visualizations

Different file formats have pros and cons. Let's compare the most common formats by saving our line plot in each format.

```python
# Save the same plot in different formats
plt.figure(figsize=(10, 6))
plt.plot(x, y, '-o', markersize=4, linewidth=2, color='royalblue', alpha=0.7)
plt.title('Same Plot in Different Formats', fontsize=16)
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.grid(True, alpha=0.3)

# Save in various formats
formats = {
    'png': 'Raster format, good for web, supports transparency',
    'jpg': 'Raster format, smaller file size, no transparency',
    'svg': 'Vector format, scalable, best for publication',
    'pdf': 'Document format, preserves vector graphics, good for publication',
    'eps': 'Vector format used in scientific publishing',
    'tiff': 'High-quality raster format used in publishing'
}

for fmt, description in formats.items():
    save_path = os.path.join(save_dir, f'format_comparison.{fmt}')
    plt.savefig(save_path, format=fmt, dpi=300, bbox_inches='tight')
    print(f"Saved as {fmt}: {description}")

plt.close()
```

## 5. Controlling Resolution and DPI

The resolution of your saved images is controlled by the `dpi` parameter (dots per inch). Higher DPI means higher resolution but larger file size.

```python
# Compare different DPI settings
plt.figure(figsize=(6, 4))
plt.bar(categories, values, color='tomato', alpha=0.7)
plt.title('DPI Comparison', fontsize=14)
plt.xlabel('Categories', fontsize=10)
plt.ylabel('Values', fontsize=10)

# Save with different DPI values
dpi_values = [72, 150, 300, 600]
for dpi in dpi_values:
    save_path = os.path.join(save_dir, f'bar_chart_dpi_{dpi}.png')
    plt.savefig(save_path, dpi=dpi, bbox_inches='tight')
    
    # Get file size in KB
    file_size = os.path.getsize(save_path) / 1024
    print(f"DPI: {dpi}, File Size: {file_size:.1f} KB")

plt.close()
```

### 5.1 When to Use Different DPI Settings

- **72 DPI**: Standard for web/screen display
- **150-200 DPI**: Good for presentations and general printing
- **300 DPI**: Professional printing standard
- **600+ DPI**: High-quality printing for scientific publications

Remember that DPI only matters for raster formats (PNG, JPG, TIFF). Vector formats (SVG, PDF, EPS) are resolution-independent.

### 7.2 Converting Plotly to Matplotlib for Saving

Sometimes, you might prefer to convert a Plotly figure to a static matplotlib figure before saving:

```python
# Define a function to convert Plotly figures to Matplotlib
def plotly_to_matplotlib(fig):
    """Convert a Plotly figure to a Matplotlib figure"""
    # Create a new Matplotlib figure
    fig_mpl, ax = plt.subplots(figsize=(10, 6))
    
    # Extract data from the first trace
    for trace in fig.data:
        if hasattr(trace, 'x') and hasattr(trace, 'y'):
            x_values = trace.x
            y_values = trace.y
            
            # Get trace properties
            mode = trace.mode if hasattr(trace, 'mode') else 'lines'
            
            # Plot data based on mode
            if 'markers' in mode and 'lines' in mode:
                ax.plot(x_values, y_values, 'o-')
            elif 'markers' in mode:
                ax.scatter(x_values, y_values)
            else:
                ax.plot(x_values, y_values)
    
    # Set title and labels
    if fig.layout.title:
        if hasattr(fig.layout.title, 'text'):
            ax.set_title(fig.layout.title.text)
        else:
            ax.set_title(fig.layout.title)
    
    if fig.layout.xaxis and fig.layout.xaxis.title:
        if hasattr(fig.layout.xaxis.title, 'text'):
            ax.set_xlabel(fig.layout.xaxis.title.text)
    
    if fig.layout.yaxis and fig.layout.yaxis.title:
        if hasattr(fig.layout.yaxis.title, 'text'):
            ax.set_ylabel(fig.layout.yaxis.title.text)
    
    plt.tight_layout()
    return fig_mpl

# Convert Plotly figure to Matplotlib
mpl_fig = plotly_to_matplotlib(fig)

# Save the converted figure
save_path = os.path.join(save_dir, 'plotly_converted_to_matplotlib.png')
mpl_fig.savefig(save_path, dpi=300, bbox_inches='tight')
print(f"Saved Plotly figure converted to Matplotlib: {save_path}")
plt.close()
```

## 8. Batch Saving Multiple Plots

When working with multiple visualizations, it's efficient to automate the saving process.

```python
# Create a function to generate multiple plots and save them
def create_and_save_multiple_plots(save_directory, n_plots=4, base_filename='plot', 
                                  format='png', dpi=300):
    """Create and save multiple plots at once"""
    # Ensure directory exists
    os.makedirs(save_directory, exist_ok=True)
    
    saved_files = []
    
    for i in range(n_plots):
        # Create a simple plot with different data
        plt.figure(figsize=(8, 5))
        x_data = np.linspace(0, 10, 100)
        y_data = np.sin(x_data + i) * np.exp(-0.1 * i * x_data)
        
        plt.plot(x_data, y_data, label=f'Series {i+1}')
        plt.title(f'Plot {i+1}: Damped Sine Wave', fontsize=14)
        plt.xlabel('X Values', fontsize=12)
        plt.ylabel('Y Values', fontsize=12)
        plt.grid(True, alpha=0.3)
        plt.legend()
        
        # Generate filename with index padding (e.g., plot_001.png)
        filename = f"{base_filename}_{i+1:03d}.{format}"
        filepath = os.path.join(save_directory, filename)
        
        # Save the figure
        plt.savefig(filepath, format=format, dpi=dpi, bbox_inches='tight')
        plt.close()
        
        saved_files.append(filepath)
        
    return saved_files

# Create a subdirectory for batch plots
batch_dir = os.path.join(save_dir, 'batch_plots')
saved_files = create_and_save_multiple_plots(batch_dir, n_plots=4)

print(f"Created and saved {len(saved_files)} plots:")
for file in saved_files:
    print(f" - {file}")
```

### 8.1 Creating a Figure Manager for Multiple Plots

```python
# Create a class to manage multiple figures and save them
class FigureManager:
    """A class to manage and save multiple matplotlib figures"""
    
    def __init__(self, save_directory):
        """Initialize the figure manager with a save directory"""
        self.save_directory = save_directory
        os.makedirs(save_directory, exist_ok=True)
        self.figures = {}
    
    def add_figure(self, name, fig=None):
        """Add a figure to the manager, either by providing one or creating new"""
        if fig is None:
            fig = plt.figure(figsize=(8, 6))
        
        self.figures[name] = fig
        return fig
    
    def get_figure(self, name):
        """Get a figure by name"""
        if name in self.figures:
            return self.figures[name]
        else:
            return self.add_figure(name)
    
    def save_all(self, format='png', dpi=300, **kwargs):
        """Save all figures to the save directory"""
        saved_files = {}
        
        for name, fig in self.figures.items():
            # Generate safe filename
            safe_name = name.replace(' ', '_').replace('.', '_').lower()
            filename = f"{safe_name}.{format}"
            filepath = os.path.join(self.save_directory, filename)
            
            # Save the figure
            fig.savefig(filepath, format=format, dpi=dpi, **kwargs)
            saved_files[name] = filepath
        
        return saved_files
    
    def close_all(self):
        """Close all figures to free memory"""
        for fig in self.figures.values():
            plt.close(fig)
        self.figures = {}

# Demo the FigureManager
manager = FigureManager(os.path.join(save_dir, 'managed_plots'))

# Add some figures
fig1 = manager.add_figure("Line Plot")
plt.plot(x, np.sin(x), '-o')
plt.title("Simple Line Plot")

fig2 = manager.add_figure("Bar Chart")
plt.bar(categories, values)
plt.title("Simple Bar Chart")

fig3 = manager.add_figure("Scatter Plot")
plt.scatter(df['x'][:50], df['y'][:50], c=np.arange(50), cmap='viridis')
plt.title("Simple Scatter Plot")

# Save all figures
saved_files = manager.save_all(format='png', dpi=200, bbox_inches='tight')

print(f"Saved {len(saved_files)} figures:")
for name, filepath in saved_files.items():
    print(f" - {name}: {filepath}")

# Close all figures to free memory
manager.close_all()
```

## 9. Automating Visualization Exports

Let's create a more advanced system for automatically generating and saving visualizations as part of a data analysis pipeline.

```python
class AutomatedVisualizationExporter:
    """Class for automating the creation and export of visualizations"""
    
    def __init__(self, base_directory, dataset=None):
        """
        Initialize the exporter with a base directory and optional dataset
        
        Parameters:
        -----------
        base_directory : str
            Directory where visualizations will be saved
        dataset : pandas.DataFrame, optional
            Dataset to use for visualizations
        """
        self.base_directory = base_directory
        self.dataset = dataset
        self.visualization_count = {}
        
        # Create the directory structure
        os.makedirs(base_directory, exist_ok=True)
        
        # Create subdirectories for different visualization types
        self.dirs = {
            'matplotlib': os.path.join(base_directory, 'matplotlib'),
            'seaborn': os.path.join(base_directory, 'seaborn'),
            'plotly': os.path.join(base_directory, 'plotly')
        }
        
        for directory in self.dirs.values():
            os.makedirs(directory, exist_ok=True)
    
    def _get_filename(self, viz_type, name, format='png'):
        """Generate a unique filename for a visualization"""
        # Track the count for each type to ensure unique names
        if viz_type not in self.visualization_count:
            self.visualization_count[viz_type] = 0
        
        self.visualization_count[viz_type] += 1
        count = self.visualization_count[viz_type]
        
        # Create a safe filename
        safe_name = name.replace(' ', '_').replace('.', '_').lower()
        filename = f"{viz_type}_{count:03d}_{safe_name}.{format}"
        
        return os.path.join(self.dirs[viz_type], filename)
    
    def save_matplotlib(self, fig, name, format='png', dpi=300, **kwargs):
        """Save a matplotlib figure"""
        filepath = self._get_filename('matplotlib', name, format)
        fig.savefig(filepath, format=format, dpi=dpi, **kwargs)
        plt.close(fig)
        return filepath
    
    def save_seaborn(self, fig, name, format='png', dpi=300, **kwargs):
        """Save a seaborn figure (which is just a matplotlib figure)"""
        filepath = self._get_filename('seaborn', name, format)
        fig.savefig(filepath, format=format, dpi=dpi, **kwargs)
        plt.close(fig)
        return filepath
    
    def save_plotly(self, fig, name, format='png', scale=2):
        """Save a plotly figure as both static image and HTML"""
        # Save as static image
        img_path = self._get_filename('plotly', name, format)
        try:
            fig.write_image(img_path, scale=scale)
        except Exception as e:
            print(f"Could not save plotly image: {e}")
        
        # Save as interactive HTML
        html_path = self._get_filename('plotly', name, 'html')
        fig.write_html(html_path, include_plotlyjs='cdn')
        
        return {'image': img_path, 'html': html_path}
    
    def generate_standard_plots(self, df=None):
        """Generate and save a set of standard plots for a dataset"""
        if df is None:
            if self.dataset is None:
                raise ValueError("No dataset provided for visualization")
            df = self.dataset
        
        results = {}
        
        # Generate a histogram for each numeric column
        numeric_cols = df.select_dtypes(include=['number']).columns
        for col in numeric_cols:
            # Matplotlib histogram
            fig, ax = plt.subplots(figsize=(8, 5))
            ax.hist(df[col].dropna(), bins=20, alpha=0.7)
            ax.set_title(f'Distribution of {col}')
            ax.set_xlabel(col)
            ax.set_ylabel('Frequency')
            results[f'hist_{col}'] = self.save_matplotlib(fig, f'histogram_{col}')
            
            # Seaborn distribution plot 
            fig, ax = plt.subplots(figsize=(8, 5))
            sns.histplot(df[col].dropna(), kde=True, ax=ax)
            ax.set_title(f'Distribution of {col} (Seaborn)')
            results[f'sns_dist_{col}'] = self.save_seaborn(fig, f'seaborn_dist_{col}')
            
            # Plotly histogram
            fig = px.histogram(df, x=col, marginal='box', title=f'Distribution of {col} (Plotly)')
            results[f'plotly_hist_{col}'] = self.save_plotly(fig, f'plotly_hist_{col}')
        
        # Generate correlation heatmap if multiple numeric columns
        if len(numeric_cols) > 1:
            # Seaborn correlation heatmap
            corr = df[numeric_cols].corr()
            fig, ax = plt.subplots(figsize=(10, 8))
            sns.heatmap(corr, annot=True, cmap='coolwarm', ax=ax)
            ax.set_title('Correlation Matrix')
            results['correlation_heatmap'] = self.save_seaborn(fig, 'correlation_heatmap')
            
            # Plotly correlation heatmap
            fig = px.imshow(corr, text_auto=True, aspect='auto', title='Correlation Matrix (Plotly)')
            results['plotly_correlation'] = self.save_plotly(fig, 'plotly_correlation')
            
        return results

# Create a test dataset
np.random.seed(42)
test_df = pd.DataFrame({
    'age': np.random.normal(35, 10, 200).astype(int),
    'income': np.random.lognormal(10, 1, 200),
    'spending': np.random.lognormal(8, 0.7, 200),
    'satisfaction': np.random.randint(1, 11, 200),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 200)
})

# Initialize the exporter and generate standard plots
exporter = AutomatedVisualizationExporter(
    os.path.join(save_dir, 'automated_exports'),
    dataset=test_df
)

# Generate standard plots
results = exporter.generate_standard_plots()

print(f"Generated {len(results)} standard plots:")
for name, path in results.items():
    if isinstance(path, dict):
        print(f" - {name}: {path['image']} and {path['html']}")
    else:
        print(f" - {name}: {path}")
```

### 9.1 Custom Visualization Pipeline

Now, let's create a complete visualization pipeline that can be integrated into a data processing workflow:

```python
class VisualizationPipeline:
    """A complete pipeline for generating, saving, and managing visualizations"""
    
    def __init__(self, output_directory, report_name="Visualization_Report"):
        """Initialize the pipeline with output directory"""
        self.output_directory = output_directory
        self.report_name = report_name
        self.visualizations = []
        
        # Create output directory structure
        self.dirs = {
            'images': os.path.join(output_directory, 'images'),
            'interactive': os.path.join(output_directory, 'interactive'),
            'reports': os.path.join(output_directory, 'reports')
        }
        
        for directory in self.dirs.values():
            os.makedirs(directory, exist_ok=True)
    
    def process_dataset(self, df, dataset_name="dataset"):
        """Process a dataset and generate standard visualizations"""
        print(f"Processing dataset: {dataset_name}")
        
        # Create subdirectory for this dataset
        dataset_dir = os.path.join(self.output_directory, f"{dataset_name}")
        os.makedirs(dataset_dir, exist_ok=True)
        
        # Generate visualizations based on data types
        numeric_columns = df.select_dtypes(include=['number']).columns
        categorical_columns = df.select_dtypes(include=['object', 'category']).columns
        
        # Generate descriptive statistics summary
        stats_file = os.path.join(dataset_dir, "statistics_summary.csv")
        df.describe().to_csv(stats_file)
        print(f"Saved statistics to: {stats_file}")
        
        # Generate numeric visualizations
        for col in numeric_columns:
            self._create_numeric_visualizations(df, col, dataset_dir)
            
        # Generate categorical visualizations
        for col in categorical_columns:
            self._create_categorical_visualizations(df, col, dataset_dir)
        
        # Generate relationships (if multiple columns)
        if len(numeric_columns) >= 2:
            self._create_relationship_visualizations(df, numeric_columns, dataset_dir)
        
        # Generate HTML summary report
        self._generate_html_report(df, dataset_name, dataset_dir)
        
        return dataset_dir
    
    def _create_numeric_visualizations(self, df, column, save_dir):
        """Create visualizations for numeric columns"""
        # Histogram
        plt.figure(figsize=(8, 5))
        plt.hist(df[column].dropna(), bins=20, alpha=0.7, color='steelblue')
        plt.title(f'Distribution of {column}')
        plt.grid(alpha=0.3)
        plt.savefig(os.path.join(save_dir, f"{column}_histogram.png"), dpi=300, bbox_inches='tight')
        plt.close()
        
        # Boxplot
        plt.figure(figsize=(8, 5))
        plt.boxplot(df[column].dropna())
        plt.title(f'Boxplot of {column}')
        plt.grid(alpha=0.3)
        plt.savefig(os.path.join(save_dir, f"{column}_boxplot.png"), dpi=300, bbox_inches='tight')
        plt.close()
    
    def _create_categorical_visualizations(self, df, column, save_dir):
        """Create visualizations for categorical columns"""
        # Bar chart of value counts
        plt.figure(figsize=(10, 6))
        df[column].value_counts().plot(kind='bar')
        plt.title(f'Counts of {column}')
        plt.xticks(rotation=45)
        plt.grid(axis='y', alpha=0.3)
        plt.tight_layout()
        plt.savefig(os.path.join(save_dir, f"{column}_counts.png"), dpi=300, bbox_inches='tight')
        plt.close()
    
    def _create_relationship_visualizations(self, df, columns, save_dir):
        """Create visualizations showing relationships between variables"""
        # Correlation matrix
        if len(columns) > 1:
            plt.figure(figsize=(10, 8))
            corr = df[columns].corr()
            sns.heatmap(corr, annot=True, cmap='coolwarm')
            plt.title('Correlation Matrix')
            plt.tight_layout()
            plt.savefig(os.path.join(save_dir, "correlation_matrix.png"), dpi=300, bbox_inches='tight')
            plt.close()
        
        # Scatter plot matrix (limit to first 5 columns to avoid too many plots)
        if len(columns) > 1:
            plot_cols = columns[:min(5, len(columns))]
            fig = sns.pairplot(df[plot_cols], corner=True)
            fig.savefig(os.path.join(save_dir, "scatter_matrix.png"), dpi=300, bbox_inches='tight')
            plt.close()
    
    def _generate_html_report(self, df, dataset_name, dataset_dir):
        """Generate an HTML report summarizing the dataset and visualizations"""
        html_content = f"""
        <html>
        <head>
            <title>Visualization Report: {dataset_name}</title>
            <style>
                body {{ font-family: Arial, sans-serif; margin: 20px; }}
                h1, h2 {{ color: #2c3e50; }}
                .stats {{ border-collapse: collapse; width: 100%; }}
                .stats th, .stats td {{ border: 1px solid #ddd; padding: 8px; text-align: left; }}
                .stats tr:nth-child(even) {{ background-color: #f2f2f2; }}
                .img-container {{ margin: 20px 0; }}
                img {{ max-width: 100%; border: 1px solid #ddd; }}
            </style>
        </head>
        <body>
            <h1>Visualization Report: {dataset_name}</h1>
            <p>Generated on {pd.Timestamp.now().strftime('%Y-%m-%d %H:%M:%S')}</p>
            
            <h2>Dataset Overview</h2>
            <p>Rows: {df.shape[0]}, Columns: {df.shape[1]}</p>
            
            <h2>Data Types</h2>
            <table class="stats">
                <tr><th>Column</th><th>Type</th></tr>
        """
        
        # Add data types
        for col, dtype in df.dtypes.items():
            html_content += f"<tr><td>{col}</td><td>{dtype}</td></tr>\n"
        
        html_content += "</table>\n"
        
        # Add visualizations
        html_content += "<h2>Visualizations</h2>\n"
        
        # Find all PNG files in the directory
        image_files = [f for f in os.listdir(dataset_dir) if f.endswith('.png')]
        
        for image_file in image_files:
            img_path = os.path.join(os.path.basename(dataset_dir), image_file)
            html_content += f"""
            <div class="img-container">
                <h3>{image_file.split('.')[0]}</h3>
                <img src="{img_path}" alt="{image_file}">
            </div>
            """
        
        html_content += """
        </body>
        </html>
        """
        
        # Write the HTML file
        html_file = os.path.join(dataset_dir, f"{dataset_name}_report.html")
        with open(html_file, 'w') as f:
            f.write(html_content)
        
        print(f"Generated HTML report: {html_file}")
        
        return html_file

# Example usage of the pipeline
pipeline = VisualizationPipeline(os.path.join(save_dir, 'pipeline_exports'))
pipeline_output = pipeline.process_dataset(test_df, "customer_data")
print(f"Pipeline completed, output saved to: {pipeline_output}")
```
