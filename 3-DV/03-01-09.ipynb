# Visualizing Categorical Data

Categorical data is non-numeric data that can be divided into groups or categories. Effective visualization of categorical data helps to identify patterns, distributions, and relationships between categories. This notebook explores various techniques for visualizing categorical data using different Python libraries like Matplotlib, Seaborn, and Plotly.

## 1. Import Required Libraries
We'll start by importing the necessary libraries for data manipulation and visualization:
- Pandas for data manipulation
- NumPy for numerical operations
- Matplotlib for basic plotting
- Seaborn for statistical visualizations
- Plotly for interactive visualizations

```python
# Import basic libraries
import pandas as pd
import numpy as np

# Import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Import plotly for interactive visualizations
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set style for seaborn
sns.set(style="whitegrid")

# Configure matplotlib for better resolution and style
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['figure.dpi'] = 100
```

## 2. Generate Sample Categorical Data
Let's create some sample categorical datasets that we'll use throughout this notebook:
1. Single category data (e.g., product categories)
2. Multi-category data (e.g., product categories by region)
3. Categorical data with associated numerical values (e.g., sales by product category)

```python
# Create sample categorical data - Single category
product_categories = ['Electronics', 'Clothing', 'Home & Garden', 'Books', 'Toys', 'Sports', 'Automotive']
product_counts = [120, 85, 63, 45, 30, 25, 18]

# Create a DataFrame for single category data
df_products = pd.DataFrame({
    'Category': product_categories,
    'Count': product_counts
})

# Create sample categorical data - Multi-category with regions
regions = ['North', 'South', 'East', 'West']
multi_category_data = {
    'Category': [],
    'Region': [],
    'Sales': []
}

# Generate synthetic data
np.random.seed(42)
for category in product_categories:
    for region in regions:
        multi_category_data['Category'].extend([category] * 3)  # 3 data points per category-region
        multi_category_data['Region'].extend([region] * 3)
        multi_category_data['Sales'].extend(np.random.randint(500, 5000, 3))

# Create DataFrame for multi-category data
df_multi = pd.DataFrame(multi_category_data)

# Create sample categorical data with multiple measures
survey_data = {
    'Age Group': ['18-24', '25-34', '35-44', '45-54', '55-64', '65+'],
    'Satisfaction': [3.2, 3.8, 4.1, 3.9, 4.3, 4.6],
    'Usage Frequency': [12, 10, 7, 5, 4, 2],
    'Recommendation Score': [7.5, 8.2, 8.0, 7.8, 8.5, 8.7]
}

df_survey = pd.DataFrame(survey_data)

# Display the first few rows of each DataFrame
print("Product Categories Dataset:")
print(df_products.head())
print("\nMulti-category Dataset:")
print(df_multi.head())
print("\nSurvey Dataset:")
print(df_survey.head())
```

## 3. Bar Charts for Categorical Data
Bar charts are one of the most common and effective ways to visualize categorical data. They work by representing each category as a bar, with the height (or length for horizontal bars) of the bar representing the value associated with that category. Let's create some basic bar charts using Matplotlib.

```python
# Basic vertical bar chart with Matplotlib
plt.figure(figsize=(12, 6))

# Create the bar chart
plt.bar(df_products['Category'], df_products['Count'], color='skyblue', edgecolor='navy')

# Add labels and title
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Count of Products by Category', fontsize=14)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Add data labels on top of each bar
for i, count in enumerate(df_products['Count']):
    plt.text(i, count + 1, str(count), ha='center')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Horizontal bar chart - often better for long category names
plt.figure(figsize=(10, 6))

# Sort the data by count for better visualization
df_sorted = df_products.sort_values('Count')

# Create horizontal bar chart
plt.barh(df_sorted['Category'], df_sorted['Count'], color='lightgreen', edgecolor='darkgreen')

# Add labels and title
plt.xlabel('Count', fontsize=12)
plt.ylabel('Product Category', fontsize=12)
plt.title('Count of Products by Category (Horizontal)', fontsize=14)

# Add data labels inside each bar
for i, count in enumerate(df_sorted['Count']):
    plt.text(count - 8, i, str(count), va='center', color='white', fontweight='bold')

# Adjust layout
plt.tight_layout()

# Show the plot
plt.show()
```

## 4. Count Plots with Seaborn
Seaborn's countplot is specifically designed to show the counts of observations in each categorical bin using bars. It simplifies the process of creating count-based visualizations of categorical data.

```python
# Basic count plot with Seaborn
plt.figure(figsize=(12, 6))

# Create count plot
sns.countplot(x='Region', data=df_multi, palette='viridis')

# Add labels and title
plt.xlabel('Region', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Count of Observations by Region', fontsize=14)

# Display the plot
plt.show()

# Count plot with categories and a hue (multiple variables)
plt.figure(figsize=(14, 7))

# Create a count plot with hue
ax = sns.countplot(x='Category', hue='Region', data=df_multi, palette='Set2')

# Add labels and title
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Count of Products by Category and Region', fontsize=14)

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Adjust legend position
plt.legend(title='Region', loc='upper right')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

## 5. Grouped Bar Charts for Multiple Categories
Grouped bar charts (also called clustered bar charts) are useful for comparing values across multiple categories and sub-categories. Each group of bars represents a main category, with individual bars within the group representing sub-categories.

```python
# Calculate the average sales by category and region
avg_sales = df_multi.groupby(['Category', 'Region'])['Sales'].mean().reset_index()

# Create a grouped bar chart
plt.figure(figsize=(14, 7))

# Get unique categories and regions
categories = avg_sales['Category'].unique()
regions = avg_sales['Region'].unique()

# Set width of bars
bar_width = 0.2
index = np.arange(len(categories))

# Plot bars for each region
for i, region in enumerate(regions):
    region_data = avg_sales[avg_sales['Region'] == region]
    plt.bar(index + i * bar_width, region_data['Sales'], 
           width=bar_width, label=region)

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Average Sales', fontsize=12)
plt.title('Average Sales by Product Category and Region', fontsize=14)
plt.xticks(index + bar_width * (len(regions)-1)/2, categories, rotation=45, ha='right')
plt.legend(title='Region')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add annotations (optional for clarity)
plt.annotate('Higher sales in Electronics category', 
            xy=(0, 3000), 
            xytext=(1, 3500), 
            arrowprops=dict(facecolor='black', shrink=0.05, width=1.5),
            fontsize=10)

plt.tight_layout()
plt.show()
```

## 6. Stacked Bar Charts
Stacked bar charts show the total value across categories, while also displaying how that total is divided into subcategories. They're useful for showing part-to-whole relationships within categories.

```python
# Calculate the total sales by category and region for stacked bar chart
region_sales = df_multi.groupby(['Category', 'Region'])['Sales'].sum().unstack()

# Create a stacked bar chart
plt.figure(figsize=(14, 7))

# Plot the stacked bar chart
region_sales.plot(kind='bar', stacked=True, ax=plt.gca(), colormap='viridis')

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Total Sales by Product Category and Region', fontsize=14)
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()

# Create a 100% stacked bar chart to show proportion
plt.figure(figsize=(14, 7))

# Calculate the percentage for each region within each category
region_sales_pct = region_sales.div(region_sales.sum(axis=1), axis=0) * 100

# Plot the 100% stacked bar chart
region_sales_pct.plot(kind='bar', stacked=True, ax=plt.gca(), colormap='tab10')

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Percentage of Sales', fontsize=12)
plt.title('Percentage of Sales by Product Category and Region', fontsize=14)
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.3)

# Add percentage sign to y-axis labels
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(x)}%'))

plt.tight_layout()
plt.show()
```

## 7. Pie Charts for Categorical Data
Pie charts are useful for showing proportions of a whole, especially when the number of categories is small (typically 5-7 or fewer categories). Each slice of the pie represents a category's proportion of the total.

```python
# Create a basic pie chart
plt.figure(figsize=(10, 8))

# Sort data by values for better visual representation
df_sorted = df_products.sort_values('Count', ascending=False)

# Create the pie chart
plt.pie(df_sorted['Count'], 
        labels=df_sorted['Category'], 
        autopct='%1.1f%%',
        startangle=90, 
        shadow=True, 
        explode=[0.1 if i == 0 else 0 for i in range(len(df_sorted))],  # Explode the largest slice
        colors=plt.cm.Paired(np.linspace(0, 1, len(df_sorted))))

# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  
plt.title('Distribution of Products by Category', fontsize=14)
plt.tight_layout()
plt.show()

# Create a donut chart (pie chart with a hole in the middle)
plt.figure(figsize=(10, 8))

# Create the pie chart with a hole
plt.pie(df_sorted['Count'], 
        labels=df_sorted['Category'], 
        autopct='%1.1f%%',
        startangle=90, 
        wedgeprops=dict(width=0.5),  # This creates the donut hole
        colors=plt.cm.tab10(np.linspace(0, 1, len(df_sorted))))

# Add a circle at the center to create a cleaner donut hole
circle = plt.Circle((0,0), 0.3, fc='white')
plt.gca().add_artist(circle)

plt.axis('equal')
plt.title('Distribution of Products by Category (Donut Chart)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 8. Box Plots for Categorical vs Numerical Data
Box plots (also known as box-and-whisker plots) are a great way to visualize the distribution of numerical data across categories. They show the median, quartiles, and potential outliers for each category.

```python
# Create box plots to compare sales distribution across categories
plt.figure(figsize=(14, 8))

# Create box plot
sns.boxplot(x='Category', y='Sales', data=df_multi, palette='Set3')

# Add jitter points to see the actual data distribution
sns.stripplot(x='Category', y='Sales', data=df_multi, color='black', size=4, alpha=0.3)

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Distribution of Sales by Product Category', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

# Create box plots with region as the hue
plt.figure(figsize=(14, 8))

# Create box plot with hue
sns.boxplot(x='Category', y='Sales', hue='Region', data=df_multi, palette='Set2')

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Distribution of Sales by Product Category and Region', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Region', loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## 9. Violin Plots for Distribution Comparison
Violin plots combine aspects of box plots with kernel density plots to show the distribution shape of data across categories. They're especially useful for comparing distributions between categories.

```python
# Create violin plots to visualize the distribution of sales across categories
plt.figure(figsize=(14, 8))

# Create violin plot
sns.violinplot(x='Category', y='Sales', data=df_multi, palette='Pastel1', inner='quartile')

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Sales Distribution by Product Category (Violin Plot)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()

# Split violin plot by region
plt.figure(figsize=(14, 8))

# Create split violin plot
sns.violinplot(x='Category', y='Sales', hue='Region', data=df_multi, 
               palette='muted', split=True, inner='quartile')

# Customize the plot
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.title('Sales Distribution by Product Category and Region (Split Violin Plot)', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Region')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## 10. Categorical Heatmaps
Heatmaps can be used to show relationships between two categorical variables, with color intensity representing a third variable (typically a count or average value). They're useful for showing patterns in two-way categorical data.

```python
# Create a pivot table for the heatmap
heatmap_data = df_multi.pivot_table(
    values='Sales', 
    index='Category',
    columns='Region', 
    aggfunc='mean'
)

plt.figure(figsize=(12, 8))

# Create heatmap
sns.heatmap(heatmap_data, annot=True, fmt='.0f', cmap='YlGnBu', linewidths=0.5)

# Customize the plot
plt.title('Average Sales by Product Category and Region', fontsize=14)
plt.tight_layout()
plt.show()

# Create a more complex heatmap with normalized data
# Normalize data by row (category) to show relative performance across regions
normalized_heatmap = heatmap_data.div(heatmap_data.max(axis=1), axis=0)

plt.figure(figsize=(12, 8))

# Create normalized heatmap
sns.heatmap(normalized_heatmap, annot=True, fmt='.2f', cmap='RdYlGn', linewidths=0.5, vmin=0, vmax=1)

# Customize the plot
plt.title('Normalized Sales Performance by Product Category and Region', fontsize=14)
plt.tight_layout()
plt.show()
```

## 11. Interactive Categorical Plots with Plotly
Plotly allows us to create interactive visualizations that users can explore by hovering, zooming, and filtering. These are particularly useful for dashboard applications and presentations.

```python
# Create an interactive bar chart with Plotly Express
fig = px.bar(
    df_products, 
    x='Category', 
    y='Count',
    title='Count of Products by Category',
    color='Count',
    color_continuous_scale='Viridis',
    text='Count'
)

# Update layout
fig.update_layout(
    xaxis_title='Product Category',
    yaxis_title='Count',
    height=500,
    width=800
)

# Show the figure
fig.show()

# Create an interactive grouped bar chart for region comparison
avg_sales_pivot = df_multi.groupby(['Category', 'Region'])['Sales'].mean().reset_index()

fig = px.bar(
    avg_sales_pivot,
    x='Category',
    y='Sales',
    color='Region',
    barmode='group',
    title='Average Sales by Product Category and Region',
    text_auto='.0f'
)

# Update layout
fig.update_layout(
    xaxis_title='Product Category',
    yaxis_title='Average Sales',
    legend_title='Region',
    height=600,
    width=900
)

# Show the figure
fig.show()

# Create an interactive stacked bar chart
fig = px.bar(
    df_multi,
    x='Category',
    y='Sales',
    color='Region',
    barmode='stack',
    title='Total Sales by Product Category and Region'
)

# Update layout
fig.update_layout(
    xaxis_title='Product Category',
    yaxis_title='Total Sales',
    legend_title='Region',
    height=600,
    width=900
)

# Show the figure
fig.show()

# Create an interactive tree map
fig = px.treemap(
    df_multi,
    path=['Category', 'Region'],
    values='Sales',
    color='Sales',
    color_continuous_scale='RdBu',
    title='Hierarchical View of Sales by Category and Region'
)

# Update layout
fig.update_layout(height=700, width=900)

# Show the figure
fig.show()

# Create an interactive sunburst chart
fig = px.sunburst(
    df_multi,
    path=['Category', 'Region'],
    values='Sales',
    color='Sales',
    color_continuous_scale='Viridis',
    title='Hierarchical View of Sales by Category and Region (Sunburst Chart)'
)

# Update layout
fig.update_layout(height=700, width=900)

# Show the figure
fig.show()
```

## Summary

In this notebook, we've explored various techniques for visualizing categorical data:

1. **Bar Charts**: Basic vertical and horizontal bar charts for single categorical variables
2. **Count Plots**: Specialized plots for counting occurrences of categorical values
3. **Grouped Bar Charts**: Comparing multiple categories side-by-side
4. **Stacked Bar Charts**: Showing composition within categories
5. **Pie Charts**: Displaying proportional relationships between categories
6. **Box Plots**: Comparing distributions of numerical data across categories
7. **Violin Plots**: Showing full distribution shapes across categories
8. **Heatmaps**: Visualizing relationships between two categorical variables
9. **Interactive Plots**: Creating dynamic, interactive visualizations with Plotly

The choice of visualization depends on:
- Number of categories
- Number of variables
- Type of comparison (absolute values vs. proportions)
- Audience and context

When visualizing categorical data, always consider:
- Visual clarity and simplicity
- Appropriate color schemes (especially for color-blind accessibility)
- Proper labeling and annotations
- Whether to sort categories (alphabetically, by value, etc.)
- The story you're trying to tell with the data
